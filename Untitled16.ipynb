```python
# =====================================================================
#  01_phase0.py  --  PlantSeg Phase 0: dataset audit, in one pass
#  Writes: plantseg_phase0.json, plantseg_urlaudit.json,
#          plantseg_polygon_validation.json, plantseg_dupscan.json
#  Runtime: roughly 12 minutes on a Colab CPU runtime
# =====================================================================
import os, re, csv, json, collections, subprocess
import numpy as np
from PIL import Image, ImageDraw
from scipy.fft import dctn
from google.colab import drive

Image.MAX_IMAGE_PIXELS = None
drive.mount('/content/drive', force_remount=True)

ZIP    = "/content/drive/MyDrive/Colab Notebooks/plantseg.zip"
ROOT   = "/content/my_images"
DS     = f"{ROOT}/plantseg"
SAVE   = "/content/drive/MyDrive"
SPLITS = ["train", "val", "test"]
THRESH, WIDEST = 6, 10          # pHash decision threshold; widest collected

# ---------- 0. extract ------------------------------------------------
if not os.path.isdir(DS) or not os.listdir(DS):
    print("extracting ...")
    r = subprocess.run(["unzip", "-q", ZIP, "-d", ROOT], capture_output=True, text=True)
    print("  unzip exit code:", r.returncode)
    if r.returncode != 0:
        print(r.stdout[-1500:]); print(r.stderr[-1500:]); raise SystemExit
else:
    print("already extracted")
print("size:", subprocess.run(["du", "-sh", ROOT], capture_output=True,
                              text=True).stdout.strip())

# ---------- 1. inventory + metadata -----------------------------------
print("=" * 70); print("1. INVENTORY")
disk = {s: ({os.path.splitext(f)[0] for f in os.listdir(f"{DS}/images/{s}")
             if f.lower().endswith(".jpg")}
            & {os.path.splitext(f)[0] for f in os.listdir(f"{DS}/annotations/{s}")
               if f.lower().endswith(".png")}) for s in SPLITS}
split_of = {st: s for s in SPLITS for st in disk[s]}
N_DISK = len(split_of)
for s in SPLITS: print(f"  {s:5s} paired {len(disk[s]):5d}")
print(f"  TOTAL {N_DISK}")
for i in range(3):
    for j in range(i + 1, 3):
        ov = disk[SPLITS[i]] & disk[SPLITS[j]]
        if ov: print(f"  !! {SPLITS[i]}/{SPLITS[j]} share {len(ov)}: {sorted(ov)[:4]}")

meta = {os.path.splitext(r["Name"])[0]: r for r in
        csv.DictReader(open(f"{DS}/Metadata.csv", encoding="utf-8-sig", newline=""))}
frac = {k: float(v["Mask ratio"]) for k, v in meta.items()}
print(f"  metadata rows {len(meta)}   rows without a file "
      f"{len(set(meta) - set(split_of))}   files without a row "
      f"{len(set(split_of) - set(meta))}")
print(f"  plants {len({r['Plant'] for r in meta.values()})}   "
      f"diseases {len({r['Disease'] for r in meta.values()})}   "
      f"licences {dict(collections.Counter(r['License'] for r in meta.values()))}")
res = np.array([[int(x) for x in r["Resolution"].lower().split("x")] for r in meta.values()])
px = res[:, 0] * res[:, 1]
print(f"  resolution {res[:,0].min()}x{res[:,1].min()} .. {res[:,0].max()}x{res[:,1].max()}"
      f"   median {int(np.median(res[:,0]))}x{int(np.median(res[:,1]))}"
      f"   under 0.1MP {int((px<1e5).sum())}   over 2MP {int((px>2e6).sum())}")

# ---------- 2. COCO annotations ---------------------------------------
print("=" * 70); print("2. COCO ANNOTATIONS")
n_lesion, img_cat, poly_of = {}, {}, {}
for s in SPLITS:
    J = json.load(open(f"{DS}/annotation_{s}.json"))
    ids = {im["id"]: (os.path.splitext(im["file_name"])[0], im["width"], im["height"])
           for im in J["images"]}
    orph = 0
    for a in J["annotations"]:
        if a["image_id"] not in ids:
            orph += 1; continue
        stem = ids[a["image_id"]][0]
        n_lesion[stem] = n_lesion.get(stem, 0) + 1
        img_cat.setdefault(stem, set()).add(int(a["category_id"]))
        poly_of.setdefault(stem, []).append(a["segmentation"])
    for iid, (stem, W, H) in ids.items():
        poly_of.setdefault(stem, []); n_lesion.setdefault(stem, 0)
        meta[stem]["_wh"] = (W, H)
    print(f"  {s:5s} imgs {len(ids):5d}  anns {len(J['annotations']):6d}"
          f"  kept {len(J['annotations'])-orph:6d}  orphan {orph:6d}"
          f"  json-not-on-disk {len({v[0] for v in ids.values()} - disk[s]):4d}")
nl = np.array(sorted(n_lesion.values()))
print(f"  lesions/image min {nl.min()}  median {np.median(nl):.0f}  mean {nl.mean():.2f}"
      f"  max {nl.max()}   exactly 1: {int((nl==1).sum())}   over 10: {int((nl>10).sum())}")

# ---------- 3. full mask scan -----------------------------------------
print("=" * 70); print("3. MASK SCAN")
present = np.zeros(4096, dtype=np.int64)
ndist, true_frac, fg_of, bad = collections.Counter(), {}, {}, []
for s in SPLITS:
    for k, stem in enumerate(sorted(disk[s]), 1):
        a = np.asarray(Image.open(f"{DS}/annotations/{s}/{stem}.png"))
        if a.ndim != 2 or int(a.max()) >= 4096:
            bad.append((stem, str(a.shape))); continue
        h = np.bincount(a.ravel(), minlength=4096)
        nz = np.nonzero(h)[0]
        present[nz] += 1; ndist[len(nz)] += 1
        true_frac[stem] = float(h[1:].sum()) / float(h.sum())
        fg_of[stem] = {int(v) for v in nz if v > 0}
        if k % 2000 == 0: print(f"    {s} {k}/{len(disk[s])}")
vals = np.nonzero(present)[0]
print(f"  values {int(vals.min())}..{int(vals.max())} ({len(vals)} distinct)"
      f"   255 present in {int(present[255])} masks"
      f"   distinct-per-mask {dict(sorted(ndist.items()))}   odd {bad[:3]}")
d = np.array([abs(true_frac[k] - frac[k]) for k in true_frac])
print(f"  mask fraction vs Metadata 'Mask ratio': median {np.median(d):.2e}  max {d.max():.2e}")
off = collections.Counter()
for stem, cs in img_cat.items():
    if stem in fg_of and len(cs) == 1 and len(fg_of[stem]) == 1:
        off[next(iter(fg_of[stem])) - next(iter(cs))] += 1
print(f"  (mask value - category_id): {dict(off.most_common(3))}")

# ---------- 4. lesion area, per-disease -------------------------------
print("=" * 70); print("4. TRAIN LESION AREA + CLASS SIZES")
tr = np.array([frac[s] for s in sorted(disk["train"])])
q = np.percentile(tr, [25, 50, 75]).tolist()
print(f"  n {len(tr)}  mean {tr.mean():.4f}  median {np.median(tr):.4f}"
      f"  min {tr.min():.5f}  max {tr.max():.4f}")
print(f"  QUARTILE CUTS (FREEZE): {q[0]:.6f}  {q[1]:.6f}  {q[2]:.6f}")
for t in (0.01, 0.02, 0.05, 0.10):
    print(f"    under {t*100:4.0f}%: {(tr<t).mean()*100:5.1f}% ({int((tr<t).sum())})")
dis_tr = collections.Counter(meta[s]["Disease"] for s in disk["train"])
cc = np.array(sorted(dis_tr.values()))
print(f"  diseases in train {len(dis_tr)}   images/disease min {cc.min()}"
      f" median {np.median(cc):.0f} max {cc.max()}"
      f"   under 10: {int((cc<10).sum())}   under 20: {int((cc<20).sum())}")
cov = collections.defaultdict(collections.Counter)
for st, sp in split_of.items(): cov[meta[st]["Disease"]][sp] += 1
for sp in SPLITS:
    gone = [k for k, v in cov.items() if v[sp] == 0]
    if gone: print(f"  diseases with no {sp} image: {gone}")

# ---------- 5. polygons vs mask ---------------------------------------
print("=" * 70); print("5. POLYGONS vs MASK")
pv, over, big = {}, [], []
for k, stem in enumerate(sorted(split_of), 1):
    W, H = meta[stem]["_wh"]
    xs, ys = [], []
    canvas = Image.new("L", (W, H), 0); dr = ImageDraw.Draw(canvas)
    for seg in poly_of.get(stem, []):
        for p in seg:
            if len(p) >= 6:
                xs += p[0::2]; ys += p[1::2]
                dr.polygon(list(zip(p[0::2], p[1::2])), fill=1)
    pf = float(np.asarray(canvas, dtype=np.uint8).sum()) / (W * H)
    fit = bool(xs) and max(xs) <= W + 1 and max(ys) <= H + 1 and min(xs) >= -1 and min(ys) >= -1
    diff = pf - frac[stem]
    pv[stem] = dict(split=split_of[stem], n=len(poly_of.get(stem, [])),
                    poly=round(pf, 6), mask=frac[stem], diff=round(diff, 6),
                    fit=fit, usable=bool(fit and abs(diff) <= 0.01 and xs))
    if not fit: over.append(stem)
    elif abs(diff) > 0.01: big.append(stem)
    if k % 2000 == 0: print(f"    {k}/{len(split_of)}")
usable = [k for k, v in pv.items() if v["usable"]]
print(f"  usable {len(usable)}/{len(pv)}   coord overflow {len(over)}   "
      f"large disagreement {len(big)}")
print(f"  no polygons: {[k for k,v in pv.items() if v['n']==0]}")
print(f"  mask covers whole frame: {[k for k,v in frac.items() if v>=0.99999]}")
mism = [(s, meta[s]["Disease"]) for s in sorted(split_of)
        if s in fg_of and len(fg_of[s]) == 1 and s in img_cat and len(img_cat[s]) == 1
        and next(iter(fg_of[s])) - next(iter(img_cat[s])) != 1]
print(f"  mask index inconsistent with category_id: {len(mism)}")

# ---------- 6. source-URL screen --------------------------------------
print("=" * 70); print("6. SOURCE-URL SCREEN")
def nurl(u):
    u = re.sub(r"^www\.", "", re.sub(r"^https?://", "", str(u).strip().lower()))
    return u.rstrip("/") or None
by_url = collections.defaultdict(list)
for st in split_of:
    u = nurl(meta[st]["URL"])
    if u: by_url[u].append(st)
cross_url = {u: v for u, v in by_url.items()
             if len(v) > 1 and len({split_of[s] for s in v}) > 1}
aff = sorted({s for v in cross_url.values() for s in v})
per = collections.Counter(split_of[s] for s in aff)
print(f"  distinct URLs {len(by_url)}   groups spanning splits {len(cross_url)}"
      f"   images affected {len(aff)}")
for sp in SPLITS:
    print(f"    {sp:5s} {per.get(sp,0):4d}/{len(disk[sp]):5d}"
          f" = {100*per.get(sp,0)/len(disk[sp]):5.2f}%")

# ---------- 7. perceptual-hash duplicate scan -------------------------
print("=" * 70); print("7. pHASH DUPLICATE SCAN")
stems = [s for sp in SPLITS for s in sorted(disk[sp])]
N = len(stems)

def ph(a):
    D = dctn(a, norm="ortho")[:8, :8].copy(); D[0, 0] = 0.0
    return (D > np.median(D)).ravel()

H = np.zeros((N, 64), bool); HF = np.zeros((N, 64), bool)
for i, s in enumerate(stems):                 # no im.draft: it breaks cross-size matching
    a = np.asarray(Image.open(f"{DS}/images/{split_of[s]}/{s}.jpg")
                   .convert("L").resize((32, 32), Image.BILINEAR), dtype=np.float32)
    H[i], HF[i] = ph(a), ph(a[:, ::-1])
    if (i + 1) % 2000 == 0: print(f"    hashed {i+1}/{N}")

PK, PKF = np.packbits(H, axis=1), np.packbits(HF, axis=1)
POP = np.unpackbits(np.arange(256, dtype=np.uint8)[:, None], axis=1).sum(1).astype(np.uint8)
pairs = []
for i0 in range(0, N, 512):
    sl = PK[i0:i0 + 512]
    dd = np.minimum(POP[sl[:, None, :] ^ PK[None, :, :]].sum(2),
                    POP[sl[:, None, :] ^ PKF[None, :, :]].sum(2))
    ii, jj = np.nonzero(dd <= WIDEST)
    for a_, b_ in zip(ii, jj):
        g = i0 + int(a_)
        if g < int(b_): pairs.append((g, int(b_), int(dd[a_, b_])))
    print(f"    compared {min(i0+512, N)}/{N}")

def cluster(thr):
    par = list(range(N))
    def find(x):
        while par[x] != x: par[x] = par[par[x]]; x = par[x]
        return x
    for a_, b_, dd in pairs:
        if dd <= thr:
            ra, rb = find(a_), find(b_)
            if ra != rb: par[rb] = ra
    comp = collections.defaultdict(list)
    for i in range(N): comp[find(i)].append(i)
    return comp

print(f"  pairs at Hamming <= {WIDEST}: {len(pairs)}")
print(f"  {'thr':>4} {'clusters':>9} {'uniq':>7} {'test %':>8} {'val %':>8}")
sens = {}
for thr in (2, 4, 6, 8, 10):
    comp = cluster(thr)
    lk = collections.Counter()
    for c in comp.values():
        if len(c) > 1 and len({split_of[stems[i]] for i in c}) > 1:
            for i in c: lk[split_of[stems[i]]] += 1
    sens[thr] = dict(clusters=sum(1 for v in comp.values() if len(v) > 1),
                     unique=len(comp), leak=dict(lk))
    print(f"  {thr:>4} {sens[thr]['clusters']:>9} {len(comp):>7}"
          f" {100*lk.get('test',0)/len(disk['test']):>7.2f}%"
          f" {100*lk.get('val',0)/len(disk['val']):>7.2f}%")

comp = cluster(THRESH)
clus = sorted([v for v in comp.values() if len(v) > 1], key=len, reverse=True)
cross, leak = [], collections.Counter()
for c in clus:
    if len({split_of[stems[i]] for i in c}) > 1:
        cross.append([stems[i] for i in c])
        for i in c: leak[split_of[stems[i]]] += 1
print(f"  THRESH={THRESH}: clusters {len(clus)}  cross-split {len(cross)}"
      f"  unique photographs {len(comp)}/{N}")
for sp in SPLITS:
    print(f"    {sp:5s} {leak.get(sp,0):4d}/{len(disk[sp]):5d}"
          f" = {100*leak.get(sp,0)/len(disk[sp]):5.2f}%")
print("  largest cross-split clusters:")
for c in cross[:6]:
    print("   ", [(split_of[s], s) for s in c][:4])
same = sum(1 for c in cross if len({nurl(meta[s]["URL"]) for s in c}) == 1)
print(f"  cross-split clusters also sharing one source URL: {same}/{len(cross)}")

# ---------- 8. save + verdict -----------------------------------------
for st in meta.values(): st.pop("_wh", None)
json.dump(dict(paired={s: len(disk[s]) for s in SPLITS},
               splits={s: sorted(disk[s]) for s in SPLITS},
               mask_values=[int(v) for v in vals], has_255=bool(present[255] > 0),
               diseased_fraction=frac, n_lesions=n_lesion, quartile_cuts=q),
          open(f"{SAVE}/plantseg_phase0.json", "w"))
json.dump(dict(url_groups={k: v for k, v in by_url.items()},
               cross_split_groups=cross_url), open(f"{SAVE}/plantseg_urlaudit.json", "w"))
json.dump(dict(counts=dict(total=len(pv), usable=len(usable), overflow=len(over),
                           large_disagreement=len(big)),
               per_image=pv), open(f"{SAVE}/plantseg_polygon_validation.json", "w"))
json.dump(dict(threshold=THRESH, sensitivity={str(k): v for k, v in sens.items()},
               clusters=[[stems[i] for i in c] for c in clus],
               cross_split_clusters=cross, counts=dict(leak),
               unique_photographs=len(comp)), open(f"{SAVE}/plantseg_dupscan.json", "w"))

g = 100 * leak.get("test", 0) / len(disk["test"])
print("=" * 70); print("DECISION NUMBERS")
print(f"  images on disk        : {N_DISK}")
print(f"  metadata rows         : {len(meta)}")
print(f"  masks with 255        : {int(present[255])}")
print(f"  max mask value        : {int(vals.max())}")
print(f"  multi-class masks     : {sum(v for k, v in ndist.items() if k > 2)}")
print(f"  usable polygons       : {len(usable)}/{len(pv)}")
print(f"  quartile cuts         : {[round(x,6) for x in q]}")
print(f"  test sharing a URL    : {100*per.get('test',0)/len(disk['test']):.2f}%")
print(f"  TEST NEAR-DUPLICATES  : {g:.2f}%")
print("  VERDICT               :",
      "official split only; one-paragraph audit" if g < 2 else
      "official split primary + source-grouped split as robustness" if g <= 15 else
      "both splits plus a figure; audit promoted to a headline")
print("  wrote 4 json files to", SAVE)
print("=" * 70)
```

```python
# ===== 02_dup_verify.py  --  is the 14.73% real? =====
import os, json, csv, collections, random
from PIL import Image, ImageDraw, ImageFont

DS, SAVE = "/content/my_images/plantseg", "/content/drive/MyDrive"
SPLITS = ["train", "val", "test"]
D = json.load(open(f"{SAVE}/plantseg_dupscan.json"))
cross = D["cross_split_clusters"]
meta = {os.path.splitext(r["Name"])[0]: r for r in
        csv.DictReader(open(f"{DS}/Metadata.csv", encoding="utf-8-sig", newline=""))}
disk = {s: {os.path.splitext(f)[0] for f in os.listdir(f"{DS}/images/{s}")
            if f.lower().endswith(".jpg")} for s in SPLITS}
split_of = {st: s for s in SPLITS for st in disk[s]}

# ---------- 1. same-disease vs cross-disease ----------
same, diff = [], []
for c in cross:
    (same if len({meta[s]["Disease"] for s in c}) == 1 else diff).append(c)
print(f"cross-split clusters: {len(cross)}   same disease: {len(same)}   "
      f"different diseases: {len(diff)}")
print(f"  same-disease also share one URL: "
      f"{sum(1 for c in same if len({meta[s]['URL'] for s in c})==1)}/{len(same)}")
print(f"  cross-disease also share one URL: "
      f"{sum(1 for c in diff if len({meta[s]['URL'] for s in c})==1)}/{len(diff)}")

for lab, grp in (("ALL", cross), ("SAME-DISEASE ONLY", same)):
    lk = collections.Counter()
    for c in grp:
        for s in c: lk[split_of[s]] += 1
    print(f"  {lab:18s} test {lk['test']:4d}/{len(disk['test'])} "
          f"= {100*lk['test']/len(disk['test']):5.2f}%   "
          f"val {100*lk['val']/len(disk['val']):5.2f}%")

print("\n  cross-disease examples (these are the ones in question):")
for c in diff[:10]:
    print("   ", [(split_of[s], meta[s]["Disease"]) for s in c][:4])

# ---------- 2. contact sheets ----------
TH, LAB, PER = 190, 46, 5
try:
    F = ImageFont.truetype("/usr/share/fonts/truetype/dejavu/DejaVuSans.ttf", 13)
except Exception:
    F = ImageFont.load_default()

def sheet(clusters, path, title):
    rows = clusters[:12]
    W, H = PER * TH, len(rows) * (TH + LAB) + 30
    sh = Image.new("RGB", (W, H), "white")
    dr = ImageDraw.Draw(sh)
    dr.text((6, 8), title, fill="black", font=F)
    for r, c in enumerate(rows):
        y = 30 + r * (TH + LAB)
        for k, s in enumerate(c[:PER]):
            im = Image.open(f"{DS}/images/{split_of[s]}/{s}.jpg").convert("RGB")
            im.thumbnail((TH - 8, TH - 8))
            sh.paste(im, (k * TH + 4, y + 4))
            dr.text((k * TH + 4, y + TH), f"{split_of[s]}", fill="red", font=F)
            dr.text((k * TH + 4, y + TH + 15), meta[s]["Disease"][:26], fill="black", font=F)
            dr.text((k * TH + 4, y + TH + 30), s[:28], fill="gray", font=F)
    sh.save(path, quality=88)
    print("  wrote", path, f"({os.path.getsize(path)/1e6:.2f} MB)")

random.seed(0)
sheet(random.sample(same, min(12, len(same))), f"{SAVE}/dup_same_disease.jpg",
      "CROSS-SPLIT DUPLICATE CANDIDATES - same disease label")
sheet(random.sample(diff, min(12, len(diff))), f"{SAVE}/dup_cross_disease.jpg",
      "CROSS-SPLIT DUPLICATE CANDIDATES - different disease labels")
print("\nupload both jpgs for inspection")
```
