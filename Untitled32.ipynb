```python
# =====================================================================
#  plantseg_stage0_reads.py -- the three reads that come before any
#  code or GPU time. CPU only, about four minutes, almost all unzip.
#
#  1. REALIZED ZOOM. Component 1 assumes a 768 cache shows the network
#     detail that a 384 resize destroyed. That is only true where the
#     ORIGINAL has the detail. For an image whose short side is 400 px,
#     a 768 cache is an upsample and carries no new information. The
#     quantity that decides component 1 is therefore
#         min(short side, CACHE) / 384
#     over Q0 images specifically -- not the plan's yes/no on 768.
#  2. LOSS FLOOR. The under-1% / 2% / 5% lesion-area counts the audit
#     already computed, which say how often the floor binds.
#  3. CACHE SHAPE. Read from the cache index, never inferred from size.
# =====================================================================
import os, csv, json, glob, zipfile, subprocess
import numpy as np

SAVE   = "/content/drive/MyDrive"
ROOT   = "/content/my_images"
DS     = f"{ROOT}/plantseg"
CACHE  = "/content/cache"
TRAIN  = 384                       # current training input
FROZEN = [0.059468, 0.138448, 0.300831]
try:
    from google.colab import drive; drive.mount("/content/drive", force_remount=False)
except Exception:
    pass

# ---------------------- get Metadata.csv available ---------------------
if not os.path.exists(f"{DS}/Metadata.csv"):
    hits = [p for p in glob.glob(f"{SAVE}/**/*.zip", recursive=True)
            if "plantseg" in os.path.basename(p).lower()
            and "result" not in os.path.basename(p).lower()]
    if not hits:
        raise SystemExit(f"no plantseg zip under {SAVE}")
    zp = max(hits, key=os.path.getsize)
    print(f"unzipping {zp} ({os.path.getsize(zp)/1e9:.1f} GB) ...", flush=True)
    os.makedirs(ROOT, exist_ok=True)
    subprocess.run(["unzip", "-q", "-o", zp, "-d", ROOT], check=True)
if not os.path.exists(f"{DS}/Metadata.csv"):
    found = next((r for r, _, f in os.walk(ROOT) if "Metadata.csv" in f), None)
    if found is None: raise SystemExit("Metadata.csv not found")
    DS = found
print(f"metadata at {DS}/Metadata.csv")

rows = list(csv.DictReader(open(f"{DS}/Metadata.csv", encoding="utf-8-sig", newline="")))
meta = {os.path.splitext(r["Name"])[0]: r for r in rows}
frac = {k: float(v["Mask ratio"]) for k, v in meta.items()}
rescol = next((c for c in rows[0] if c.strip().lower() == "resolution"), None)
if rescol is None:
    raise SystemExit(f"no Resolution column; columns are {list(rows[0])}")

def short_side(s):
    try:
        w, h = (int(x) for x in str(meta[s][rescol]).lower().split("x")[:2])
        return min(w, h)
    except Exception:
        return None

def quart(s):
    f = frac[s]
    return 0 if f <= FROZEN[0] else 1 if f <= FROZEN[1] else 2 if f <= FROZEN[2] else 3

# split membership: prefer the audit's record, else the folder layout
SP = None
if os.path.exists(f"{SAVE}/plantseg_splits.json"):
    SP = json.load(open(f"{SAVE}/plantseg_splits.json"))
    split_of = {s: sp for sp, lst in SP["official_splits"].items() for s in lst}
    if max(abs(a - b) for a, b in zip(SP["quartile_cuts"], FROZEN)) > 1e-6:
        raise SystemExit(f"cuts {SP['quartile_cuts']} != frozen {FROZEN}")
else:
    split_of = {}
    for root, _, files in os.walk(DS):
        low = root.lower().replace("\\", "/").split("/")
        sp = ("train" if "train" in low else "val" if "val" in low
              else "test" if "test" in low else None)
        if sp:
            for f in files:
                st = os.path.splitext(f)[0]
                if st in frac: split_of[st] = sp
    print("   (plantseg_splits.json absent -- splits read from folder names)")

stems = [s for s in frac if short_side(s)]
print(f"{len(stems)} images with a parseable resolution of {len(frac)} total\n")

# =================== 1. realized zoom, by quartile =====================
print("=" * 78)
print("1. REALIZED DETAIL GAIN -- does a bigger cache actually show more?")
print("=" * 78)
print(f"   ratio = min(original short side, CACHE) / {TRAIN}")
print(f"   a ratio of 1.00 or less means the cache upsamples past the original,")
print(f"   so the network sees NO detail it does not already see at {TRAIN}.\n")

ss = {s: short_side(s) for s in stems}
print(f"   {'':14s}{'n':>6}{'median':>9}{'  short side':>13}"
      f"{'ratio<=1':>10}{'ratio>=1.5':>12}{'ratio>=2':>10}")
verdict = {}
for CACHE_PX in (768, 512):
    print(f"\n   --- cache at {CACHE_PX} ---")
    for q in range(4):
        sel = [s for s in stems if quart(s) == q]
        if not sel: continue
        r = np.array([min(ss[s], CACHE_PX) / TRAIN for s in sel])
        med_short = int(np.median([ss[s] for s in sel]))
        print(f"   Q{q}{'':11s}{len(sel):>6}{np.median(r):>9.2f}{med_short:>13}"
              f"{100*(r <= 1.0).mean():>9.1f}%{100*(r >= 1.5).mean():>11.1f}%"
              f"{100*(r >= 2.0).mean():>9.1f}%")
        if q == 0: verdict[CACHE_PX] = dict(median=float(np.median(r)),
                                            nogain=float((r <= 1.0).mean()),
                                            full=float((r >= 2.0).mean()))

print("\n   the plan's own rule: work at 768 if at least half of Q0 originals")
print("   reach 768 on the short side.")
q0 = [s for s in stems if quart(s) == 0]
reach768 = np.mean([ss[s] >= 768 for s in q0])
print(f"   Q0 originals with short side >= 768: {100*reach768:.1f}%"
      f"  ->  {'768' if reach768 >= 0.5 else '512'}")

v7, v5 = verdict.get(768, {}), verdict.get(512, {})
print("\n   VERDICT on component 1")
if v7.get("median", 0) <= 1.15:
    print(f"   !! Q0 median realized zoom at 768 is {v7['median']:.2f}, and "
          f"{100*v7['nogain']:.0f}% of Q0 images gain nothing at all.")
    print("      A 768 cache cannot show the network detail these originals do")
    print("      not have. Component 1 is unlikely to work and 21 GPU-hours of")
    print("      Stage 2 would be spent proving it. Consider dropping crops and")
    print("      going straight to component 2, which needs no new cache.")
elif v7.get("median", 0) >= 1.6:
    print(f"   Q0 median realized zoom at 768 is {v7['median']:.2f} with "
          f"{100*v7['full']:.0f}% at the full 2x. Component 1 has room to work.")
else:
    print(f"   Q0 median realized zoom at 768 is {v7.get('median', 0):.2f} "
          f"(512: {v5.get('median', 0):.2f}).")
    print("      Partial headroom. Gate A stands, but expect a smaller effect")
    print("      than a clean 2x would give, and say so when setting the gate.")

# ================= 2. the loss floor, from the audit ===================
print("\n" + "=" * 78)
print("2. LOSS FLOOR -- how often does it bind?")
print("=" * 78)
AU = f"{SAVE}/plantseg_audit.json"
tr = np.array([frac[s] for s in stems if split_of.get(s) == "train"])
if os.path.exists(AU):
    A = json.load(open(AU)).get("train_area", {})
    print(f"   from plantseg_audit.json: {A}")
if len(tr):
    print(f"\n   train images below each lesion-area threshold (n={len(tr)}):")
    for p in (0.002, 0.005, 0.01, 0.02, 0.05):
        print(f"      under {100*p:>4.1f}% : {int((tr < p).sum()):>5}  "
              f"({100*(tr < p).mean():>4.1f}% of train)")
    print(f"\n   the floor is a count of pixels on a {TRAIN}x{TRAIN} input:")
    for p in (0.002, 0.005, 0.01, 0.02):
        print(f"      floor at {100*p:>4.1f}% = {int(p*TRAIN*TRAIN):>6} px  "
              f"-> binds on {100*(tr < p).mean():>4.1f}% of train images")
    print("\n   pick the floor so it binds on the outliers you mean to guard")
    print("   against, not on the sparse images the component is built for.")

# ====================== 3. the cache's shape ===========================
print("\n" + "=" * 78)
print("3. EXISTING CACHE SHAPE -- read, never inferred from file size")
print("=" * 78)
IX = f"{CACHE}/index.json"
if os.path.exists(IX):
    I = json.load(open(IX))
    a = np.load(I["img_path"], mmap_mode="r"); m = np.load(I["msk_path"], mmap_mode="r")
    gb = (a.nbytes + m.nbytes) / 1e9
    print(f"   images {a.shape} {a.dtype}   masks {m.shape} {m.dtype}")
    print(f"   cache_size recorded: {I.get('cache_size', 'not recorded')}")
    print(f"   total on disk: {gb:.2f} GB")
    print(f"   -> build the 768 cache with the SAME resize ops at {a.shape[1]} -> 768")
else:
    print(f"   {IX} not present (/content is wiped on a runtime restart).")
    print("   Do NOT infer it from the 6 GB figure: that fits 448 with masks")
    print("   (6.24 GB) as well as 512 images alone (6.11 GB).")
    print("   Read it off your own plantseg_prep.py instead -- the resize size,")
    print("   and the resample filter used for images and for masks.")

print("\n" + "=" * 78)
print("Nothing above needs a GPU. Decide 768 vs 512 vs dropping component 1")
print("from section 1 before writing cell A or building any cache.")
print("=" * 78)
```

```python
# =====================================================================
#  cell B -- the Stage 2 cache at 768.
#
#  Only the 537 labelled + 846 validation images (1,383 of 7,774), so it
#  is 3.26 GB and lives on Drive. No per-session rebuild, and if gate A
#  fails the full 18.3 GB cache is never built.
#
#  Every operation is copied from plantseg_prep.py so the only thing
#  that differs between the 441 and 768 caches is the size:
#      images : Image.open(...).convert("RGB").resize((S,S), BILINEAR)
#      masks  : Image.open(...).resize((S,S), NEAREST), then (> 0) -> 0/1
#  The labelled set is READ from plantseg_splits.json["budgets"], not
#  recomputed, so it cannot drift from the audit's stratified ordering.
# =====================================================================
import os, csv, json, time, hashlib, shutil
import numpy as np
from PIL import Image
Image.MAX_IMAGE_PIXELS = None

DS     = "/content/my_images/plantseg"
SAVE   = "/content/drive/MyDrive"
LOCAL  = "/content/cache768"
DRIVE  = f"{SAVE}/cache768"
LOAD   = 768                 # the 441 cache stays untouched
BUDGET = "10pct"             # Stage 2 is the 10% budget only
SPLITS = ["train", "val", "test"]
FROZEN = [0.059468, 0.138448, 0.300831]
try:
    from google.colab import drive; drive.mount("/content/drive", force_remount=False)
except Exception:
    pass
os.makedirs(LOCAL, exist_ok=True); os.makedirs(DRIVE, exist_ok=True)

# ---------------------- inputs, all cross-checked ----------------------
SP = json.load(open(f"{SAVE}/plantseg_splits.json"))
if max(abs(a-b) for a, b in zip(SP["quartile_cuts"], FROZEN)) > 1e-6:
    raise SystemExit(f"cuts {SP['quartile_cuts']} != frozen {FROZEN}")
disk = {s: sorted({os.path.splitext(f)[0] for f in os.listdir(f"{DS}/images/{s}")
                   if f.lower().endswith(".jpg")}
                  & {os.path.splitext(f)[0] for f in os.listdir(f"{DS}/annotations/{s}")
                     if f.lower().endswith(".png")}) for s in SPLITS}
if {k: len(v) for k, v in disk.items()} != {"train": 5367, "val": 846, "test": 1561}:
    raise SystemExit(f"split sizes {[len(disk[s]) for s in SPLITS]} are not 5367/846/1561")
meta = {os.path.splitext(r["Name"])[0]: r for r in csv.DictReader(
        open(f"{DS}/Metadata.csv", encoding="utf-8-sig", newline=""))}
frac = {k: float(v["Mask ratio"]) for k, v in meta.items()}
def quart(s):
    f = frac[s]
    return 0 if f <= FROZEN[0] else 1 if f <= FROZEN[1] else 2 if f <= FROZEN[2] else 3

labelled = sorted(SP["budgets"][BUDGET])
if len(labelled) != 537:
    raise SystemExit(f"{BUDGET} holds {len(labelled)} images, expected 537")
if not set(labelled) <= set(disk["train"]):
    raise SystemExit("the labelled budget is not a subset of train")
SEL      = labelled + sorted(disk["val"])          # 537 + 846
split_of = {s: ("train" if s in set(labelled) else "val") for s in SEL}
row_of   = {s: i for i, s in enumerate(SEL)}
N = len(SEL)
qc = [sum(1 for s in labelled if quart(s) == q) for q in range(4)]
print(f"labelled {len(labelled)} (quartiles {qc})  +  val {len(disk['val'])}  =  {N}")
print(f"target {N*LOAD*LOAD*4/1e9:.2f} GB at {LOAD}x{LOAD}\n")

IMG_P, MSK_P = f"{LOCAL}/img_{LOAD}.npy", f"{LOCAL}/msk_{LOAD}.npy"
IX           = f"{LOCAL}/index.json"
D_IMG, D_MSK = f"{DRIVE}/img_{LOAD}.npy", f"{DRIVE}/msk_{LOAD}.npy"
D_IX         = f"{DRIVE}/index.json"

def sha(path, chunk=1 << 24):
    h = hashlib.sha256()
    with open(path, "rb") as f:
        while (b := f.read(chunk)): h.update(b)
    return h.hexdigest()[:16]

# -------------- restore from Drive if it is already there --------------
built = False
if all(os.path.exists(p) for p in (D_IMG, D_MSK, D_IX)):
    DI = json.load(open(D_IX))
    if DI.get("stems") == SEL and DI.get("load_size") == LOAD:
        print("Stage 2 cache found on Drive -- copying to local disk ...", flush=True)
        t = time.time()
        for src, dst in ((D_IMG, IMG_P), (D_MSK, MSK_P)):
            if not os.path.exists(dst) or os.path.getsize(dst) != os.path.getsize(src):
                shutil.copyfile(src, dst)
        print(f"   copied in {time.time()-t:.0f}s")
        ok = (sha(IMG_P) == DI["sha_img"] and sha(MSK_P) == DI["sha_msk"])
        print(f"   checksum {'MATCHES' if ok else 'MISMATCH -- rebuilding'}")
        built = ok
    else:
        print("Drive cache does not match this selection -- rebuilding")

# ------------------------------ build ----------------------------------
if not built:
    print(f"building {N} images at {LOAD}x{LOAD} ...", flush=True)
    imgs = np.lib.format.open_memmap(IMG_P, mode="w+", dtype=np.uint8,
                                     shape=(N, LOAD, LOAD, 3))
    msks = np.lib.format.open_memmap(MSK_P, mode="w+", dtype=np.uint8,
                                     shape=(N, LOAD, LOAD))
    t0 = time.time()
    for i, s in enumerate(SEL):
        sp = split_of[s]
        imgs[i] = np.asarray(Image.open(f"{DS}/images/{sp}/{s}.jpg").convert("RGB")
                             .resize((LOAD, LOAD), Image.BILINEAR), dtype=np.uint8)
        m = Image.open(f"{DS}/annotations/{sp}/{s}.png").resize((LOAD, LOAD), Image.NEAREST)
        msks[i] = (np.asarray(m) > 0).astype(np.uint8)      # BINARY RULE, applied once
        if (i + 1) % 250 == 0:
            el = time.time() - t0
            print(f"    {i+1}/{N}   {el:.0f}s elapsed, ~{el/(i+1)*(N-i-1):.0f}s left",
                  flush=True)
    imgs.flush(); msks.flush(); del imgs, msks
    print(f"  built in {time.time()-t0:.0f}s")
    si, sm = sha(IMG_P), sha(MSK_P)
    json.dump(dict(load_size=LOAD, stems=SEL, row_of=row_of, split_of=split_of,
                   img_path=IMG_P, msk_path=MSK_P, budget=BUDGET,
                   labelled=labelled, sha_img=si, sha_msk=sm,
                   built_from="plantseg_prep.py operations: BILINEAR images, "
                              "NEAREST masks, (mask > 0) -> uint8 0/1"),
              open(IX, "w"))
    print(f"  checksums  img {si}  msk {sm}")
    print("  copying to Drive ...", flush=True)
    t = time.time()
    for src, dst in ((IMG_P, D_IMG), (MSK_P, D_MSK), (IX, D_IX)):
        shutil.copyfile(src, dst)
    print(f"   copied in {time.time()-t:.0f}s -> {DRIVE}")

# ----------------------------- verify ----------------------------------
I = json.load(open(IX))
imgs = np.load(I["img_path"], mmap_mode="r"); msks = np.load(I["msk_path"], mmap_mode="r")
print("\n" + "=" * 72)
print(f"images {imgs.shape} {imgs.dtype}   masks {msks.shape} {msks.dtype}")
print(f"on disk: {(os.path.getsize(IMG_P)+os.path.getsize(MSK_P))/1e9:.2f} GB")
vals = sorted(np.unique(msks[::50]).tolist())
print(f"mask values present: {vals}   {'OK' if vals in ([0,1],[0],[1]) else 'NOT BINARY'}")
rng = np.random.default_rng(0)
sel = rng.choice(N, min(400, N), replace=False)
got  = np.array([msks[i].mean() for i in sel])
want = np.array([frac[SEL[i]] for i in sel])
print(f"mask area vs Metadata: corr {np.corrcoef(got, want)[0,1]:.5f}   "
      f"median |diff| {np.median(np.abs(got-want)):.5f}   max {np.abs(got-want).max():.4f}")
print(f"blank images: {int(sum(imgs[i].max()==0 for i in sel))}/{len(sel)}")

# realized detail gain, per axis -- short side is the limiting one
print("\nrealized detail gain of this cache over the 384 training input")
print("  (square resize scales each axis on its own; the SHORT side binds)")
res = {k: tuple(int(x) for x in v["Resolution"].lower().split("x")[:2]) for k, v in meta.items()}
print(f"  {'':12}{'n':>5}{'worst axis':>12}{'best axis':>11}{'no gain':>10}{'full 2x':>9}")
for name, grp in (("labelled Q0", [s for s in labelled if quart(s) == 0]),
                  ("labelled all", labelled),
                  ("val Q0", [s for s in disk["val"] if quart(s) == 0]),
                  ("val all", sorted(disk["val"]))):
    lo = np.array([min(min(res[s]), LOAD)/384 for s in grp])
    hi = np.array([min(max(res[s]), LOAD)/384 for s in grp])
    print(f"  {name:12}{len(grp):>5}{np.median(lo):>12.2f}{np.median(hi):>11.2f}"
          f"{100*(lo <= 1.0).mean():>9.1f}%{100*(lo >= 2.0).mean():>8.1f}%")

print("\n" + "=" * 72)
print(f"Stage 2 cache ready. Local {LOCAL}, Drive copy {DRIVE}.")
print("Later sessions re-run this cell: it copies from Drive and verifies the")
print("checksum instead of decoding 1,383 JPEGs again.")
print("The 441 cache is untouched -- S-384 runs still read it.")
print("=" * 72)
```

```python
# =====================================================================
#  session warm-up -- run this first in every new Colab session.
#  /content is wiped on restart; Drive is not. This re-extracts the
#  dataset and copies the 768 cache back from Drive (a copy, not a
#  rebuild), then verifies both against their checksums.
# =====================================================================
import os, json, glob, time, shutil, hashlib, subprocess
import numpy as np

SAVE  = "/content/drive/MyDrive"
ROOT  = "/content/my_images"
DS    = f"{ROOT}/plantseg"
LOCAL = "/content/cache768"
DRIVE = f"{SAVE}/cache768"
try:
    from google.colab import drive; drive.mount("/content/drive", force_remount=False)
except Exception: pass

# ------------------------------ dataset --------------------------------
if os.path.exists(f"{DS}/Metadata.csv"):
    print("dataset already extracted")
else:
    hits = [p for p in glob.glob(f"{SAVE}/**/*.zip", recursive=True)
            if "plantseg" in os.path.basename(p).lower()
            and "result" not in os.path.basename(p).lower()]
    if not hits: raise SystemExit(f"no plantseg zip under {SAVE}")
    zp = max(hits, key=os.path.getsize)
    print(f"extracting {zp} ({os.path.getsize(zp)/1e9:.1f} GB) ...", flush=True)
    os.makedirs(ROOT, exist_ok=True); t = time.time()
    subprocess.run(["unzip", "-q", "-o", zp, "-d", ROOT], check=True)
    print(f"   extracted in {time.time()-t:.0f}s")
if not os.path.exists(f"{DS}/Metadata.csv"):
    found = next((r for r, _, f in os.walk(ROOT) if "Metadata.csv" in f), None)
    if found is None: raise SystemExit("Metadata.csv not found after extraction")
    print(f"   note: Metadata.csv is at {found}, not {DS}")

# ---------------------------- 768 cache --------------------------------
def sha(p, chunk=1 << 24):
    h = hashlib.sha256()
    with open(p, "rb") as f:
        while (b := f.read(chunk)): h.update(b)
    return h.hexdigest()[:16]

if not os.path.exists(f"{DRIVE}/index.json"):
    raise SystemExit(f"no cache on Drive at {DRIVE} -- re-run cell B to build it")
os.makedirs(LOCAL, exist_ok=True)
D = json.load(open(f"{DRIVE}/index.json"))
S = D["load_size"]
print(f"\nrestoring the {S} cache from Drive ({len(D['stems'])} images) ...", flush=True)
t = time.time()
for name in (f"img_{S}.npy", f"msk_{S}.npy", "index.json"):
    src, dst = f"{DRIVE}/{name}", f"{LOCAL}/{name}"
    if not os.path.exists(dst) or os.path.getsize(dst) != os.path.getsize(src):
        shutil.copyfile(src, dst)
print(f"   copied in {time.time()-t:.0f}s")

I = json.load(open(f"{LOCAL}/index.json"))
ok_i = sha(I["img_path"]) == I["sha_img"]
ok_m = sha(I["msk_path"]) == I["sha_msk"]
imgs = np.load(I["img_path"], mmap_mode="r"); msks = np.load(I["msk_path"], mmap_mode="r")
print(f"   images {imgs.shape}  masks {msks.shape}")
print(f"   checksum img {'OK' if ok_i else 'MISMATCH'}   msk {'OK' if ok_m else 'MISMATCH'}")
if not (ok_i and ok_m):
    raise SystemExit("checksum mismatch -- the Drive copy is damaged; re-run cell B")
print(f"   labelled {len(I['labelled'])}   total cached {len(I['stems'])}")
print(f"\nready. Next: cell D (Stage 2 training).")
```

```python
# =====================================================================
#  cell D -- Stage 2: supervised tests of components 1 and 2.
#  One cell, three switches, 25,000 steps, resume-safe. Validation only;
#  the test split is not touched until Stage 5.
#
#      INPUT_MODE in {full_384, crop_768, crop_768_up}
#      FG_LOSS    in {pooled, class_balanced, image_balanced}
#      JITTER     in {False, True}
#
#  (full_384, pooled, False) is the S-384 baseline.
#
#  crop_768_up is the control the crop statistics made necessary: uniform
#  384 crops of a 768 image carry 1.50x the Q0 lesion density of the
#  whole image, because uniform crop ORIGINS over-sample the centre (a
#  centre pixel sits in 384 of 385 origins per axis, a corner pixel in 1).
#  So crop_768 changes resolution AND density at once. crop_768_up takes
#  the 768 image, downsamples it to 384 and upsamples it back, destroying
#  exactly the detail above 384 and nothing else, then crops identically.
#  crop_768 beating crop_768_up isolates resolution; a tie says the gain
#  is crop geometry.
#
#  Everything reads ONE cache: /content/cache768 (537 labelled + 846 val).
#  The 441 cache is not needed, so prep.py need not be run this session.
# =====================================================================
import os, csv, json, math, time, random, shutil, zipfile, collections
import numpy as np
import matplotlib; matplotlib.use("Agg")
import matplotlib.pyplot as plt
from PIL import Image, ImageFilter
import torch, torch.nn as nn, torch.optim as optim, torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models
import torchvision.transforms.functional as TF

# ------------------------- the three switches --------------------------
INPUT_MODE = "full_384"        # full_384 | crop_768 | crop_768_up
FG_LOSS    = "pooled"          # pooled | class_balanced | image_balanced
JITTER     = False             # True only for S-768c-jit
TRAIN_SEED = 42
# -----------------------------------------------------------------------
FLOOR_FRAC = 0.01              # declared from check 0f: binds on 8.9% of
                               # non-empty Q0 crops, caps amplification 21.5x
SELECT_ON  = {"full_384": "whole384", "crop_768": "full768",
              "crop_768_up": "full768"}[INPUT_MODE]
MAX_STEPS  = 25000
EVAL_EVERY = 1000
SHARE_EVERY = 500              # per-quartile lesion-loss share
CKPT_EVERY = 2500
LR_PATIENCE = 3                # patience counts EVAL_EVERY blocks, now 1000
TRAIN_RES  = 384
FULL_RES   = 768
BATCH      = 4                 # matches the existing pipeline; batch 8
                               # doubles every run to ~2.2h and Stage 2 to ~42h
BATCH_EVAL = 4                 # 768 input on a T4
LR_ENCODER, LR_DECODER, WEIGHT_DECAY = 1e-5, 1e-4, 1e-4
FROZEN     = [0.059468, 0.138448, 0.300831]
SAVE, C768 = "/content/drive/MyDrive", "/content/cache768"
NUM_WORKERS = min(2, os.cpu_count() or 1)
FRESH      = True              # wipes this run's folder. Set back to False
                               # once it is past step 2500 so a dropped
                               # session resumes instead of restarting.

TAG = {"full_384": "384", "crop_768": "768c", "crop_768_up": "768up"}[INPUT_MODE]
if FG_LOSS == "class_balanced":  TAG += "-cb"
if FG_LOSS == "image_balanced":  TAG += "-bal"
if JITTER:                       TAG += "-jit"
RUN = f"S-{TAG}_seed{TRAIN_SEED}"
OUT = f"{SAVE}/stage2/{RUN}"
if FRESH and os.path.isdir(OUT): shutil.rmtree(OUT, ignore_errors=True)
os.makedirs(OUT, exist_ok=True)

random.seed(TRAIN_SEED); np.random.seed(TRAIN_SEED)
torch.manual_seed(TRAIN_SEED); torch.cuda.manual_seed_all(TRAIN_SEED)
torch.backends.cudnn.deterministic = True; torch.backends.cudnn.benchmark = False
DEV = torch.device("cuda" if torch.cuda.is_available() else "cpu")
try:
    from google.colab import drive; drive.mount("/content/drive", force_remount=False)
except Exception: pass
print(f"{RUN}\ndevice {DEV} {torch.cuda.get_device_name(0) if DEV.type=='cuda' else ''}")
print(f"input {INPUT_MODE} | loss {FG_LOSS} | jitter {JITTER} | "
      f"select on {SELECT_ON} | floor {FLOOR_FRAC} | batch {BATCH}")

# --------------------------- data and asserts --------------------------
SP = json.load(open(f"{SAVE}/plantseg_splits.json"))
if max(abs(a-b) for a, b in zip(SP["quartile_cuts"], FROZEN)) > 1e-6:
    raise SystemExit(f"cuts {SP['quartile_cuts']} != frozen {FROZEN}")
if {k: len(v) for k, v in SP["official_splits"].items()} != \
   {"train": 5367, "val": 846, "test": 1561}:
    raise SystemExit("split sizes are not 5367/846/1561")
I768 = json.load(open(f"{C768}/index.json"))
if I768["load_size"] != FULL_RES:
    raise SystemExit(f"cache is {I768['load_size']}, expected {FULL_RES}")
row = I768["row_of"]
LAB = sorted(I768["labelled"])
CLEAN_VAL = sorted(set(SP["clean_val"]) & set(I768["stems"]))
if len(LAB) != 537: raise SystemExit(f"{len(LAB)} labelled, expected 537")
DSDIR = "/content/my_images/plantseg"
meta = {os.path.splitext(r["Name"])[0]: r for r in csv.DictReader(
        open(f"{DSDIR}/Metadata.csv", encoding="utf-8-sig", newline=""))}
frac = {k: float(v["Mask ratio"]) for k, v in meta.items()}
def quart(s):
    f = frac[s]
    return 0 if f <= FROZEN[0] else 1 if f <= FROZEN[1] else 2 if f <= FROZEN[2] else 3
QV = np.array([quart(s) for s in CLEAN_VAL])
print(f"labelled {len(LAB)}   clean_val {len(CLEAN_VAL)} "
      f"(quartiles {[int((QV==q).sum()) for q in range(4)]})")
print("   gates are judged on clean_val, not the full 846: full val carries")
print("   17.14% near-duplicates of training images.")

MEAN, STD = [0.485,0.456,0.406], [0.229,0.224,0.225]
NORM = transforms.Normalize(MEAN, STD)
JIT  = transforms.ColorJitter(0.4, 0.4, 0.4, 0.1)
_mm = {}
def cache():
    if not _mm:
        _mm["i"] = np.load(I768["img_path"], mmap_mode="r")
        _mm["m"] = np.load(I768["msk_path"], mmap_mode="r")
    return _mm["i"], _mm["m"]

def strip_detail(im):
    """768 -> 384 -> 768. Destroys exactly the detail above 384, keeps
       geometry, size and lesion density identical."""
    return im.resize((TRAIN_RES,)*2, Image.BILINEAR).resize((FULL_RES,)*2, Image.BILINEAR)

def photometric(img):
    if random.random() < 0.8: img = JIT(img)
    if random.random() < 0.2: img = TF.to_grayscale(img, num_output_channels=3)
    if random.random() < 0.3:
        img = img.filter(ImageFilter.GaussianBlur(radius=random.uniform(0.1, 1.5)))
    return img

def geometric(img, msk):
    if random.random() < 0.5: img, msk = TF.hflip(img), TF.hflip(msk)
    if random.random() < 0.5: img, msk = TF.vflip(img), TF.vflip(msk)
    k = random.randint(0, 3)
    if k:
        img = img.rotate(90*k, expand=True, resample=Image.BILINEAR)
        msk = msk.rotate(90*k, expand=True, resample=Image.NEAREST)
    return img, msk

def rrc(img, msk, scale=(0.5, 1.0), ratio=(3/4, 4/3)):
    w, h = img.size; area = w*h
    for _ in range(10):
        t = area*random.uniform(*scale)
        ar = math.exp(random.uniform(math.log(ratio[0]), math.log(ratio[1])))
        cw, ch = int(round(math.sqrt(t*ar))), int(round(math.sqrt(t/ar)))
        if cw <= w and ch <= h:
            x, y = random.randint(0, w-cw), random.randint(0, h-ch)
            return img.crop((x, y, x+cw, y+ch)), msk.crop((x, y, x+cw, y+ch))
    return img, msk

class Labelled(Dataset):
    """full_384   : whole image, rrc + geometric, resized to 384
       crop_768   : uniform 384 window of the 768 image (the crop IS the
                    spatial augmentation, so rrc is not applied on top)
       crop_768_up: the same window of a detail-stripped 768 image"""
    def __init__(s, st): s.st = list(st)
    def __len__(s): return len(s.st)
    def __getitem__(s, i):
        I, M = cache(); r = row[s.st[i]]
        img = Image.fromarray(I[r]); msk = Image.fromarray(M[r])
        if INPUT_MODE == "crop_768_up": img = strip_detail(img)
        if INPUT_MODE == "full_384":
            img, msk = rrc(img, msk)
            img, msk = geometric(img, msk)
            img = img.resize((TRAIN_RES,)*2, Image.BILINEAR)
            msk = msk.resize((TRAIN_RES,)*2, Image.NEAREST)
        else:
            side = (random.randint(256, FULL_RES) if JITTER else TRAIN_RES)
            x = random.randint(0, FULL_RES-side); y = random.randint(0, FULL_RES-side)
            img = img.crop((x, y, x+side, y+side)); msk = msk.crop((x, y, x+side, y+side))
            img, msk = geometric(img, msk)
            if side != TRAIN_RES:
                img = img.resize((TRAIN_RES,)*2, Image.BILINEAR)
                msk = msk.resize((TRAIN_RES,)*2, Image.NEAREST)
        y_ = torch.from_numpy((np.array(msk) > 0).astype(np.float32)).unsqueeze(0)
        return NORM(TF.to_tensor(photometric(img))), y_, quart(s.st[i])

class EvalSet(Dataset):
    """input follows INPUT_MODE; the TARGET is always the 768 mask."""
    def __init__(s, st): s.st = list(st)
    def __len__(s): return len(s.st)
    def __getitem__(s, i):
        I, M = cache(); r = row[s.st[i]]
        img = Image.fromarray(I[r])
        if INPUT_MODE == "crop_768_up": img = strip_detail(img)
        y = torch.from_numpy((np.asarray(M[r]) > 0).astype(np.float32)).unsqueeze(0)
        return NORM(TF.to_tensor(img)), y            # image at 768, mask at 768

def seed_worker(_):
    s = torch.initial_seed() % 2**32; np.random.seed(s); random.seed(s)
LK = dict(num_workers=NUM_WORKERS, pin_memory=DEV.type == "cuda",
          worker_init_fn=seed_worker, persistent_workers=NUM_WORKERS > 0)
g = torch.Generator(); g.manual_seed(TRAIN_SEED)
loader_l = DataLoader(Labelled(LAB), BATCH, shuffle=True, drop_last=True,
                      generator=g, **LK)
val_loader = DataLoader(EvalSet(CLEAN_VAL), BATCH_EVAL, shuffle=False, **LK)
def cycle(dl):
    while True:
        for b in dl: yield b

# ------------------------------- model ---------------------------------
SPEC = dict(channels=(64,256,512,1024,2048), decoder=(512,256,128,64))
class DoubleConv(nn.Module):
    def __init__(s,i,o):
        super().__init__()
        s.net = nn.Sequential(nn.Conv2d(i,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True),
                              nn.Conv2d(o,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True))
    def forward(s,x): return s.net(x)
class Up(nn.Module):
    def __init__(s,i,k,o):
        super().__init__(); s.reduce=nn.Conv2d(i,i//2,1,bias=False); s.conv=DoubleConv(i//2+k,o)
    def forward(s,x,skip):
        x = F.interpolate(x, size=skip.shape[-2:], mode="bilinear", align_corners=False)
        return s.conv(torch.cat([s.reduce(x), skip], 1))
class ResUNet(nn.Module):
    def __init__(s):
        super().__init__()
        base = models.resnet50(weights=models.ResNet50_Weights.DEFAULT)
        c0,c1,c2,c3,c4 = SPEC["channels"]; d1,d2,d3,d4 = SPEC["decoder"]
        L = list(base.children())
        s.stem=nn.Sequential(*L[:3]); s.pool=L[3]
        s.l1,s.l2,s.l3,s.l4 = L[4],L[5],L[6],L[7]
        s.u1,s.u2,s.u3,s.u4 = Up(c4,c3,d1),Up(d1,c2,d2),Up(d2,c1,d3),Up(d3,c0,d4)
        s.head = nn.Sequential(nn.Conv2d(d4,32,3,padding=1,bias=False),
                               nn.BatchNorm2d(32), nn.ReLU(True), nn.Conv2d(32,1,1))
    def enc_params(s):
        for m in (s.stem,s.l1,s.l2,s.l3,s.l4): yield from m.parameters()
    def dec_params(s):
        for m in (s.u1,s.u2,s.u3,s.u4,s.head): yield from m.parameters()
    def forward(s,x):
        size=x.shape[-2:]
        x0=s.stem(x); x1=s.l1(s.pool(x0)); x2=s.l2(x1); x3=s.l3(x2); x4=s.l4(x3)
        d=s.u1(x4,x3); d=s.u2(d,x2); d=s.u3(d,x1); d=s.u4(d,x0)
        return s.head(F.interpolate(d,size=size,mode="bilinear",align_corners=False))

# ------------------------------- losses --------------------------------
bce_none = nn.BCEWithLogitsLoss(reduction="none")
def tversky(logits, tgt, eps=1.0):
    p = torch.sigmoid(logits)
    tp = (p*tgt).sum((1,2,3)); fp = (p*(1-tgt)).sum((1,2,3)); fn = ((1-p)*tgt).sum((1,2,3))
    return 1 - (tp+eps)/(tp+0.5*fp+0.5*fn+eps)

def fg_term(logits, tgt):
    """returns (per-image BCE term, per-image lesion-only contribution).
       All three forms are scaled so a uniform per-pixel loss c gives c,
       so the learning rate and the BCE/Tversky balance do not shift
       between configurations. The plan's formula writes the sum of the
       two class means; using their average is the same comparison at the
       same scale."""
    e = bce_none(logits, tgt)
    sf = tgt.sum((1,2,3)); sb = (1-tgt).sum((1,2,3))
    ef = (e*tgt).sum((1,2,3)); eb = (e*(1-tgt)).sum((1,2,3))
    if FG_LOSS == "pooled":
        return e.mean((1,2,3)), ef/tgt[0].numel()
    if FG_LOSS == "class_balanced":                 # batch-level class means
        n = logits.shape[0]
        mf = ef.sum()/sf.sum().clamp(min=1.0); mb = eb.sum()/sb.sum().clamp(min=1.0)
        return (0.5*(mf+mb)).expand(n), 0.5*ef/sf.sum().clamp(min=1.0)
    floor = FLOOR_FRAC*tgt[0].numel()               # per-image, with the floor
    tf = ef/torch.clamp(sf, min=floor)
    tb = eb/sb.clamp(min=1.0)
    return 0.5*(tf+tb), 0.5*tf

def sup_loss(logits, tgt):
    b, lesion = fg_term(logits, tgt)
    return b + tversky(logits, tgt), lesion

# --------------------------- 768 inference -----------------------------
@torch.no_grad()
def prob_768(net, x, mode):
    """x is the model's input image at 768. Returns a 768 probability map."""
    if mode == "full768":
        return torch.sigmoid(net(x).float())
    if mode == "whole384":
        d = F.interpolate(x, size=(TRAIN_RES,)*2, mode="bilinear", align_corners=False)
        p = torch.sigmoid(net(d).float())
        return F.interpolate(p, size=(FULL_RES,)*2, mode="bilinear", align_corners=False)
    # genuinely overlapping: stride 192 gives 3x3 tiles at 50% overlap.
    # Origins (0, 384) alone would ABUT, not overlap, and blending would
    # be a no-op -- the seam is exactly where a tiled model is weakest.
    acc = torch.zeros(x.shape[0], 1, FULL_RES, FULL_RES, device=x.device)
    cnt = torch.zeros_like(acc)
    ORIG = (0, (FULL_RES-TRAIN_RES)//2, FULL_RES-TRAIN_RES)
    for oy in ORIG:
        for ox in ORIG:
            t = x[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES]
            acc[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += torch.sigmoid(net(t).float())
            cnt[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += 1
    return acc/cnt.clamp(min=1)                      # blend, threshold afterwards

THS = [round(t, 2) for t in np.arange(0.10, 0.91, 0.05)]
@torch.no_grad()
def evaluate(net, mode):
    net.eval()
    nI = len(CLEAN_VAL); per = np.zeros((nI, len(THS))); P = np.zeros((nI, len(THS)))
    Ipx = np.zeros((nI, len(THS))); G = np.zeros(nI); k = 0
    for x, y in val_loader:
        x, y = x.to(DEV, non_blocking=True), y.to(DEV, non_blocking=True)
        with torch.amp.autocast("cuda", enabled=DEV.type == "cuda"):
            p = prob_768(net, x, mode)
        b = x.shape[0]; yf = y.view(b, -1)
        G[k:k+b] = yf.sum(1).cpu().numpy()
        pf = p.view(b, -1)
        for j, t in enumerate(THS):
            pr = (pf > t).float()
            it = (pr*yf).sum(1); pp = pr.sum(1)
            Ipx[k:k+b, j] = it.cpu().numpy(); P[k:k+b, j] = pp.cpu().numpy()
            per[k:k+b, j] = (it/(pp+yf.sum(1)-it).clamp(min=1)).cpu().numpy()
        k += b
    pooled = Ipx.sum(0)/np.maximum(P.sum(0)+G.sum()-Ipx.sum(0), 1)
    j = int(np.argmax(pooled)); t = THS[j]
    byq = {q: float(per[QV == q, j].mean()) for q in range(4)}
    return dict(th=t, pooled=float(pooled[j]), perimg=float(per[:, j].mean()),
                byq=byq, pred_frac=float(P[:, j].sum()/(len(CLEAN_VAL)*FULL_RES*FULL_RES)))

# --------------------------- resume or start ---------------------------
model = ResUNet().to(DEV)
opt = optim.AdamW([{"params": list(model.enc_params()), "lr": LR_ENCODER},
                   {"params": list(model.dec_params()), "lr": LR_DECODER}],
                  weight_decay=WEIGHT_DECAY)
sched = optim.lr_scheduler.ReduceLROnPlateau(opt, mode="max", patience=LR_PATIENCE, factor=0.5)
scaler = torch.amp.GradScaler("cuda")
step, best, history, shares = 0, -1.0, [], []
CK, BEST = f"{OUT}/ckpt.pt", f"{OUT}/best.pt"
if os.path.exists(CK):
    c = torch.load(CK, map_location=DEV, weights_only=False)
    model.load_state_dict(c["model"]); opt.load_state_dict(c["opt"])
    sched.load_state_dict(c["sched"]); scaler.load_state_dict(c["scaler"])
    step, best, history, shares = c["step"], c["best"], c["history"], c.get("shares", [])
    random.setstate(c["py_rng"]); np.random.set_state(c["np_rng"])
    torch.set_rng_state(c["t_rng"].cpu())
    if c.get("cuda_rng") is not None and DEV.type == "cuda":
        torch.cuda.set_rng_state(c["cuda_rng"].cpu())
    print(f"RESUMED from step {step} (best {best:.4f})")
else:
    print("starting fresh")

def save_ckpt():
    torch.save(dict(model=model.state_dict(), opt=opt.state_dict(), sched=sched.state_dict(),
                    scaler=scaler.state_dict(), step=step, best=best, history=history,
                    shares=shares, py_rng=random.getstate(), np_rng=np.random.get_state(),
                    t_rng=torch.get_rng_state(),
                    cuda_rng=torch.cuda.get_rng_state() if DEV.type == "cuda" else None), CK)

# ------------------------------- train ---------------------------------
it = cycle(loader_l)
print("=" * 76); print(f"{RUN}  --  {len(LAB)} labelled, {MAX_STEPS} steps"); print("=" * 76)
t0 = t_last = time.time()
r_loss = r_n = 0.0; q_les = np.zeros(4); q_cnt = np.zeros(4)
while step < MAX_STEPS:
    model.train()
    x, y, q = next(it)
    x, y = x.to(DEV, non_blocking=True), y.to(DEV, non_blocking=True)
    opt.zero_grad(set_to_none=True)
    with torch.amp.autocast("cuda", enabled=DEV.type == "cuda"):
        lg = model(x)
        vec, lesion = sup_loss(lg.float(), y)
        loss = vec.mean()
    scaler.scale(loss).backward(); scaler.step(opt); scaler.update()
    step += 1
    r_loss += loss.item(); r_n += 1
    lz = lesion.detach().float().cpu().numpy(); qz = q.numpy()
    for b in range(len(qz)): q_les[qz[b]] += lz[b]; q_cnt[qz[b]] += 1

    if step % SHARE_EVERY == 0:
        tot = max(q_les.sum(), 1e-12)
        sh = (q_les/tot).round(4).tolist()
        shares.append(dict(step=step, share=sh, seen=q_cnt.astype(int).tolist()))
        print(f"   [step {step}] lesion-loss share by quartile {sh}"
              f"  (images seen {q_cnt.astype(int).tolist()})")
        q_les[:] = 0; q_cnt[:] = 0

    if step % EVAL_EVERY == 0:
        m = evaluate(model, SELECT_ON)
        prev = opt.param_groups[-1]["lr"]; sched.step(m["pooled"])
        cur = opt.param_groups[-1]["lr"]
        saved = ""
        if m["pooled"] > best:
            best = m["pooled"]; torch.save(model.state_dict(), BEST); saved = "  <-- saved"
        history.append(dict(step=step, train_loss=r_loss/max(r_n, 1), lr=cur, **m))
        sps = EVAL_EVERY/(time.time()-t_last); t_last = time.time()
        print(f"step {step:6d} | loss {r_loss/max(r_n,1):.4f} | val pooled {m['pooled']:.4f} "
              f"@{m['th']} | per-img {m['perimg']:.4f} | Q0 {m['byq'][0]:.4f} "
              f"Q3 {m['byq'][3]:.4f} | best {best:.4f} | {sps:.2f} it/s | "
              f"ETA {(MAX_STEPS-step)/sps/3600:.1f}h{saved}"
              + (f"  [lr {prev:.1e}->{cur:.1e}]" if cur != prev else ""))
        r_loss = r_n = 0.0
        json.dump(dict(run=RUN, input=INPUT_MODE, loss=FG_LOSS, jitter=JITTER,
                       seed=TRAIN_SEED, floor=FLOOR_FRAC, select_on=SELECT_ON,
                       batch=BATCH, history=history, shares=shares),
                  open(f"{OUT}/history.json", "w"), indent=1)
    if step % CKPT_EVERY == 0:
        save_ckpt(); print(f"    checkpointed at step {step}")
save_ckpt()
print(f"\ntraining done in {(time.time()-t0)/3600:.2f}h, best val pooled {best:.4f}")

# ------------------- final: all three inference modes ------------------
net = ResUNet().to(DEV); net.load_state_dict(torch.load(BEST, map_location=DEV)); net.eval()
print("\n" + "=" * 76)
print("clean_val under each inference mode (best.pt was selected on "
      f"{SELECT_ON})")
print("=" * 76)
print(f"{'mode':12}{'th':>6}{'pooled':>9}{'per-img':>9}"
      f"{'Q0':>8}{'Q1':>8}{'Q2':>8}{'Q3':>8}{'Q3-Q0':>8}")
modes = ["whole384", "full768", "tiles384"]
final = {}
for md in modes:
    m = evaluate(net, md); final[md] = m
    print(f"{md:12}{m['th']:>6.2f}{m['pooled']:>9.4f}{m['perimg']:>9.4f}"
          + "".join(f"{m['byq'][q]:>8.4f}" for q in range(4))
          + f"{m['byq'][3]-m['byq'][0]:>8.4f}")

# the stratified check: does any Q0 gain track the realized detail gain?
res = {k: tuple(int(v) for v in meta[k]["Resolution"].lower().split("x")[:2])
       for k in CLEAN_VAL}
gain = np.array([min(min(res[s]), FULL_RES)/TRAIN_RES for s in CLEAN_VAL])
q0 = QV == 0
print(f"\nclean_val Q0 split by realized worst-axis detail gain at {FULL_RES}:")
for lab, sel in (("<= 1.0 (no gain)", q0 & (gain <= 1.0)),
                 ("1.0 - 1.5", q0 & (gain > 1.0) & (gain < 1.5)),
                 (">= 1.5", q0 & (gain >= 1.5))):
    print(f"   {lab:18} n={int(sel.sum()):4d}")
print("   per-image Q0 IoU within each band is in the saved json; compare it")
print("   against the matching S-384 run, paired by image.")

np.save(f"{OUT}/clean_val_gain.npy", gain)
json.dump(dict(run=RUN, input=INPUT_MODE, loss=FG_LOSS, jitter=JITTER, seed=TRAIN_SEED,
               floor=FLOOR_FRAC, select_on=SELECT_ON, batch=BATCH, best_val_pooled=best,
               clean_val=CLEAN_VAL, quartiles=QV.tolist(), final=final,
               history=history, shares=shares),
          open(f"{OUT}/summary.json", "w"), indent=1)
xs = [h["step"] for h in history]
plt.figure(figsize=(14, 4))
plt.subplot(1, 3, 1); plt.plot(xs, [h["train_loss"] for h in history]); plt.title("train loss")
plt.subplot(1, 3, 2); plt.plot(xs, [h["pooled"] for h in history]); plt.title("val pooled IoU")
plt.subplot(1, 3, 3)
for q in range(4): plt.plot(xs, [h["byq"][q] for h in history], label=f"Q{q}")
plt.legend(fontsize=7); plt.title("val per-image IoU by quartile")
plt.tight_layout(); plt.savefig(f"{OUT}/curves.png", dpi=110); plt.close()
print(f"\nwrote {OUT}/summary.json, best.pt, curves.png")
print(">>> download best.pt to local disk before the session ends <<<")
```

```python
# =====================================================================
#  cell F -- run this ONCE after every cell D training. Standalone: it
#  reads nothing from cell D's variables, so it also works in a fresh
#  session or after a runtime restart.
#
#  For the target run it:
#    1. scores it per image on clean_val  -> per_image.npz
#       (the vectors every paired test needs; cell D saves only means)
#    2. archives summary / history / curves / gain / per_image to Drive
#    3. records config, GPU and a checksum in weights_manifest.json
#    4. deletes ckpt.pt, but ONLY if the run completed. 490 MB each, and
#       19 of them would overrun a free Drive.
#    5. prints what Drive is holding
#    6. downloads the archive and best.pt
#
#  Any OTHER run still missing per_image.npz or an archive is also
#  brought up to date, without being downloaded again.
# =====================================================================
import os, csv, json, glob, time, zipfile, shutil, hashlib
import numpy as np
from PIL import Image
import torch, torch.nn as nn, torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models
import torchvision.transforms.functional as TF

RUN_NAME    = None       # None = the most recently finished run in stage2/
DELETE_CKPT = True       # False if you intend to extend a run later
DOWNLOAD    = True
SAVE, C768  = "/content/drive/MyDrive", "/content/cache768"
STAGE, ARCH = f"{SAVE}/stage2", f"{SAVE}/stage2_archive"
TRAIN_RES, FULL_RES = 384, 768
BATCH_EVAL  = 12 if torch.cuda.is_available() else 2
FROZEN      = [0.059468, 0.138448, 0.300831]
try:
    from google.colab import drive; drive.mount("/content/drive", force_remount=False)
except Exception: pass
DEV = torch.device("cuda" if torch.cuda.is_available() else "cpu")
os.makedirs(ARCH, exist_ok=True)

# ------------------------------- context -------------------------------
SP   = json.load(open(f"{SAVE}/plantseg_splits.json"))
I768 = json.load(open(f"{C768}/index.json"))
row  = I768["row_of"]
CLEAN_VAL = sorted(set(SP["clean_val"]) & set(I768["stems"]))
DSDIR = "/content/my_images/plantseg"
meta = {os.path.splitext(r["Name"])[0]: r for r in csv.DictReader(
        open(f"{DSDIR}/Metadata.csv", encoding="utf-8-sig", newline=""))}
frac = {k: float(v["Mask ratio"]) for k, v in meta.items()}
def quart(s):
    f = frac[s]
    return 0 if f <= FROZEN[0] else 1 if f <= FROZEN[1] else 2 if f <= FROZEN[2] else 3
QV = np.array([quart(s) for s in CLEAN_VAL])

MEAN, STD = [0.485,0.456,0.406], [0.229,0.224,0.225]
NORM = transforms.Normalize(MEAN, STD)
_mm = {}
def cache():
    if not _mm:
        _mm["i"] = np.load(I768["img_path"], mmap_mode="r")
        _mm["m"] = np.load(I768["msk_path"], mmap_mode="r")
    return _mm["i"], _mm["m"]
def strip_detail(im):
    return im.resize((TRAIN_RES,)*2, Image.BILINEAR).resize((FULL_RES,)*2, Image.BILINEAR)

class EvalSet(Dataset):
    def __init__(s, st, up): s.st, s.up = list(st), up
    def __len__(s): return len(s.st)
    def __getitem__(s, i):
        I, M = cache(); r = row[s.st[i]]
        img = Image.fromarray(I[r])
        if s.up: img = strip_detail(img)
        y = torch.from_numpy((np.asarray(M[r]) > 0).astype(np.float32)).unsqueeze(0)
        return NORM(TF.to_tensor(img)), y

SPEC = dict(channels=(64,256,512,1024,2048), decoder=(512,256,128,64))
class DoubleConv(nn.Module):
    def __init__(s,i,o):
        super().__init__()
        s.net = nn.Sequential(nn.Conv2d(i,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True),
                              nn.Conv2d(o,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True))
    def forward(s,x): return s.net(x)
class Up(nn.Module):
    def __init__(s,i,k,o):
        super().__init__(); s.reduce=nn.Conv2d(i,i//2,1,bias=False); s.conv=DoubleConv(i//2+k,o)
    def forward(s,x,skip):
        x = F.interpolate(x, size=skip.shape[-2:], mode="bilinear", align_corners=False)
        return s.conv(torch.cat([s.reduce(x), skip], 1))
class ResUNet(nn.Module):
    def __init__(s):
        super().__init__()
        base = models.resnet50(weights=None)
        c0,c1,c2,c3,c4 = SPEC["channels"]; d1,d2,d3,d4 = SPEC["decoder"]
        L = list(base.children())
        s.stem=nn.Sequential(*L[:3]); s.pool=L[3]
        s.l1,s.l2,s.l3,s.l4 = L[4],L[5],L[6],L[7]
        s.u1,s.u2,s.u3,s.u4 = Up(c4,c3,d1),Up(d1,c2,d2),Up(d2,c1,d3),Up(d3,c0,d4)
        s.head = nn.Sequential(nn.Conv2d(d4,32,3,padding=1,bias=False),
                               nn.BatchNorm2d(32), nn.ReLU(True), nn.Conv2d(32,1,1))
    def forward(s,x):
        size=x.shape[-2:]
        x0=s.stem(x); x1=s.l1(s.pool(x0)); x2=s.l2(x1); x3=s.l3(x2); x4=s.l4(x3)
        d=s.u1(x4,x3); d=s.u2(d,x2); d=s.u3(d,x1); d=s.u4(d,x0)
        return s.head(F.interpolate(d,size=size,mode="bilinear",align_corners=False))

@torch.no_grad()
def prob_768(net, x, mode):
    if mode == "full768":
        return torch.sigmoid(net(x).float())
    if mode == "whole384":
        d = F.interpolate(x, size=(TRAIN_RES,)*2, mode="bilinear", align_corners=False)
        p = torch.sigmoid(net(d).float())
        return F.interpolate(p, size=(FULL_RES,)*2, mode="bilinear", align_corners=False)
    acc = torch.zeros(x.shape[0], 1, FULL_RES, FULL_RES, device=x.device)
    cnt = torch.zeros_like(acc)
    ORIG = (0, (FULL_RES-TRAIN_RES)//2, FULL_RES-TRAIN_RES)
    for oy in ORIG:
        for ox in ORIG:
            t = x[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES]
            acc[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += torch.sigmoid(net(t).float())
            cnt[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += 1
    return acc/cnt.clamp(min=1)

THS = [round(t, 2) for t in np.arange(0.10, 0.91, 0.05)]
@torch.no_grad()
def score_per_image(bp, mode, up):
    net = ResUNet().to(DEV); net.load_state_dict(torch.load(bp, map_location=DEV)); net.eval()
    dl = DataLoader(EvalSet(CLEAN_VAL, up), BATCH_EVAL, shuffle=False,
                    num_workers=min(2, os.cpu_count() or 1), pin_memory=DEV.type == "cuda")
    n = len(CLEAN_VAL)
    per = np.zeros((n, len(THS))); Ipx = np.zeros((n, len(THS)))
    P = np.zeros((n, len(THS))); G = np.zeros(n); k = 0
    for x, y in dl:
        x, y = x.to(DEV, non_blocking=True), y.to(DEV, non_blocking=True)
        with torch.amp.autocast("cuda", enabled=DEV.type == "cuda"):
            p = prob_768(net, x, mode)
        b = x.shape[0]; yf = y.view(b, -1); pf = p.view(b, -1)
        G[k:k+b] = yf.sum(1).cpu().numpy()
        for j, t in enumerate(THS):
            pr = (pf > t).float(); it = (pr*yf).sum(1); pp = pr.sum(1)
            Ipx[k:k+b, j] = it.cpu().numpy(); P[k:k+b, j] = pp.cpu().numpy()
            per[k:k+b, j] = (it/(pp+yf.sum(1)-it).clamp(min=1)).cpu().numpy()
        k += b
    pooled = Ipx.sum(0)/np.maximum(P.sum(0)+G.sum()-Ipx.sum(0), 1)
    j = int(np.argmax(pooled))
    del net; torch.cuda.empty_cache()
    return dict(th=THS[j], pooled=float(pooled[j]), per=per[:, j])

def sha(p, chunk=1 << 24):
    h = hashlib.sha256()
    with open(p, "rb") as f:
        while (b := f.read(chunk)): h.update(b)
    return h.hexdigest()[:16]

# -------------------------- which runs to handle -----------------------
# summary.json is written only at the END of cell D, so its presence is
# the completion test -- safer than trusting a step count.
done = sorted(glob.glob(f"{STAGE}/*/summary.json"), key=os.path.getmtime)
if not done:
    raise SystemExit(f"no completed run under {STAGE}")
target = RUN_NAME or os.path.basename(os.path.dirname(done[-1]))
print(f"target run: {target}   ({len(done)} completed run(s) on Drive)\n")

man_p = f"{SAVE}/weights_manifest.json"
man = json.load(open(man_p)) if os.path.exists(man_p) else {}

for sm in done:
    d = os.path.dirname(sm); name = os.path.basename(d)
    J = json.load(open(sm))
    bp, ck = f"{d}/best.pt", f"{d}/ckpt.pt"
    if not os.path.exists(bp):
        print(f"{name}: no best.pt, skipped"); continue
    is_target = (name == target)

    # 1. per-image vectors
    npz = f"{d}/per_image.npz"
    need = True
    if os.path.exists(npz):
        try:
            z = np.load(npz, allow_pickle=True); need = list(z["stems"]) != CLEAN_VAL
        except Exception: need = True
    if need:
        t = time.time()
        r = score_per_image(bp, J["select_on"], J["input"] == "crop_768_up")
        np.savez(npz, per=r["per"], th=r["th"], pooled=r["pooled"],
                 mode=J["select_on"], stems=np.array(CLEAN_VAL))
        print(f"{name}: scored per image in {time.time()-t:.0f}s  "
              f"(mode {J['select_on']}, th {r['th']}, pooled {r['pooled']:.4f})")
    else:
        print(f"{name}: per_image.npz already current")

    # 2. archive
    zp = f"{ARCH}/{name}_results.zip"
    if is_target or not os.path.exists(zp):
        with zipfile.ZipFile(zp, "w", zipfile.ZIP_DEFLATED) as z:
            for f in ("summary.json", "history.json", "curves.png",
                      "clean_val_gain.npy", "per_image.npz"):
                p = f"{d}/{f}"
                if os.path.exists(p): z.write(p, arcname=f"{name}/{f}")
        print(f"{name}: archive {os.path.getsize(zp)/1e6:.2f} MB")

    # 3. manifest
    if is_target or name not in man:
        man[name] = dict(file=f"{name}__best.pt", bytes=os.path.getsize(bp),
                         sha256_16=sha(bp), best_val_pooled=J.get("best_val_pooled"),
                         gpu=J.get("gpu", "not recorded"), batch=J.get("batch"),
                         input=J["input"], loss=J["loss"], jitter=J["jitter"],
                         seed=J["seed"], floor=J.get("floor"))

    # 4. prune -- only a COMPLETED run, which summary.json proves
    if DELETE_CKPT and os.path.exists(ck):
        mb = os.path.getsize(ck)/1e6
        os.remove(ck); print(f"{name}: removed ckpt.pt (freed {mb:.0f} MB)")

json.dump(man, open(man_p, "w"), indent=1)
print(f"\nmanifest: {man_p}  ({len(man)} run(s))")

# ------------------------------ drive usage ----------------------------
def dsz(p):
    return sum(os.path.getsize(os.path.join(r, f))
               for r, _, fs in os.walk(p) for f in fs) if os.path.isdir(p) else 0
tot = 0
print("\nDrive usage")
for lab, p in (("stage2", STAGE), ("stage2_archive", ARCH),
               ("runs (stage 1)", f"{SAVE}/runs"), ("cache768", f"{SAVE}/cache768")):
    s = dsz(p); tot += s
    print(f"   {lab:18}{s/1e9:7.2f} GB")
print(f"   {'TRACKED TOTAL':18}{tot/1e9:7.2f} GB"
      + ("   !! approaching a 15 GB free Drive" if tot/1e9 > 11 else ""))

# ------------------------------- download ------------------------------
if DOWNLOAD:
    d = f"{STAGE}/{target}"
    loc = f"/content/{target}__best.pt"
    if not os.path.exists(loc): shutil.copyfile(f"{d}/best.pt", loc)
    print(f"\n{target}  best.pt {os.path.getsize(loc)/1e6:.1f} MB  "
          f"sha256[:16] {man[target]['sha256_16']}")
    try:
        from google.colab import files
        files.download(f"{ARCH}/{target}_results.zip")
        time.sleep(2)
        files.download(loc)
        print("downloads queued. Verify locally:")
        print(f"   shasum -a 256 {target}__best.pt | cut -c1-16"
              f"   ->  {man[target]['sha256_16']}")
    except Exception as e:
        print(f"(browser download unavailable: {type(e).__name__})")
    print(f"also on Drive:\n   {ARCH}/{target}_results.zip\n   {d}/best.pt")
```

```python
# =====================================================================
#  cell E -- score every finished Stage 2 run per image, then compare.
#
#  Cell D saves per-quartile MEANS, which cannot support a paired test.
#  This reloads each best.pt and writes the per-image IoU vector over
#  clean_val, at that run's own selected mode and threshold. No retraining.
#
#  Then, against the S-384 baseline, for each other run:
#    - paired per-image differences over the same 688 images
#    - an image-level paired bootstrap on Q0  (the PRIMARY inference:
#      688 paired images carry far more power than a 3-seed t-interval,
#      which at 3 seeds needs a seed-sd below 0.008 to clear +0.02)
#    - the same split by realized detail gain, declared in advance
#    - the PILOT rule: kill only if the mean Q0 difference is <= 0.
#      Killing anything that misses the +0.02 gate at one seed discards
#      a real +0.02 effect half the time; the <= 0 rule, 10% of the time.
#
#  A single-seed result here is a FILTER, never a result. Nothing from
#  this cell goes in the paper without seeds 43 and 44.
# =====================================================================
import os, csv, json, glob, time
import numpy as np
from PIL import Image
import torch, torch.nn as nn, torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models
import torchvision.transforms.functional as TF

SAVE, C768 = "/content/drive/MyDrive", "/content/cache768"
STAGE   = f"{SAVE}/stage2"
BASE    = "S-384_seed42"           # the baseline every run is paired against
TRAIN_RES, FULL_RES = 384, 768
BATCH_EVAL = 4
NBOOT   = 10000
FROZEN  = [0.059468, 0.138448, 0.300831]
KILL_AT = 0.0                      # pilot rule, declared before looking
try:
    from google.colab import drive; drive.mount("/content/drive", force_remount=False)
except Exception: pass
DEV = torch.device("cuda" if torch.cuda.is_available() else "cpu")

SP = json.load(open(f"{SAVE}/plantseg_splits.json"))
I768 = json.load(open(f"{C768}/index.json"))
row = I768["row_of"]
CLEAN_VAL = sorted(set(SP["clean_val"]) & set(I768["stems"]))
DSDIR = "/content/my_images/plantseg"
meta = {os.path.splitext(r["Name"])[0]: r for r in csv.DictReader(
        open(f"{DSDIR}/Metadata.csv", encoding="utf-8-sig", newline=""))}
frac = {k: float(v["Mask ratio"]) for k, v in meta.items()}
def quart(s):
    f = frac[s]
    return 0 if f <= FROZEN[0] else 1 if f <= FROZEN[1] else 2 if f <= FROZEN[2] else 3
QV = np.array([quart(s) for s in CLEAN_VAL])
res = {k: tuple(int(v) for v in meta[k]["Resolution"].lower().split("x")[:2])
       for k in CLEAN_VAL}
GAIN = np.array([min(min(res[s]), FULL_RES)/TRAIN_RES for s in CLEAN_VAL])
print(f"clean_val {len(CLEAN_VAL)}  quartiles {[int((QV==q).sum()) for q in range(4)]}")

MEAN, STD = [0.485,0.456,0.406], [0.229,0.224,0.225]
NORM = transforms.Normalize(MEAN, STD)
_mm = {}
def cache():
    if not _mm:
        _mm["i"] = np.load(I768["img_path"], mmap_mode="r")
        _mm["m"] = np.load(I768["msk_path"], mmap_mode="r")
    return _mm["i"], _mm["m"]
def strip_detail(im):
    return im.resize((TRAIN_RES,)*2, Image.BILINEAR).resize((FULL_RES,)*2, Image.BILINEAR)

class EvalSet(Dataset):
    def __init__(s, st, up): s.st, s.up = list(st), up
    def __len__(s): return len(s.st)
    def __getitem__(s, i):
        I, M = cache(); r = row[s.st[i]]
        img = Image.fromarray(I[r])
        if s.up: img = strip_detail(img)
        y = torch.from_numpy((np.asarray(M[r]) > 0).astype(np.float32)).unsqueeze(0)
        return NORM(TF.to_tensor(img)), y

SPEC = dict(channels=(64,256,512,1024,2048), decoder=(512,256,128,64))
class DoubleConv(nn.Module):
    def __init__(s,i,o):
        super().__init__()
        s.net = nn.Sequential(nn.Conv2d(i,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True),
                              nn.Conv2d(o,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True))
    def forward(s,x): return s.net(x)
class Up(nn.Module):
    def __init__(s,i,k,o):
        super().__init__(); s.reduce=nn.Conv2d(i,i//2,1,bias=False); s.conv=DoubleConv(i//2+k,o)
    def forward(s,x,skip):
        x = F.interpolate(x, size=skip.shape[-2:], mode="bilinear", align_corners=False)
        return s.conv(torch.cat([s.reduce(x), skip], 1))
class ResUNet(nn.Module):
    def __init__(s):
        super().__init__()
        base = models.resnet50(weights=None)
        c0,c1,c2,c3,c4 = SPEC["channels"]; d1,d2,d3,d4 = SPEC["decoder"]
        L = list(base.children())
        s.stem=nn.Sequential(*L[:3]); s.pool=L[3]
        s.l1,s.l2,s.l3,s.l4 = L[4],L[5],L[6],L[7]
        s.u1,s.u2,s.u3,s.u4 = Up(c4,c3,d1),Up(d1,c2,d2),Up(d2,c1,d3),Up(d3,c0,d4)
        s.head = nn.Sequential(nn.Conv2d(d4,32,3,padding=1,bias=False),
                               nn.BatchNorm2d(32), nn.ReLU(True), nn.Conv2d(32,1,1))
    def forward(s,x):
        size=x.shape[-2:]
        x0=s.stem(x); x1=s.l1(s.pool(x0)); x2=s.l2(x1); x3=s.l3(x2); x4=s.l4(x3)
        d=s.u1(x4,x3); d=s.u2(d,x2); d=s.u3(d,x1); d=s.u4(d,x0)
        return s.head(F.interpolate(d,size=size,mode="bilinear",align_corners=False))

@torch.no_grad()
def prob_768(net, x, mode):
    if mode == "full768":
        return torch.sigmoid(net(x).float())
    if mode == "whole384":
        d = F.interpolate(x, size=(TRAIN_RES,)*2, mode="bilinear", align_corners=False)
        p = torch.sigmoid(net(d).float())
        return F.interpolate(p, size=(FULL_RES,)*2, mode="bilinear", align_corners=False)
    acc = torch.zeros(x.shape[0], 1, FULL_RES, FULL_RES, device=x.device)
    cnt = torch.zeros_like(acc)
    ORIG = (0, (FULL_RES-TRAIN_RES)//2, FULL_RES-TRAIN_RES)
    for oy in ORIG:
        for ox in ORIG:
            t = x[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES]
            acc[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += torch.sigmoid(net(t).float())
            cnt[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += 1
    return acc/cnt.clamp(min=1)

THS = [round(t, 2) for t in np.arange(0.10, 0.91, 0.05)]
@torch.no_grad()
def score(bp, mode, up):
    net = ResUNet().to(DEV); net.load_state_dict(torch.load(bp, map_location=DEV)); net.eval()
    dl = DataLoader(EvalSet(CLEAN_VAL, up), BATCH_EVAL, shuffle=False,
                    num_workers=min(2, os.cpu_count() or 1), pin_memory=DEV.type == "cuda")
    n = len(CLEAN_VAL)
    per = np.zeros((n, len(THS))); Ipx = np.zeros((n, len(THS)))
    P = np.zeros((n, len(THS))); G = np.zeros(n); k = 0
    for x, y in dl:
        x, y = x.to(DEV, non_blocking=True), y.to(DEV, non_blocking=True)
        with torch.amp.autocast("cuda", enabled=DEV.type == "cuda"):
            p = prob_768(net, x, mode)
        b = x.shape[0]; yf = y.view(b, -1); pf = p.view(b, -1)
        G[k:k+b] = yf.sum(1).cpu().numpy()
        for j, t in enumerate(THS):
            pr = (pf > t).float(); it = (pr*yf).sum(1); pp = pr.sum(1)
            Ipx[k:k+b, j] = it.cpu().numpy(); P[k:k+b, j] = pp.cpu().numpy()
            per[k:k+b, j] = (it/(pp+yf.sum(1)-it).clamp(min=1)).cpu().numpy()
        k += b
    pooled = Ipx.sum(0)/np.maximum(P.sum(0)+G.sum()-Ipx.sum(0), 1)
    j = int(np.argmax(pooled))
    del net; torch.cuda.empty_cache()
    return dict(th=THS[j], pooled=float(pooled[j]), per=per[:, j])

# ------------------------- score what is finished ----------------------
runs = {}
for sm in sorted(glob.glob(f"{STAGE}/*/summary.json")):
    d = os.path.dirname(sm); name = os.path.basename(d)
    if not os.path.exists(f"{d}/best.pt"):
        print(f"   {name}: no best.pt, skipped"); continue
    J = json.load(open(sm))
    npz = f"{d}/per_image.npz"
    if os.path.exists(npz):
        z = np.load(npz, allow_pickle=True)
        if list(z["stems"]) == CLEAN_VAL:
            runs[name] = dict(per=z["per"], th=float(z["th"]), pooled=float(z["pooled"]),
                              mode=str(z["mode"]), cfg=J); print(f"   {name}: cached"); continue
    t = time.time()
    r = score(f"{d}/best.pt", J["select_on"], J["input"] == "crop_768_up")
    np.savez(npz, per=r["per"], th=r["th"], pooled=r["pooled"],
             mode=J["select_on"], stems=np.array(CLEAN_VAL))
    runs[name] = dict(**r, mode=J["select_on"], cfg=J)
    print(f"   {name}: scored in {time.time()-t:.0f}s  "
          f"mode {J['select_on']} th {r['th']} pooled {r['pooled']:.4f}")

if BASE not in runs:
    raise SystemExit(f"baseline {BASE} not scored yet -- run it first")
print(f"\n{len(runs)} run(s) scored; baseline {BASE}")

# --------------------------- paired comparison -------------------------
rng = np.random.default_rng(0)
b_per = runs[BASE]["per"]; q0 = QV == 0
bands = {"Q0 all": q0, "Q0 gain<=1.0": q0 & (GAIN <= 1.0),
         "Q0 gain>=1.5": q0 & (GAIN >= 1.5), "all images": np.ones(len(QV), bool)}

def boot(d, sel):
    v = d[sel]
    if len(v) < 5: return float("nan"), float("nan"), float("nan")
    idx = rng.integers(0, len(v), (NBOOT, len(v)))
    bs = v[idx].mean(1)
    return float(v.mean()), float(np.percentile(bs, 2.5)), float(np.percentile(bs, 97.5))

print("\n" + "=" * 88)
print(f"paired per-image IoU difference vs {BASE}, clean_val, "
      f"{NBOOT} image-level bootstraps")
print("=" * 88)
out = {}
for name, r in runs.items():
    if name == BASE: continue
    d = r["per"] - b_per
    print(f"\n{name}   ({r['cfg']['input']} / {r['cfg']['loss']}"
          f"{' / jitter' if r['cfg']['jitter'] else ''})   "
          f"pooled {r['pooled']:.4f} vs {runs[BASE]['pooled']:.4f}")
    print(f"   {'band':16}{'n':>6}{'mean diff':>12}{'95% CI':>22}")
    row = {}
    for lab, sel in bands.items():
        m, lo, hi = boot(d, sel)
        star = "  *" if (lo > 0 or hi < 0) else ""
        print(f"   {lab:16}{int(sel.sum()):>6}{m:>12.4f}"
              f"   [{lo:+.4f}, {hi:+.4f}]{star}")
        row[lab] = dict(n=int(sel.sum()), mean=m, lo=lo, hi=hi)
    mq0 = row["Q0 all"]["mean"]
    verdict = ("KILL -- Q0 moves the wrong way" if mq0 <= KILL_AT else
               "KEEP -- run seeds 43 and 44")
    print(f"   PILOT ({'<=' if mq0 <= KILL_AT else '>'} {KILL_AT:+.2f}): {verdict}")
    if row["Q0 gain>=1.5"]["n"] >= 20 and row["Q0 gain<=1.0"]["n"] >= 20:
        sep = row["Q0 gain>=1.5"]["mean"] - row["Q0 gain<=1.0"]["mean"]
        print(f"   mechanism check: high-detail minus no-detail = {sep:+.4f}"
              f"   {'(consistent with resolution)' if sep > 0 else '(not resolution)'}")
        row["mechanism_sep"] = sep
    out[name] = row

print("\n" + "=" * 88)
print("* marks a CI excluding zero. That is an image-level interval: it says the")
print("difference is not sampling noise over these 688 images, NOT that it would")
print("survive another training seed. Only seeds 43 and 44 answer that, and the")
print("gate is judged on the three-seed paired mean, not on anything here.")
json.dump(dict(baseline=BASE, kill_at=KILL_AT, nboot=NBOOT, results=out),
          open(f"{SAVE}/stage2_compare.json", "w"), indent=1, default=float)
print(f"\nwrote {SAVE}/stage2_compare.json")
```

```python
# =====================================================================
#  cell D -- Stage 2: supervised tests of components 1 and 2.
#  One cell, three switches, 25,000 steps, resume-safe. Validation only;
#  the test split is not touched until Stage 5.
#
#      INPUT_MODE in {full_384, crop_768, crop_768_up}
#      FG_LOSS    in {pooled, class_balanced, image_balanced}
#      JITTER     in {False, True}
#
#  (full_384, pooled, False) is the S-384 baseline.
#
#  crop_768_up is the control the crop statistics made necessary: uniform
#  384 crops of a 768 image carry 1.50x the Q0 lesion density of the
#  whole image, because uniform crop ORIGINS over-sample the centre (a
#  centre pixel sits in 384 of 385 origins per axis, a corner pixel in 1).
#  So crop_768 changes resolution AND density at once. crop_768_up takes
#  the 768 image, downsamples it to 384 and upsamples it back, destroying
#  exactly the detail above 384 and nothing else, then crops identically.
#  crop_768 beating crop_768_up isolates resolution; a tie says the gain
#  is crop geometry.
#
#  Everything reads ONE cache: /content/cache768 (537 labelled + 846 val).
#  The 441 cache is not needed, so prep.py need not be run this session.
# =====================================================================
import os, csv, json, math, time, random, shutil, zipfile, collections
import numpy as np
import matplotlib; matplotlib.use("Agg")
import matplotlib.pyplot as plt
from PIL import Image, ImageFilter
import torch, torch.nn as nn, torch.optim as optim, torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models
import torchvision.transforms.functional as TF

# ------------------------- the three switches --------------------------
INPUT_MODE = "crop_768"        # full_384 | crop_768 | crop_768_up
FG_LOSS    = "pooled"          # pooled | class_balanced | image_balanced
JITTER     = False             # True only for S-768c-jit
TRAIN_SEED = 42
# -----------------------------------------------------------------------
FLOOR_FRAC = 0.01              # declared from check 0f: binds on 8.9% of
                               # non-empty Q0 crops, caps amplification 21.5x
SELECT_ON  = {"full_384": "whole384", "crop_768": "full768",
              "crop_768_up": "full768"}[INPUT_MODE]
MAX_STEPS  = 25000
EVAL_EVERY = 1000
SHARE_EVERY = 500              # per-quartile lesion-loss share
CKPT_EVERY = 2500
LR_PATIENCE = 3                # patience counts EVAL_EVERY blocks, now 1000
TRAIN_RES  = 384
FULL_RES   = 768
BATCH      = 4                 # matches the existing pipeline; batch 8
                               # doubles every run to ~2.2h and Stage 2 to ~42h
BATCH_EVAL = 4                 # 768 input on a T4
LR_ENCODER, LR_DECODER, WEIGHT_DECAY = 1e-5, 1e-4, 1e-4
FROZEN     = [0.059468, 0.138448, 0.300831]
SAVE, C768 = "/content/drive/MyDrive", "/content/cache768"
NUM_WORKERS = min(2, os.cpu_count() or 1)
FRESH      = True              # wipes this run's folder. Set back to False
                               # once it is past step 2500 so a dropped
                               # session resumes instead of restarting.

TAG = {"full_384": "384", "crop_768": "768c", "crop_768_up": "768up"}[INPUT_MODE]
if FG_LOSS == "class_balanced":  TAG += "-cb"
if FG_LOSS == "image_balanced":  TAG += "-bal"
if JITTER:                       TAG += "-jit"
RUN = f"S-{TAG}_seed{TRAIN_SEED}"
OUT = f"{SAVE}/stage2/{RUN}"
if FRESH and os.path.isdir(OUT): shutil.rmtree(OUT, ignore_errors=True)
os.makedirs(OUT, exist_ok=True)

random.seed(TRAIN_SEED); np.random.seed(TRAIN_SEED)
torch.manual_seed(TRAIN_SEED); torch.cuda.manual_seed_all(TRAIN_SEED)
torch.backends.cudnn.deterministic = True; torch.backends.cudnn.benchmark = False
DEV = torch.device("cuda" if torch.cuda.is_available() else "cpu")
try:
    from google.colab import drive; drive.mount("/content/drive", force_remount=False)
except Exception: pass
print(f"{RUN}\ndevice {DEV} {torch.cuda.get_device_name(0) if DEV.type=='cuda' else ''}")
print(f"input {INPUT_MODE} | loss {FG_LOSS} | jitter {JITTER} | "
      f"select on {SELECT_ON} | floor {FLOOR_FRAC} | batch {BATCH}")

# --------------------------- data and asserts --------------------------
SP = json.load(open(f"{SAVE}/plantseg_splits.json"))
if max(abs(a-b) for a, b in zip(SP["quartile_cuts"], FROZEN)) > 1e-6:
    raise SystemExit(f"cuts {SP['quartile_cuts']} != frozen {FROZEN}")
if {k: len(v) for k, v in SP["official_splits"].items()} != \
   {"train": 5367, "val": 846, "test": 1561}:
    raise SystemExit("split sizes are not 5367/846/1561")
I768 = json.load(open(f"{C768}/index.json"))
if I768["load_size"] != FULL_RES:
    raise SystemExit(f"cache is {I768['load_size']}, expected {FULL_RES}")
row = I768["row_of"]
LAB = sorted(I768["labelled"])
CLEAN_VAL = sorted(set(SP["clean_val"]) & set(I768["stems"]))
if len(LAB) != 537: raise SystemExit(f"{len(LAB)} labelled, expected 537")
DSDIR = "/content/my_images/plantseg"
meta = {os.path.splitext(r["Name"])[0]: r for r in csv.DictReader(
        open(f"{DSDIR}/Metadata.csv", encoding="utf-8-sig", newline=""))}
frac = {k: float(v["Mask ratio"]) for k, v in meta.items()}
def quart(s):
    f = frac[s]
    return 0 if f <= FROZEN[0] else 1 if f <= FROZEN[1] else 2 if f <= FROZEN[2] else 3
QV = np.array([quart(s) for s in CLEAN_VAL])
print(f"labelled {len(LAB)}   clean_val {len(CLEAN_VAL)} "
      f"(quartiles {[int((QV==q).sum()) for q in range(4)]})")
print("   gates are judged on clean_val, not the full 846: full val carries")
print("   17.14% near-duplicates of training images.")

MEAN, STD = [0.485,0.456,0.406], [0.229,0.224,0.225]
NORM = transforms.Normalize(MEAN, STD)
JIT  = transforms.ColorJitter(0.4, 0.4, 0.4, 0.1)
_mm = {}
def cache():
    if not _mm:
        _mm["i"] = np.load(I768["img_path"], mmap_mode="r")
        _mm["m"] = np.load(I768["msk_path"], mmap_mode="r")
    return _mm["i"], _mm["m"]

def strip_detail(im):
    """768 -> 384 -> 768. Destroys exactly the detail above 384, keeps
       geometry, size and lesion density identical."""
    return im.resize((TRAIN_RES,)*2, Image.BILINEAR).resize((FULL_RES,)*2, Image.BILINEAR)

def photometric(img):
    if random.random() < 0.8: img = JIT(img)
    if random.random() < 0.2: img = TF.to_grayscale(img, num_output_channels=3)
    if random.random() < 0.3:
        img = img.filter(ImageFilter.GaussianBlur(radius=random.uniform(0.1, 1.5)))
    return img

def geometric(img, msk):
    if random.random() < 0.5: img, msk = TF.hflip(img), TF.hflip(msk)
    if random.random() < 0.5: img, msk = TF.vflip(img), TF.vflip(msk)
    k = random.randint(0, 3)
    if k:
        img = img.rotate(90*k, expand=True, resample=Image.BILINEAR)
        msk = msk.rotate(90*k, expand=True, resample=Image.NEAREST)
    return img, msk

def rrc(img, msk, scale=(0.5, 1.0), ratio=(3/4, 4/3)):
    w, h = img.size; area = w*h
    for _ in range(10):
        t = area*random.uniform(*scale)
        ar = math.exp(random.uniform(math.log(ratio[0]), math.log(ratio[1])))
        cw, ch = int(round(math.sqrt(t*ar))), int(round(math.sqrt(t/ar)))
        if cw <= w and ch <= h:
            x, y = random.randint(0, w-cw), random.randint(0, h-ch)
            return img.crop((x, y, x+cw, y+ch)), msk.crop((x, y, x+cw, y+ch))
    return img, msk

class Labelled(Dataset):
    """full_384   : whole image, rrc + geometric, resized to 384
       crop_768   : uniform 384 window of the 768 image (the crop IS the
                    spatial augmentation, so rrc is not applied on top)
       crop_768_up: the same window of a detail-stripped 768 image"""
    def __init__(s, st): s.st = list(st)
    def __len__(s): return len(s.st)
    def __getitem__(s, i):
        I, M = cache(); r = row[s.st[i]]
        img = Image.fromarray(I[r]); msk = Image.fromarray(M[r])
        if INPUT_MODE == "crop_768_up": img = strip_detail(img)
        if INPUT_MODE == "full_384":
            img, msk = rrc(img, msk)
            img, msk = geometric(img, msk)
            img = img.resize((TRAIN_RES,)*2, Image.BILINEAR)
            msk = msk.resize((TRAIN_RES,)*2, Image.NEAREST)
        else:
            side = (random.randint(256, FULL_RES) if JITTER else TRAIN_RES)
            x = random.randint(0, FULL_RES-side); y = random.randint(0, FULL_RES-side)
            img = img.crop((x, y, x+side, y+side)); msk = msk.crop((x, y, x+side, y+side))
            img, msk = geometric(img, msk)
            if side != TRAIN_RES:
                img = img.resize((TRAIN_RES,)*2, Image.BILINEAR)
                msk = msk.resize((TRAIN_RES,)*2, Image.NEAREST)
        y_ = torch.from_numpy((np.array(msk) > 0).astype(np.float32)).unsqueeze(0)
        return NORM(TF.to_tensor(photometric(img))), y_, quart(s.st[i])

class EvalSet(Dataset):
    """input follows INPUT_MODE; the TARGET is always the 768 mask."""
    def __init__(s, st): s.st = list(st)
    def __len__(s): return len(s.st)
    def __getitem__(s, i):
        I, M = cache(); r = row[s.st[i]]
        img = Image.fromarray(I[r])
        if INPUT_MODE == "crop_768_up": img = strip_detail(img)
        y = torch.from_numpy((np.asarray(M[r]) > 0).astype(np.float32)).unsqueeze(0)
        return NORM(TF.to_tensor(img)), y            # image at 768, mask at 768

def seed_worker(_):
    s = torch.initial_seed() % 2**32; np.random.seed(s); random.seed(s)
LK = dict(num_workers=NUM_WORKERS, pin_memory=DEV.type == "cuda",
          worker_init_fn=seed_worker, persistent_workers=NUM_WORKERS > 0)
g = torch.Generator(); g.manual_seed(TRAIN_SEED)
loader_l = DataLoader(Labelled(LAB), BATCH, shuffle=True, drop_last=True,
                      generator=g, **LK)
val_loader = DataLoader(EvalSet(CLEAN_VAL), BATCH_EVAL, shuffle=False, **LK)
def cycle(dl):
    while True:
        for b in dl: yield b

# ------------------------------- model ---------------------------------
SPEC = dict(channels=(64,256,512,1024,2048), decoder=(512,256,128,64))
class DoubleConv(nn.Module):
    def __init__(s,i,o):
        super().__init__()
        s.net = nn.Sequential(nn.Conv2d(i,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True),
                              nn.Conv2d(o,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True))
    def forward(s,x): return s.net(x)
class Up(nn.Module):
    def __init__(s,i,k,o):
        super().__init__(); s.reduce=nn.Conv2d(i,i//2,1,bias=False); s.conv=DoubleConv(i//2+k,o)
    def forward(s,x,skip):
        x = F.interpolate(x, size=skip.shape[-2:], mode="bilinear", align_corners=False)
        return s.conv(torch.cat([s.reduce(x), skip], 1))
class ResUNet(nn.Module):
    def __init__(s):
        super().__init__()
        base = models.resnet50(weights=models.ResNet50_Weights.DEFAULT)
        c0,c1,c2,c3,c4 = SPEC["channels"]; d1,d2,d3,d4 = SPEC["decoder"]
        L = list(base.children())
        s.stem=nn.Sequential(*L[:3]); s.pool=L[3]
        s.l1,s.l2,s.l3,s.l4 = L[4],L[5],L[6],L[7]
        s.u1,s.u2,s.u3,s.u4 = Up(c4,c3,d1),Up(d1,c2,d2),Up(d2,c1,d3),Up(d3,c0,d4)
        s.head = nn.Sequential(nn.Conv2d(d4,32,3,padding=1,bias=False),
                               nn.BatchNorm2d(32), nn.ReLU(True), nn.Conv2d(32,1,1))
    def enc_params(s):
        for m in (s.stem,s.l1,s.l2,s.l3,s.l4): yield from m.parameters()
    def dec_params(s):
        for m in (s.u1,s.u2,s.u3,s.u4,s.head): yield from m.parameters()
    def forward(s,x):
        size=x.shape[-2:]
        x0=s.stem(x); x1=s.l1(s.pool(x0)); x2=s.l2(x1); x3=s.l3(x2); x4=s.l4(x3)
        d=s.u1(x4,x3); d=s.u2(d,x2); d=s.u3(d,x1); d=s.u4(d,x0)
        return s.head(F.interpolate(d,size=size,mode="bilinear",align_corners=False))

# ------------------------------- losses --------------------------------
bce_none = nn.BCEWithLogitsLoss(reduction="none")
def tversky(logits, tgt, eps=1.0):
    p = torch.sigmoid(logits)
    tp = (p*tgt).sum((1,2,3)); fp = (p*(1-tgt)).sum((1,2,3)); fn = ((1-p)*tgt).sum((1,2,3))
    return 1 - (tp+eps)/(tp+0.5*fp+0.5*fn+eps)

def fg_term(logits, tgt):
    """returns (per-image BCE term, per-image lesion-only contribution).
       All three forms are scaled so a uniform per-pixel loss c gives c,
       so the learning rate and the BCE/Tversky balance do not shift
       between configurations. The plan's formula writes the sum of the
       two class means; using their average is the same comparison at the
       same scale."""
    e = bce_none(logits, tgt)
    sf = tgt.sum((1,2,3)); sb = (1-tgt).sum((1,2,3))
    ef = (e*tgt).sum((1,2,3)); eb = (e*(1-tgt)).sum((1,2,3))
    if FG_LOSS == "pooled":
        return e.mean((1,2,3)), ef/tgt[0].numel()
    if FG_LOSS == "class_balanced":                 # batch-level class means
        n = logits.shape[0]
        mf = ef.sum()/sf.sum().clamp(min=1.0); mb = eb.sum()/sb.sum().clamp(min=1.0)
        return (0.5*(mf+mb)).expand(n), 0.5*ef/sf.sum().clamp(min=1.0)
    floor = FLOOR_FRAC*tgt[0].numel()               # per-image, with the floor
    tf = ef/torch.clamp(sf, min=floor)
    tb = eb/sb.clamp(min=1.0)
    return 0.5*(tf+tb), 0.5*tf

def sup_loss(logits, tgt):
    b, lesion = fg_term(logits, tgt)
    return b + tversky(logits, tgt), lesion

# --------------------------- 768 inference -----------------------------
@torch.no_grad()
def prob_768(net, x, mode):
    """x is the model's input image at 768. Returns a 768 probability map."""
    if mode == "full768":
        return torch.sigmoid(net(x).float())
    if mode == "whole384":
        d = F.interpolate(x, size=(TRAIN_RES,)*2, mode="bilinear", align_corners=False)
        p = torch.sigmoid(net(d).float())
        return F.interpolate(p, size=(FULL_RES,)*2, mode="bilinear", align_corners=False)
    # genuinely overlapping: stride 192 gives 3x3 tiles at 50% overlap.
    # Origins (0, 384) alone would ABUT, not overlap, and blending would
    # be a no-op -- the seam is exactly where a tiled model is weakest.
    acc = torch.zeros(x.shape[0], 1, FULL_RES, FULL_RES, device=x.device)
    cnt = torch.zeros_like(acc)
    ORIG = (0, (FULL_RES-TRAIN_RES)//2, FULL_RES-TRAIN_RES)
    for oy in ORIG:
        for ox in ORIG:
            t = x[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES]
            acc[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += torch.sigmoid(net(t).float())
            cnt[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += 1
    return acc/cnt.clamp(min=1)                      # blend, threshold afterwards

THS = [round(t, 2) for t in np.arange(0.10, 0.91, 0.05)]
@torch.no_grad()
def evaluate(net, mode):
    net.eval()
    nI = len(CLEAN_VAL); per = np.zeros((nI, len(THS))); P = np.zeros((nI, len(THS)))
    Ipx = np.zeros((nI, len(THS))); G = np.zeros(nI); k = 0
    for x, y in val_loader:
        x, y = x.to(DEV, non_blocking=True), y.to(DEV, non_blocking=True)
        with torch.amp.autocast("cuda", enabled=DEV.type == "cuda"):
            p = prob_768(net, x, mode)
        b = x.shape[0]; yf = y.view(b, -1)
        G[k:k+b] = yf.sum(1).cpu().numpy()
        pf = p.view(b, -1)
        for j, t in enumerate(THS):
            pr = (pf > t).float()
            it = (pr*yf).sum(1); pp = pr.sum(1)
            Ipx[k:k+b, j] = it.cpu().numpy(); P[k:k+b, j] = pp.cpu().numpy()
            per[k:k+b, j] = (it/(pp+yf.sum(1)-it).clamp(min=1)).cpu().numpy()
        k += b
    pooled = Ipx.sum(0)/np.maximum(P.sum(0)+G.sum()-Ipx.sum(0), 1)
    j = int(np.argmax(pooled)); t = THS[j]
    byq = {q: float(per[QV == q, j].mean()) for q in range(4)}
    return dict(th=t, pooled=float(pooled[j]), perimg=float(per[:, j].mean()),
                byq=byq, pred_frac=float(P[:, j].sum()/(len(CLEAN_VAL)*FULL_RES*FULL_RES)))

# --------------------------- resume or start ---------------------------
model = ResUNet().to(DEV)
opt = optim.AdamW([{"params": list(model.enc_params()), "lr": LR_ENCODER},
                   {"params": list(model.dec_params()), "lr": LR_DECODER}],
                  weight_decay=WEIGHT_DECAY)
sched = optim.lr_scheduler.ReduceLROnPlateau(opt, mode="max", patience=LR_PATIENCE, factor=0.5)
scaler = torch.amp.GradScaler("cuda")
step, best, history, shares = 0, -1.0, [], []
CK, BEST = f"{OUT}/ckpt.pt", f"{OUT}/best.pt"
if os.path.exists(CK):
    c = torch.load(CK, map_location=DEV, weights_only=False)
    model.load_state_dict(c["model"]); opt.load_state_dict(c["opt"])
    sched.load_state_dict(c["sched"]); scaler.load_state_dict(c["scaler"])
    step, best, history, shares = c["step"], c["best"], c["history"], c.get("shares", [])
    random.setstate(c["py_rng"]); np.random.set_state(c["np_rng"])
    torch.set_rng_state(c["t_rng"].cpu())
    if c.get("cuda_rng") is not None and DEV.type == "cuda":
        torch.cuda.set_rng_state(c["cuda_rng"].cpu())
    print(f"RESUMED from step {step} (best {best:.4f})")
else:
    print("starting fresh")

def save_ckpt():
    torch.save(dict(model=model.state_dict(), opt=opt.state_dict(), sched=sched.state_dict(),
                    scaler=scaler.state_dict(), step=step, best=best, history=history,
                    shares=shares, py_rng=random.getstate(), np_rng=np.random.get_state(),
                    t_rng=torch.get_rng_state(),
                    cuda_rng=torch.cuda.get_rng_state() if DEV.type == "cuda" else None), CK)

# ------------------------------- train ---------------------------------
it = cycle(loader_l)
print("=" * 76); print(f"{RUN}  --  {len(LAB)} labelled, {MAX_STEPS} steps"); print("=" * 76)
t0 = t_last = time.time()
r_loss = r_n = 0.0; q_les = np.zeros(4); q_cnt = np.zeros(4)
while step < MAX_STEPS:
    model.train()
    x, y, q = next(it)
    x, y = x.to(DEV, non_blocking=True), y.to(DEV, non_blocking=True)
    opt.zero_grad(set_to_none=True)
    with torch.amp.autocast("cuda", enabled=DEV.type == "cuda"):
        lg = model(x)
        vec, lesion = sup_loss(lg.float(), y)
        loss = vec.mean()
    scaler.scale(loss).backward(); scaler.step(opt); scaler.update()
    step += 1
    r_loss += loss.item(); r_n += 1
    lz = lesion.detach().float().cpu().numpy(); qz = q.numpy()
    for b in range(len(qz)): q_les[qz[b]] += lz[b]; q_cnt[qz[b]] += 1

    if step % SHARE_EVERY == 0:
        tot = max(q_les.sum(), 1e-12)
        sh = (q_les/tot).round(4).tolist()
        shares.append(dict(step=step, share=sh, seen=q_cnt.astype(int).tolist()))
        print(f"   [step {step}] lesion-loss share by quartile {sh}"
              f"  (images seen {q_cnt.astype(int).tolist()})")
        q_les[:] = 0; q_cnt[:] = 0

    if step % EVAL_EVERY == 0:
        m = evaluate(model, SELECT_ON)
        prev = opt.param_groups[-1]["lr"]; sched.step(m["pooled"])
        cur = opt.param_groups[-1]["lr"]
        saved = ""
        if m["pooled"] > best:
            best = m["pooled"]; torch.save(model.state_dict(), BEST); saved = "  <-- saved"
        history.append(dict(step=step, train_loss=r_loss/max(r_n, 1), lr=cur, **m))
        sps = EVAL_EVERY/(time.time()-t_last); t_last = time.time()
        print(f"step {step:6d} | loss {r_loss/max(r_n,1):.4f} | val pooled {m['pooled']:.4f} "
              f"@{m['th']} | per-img {m['perimg']:.4f} | Q0 {m['byq'][0]:.4f} "
              f"Q3 {m['byq'][3]:.4f} | best {best:.4f} | {sps:.2f} it/s | "
              f"ETA {(MAX_STEPS-step)/sps/3600:.1f}h{saved}"
              + (f"  [lr {prev:.1e}->{cur:.1e}]" if cur != prev else ""))
        r_loss = r_n = 0.0
        json.dump(dict(run=RUN, input=INPUT_MODE, loss=FG_LOSS, jitter=JITTER,
                       seed=TRAIN_SEED, floor=FLOOR_FRAC, select_on=SELECT_ON,
                       batch=BATCH, history=history, shares=shares),
                  open(f"{OUT}/history.json", "w"), indent=1)
    if step % CKPT_EVERY == 0:
        save_ckpt(); print(f"    checkpointed at step {step}")
save_ckpt()
print(f"\ntraining done in {(time.time()-t0)/3600:.2f}h, best val pooled {best:.4f}")

# ------------------- final: all three inference modes ------------------
net = ResUNet().to(DEV); net.load_state_dict(torch.load(BEST, map_location=DEV)); net.eval()
print("\n" + "=" * 76)
print("clean_val under each inference mode (best.pt was selected on "
      f"{SELECT_ON})")
print("=" * 76)
print(f"{'mode':12}{'th':>6}{'pooled':>9}{'per-img':>9}"
      f"{'Q0':>8}{'Q1':>8}{'Q2':>8}{'Q3':>8}{'Q3-Q0':>8}")
modes = ["whole384", "full768", "tiles384"]
final = {}
for md in modes:
    m = evaluate(net, md); final[md] = m
    print(f"{md:12}{m['th']:>6.2f}{m['pooled']:>9.4f}{m['perimg']:>9.4f}"
          + "".join(f"{m['byq'][q]:>8.4f}" for q in range(4))
          + f"{m['byq'][3]-m['byq'][0]:>8.4f}")

# the stratified check: does any Q0 gain track the realized detail gain?
res = {k: tuple(int(v) for v in meta[k]["Resolution"].lower().split("x")[:2])
       for k in CLEAN_VAL}
gain = np.array([min(min(res[s]), FULL_RES)/TRAIN_RES for s in CLEAN_VAL])
q0 = QV == 0
print(f"\nclean_val Q0 split by realized worst-axis detail gain at {FULL_RES}:")
for lab, sel in (("<= 1.0 (no gain)", q0 & (gain <= 1.0)),
                 ("1.0 - 1.5", q0 & (gain > 1.0) & (gain < 1.5)),
                 (">= 1.5", q0 & (gain >= 1.5))):
    print(f"   {lab:18} n={int(sel.sum()):4d}")
print("   per-image Q0 IoU within each band is in the saved json; compare it")
print("   against the matching S-384 run, paired by image.")

np.save(f"{OUT}/clean_val_gain.npy", gain)
json.dump(dict(run=RUN, input=INPUT_MODE, loss=FG_LOSS, jitter=JITTER, seed=TRAIN_SEED,
               floor=FLOOR_FRAC, select_on=SELECT_ON, batch=BATCH, best_val_pooled=best,
               clean_val=CLEAN_VAL, quartiles=QV.tolist(), final=final,
               history=history, shares=shares),
          open(f"{OUT}/summary.json", "w"), indent=1)
xs = [h["step"] for h in history]
plt.figure(figsize=(14, 4))
plt.subplot(1, 3, 1); plt.plot(xs, [h["train_loss"] for h in history]); plt.title("train loss")
plt.subplot(1, 3, 2); plt.plot(xs, [h["pooled"] for h in history]); plt.title("val pooled IoU")
plt.subplot(1, 3, 3)
for q in range(4): plt.plot(xs, [h["byq"][q] for h in history], label=f"Q{q}")
plt.legend(fontsize=7); plt.title("val per-image IoU by quartile")
plt.tight_layout(); plt.savefig(f"{OUT}/curves.png", dpi=110); plt.close()
print(f"\nwrote {OUT}/summary.json, best.pt, curves.png")
print(">>> download best.pt to local disk before the session ends <<<")
```

```python
# =====================================================================
#  cell F -- run this ONCE after every cell D training. Standalone: it
#  reads nothing from cell D's variables, so it also works in a fresh
#  session or after a runtime restart.
#
#  For the target run it:
#    1. scores it per image on clean_val  -> per_image.npz
#       (the vectors every paired test needs; cell D saves only means)
#    2. archives summary / history / curves / gain / per_image to Drive
#    3. records config, GPU and a checksum in weights_manifest.json
#    4. deletes ckpt.pt, but ONLY if the run completed. 490 MB each, and
#       19 of them would overrun a free Drive.
#    5. prints what Drive is holding
#    6. downloads the archive and best.pt
#
#  Any OTHER run still missing per_image.npz or an archive is also
#  brought up to date, without being downloaded again.
# =====================================================================
import os, csv, json, glob, time, zipfile, shutil, hashlib
import numpy as np
from PIL import Image
import torch, torch.nn as nn, torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models
import torchvision.transforms.functional as TF

RUN_NAME    = None       # None = the most recently finished run in stage2/
DELETE_CKPT = True       # False if you intend to extend a run later
DOWNLOAD    = True
SAVE, C768  = "/content/drive/MyDrive", "/content/cache768"
STAGE, ARCH = f"{SAVE}/stage2", f"{SAVE}/stage2_archive"
TRAIN_RES, FULL_RES = 384, 768
BATCH_EVAL  = 12 if torch.cuda.is_available() else 2
FROZEN      = [0.059468, 0.138448, 0.300831]
try:
    from google.colab import drive; drive.mount("/content/drive", force_remount=False)
except Exception: pass
DEV = torch.device("cuda" if torch.cuda.is_available() else "cpu")
os.makedirs(ARCH, exist_ok=True)

# ------------------------------- context -------------------------------
SP   = json.load(open(f"{SAVE}/plantseg_splits.json"))
I768 = json.load(open(f"{C768}/index.json"))
row  = I768["row_of"]
CLEAN_VAL = sorted(set(SP["clean_val"]) & set(I768["stems"]))
DSDIR = "/content/my_images/plantseg"
meta = {os.path.splitext(r["Name"])[0]: r for r in csv.DictReader(
        open(f"{DSDIR}/Metadata.csv", encoding="utf-8-sig", newline=""))}
frac = {k: float(v["Mask ratio"]) for k, v in meta.items()}
def quart(s):
    f = frac[s]
    return 0 if f <= FROZEN[0] else 1 if f <= FROZEN[1] else 2 if f <= FROZEN[2] else 3
QV = np.array([quart(s) for s in CLEAN_VAL])

MEAN, STD = [0.485,0.456,0.406], [0.229,0.224,0.225]
NORM = transforms.Normalize(MEAN, STD)
_mm = {}
def cache():
    if not _mm:
        _mm["i"] = np.load(I768["img_path"], mmap_mode="r")
        _mm["m"] = np.load(I768["msk_path"], mmap_mode="r")
    return _mm["i"], _mm["m"]
def strip_detail(im):
    return im.resize((TRAIN_RES,)*2, Image.BILINEAR).resize((FULL_RES,)*2, Image.BILINEAR)

class EvalSet(Dataset):
    def __init__(s, st, up): s.st, s.up = list(st), up
    def __len__(s): return len(s.st)
    def __getitem__(s, i):
        I, M = cache(); r = row[s.st[i]]
        img = Image.fromarray(I[r])
        if s.up: img = strip_detail(img)
        y = torch.from_numpy((np.asarray(M[r]) > 0).astype(np.float32)).unsqueeze(0)
        return NORM(TF.to_tensor(img)), y

SPEC = dict(channels=(64,256,512,1024,2048), decoder=(512,256,128,64))
class DoubleConv(nn.Module):
    def __init__(s,i,o):
        super().__init__()
        s.net = nn.Sequential(nn.Conv2d(i,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True),
                              nn.Conv2d(o,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True))
    def forward(s,x): return s.net(x)
class Up(nn.Module):
    def __init__(s,i,k,o):
        super().__init__(); s.reduce=nn.Conv2d(i,i//2,1,bias=False); s.conv=DoubleConv(i//2+k,o)
    def forward(s,x,skip):
        x = F.interpolate(x, size=skip.shape[-2:], mode="bilinear", align_corners=False)
        return s.conv(torch.cat([s.reduce(x), skip], 1))
class ResUNet(nn.Module):
    def __init__(s):
        super().__init__()
        base = models.resnet50(weights=None)
        c0,c1,c2,c3,c4 = SPEC["channels"]; d1,d2,d3,d4 = SPEC["decoder"]
        L = list(base.children())
        s.stem=nn.Sequential(*L[:3]); s.pool=L[3]
        s.l1,s.l2,s.l3,s.l4 = L[4],L[5],L[6],L[7]
        s.u1,s.u2,s.u3,s.u4 = Up(c4,c3,d1),Up(d1,c2,d2),Up(d2,c1,d3),Up(d3,c0,d4)
        s.head = nn.Sequential(nn.Conv2d(d4,32,3,padding=1,bias=False),
                               nn.BatchNorm2d(32), nn.ReLU(True), nn.Conv2d(32,1,1))
    def forward(s,x):
        size=x.shape[-2:]
        x0=s.stem(x); x1=s.l1(s.pool(x0)); x2=s.l2(x1); x3=s.l3(x2); x4=s.l4(x3)
        d=s.u1(x4,x3); d=s.u2(d,x2); d=s.u3(d,x1); d=s.u4(d,x0)
        return s.head(F.interpolate(d,size=size,mode="bilinear",align_corners=False))

@torch.no_grad()
def prob_768(net, x, mode):
    if mode == "full768":
        return torch.sigmoid(net(x).float())
    if mode == "whole384":
        d = F.interpolate(x, size=(TRAIN_RES,)*2, mode="bilinear", align_corners=False)
        p = torch.sigmoid(net(d).float())
        return F.interpolate(p, size=(FULL_RES,)*2, mode="bilinear", align_corners=False)
    acc = torch.zeros(x.shape[0], 1, FULL_RES, FULL_RES, device=x.device)
    cnt = torch.zeros_like(acc)
    ORIG = (0, (FULL_RES-TRAIN_RES)//2, FULL_RES-TRAIN_RES)
    for oy in ORIG:
        for ox in ORIG:
            t = x[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES]
            acc[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += torch.sigmoid(net(t).float())
            cnt[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += 1
    return acc/cnt.clamp(min=1)

THS = [round(t, 2) for t in np.arange(0.10, 0.91, 0.05)]
@torch.no_grad()
def score_per_image(bp, mode, up):
    net = ResUNet().to(DEV); net.load_state_dict(torch.load(bp, map_location=DEV)); net.eval()
    dl = DataLoader(EvalSet(CLEAN_VAL, up), BATCH_EVAL, shuffle=False,
                    num_workers=min(2, os.cpu_count() or 1), pin_memory=DEV.type == "cuda")
    n = len(CLEAN_VAL)
    per = np.zeros((n, len(THS))); Ipx = np.zeros((n, len(THS)))
    P = np.zeros((n, len(THS))); G = np.zeros(n); k = 0
    for x, y in dl:
        x, y = x.to(DEV, non_blocking=True), y.to(DEV, non_blocking=True)
        with torch.amp.autocast("cuda", enabled=DEV.type == "cuda"):
            p = prob_768(net, x, mode)
        b = x.shape[0]; yf = y.view(b, -1); pf = p.view(b, -1)
        G[k:k+b] = yf.sum(1).cpu().numpy()
        for j, t in enumerate(THS):
            pr = (pf > t).float(); it = (pr*yf).sum(1); pp = pr.sum(1)
            Ipx[k:k+b, j] = it.cpu().numpy(); P[k:k+b, j] = pp.cpu().numpy()
            per[k:k+b, j] = (it/(pp+yf.sum(1)-it).clamp(min=1)).cpu().numpy()
        k += b
    pooled = Ipx.sum(0)/np.maximum(P.sum(0)+G.sum()-Ipx.sum(0), 1)
    j = int(np.argmax(pooled))
    del net; torch.cuda.empty_cache()
    return dict(th=THS[j], pooled=float(pooled[j]), per=per[:, j])

def sha(p, chunk=1 << 24):
    h = hashlib.sha256()
    with open(p, "rb") as f:
        while (b := f.read(chunk)): h.update(b)
    return h.hexdigest()[:16]

# -------------------------- which runs to handle -----------------------
# summary.json is written only at the END of cell D, so its presence is
# the completion test -- safer than trusting a step count.
done = sorted(glob.glob(f"{STAGE}/*/summary.json"), key=os.path.getmtime)
if not done:
    raise SystemExit(f"no completed run under {STAGE}")
target = RUN_NAME or os.path.basename(os.path.dirname(done[-1]))
print(f"target run: {target}   ({len(done)} completed run(s) on Drive)\n")

man_p = f"{SAVE}/weights_manifest.json"
man = json.load(open(man_p)) if os.path.exists(man_p) else {}

for sm in done:
    d = os.path.dirname(sm); name = os.path.basename(d)
    J = json.load(open(sm))
    bp, ck = f"{d}/best.pt", f"{d}/ckpt.pt"
    if not os.path.exists(bp):
        print(f"{name}: no best.pt, skipped"); continue
    is_target = (name == target)

    # 1. per-image vectors
    npz = f"{d}/per_image.npz"
    need = True
    if os.path.exists(npz):
        try:
            z = np.load(npz, allow_pickle=True); need = list(z["stems"]) != CLEAN_VAL
        except Exception: need = True
    if need:
        t = time.time()
        r = score_per_image(bp, J["select_on"], J["input"] == "crop_768_up")
        np.savez(npz, per=r["per"], th=r["th"], pooled=r["pooled"],
                 mode=J["select_on"], stems=np.array(CLEAN_VAL))
        print(f"{name}: scored per image in {time.time()-t:.0f}s  "
              f"(mode {J['select_on']}, th {r['th']}, pooled {r['pooled']:.4f})")
    else:
        print(f"{name}: per_image.npz already current")

    # 2. archive
    zp = f"{ARCH}/{name}_results.zip"
    if is_target or not os.path.exists(zp):
        with zipfile.ZipFile(zp, "w", zipfile.ZIP_DEFLATED) as z:
            for f in ("summary.json", "history.json", "curves.png",
                      "clean_val_gain.npy", "per_image.npz"):
                p = f"{d}/{f}"
                if os.path.exists(p): z.write(p, arcname=f"{name}/{f}")
        print(f"{name}: archive {os.path.getsize(zp)/1e6:.2f} MB")

    # 3. manifest
    if is_target or name not in man:
        man[name] = dict(file=f"{name}__best.pt", bytes=os.path.getsize(bp),
                         sha256_16=sha(bp), best_val_pooled=J.get("best_val_pooled"),
                         gpu=J.get("gpu", "not recorded"), batch=J.get("batch"),
                         input=J["input"], loss=J["loss"], jitter=J["jitter"],
                         seed=J["seed"], floor=J.get("floor"))

    # 4. prune -- only a COMPLETED run, which summary.json proves
    if DELETE_CKPT and os.path.exists(ck):
        mb = os.path.getsize(ck)/1e6
        os.remove(ck); print(f"{name}: removed ckpt.pt (freed {mb:.0f} MB)")

json.dump(man, open(man_p, "w"), indent=1)
print(f"\nmanifest: {man_p}  ({len(man)} run(s))")

# ------------------------------ drive usage ----------------------------
def dsz(p):
    return sum(os.path.getsize(os.path.join(r, f))
               for r, _, fs in os.walk(p) for f in fs) if os.path.isdir(p) else 0
tot = 0
print("\nDrive usage")
for lab, p in (("stage2", STAGE), ("stage2_archive", ARCH),
               ("runs (stage 1)", f"{SAVE}/runs"), ("cache768", f"{SAVE}/cache768")):
    s = dsz(p); tot += s
    print(f"   {lab:18}{s/1e9:7.2f} GB")
print(f"   {'TRACKED TOTAL':18}{tot/1e9:7.2f} GB"
      + ("   !! approaching a 15 GB free Drive" if tot/1e9 > 11 else ""))

# ------------------------------- download ------------------------------
if DOWNLOAD:
    d = f"{STAGE}/{target}"
    loc = f"/content/{target}__best.pt"
    if not os.path.exists(loc): shutil.copyfile(f"{d}/best.pt", loc)
    print(f"\n{target}  best.pt {os.path.getsize(loc)/1e6:.1f} MB  "
          f"sha256[:16] {man[target]['sha256_16']}")
    try:
        from google.colab import files
        files.download(f"{ARCH}/{target}_results.zip")
        time.sleep(2)
        files.download(loc)
        print("downloads queued. Verify locally:")
        print(f"   shasum -a 256 {target}__best.pt | cut -c1-16"
              f"   ->  {man[target]['sha256_16']}")
    except Exception as e:
        print(f"(browser download unavailable: {type(e).__name__})")
    print(f"also on Drive:\n   {ARCH}/{target}_results.zip\n   {d}/best.pt")
```

```python
# =====================================================================
#  cell E -- score every finished Stage 2 run per image, then compare.
#
#  Cell D saves per-quartile MEANS, which cannot support a paired test.
#  This reloads each best.pt and writes the per-image IoU vector over
#  clean_val, at that run's own selected mode and threshold. No retraining.
#
#  Then, against the S-384 baseline, for each other run:
#    - paired per-image differences over the same 688 images
#    - an image-level paired bootstrap on Q0  (the PRIMARY inference:
#      688 paired images carry far more power than a 3-seed t-interval,
#      which at 3 seeds needs a seed-sd below 0.008 to clear +0.02)
#    - the same split by realized detail gain, declared in advance
#    - the PILOT rule: kill only if the mean Q0 difference is <= 0.
#      Killing anything that misses the +0.02 gate at one seed discards
#      a real +0.02 effect half the time; the <= 0 rule, 10% of the time.
#
#  A single-seed result here is a FILTER, never a result. Nothing from
#  this cell goes in the paper without seeds 43 and 44.
# =====================================================================
import os, csv, json, glob, time
import numpy as np
from PIL import Image
import torch, torch.nn as nn, torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models
import torchvision.transforms.functional as TF

SAVE, C768 = "/content/drive/MyDrive", "/content/cache768"
STAGE   = f"{SAVE}/stage2"
BASE    = "S-384_seed42"           # the baseline every run is paired against
TRAIN_RES, FULL_RES = 384, 768
BATCH_EVAL = 4
NBOOT   = 10000
FROZEN  = [0.059468, 0.138448, 0.300831]
KILL_AT = 0.0                      # pilot rule, declared before looking
try:
    from google.colab import drive; drive.mount("/content/drive", force_remount=False)
except Exception: pass
DEV = torch.device("cuda" if torch.cuda.is_available() else "cpu")

SP = json.load(open(f"{SAVE}/plantseg_splits.json"))
I768 = json.load(open(f"{C768}/index.json"))
row = I768["row_of"]
CLEAN_VAL = sorted(set(SP["clean_val"]) & set(I768["stems"]))
DSDIR = "/content/my_images/plantseg"
meta = {os.path.splitext(r["Name"])[0]: r for r in csv.DictReader(
        open(f"{DSDIR}/Metadata.csv", encoding="utf-8-sig", newline=""))}
frac = {k: float(v["Mask ratio"]) for k, v in meta.items()}
def quart(s):
    f = frac[s]
    return 0 if f <= FROZEN[0] else 1 if f <= FROZEN[1] else 2 if f <= FROZEN[2] else 3
QV = np.array([quart(s) for s in CLEAN_VAL])
res = {k: tuple(int(v) for v in meta[k]["Resolution"].lower().split("x")[:2])
       for k in CLEAN_VAL}
GAIN = np.array([min(min(res[s]), FULL_RES)/TRAIN_RES for s in CLEAN_VAL])
print(f"clean_val {len(CLEAN_VAL)}  quartiles {[int((QV==q).sum()) for q in range(4)]}")

MEAN, STD = [0.485,0.456,0.406], [0.229,0.224,0.225]
NORM = transforms.Normalize(MEAN, STD)
_mm = {}
def cache():
    if not _mm:
        _mm["i"] = np.load(I768["img_path"], mmap_mode="r")
        _mm["m"] = np.load(I768["msk_path"], mmap_mode="r")
    return _mm["i"], _mm["m"]
def strip_detail(im):
    return im.resize((TRAIN_RES,)*2, Image.BILINEAR).resize((FULL_RES,)*2, Image.BILINEAR)

class EvalSet(Dataset):
    def __init__(s, st, up): s.st, s.up = list(st), up
    def __len__(s): return len(s.st)
    def __getitem__(s, i):
        I, M = cache(); r = row[s.st[i]]
        img = Image.fromarray(I[r])
        if s.up: img = strip_detail(img)
        y = torch.from_numpy((np.asarray(M[r]) > 0).astype(np.float32)).unsqueeze(0)
        return NORM(TF.to_tensor(img)), y

SPEC = dict(channels=(64,256,512,1024,2048), decoder=(512,256,128,64))
class DoubleConv(nn.Module):
    def __init__(s,i,o):
        super().__init__()
        s.net = nn.Sequential(nn.Conv2d(i,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True),
                              nn.Conv2d(o,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True))
    def forward(s,x): return s.net(x)
class Up(nn.Module):
    def __init__(s,i,k,o):
        super().__init__(); s.reduce=nn.Conv2d(i,i//2,1,bias=False); s.conv=DoubleConv(i//2+k,o)
    def forward(s,x,skip):
        x = F.interpolate(x, size=skip.shape[-2:], mode="bilinear", align_corners=False)
        return s.conv(torch.cat([s.reduce(x), skip], 1))
class ResUNet(nn.Module):
    def __init__(s):
        super().__init__()
        base = models.resnet50(weights=None)
        c0,c1,c2,c3,c4 = SPEC["channels"]; d1,d2,d3,d4 = SPEC["decoder"]
        L = list(base.children())
        s.stem=nn.Sequential(*L[:3]); s.pool=L[3]
        s.l1,s.l2,s.l3,s.l4 = L[4],L[5],L[6],L[7]
        s.u1,s.u2,s.u3,s.u4 = Up(c4,c3,d1),Up(d1,c2,d2),Up(d2,c1,d3),Up(d3,c0,d4)
        s.head = nn.Sequential(nn.Conv2d(d4,32,3,padding=1,bias=False),
                               nn.BatchNorm2d(32), nn.ReLU(True), nn.Conv2d(32,1,1))
    def forward(s,x):
        size=x.shape[-2:]
        x0=s.stem(x); x1=s.l1(s.pool(x0)); x2=s.l2(x1); x3=s.l3(x2); x4=s.l4(x3)
        d=s.u1(x4,x3); d=s.u2(d,x2); d=s.u3(d,x1); d=s.u4(d,x0)
        return s.head(F.interpolate(d,size=size,mode="bilinear",align_corners=False))

@torch.no_grad()
def prob_768(net, x, mode):
    if mode == "full768":
        return torch.sigmoid(net(x).float())
    if mode == "whole384":
        d = F.interpolate(x, size=(TRAIN_RES,)*2, mode="bilinear", align_corners=False)
        p = torch.sigmoid(net(d).float())
        return F.interpolate(p, size=(FULL_RES,)*2, mode="bilinear", align_corners=False)
    acc = torch.zeros(x.shape[0], 1, FULL_RES, FULL_RES, device=x.device)
    cnt = torch.zeros_like(acc)
    ORIG = (0, (FULL_RES-TRAIN_RES)//2, FULL_RES-TRAIN_RES)
    for oy in ORIG:
        for ox in ORIG:
            t = x[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES]
            acc[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += torch.sigmoid(net(t).float())
            cnt[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += 1
    return acc/cnt.clamp(min=1)

THS = [round(t, 2) for t in np.arange(0.10, 0.91, 0.05)]
@torch.no_grad()
def score(bp, mode, up):
    net = ResUNet().to(DEV); net.load_state_dict(torch.load(bp, map_location=DEV)); net.eval()
    dl = DataLoader(EvalSet(CLEAN_VAL, up), BATCH_EVAL, shuffle=False,
                    num_workers=min(2, os.cpu_count() or 1), pin_memory=DEV.type == "cuda")
    n = len(CLEAN_VAL)
    per = np.zeros((n, len(THS))); Ipx = np.zeros((n, len(THS)))
    P = np.zeros((n, len(THS))); G = np.zeros(n); k = 0
    for x, y in dl:
        x, y = x.to(DEV, non_blocking=True), y.to(DEV, non_blocking=True)
        with torch.amp.autocast("cuda", enabled=DEV.type == "cuda"):
            p = prob_768(net, x, mode)
        b = x.shape[0]; yf = y.view(b, -1); pf = p.view(b, -1)
        G[k:k+b] = yf.sum(1).cpu().numpy()
        for j, t in enumerate(THS):
            pr = (pf > t).float(); it = (pr*yf).sum(1); pp = pr.sum(1)
            Ipx[k:k+b, j] = it.cpu().numpy(); P[k:k+b, j] = pp.cpu().numpy()
            per[k:k+b, j] = (it/(pp+yf.sum(1)-it).clamp(min=1)).cpu().numpy()
        k += b
    pooled = Ipx.sum(0)/np.maximum(P.sum(0)+G.sum()-Ipx.sum(0), 1)
    j = int(np.argmax(pooled))
    del net; torch.cuda.empty_cache()
    return dict(th=THS[j], pooled=float(pooled[j]), per=per[:, j])

# ------------------------- score what is finished ----------------------
runs = {}
for sm in sorted(glob.glob(f"{STAGE}/*/summary.json")):
    d = os.path.dirname(sm); name = os.path.basename(d)
    if not os.path.exists(f"{d}/best.pt"):
        print(f"   {name}: no best.pt, skipped"); continue
    J = json.load(open(sm))
    npz = f"{d}/per_image.npz"
    if os.path.exists(npz):
        z = np.load(npz, allow_pickle=True)
        if list(z["stems"]) == CLEAN_VAL:
            runs[name] = dict(per=z["per"], th=float(z["th"]), pooled=float(z["pooled"]),
                              mode=str(z["mode"]), cfg=J); print(f"   {name}: cached"); continue
    t = time.time()
    r = score(f"{d}/best.pt", J["select_on"], J["input"] == "crop_768_up")
    np.savez(npz, per=r["per"], th=r["th"], pooled=r["pooled"],
             mode=J["select_on"], stems=np.array(CLEAN_VAL))
    runs[name] = dict(**r, mode=J["select_on"], cfg=J)
    print(f"   {name}: scored in {time.time()-t:.0f}s  "
          f"mode {J['select_on']} th {r['th']} pooled {r['pooled']:.4f}")

if BASE not in runs:
    raise SystemExit(f"baseline {BASE} not scored yet -- run it first")
print(f"\n{len(runs)} run(s) scored; baseline {BASE}")

# --------------------------- paired comparison -------------------------
rng = np.random.default_rng(0)
b_per = runs[BASE]["per"]; q0 = QV == 0
bands = {"Q0 all": q0, "Q0 gain<=1.0": q0 & (GAIN <= 1.0),
         "Q0 gain>=1.5": q0 & (GAIN >= 1.5), "all images": np.ones(len(QV), bool)}

def boot(d, sel):
    v = d[sel]
    if len(v) < 5: return float("nan"), float("nan"), float("nan")
    idx = rng.integers(0, len(v), (NBOOT, len(v)))
    bs = v[idx].mean(1)
    return float(v.mean()), float(np.percentile(bs, 2.5)), float(np.percentile(bs, 97.5))

print("\n" + "=" * 88)
print(f"paired per-image IoU difference vs {BASE}, clean_val, "
      f"{NBOOT} image-level bootstraps")
print("=" * 88)
out = {}
for name, r in runs.items():
    if name == BASE: continue
    d = r["per"] - b_per
    print(f"\n{name}   ({r['cfg']['input']} / {r['cfg']['loss']}"
          f"{' / jitter' if r['cfg']['jitter'] else ''})   "
          f"pooled {r['pooled']:.4f} vs {runs[BASE]['pooled']:.4f}")
    print(f"   {'band':16}{'n':>6}{'mean diff':>12}{'95% CI':>22}")
    row = {}
    for lab, sel in bands.items():
        m, lo, hi = boot(d, sel)
        star = "  *" if (lo > 0 or hi < 0) else ""
        print(f"   {lab:16}{int(sel.sum()):>6}{m:>12.4f}"
              f"   [{lo:+.4f}, {hi:+.4f}]{star}")
        row[lab] = dict(n=int(sel.sum()), mean=m, lo=lo, hi=hi)
    mq0 = row["Q0 all"]["mean"]
    verdict = ("KILL -- Q0 moves the wrong way" if mq0 <= KILL_AT else
               "KEEP -- run seeds 43 and 44")
    print(f"   PILOT ({'<=' if mq0 <= KILL_AT else '>'} {KILL_AT:+.2f}): {verdict}")
    if row["Q0 gain>=1.5"]["n"] >= 20 and row["Q0 gain<=1.0"]["n"] >= 20:
        sep = row["Q0 gain>=1.5"]["mean"] - row["Q0 gain<=1.0"]["mean"]
        print(f"   mechanism check: high-detail minus no-detail = {sep:+.4f}"
              f"   {'(consistent with resolution)' if sep > 0 else '(not resolution)'}")
        row["mechanism_sep"] = sep
    out[name] = row

print("\n" + "=" * 88)
print("* marks a CI excluding zero. That is an image-level interval: it says the")
print("difference is not sampling noise over these 688 images, NOT that it would")
print("survive another training seed. Only seeds 43 and 44 answer that, and the")
print("gate is judged on the three-seed paired mean, not on anything here.")
json.dump(dict(baseline=BASE, kill_at=KILL_AT, nboot=NBOOT, results=out),
          open(f"{SAVE}/stage2_compare.json", "w"), indent=1, default=float)
print(f"\nwrote {SAVE}/stage2_compare.json")
```

```python
# =====================================================================
#  cell D -- Stage 2: supervised tests of components 1 and 2.
#  One cell, three switches, 25,000 steps, resume-safe. Validation only;
#  the test split is not touched until Stage 5.
#
#      INPUT_MODE in {full_384, crop_768, crop_768_up}
#      FG_LOSS    in {pooled, class_balanced, image_balanced}
#      JITTER     in {False, True}
#
#  (full_384, pooled, False) is the S-384 baseline.
#
#  crop_768_up is the control the crop statistics made necessary: uniform
#  384 crops of a 768 image carry 1.50x the Q0 lesion density of the
#  whole image, because uniform crop ORIGINS over-sample the centre (a
#  centre pixel sits in 384 of 385 origins per axis, a corner pixel in 1).
#  So crop_768 changes resolution AND density at once. crop_768_up takes
#  the 768 image, downsamples it to 384 and upsamples it back, destroying
#  exactly the detail above 384 and nothing else, then crops identically.
#  crop_768 beating crop_768_up isolates resolution; a tie says the gain
#  is crop geometry.
#
#  Everything reads ONE cache: /content/cache768 (537 labelled + 846 val).
#  The 441 cache is not needed, so prep.py need not be run this session.
# =====================================================================
import os, csv, json, math, time, random, shutil, zipfile, collections
import numpy as np
import matplotlib; matplotlib.use("Agg")
import matplotlib.pyplot as plt
from PIL import Image, ImageFilter
import torch, torch.nn as nn, torch.optim as optim, torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models
import torchvision.transforms.functional as TF

# ------------------------- the three switches --------------------------
INPUT_MODE = "crop_768_up"        # full_384 | crop_768 | crop_768_up
FG_LOSS    = "pooled"          # pooled | class_balanced | image_balanced
JITTER     = False             # True only for S-768c-jit
TRAIN_SEED = 42
# -----------------------------------------------------------------------
FLOOR_FRAC = 0.01              # declared from check 0f: binds on 8.9% of
                               # non-empty Q0 crops, caps amplification 21.5x
SELECT_ON  = {"full_384": "whole384", "crop_768": "full768",
              "crop_768_up": "full768"}[INPUT_MODE]
MAX_STEPS  = 25000
EVAL_EVERY = 1000
SHARE_EVERY = 500              # per-quartile lesion-loss share
CKPT_EVERY = 2500
LR_PATIENCE = 3                # patience counts EVAL_EVERY blocks, now 1000
TRAIN_RES  = 384
FULL_RES   = 768
BATCH      = 4                 # matches the existing pipeline; batch 8
                               # doubles every run to ~2.2h and Stage 2 to ~42h
BATCH_EVAL = 4                 # 768 input on a T4
LR_ENCODER, LR_DECODER, WEIGHT_DECAY = 1e-5, 1e-4, 1e-4
FROZEN     = [0.059468, 0.138448, 0.300831]
SAVE, C768 = "/content/drive/MyDrive", "/content/cache768"
NUM_WORKERS = min(2, os.cpu_count() or 1)
FRESH      = True              # wipes this run's folder. Set back to False
                               # once it is past step 2500 so a dropped
                               # session resumes instead of restarting.

TAG = {"full_384": "384", "crop_768": "768c", "crop_768_up": "768up"}[INPUT_MODE]
if FG_LOSS == "class_balanced":  TAG += "-cb"
if FG_LOSS == "image_balanced":  TAG += "-bal"
if JITTER:                       TAG += "-jit"
RUN = f"S-{TAG}_seed{TRAIN_SEED}"
OUT = f"{SAVE}/stage2/{RUN}"
if FRESH and os.path.isdir(OUT): shutil.rmtree(OUT, ignore_errors=True)
os.makedirs(OUT, exist_ok=True)

random.seed(TRAIN_SEED); np.random.seed(TRAIN_SEED)
torch.manual_seed(TRAIN_SEED); torch.cuda.manual_seed_all(TRAIN_SEED)
torch.backends.cudnn.deterministic = True; torch.backends.cudnn.benchmark = False
DEV = torch.device("cuda" if torch.cuda.is_available() else "cpu")
try:
    from google.colab import drive; drive.mount("/content/drive", force_remount=False)
except Exception: pass
print(f"{RUN}\ndevice {DEV} {torch.cuda.get_device_name(0) if DEV.type=='cuda' else ''}")
print(f"input {INPUT_MODE} | loss {FG_LOSS} | jitter {JITTER} | "
      f"select on {SELECT_ON} | floor {FLOOR_FRAC} | batch {BATCH}")

# --------------------------- data and asserts --------------------------
SP = json.load(open(f"{SAVE}/plantseg_splits.json"))
if max(abs(a-b) for a, b in zip(SP["quartile_cuts"], FROZEN)) > 1e-6:
    raise SystemExit(f"cuts {SP['quartile_cuts']} != frozen {FROZEN}")
if {k: len(v) for k, v in SP["official_splits"].items()} != \
   {"train": 5367, "val": 846, "test": 1561}:
    raise SystemExit("split sizes are not 5367/846/1561")
I768 = json.load(open(f"{C768}/index.json"))
if I768["load_size"] != FULL_RES:
    raise SystemExit(f"cache is {I768['load_size']}, expected {FULL_RES}")
row = I768["row_of"]
LAB = sorted(I768["labelled"])
CLEAN_VAL = sorted(set(SP["clean_val"]) & set(I768["stems"]))
if len(LAB) != 537: raise SystemExit(f"{len(LAB)} labelled, expected 537")
DSDIR = "/content/my_images/plantseg"
meta = {os.path.splitext(r["Name"])[0]: r for r in csv.DictReader(
        open(f"{DSDIR}/Metadata.csv", encoding="utf-8-sig", newline=""))}
frac = {k: float(v["Mask ratio"]) for k, v in meta.items()}
def quart(s):
    f = frac[s]
    return 0 if f <= FROZEN[0] else 1 if f <= FROZEN[1] else 2 if f <= FROZEN[2] else 3
QV = np.array([quart(s) for s in CLEAN_VAL])
print(f"labelled {len(LAB)}   clean_val {len(CLEAN_VAL)} "
      f"(quartiles {[int((QV==q).sum()) for q in range(4)]})")
print("   gates are judged on clean_val, not the full 846: full val carries")
print("   17.14% near-duplicates of training images.")

MEAN, STD = [0.485,0.456,0.406], [0.229,0.224,0.225]
NORM = transforms.Normalize(MEAN, STD)
JIT  = transforms.ColorJitter(0.4, 0.4, 0.4, 0.1)
_mm = {}
def cache():
    if not _mm:
        _mm["i"] = np.load(I768["img_path"], mmap_mode="r")
        _mm["m"] = np.load(I768["msk_path"], mmap_mode="r")
    return _mm["i"], _mm["m"]

def strip_detail(im):
    """768 -> 384 -> 768. Destroys exactly the detail above 384, keeps
       geometry, size and lesion density identical."""
    return im.resize((TRAIN_RES,)*2, Image.BILINEAR).resize((FULL_RES,)*2, Image.BILINEAR)

def photometric(img):
    if random.random() < 0.8: img = JIT(img)
    if random.random() < 0.2: img = TF.to_grayscale(img, num_output_channels=3)
    if random.random() < 0.3:
        img = img.filter(ImageFilter.GaussianBlur(radius=random.uniform(0.1, 1.5)))
    return img

def geometric(img, msk):
    if random.random() < 0.5: img, msk = TF.hflip(img), TF.hflip(msk)
    if random.random() < 0.5: img, msk = TF.vflip(img), TF.vflip(msk)
    k = random.randint(0, 3)
    if k:
        img = img.rotate(90*k, expand=True, resample=Image.BILINEAR)
        msk = msk.rotate(90*k, expand=True, resample=Image.NEAREST)
    return img, msk

def rrc(img, msk, scale=(0.5, 1.0), ratio=(3/4, 4/3)):
    w, h = img.size; area = w*h
    for _ in range(10):
        t = area*random.uniform(*scale)
        ar = math.exp(random.uniform(math.log(ratio[0]), math.log(ratio[1])))
        cw, ch = int(round(math.sqrt(t*ar))), int(round(math.sqrt(t/ar)))
        if cw <= w and ch <= h:
            x, y = random.randint(0, w-cw), random.randint(0, h-ch)
            return img.crop((x, y, x+cw, y+ch)), msk.crop((x, y, x+cw, y+ch))
    return img, msk

class Labelled(Dataset):
    """full_384   : whole image, rrc + geometric, resized to 384
       crop_768   : uniform 384 window of the 768 image (the crop IS the
                    spatial augmentation, so rrc is not applied on top)
       crop_768_up: the same window of a detail-stripped 768 image"""
    def __init__(s, st): s.st = list(st)
    def __len__(s): return len(s.st)
    def __getitem__(s, i):
        I, M = cache(); r = row[s.st[i]]
        img = Image.fromarray(I[r]); msk = Image.fromarray(M[r])
        if INPUT_MODE == "crop_768_up": img = strip_detail(img)
        if INPUT_MODE == "full_384":
            img, msk = rrc(img, msk)
            img, msk = geometric(img, msk)
            img = img.resize((TRAIN_RES,)*2, Image.BILINEAR)
            msk = msk.resize((TRAIN_RES,)*2, Image.NEAREST)
        else:
            side = (random.randint(256, FULL_RES) if JITTER else TRAIN_RES)
            x = random.randint(0, FULL_RES-side); y = random.randint(0, FULL_RES-side)
            img = img.crop((x, y, x+side, y+side)); msk = msk.crop((x, y, x+side, y+side))
            img, msk = geometric(img, msk)
            if side != TRAIN_RES:
                img = img.resize((TRAIN_RES,)*2, Image.BILINEAR)
                msk = msk.resize((TRAIN_RES,)*2, Image.NEAREST)
        y_ = torch.from_numpy((np.array(msk) > 0).astype(np.float32)).unsqueeze(0)
        return NORM(TF.to_tensor(photometric(img))), y_, quart(s.st[i])

class EvalSet(Dataset):
    """input follows INPUT_MODE; the TARGET is always the 768 mask."""
    def __init__(s, st): s.st = list(st)
    def __len__(s): return len(s.st)
    def __getitem__(s, i):
        I, M = cache(); r = row[s.st[i]]
        img = Image.fromarray(I[r])
        if INPUT_MODE == "crop_768_up": img = strip_detail(img)
        y = torch.from_numpy((np.asarray(M[r]) > 0).astype(np.float32)).unsqueeze(0)
        return NORM(TF.to_tensor(img)), y            # image at 768, mask at 768

def seed_worker(_):
    s = torch.initial_seed() % 2**32; np.random.seed(s); random.seed(s)
LK = dict(num_workers=NUM_WORKERS, pin_memory=DEV.type == "cuda",
          worker_init_fn=seed_worker, persistent_workers=NUM_WORKERS > 0)
g = torch.Generator(); g.manual_seed(TRAIN_SEED)
loader_l = DataLoader(Labelled(LAB), BATCH, shuffle=True, drop_last=True,
                      generator=g, **LK)
val_loader = DataLoader(EvalSet(CLEAN_VAL), BATCH_EVAL, shuffle=False, **LK)
def cycle(dl):
    while True:
        for b in dl: yield b

# ------------------------------- model ---------------------------------
SPEC = dict(channels=(64,256,512,1024,2048), decoder=(512,256,128,64))
class DoubleConv(nn.Module):
    def __init__(s,i,o):
        super().__init__()
        s.net = nn.Sequential(nn.Conv2d(i,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True),
                              nn.Conv2d(o,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True))
    def forward(s,x): return s.net(x)
class Up(nn.Module):
    def __init__(s,i,k,o):
        super().__init__(); s.reduce=nn.Conv2d(i,i//2,1,bias=False); s.conv=DoubleConv(i//2+k,o)
    def forward(s,x,skip):
        x = F.interpolate(x, size=skip.shape[-2:], mode="bilinear", align_corners=False)
        return s.conv(torch.cat([s.reduce(x), skip], 1))
class ResUNet(nn.Module):
    def __init__(s):
        super().__init__()
        base = models.resnet50(weights=models.ResNet50_Weights.DEFAULT)
        c0,c1,c2,c3,c4 = SPEC["channels"]; d1,d2,d3,d4 = SPEC["decoder"]
        L = list(base.children())
        s.stem=nn.Sequential(*L[:3]); s.pool=L[3]
        s.l1,s.l2,s.l3,s.l4 = L[4],L[5],L[6],L[7]
        s.u1,s.u2,s.u3,s.u4 = Up(c4,c3,d1),Up(d1,c2,d2),Up(d2,c1,d3),Up(d3,c0,d4)
        s.head = nn.Sequential(nn.Conv2d(d4,32,3,padding=1,bias=False),
                               nn.BatchNorm2d(32), nn.ReLU(True), nn.Conv2d(32,1,1))
    def enc_params(s):
        for m in (s.stem,s.l1,s.l2,s.l3,s.l4): yield from m.parameters()
    def dec_params(s):
        for m in (s.u1,s.u2,s.u3,s.u4,s.head): yield from m.parameters()
    def forward(s,x):
        size=x.shape[-2:]
        x0=s.stem(x); x1=s.l1(s.pool(x0)); x2=s.l2(x1); x3=s.l3(x2); x4=s.l4(x3)
        d=s.u1(x4,x3); d=s.u2(d,x2); d=s.u3(d,x1); d=s.u4(d,x0)
        return s.head(F.interpolate(d,size=size,mode="bilinear",align_corners=False))

# ------------------------------- losses --------------------------------
bce_none = nn.BCEWithLogitsLoss(reduction="none")
def tversky(logits, tgt, eps=1.0):
    p = torch.sigmoid(logits)
    tp = (p*tgt).sum((1,2,3)); fp = (p*(1-tgt)).sum((1,2,3)); fn = ((1-p)*tgt).sum((1,2,3))
    return 1 - (tp+eps)/(tp+0.5*fp+0.5*fn+eps)

def fg_term(logits, tgt):
    """returns (per-image BCE term, per-image lesion-only contribution).
       All three forms are scaled so a uniform per-pixel loss c gives c,
       so the learning rate and the BCE/Tversky balance do not shift
       between configurations. The plan's formula writes the sum of the
       two class means; using their average is the same comparison at the
       same scale."""
    e = bce_none(logits, tgt)
    sf = tgt.sum((1,2,3)); sb = (1-tgt).sum((1,2,3))
    ef = (e*tgt).sum((1,2,3)); eb = (e*(1-tgt)).sum((1,2,3))
    if FG_LOSS == "pooled":
        return e.mean((1,2,3)), ef/tgt[0].numel()
    if FG_LOSS == "class_balanced":                 # batch-level class means
        n = logits.shape[0]
        mf = ef.sum()/sf.sum().clamp(min=1.0); mb = eb.sum()/sb.sum().clamp(min=1.0)
        return (0.5*(mf+mb)).expand(n), 0.5*ef/sf.sum().clamp(min=1.0)
    floor = FLOOR_FRAC*tgt[0].numel()               # per-image, with the floor
    tf = ef/torch.clamp(sf, min=floor)
    tb = eb/sb.clamp(min=1.0)
    return 0.5*(tf+tb), 0.5*tf

def sup_loss(logits, tgt):
    b, lesion = fg_term(logits, tgt)
    return b + tversky(logits, tgt), lesion

# --------------------------- 768 inference -----------------------------
@torch.no_grad()
def prob_768(net, x, mode):
    """x is the model's input image at 768. Returns a 768 probability map."""
    if mode == "full768":
        return torch.sigmoid(net(x).float())
    if mode == "whole384":
        d = F.interpolate(x, size=(TRAIN_RES,)*2, mode="bilinear", align_corners=False)
        p = torch.sigmoid(net(d).float())
        return F.interpolate(p, size=(FULL_RES,)*2, mode="bilinear", align_corners=False)
    # genuinely overlapping: stride 192 gives 3x3 tiles at 50% overlap.
    # Origins (0, 384) alone would ABUT, not overlap, and blending would
    # be a no-op -- the seam is exactly where a tiled model is weakest.
    acc = torch.zeros(x.shape[0], 1, FULL_RES, FULL_RES, device=x.device)
    cnt = torch.zeros_like(acc)
    ORIG = (0, (FULL_RES-TRAIN_RES)//2, FULL_RES-TRAIN_RES)
    for oy in ORIG:
        for ox in ORIG:
            t = x[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES]
            acc[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += torch.sigmoid(net(t).float())
            cnt[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += 1
    return acc/cnt.clamp(min=1)                      # blend, threshold afterwards

THS = [round(t, 2) for t in np.arange(0.10, 0.91, 0.05)]
@torch.no_grad()
def evaluate(net, mode):
    net.eval()
    nI = len(CLEAN_VAL); per = np.zeros((nI, len(THS))); P = np.zeros((nI, len(THS)))
    Ipx = np.zeros((nI, len(THS))); G = np.zeros(nI); k = 0
    for x, y in val_loader:
        x, y = x.to(DEV, non_blocking=True), y.to(DEV, non_blocking=True)
        with torch.amp.autocast("cuda", enabled=DEV.type == "cuda"):
            p = prob_768(net, x, mode)
        b = x.shape[0]; yf = y.view(b, -1)
        G[k:k+b] = yf.sum(1).cpu().numpy()
        pf = p.view(b, -1)
        for j, t in enumerate(THS):
            pr = (pf > t).float()
            it = (pr*yf).sum(1); pp = pr.sum(1)
            Ipx[k:k+b, j] = it.cpu().numpy(); P[k:k+b, j] = pp.cpu().numpy()
            per[k:k+b, j] = (it/(pp+yf.sum(1)-it).clamp(min=1)).cpu().numpy()
        k += b
    pooled = Ipx.sum(0)/np.maximum(P.sum(0)+G.sum()-Ipx.sum(0), 1)
    j = int(np.argmax(pooled)); t = THS[j]
    byq = {q: float(per[QV == q, j].mean()) for q in range(4)}
    return dict(th=t, pooled=float(pooled[j]), perimg=float(per[:, j].mean()),
                byq=byq, pred_frac=float(P[:, j].sum()/(len(CLEAN_VAL)*FULL_RES*FULL_RES)))

# --------------------------- resume or start ---------------------------
model = ResUNet().to(DEV)
opt = optim.AdamW([{"params": list(model.enc_params()), "lr": LR_ENCODER},
                   {"params": list(model.dec_params()), "lr": LR_DECODER}],
                  weight_decay=WEIGHT_DECAY)
sched = optim.lr_scheduler.ReduceLROnPlateau(opt, mode="max", patience=LR_PATIENCE, factor=0.5)
scaler = torch.amp.GradScaler("cuda")
step, best, history, shares = 0, -1.0, [], []
CK, BEST = f"{OUT}/ckpt.pt", f"{OUT}/best.pt"
if os.path.exists(CK):
    c = torch.load(CK, map_location=DEV, weights_only=False)
    model.load_state_dict(c["model"]); opt.load_state_dict(c["opt"])
    sched.load_state_dict(c["sched"]); scaler.load_state_dict(c["scaler"])
    step, best, history, shares = c["step"], c["best"], c["history"], c.get("shares", [])
    random.setstate(c["py_rng"]); np.random.set_state(c["np_rng"])
    torch.set_rng_state(c["t_rng"].cpu())
    if c.get("cuda_rng") is not None and DEV.type == "cuda":
        torch.cuda.set_rng_state(c["cuda_rng"].cpu())
    print(f"RESUMED from step {step} (best {best:.4f})")
else:
    print("starting fresh")

def save_ckpt():
    torch.save(dict(model=model.state_dict(), opt=opt.state_dict(), sched=sched.state_dict(),
                    scaler=scaler.state_dict(), step=step, best=best, history=history,
                    shares=shares, py_rng=random.getstate(), np_rng=np.random.get_state(),
                    t_rng=torch.get_rng_state(),
                    cuda_rng=torch.cuda.get_rng_state() if DEV.type == "cuda" else None), CK)

# ------------------------------- train ---------------------------------
it = cycle(loader_l)
print("=" * 76); print(f"{RUN}  --  {len(LAB)} labelled, {MAX_STEPS} steps"); print("=" * 76)
t0 = t_last = time.time()
r_loss = r_n = 0.0; q_les = np.zeros(4); q_cnt = np.zeros(4)
while step < MAX_STEPS:
    model.train()
    x, y, q = next(it)
    x, y = x.to(DEV, non_blocking=True), y.to(DEV, non_blocking=True)
    opt.zero_grad(set_to_none=True)
    with torch.amp.autocast("cuda", enabled=DEV.type == "cuda"):
        lg = model(x)
        vec, lesion = sup_loss(lg.float(), y)
        loss = vec.mean()
    scaler.scale(loss).backward(); scaler.step(opt); scaler.update()
    step += 1
    r_loss += loss.item(); r_n += 1
    lz = lesion.detach().float().cpu().numpy(); qz = q.numpy()
    for b in range(len(qz)): q_les[qz[b]] += lz[b]; q_cnt[qz[b]] += 1

    if step % SHARE_EVERY == 0:
        tot = max(q_les.sum(), 1e-12)
        sh = (q_les/tot).round(4).tolist()
        shares.append(dict(step=step, share=sh, seen=q_cnt.astype(int).tolist()))
        print(f"   [step {step}] lesion-loss share by quartile {sh}"
              f"  (images seen {q_cnt.astype(int).tolist()})")
        q_les[:] = 0; q_cnt[:] = 0

    if step % EVAL_EVERY == 0:
        m = evaluate(model, SELECT_ON)
        prev = opt.param_groups[-1]["lr"]; sched.step(m["pooled"])
        cur = opt.param_groups[-1]["lr"]
        saved = ""
        if m["pooled"] > best:
            best = m["pooled"]; torch.save(model.state_dict(), BEST); saved = "  <-- saved"
        history.append(dict(step=step, train_loss=r_loss/max(r_n, 1), lr=cur, **m))
        sps = EVAL_EVERY/(time.time()-t_last); t_last = time.time()
        print(f"step {step:6d} | loss {r_loss/max(r_n,1):.4f} | val pooled {m['pooled']:.4f} "
              f"@{m['th']} | per-img {m['perimg']:.4f} | Q0 {m['byq'][0]:.4f} "
              f"Q3 {m['byq'][3]:.4f} | best {best:.4f} | {sps:.2f} it/s | "
              f"ETA {(MAX_STEPS-step)/sps/3600:.1f}h{saved}"
              + (f"  [lr {prev:.1e}->{cur:.1e}]" if cur != prev else ""))
        r_loss = r_n = 0.0
        json.dump(dict(run=RUN, input=INPUT_MODE, loss=FG_LOSS, jitter=JITTER,
                       seed=TRAIN_SEED, floor=FLOOR_FRAC, select_on=SELECT_ON,
                       batch=BATCH, history=history, shares=shares),
                  open(f"{OUT}/history.json", "w"), indent=1)
    if step % CKPT_EVERY == 0:
        save_ckpt(); print(f"    checkpointed at step {step}")
save_ckpt()
print(f"\ntraining done in {(time.time()-t0)/3600:.2f}h, best val pooled {best:.4f}")

# ------------------- final: all three inference modes ------------------
net = ResUNet().to(DEV); net.load_state_dict(torch.load(BEST, map_location=DEV)); net.eval()
print("\n" + "=" * 76)
print("clean_val under each inference mode (best.pt was selected on "
      f"{SELECT_ON})")
print("=" * 76)
print(f"{'mode':12}{'th':>6}{'pooled':>9}{'per-img':>9}"
      f"{'Q0':>8}{'Q1':>8}{'Q2':>8}{'Q3':>8}{'Q3-Q0':>8}")
modes = ["whole384", "full768", "tiles384"]
final = {}
for md in modes:
    m = evaluate(net, md); final[md] = m
    print(f"{md:12}{m['th']:>6.2f}{m['pooled']:>9.4f}{m['perimg']:>9.4f}"
          + "".join(f"{m['byq'][q]:>8.4f}" for q in range(4))
          + f"{m['byq'][3]-m['byq'][0]:>8.4f}")

# the stratified check: does any Q0 gain track the realized detail gain?
res = {k: tuple(int(v) for v in meta[k]["Resolution"].lower().split("x")[:2])
       for k in CLEAN_VAL}
gain = np.array([min(min(res[s]), FULL_RES)/TRAIN_RES for s in CLEAN_VAL])
q0 = QV == 0
print(f"\nclean_val Q0 split by realized worst-axis detail gain at {FULL_RES}:")
for lab, sel in (("<= 1.0 (no gain)", q0 & (gain <= 1.0)),
                 ("1.0 - 1.5", q0 & (gain > 1.0) & (gain < 1.5)),
                 (">= 1.5", q0 & (gain >= 1.5))):
    print(f"   {lab:18} n={int(sel.sum()):4d}")
print("   per-image Q0 IoU within each band is in the saved json; compare it")
print("   against the matching S-384 run, paired by image.")

np.save(f"{OUT}/clean_val_gain.npy", gain)
json.dump(dict(run=RUN, input=INPUT_MODE, loss=FG_LOSS, jitter=JITTER, seed=TRAIN_SEED,
               floor=FLOOR_FRAC, select_on=SELECT_ON, batch=BATCH, best_val_pooled=best,
               clean_val=CLEAN_VAL, quartiles=QV.tolist(), final=final,
               history=history, shares=shares),
          open(f"{OUT}/summary.json", "w"), indent=1)
xs = [h["step"] for h in history]
plt.figure(figsize=(14, 4))
plt.subplot(1, 3, 1); plt.plot(xs, [h["train_loss"] for h in history]); plt.title("train loss")
plt.subplot(1, 3, 2); plt.plot(xs, [h["pooled"] for h in history]); plt.title("val pooled IoU")
plt.subplot(1, 3, 3)
for q in range(4): plt.plot(xs, [h["byq"][q] for h in history], label=f"Q{q}")
plt.legend(fontsize=7); plt.title("val per-image IoU by quartile")
plt.tight_layout(); plt.savefig(f"{OUT}/curves.png", dpi=110); plt.close()
print(f"\nwrote {OUT}/summary.json, best.pt, curves.png")
print(">>> download best.pt to local disk before the session ends <<<")
```

```python
# =====================================================================
#  cell F -- run this ONCE after every cell D training. Standalone: it
#  reads nothing from cell D's variables, so it also works in a fresh
#  session or after a runtime restart.
#
#  For the target run it:
#    1. scores it per image on clean_val  -> per_image.npz
#       (the vectors every paired test needs; cell D saves only means)
#    2. archives summary / history / curves / gain / per_image to Drive
#    3. records config, GPU and a checksum in weights_manifest.json
#    4. deletes ckpt.pt, but ONLY if the run completed. 490 MB each, and
#       19 of them would overrun a free Drive.
#    5. prints what Drive is holding
#    6. downloads the archive and best.pt
#
#  Any OTHER run still missing per_image.npz or an archive is also
#  brought up to date, without being downloaded again.
# =====================================================================
import os, csv, json, glob, time, zipfile, shutil, hashlib
import numpy as np
from PIL import Image
import torch, torch.nn as nn, torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models
import torchvision.transforms.functional as TF

RUN_NAME    = None       # None = the most recently finished run in stage2/
DELETE_CKPT = True       # False if you intend to extend a run later
DOWNLOAD    = True
SAVE, C768  = "/content/drive/MyDrive", "/content/cache768"
STAGE, ARCH = f"{SAVE}/stage2", f"{SAVE}/stage2_archive"
TRAIN_RES, FULL_RES = 384, 768
BATCH_EVAL  = 12 if torch.cuda.is_available() else 2
FROZEN      = [0.059468, 0.138448, 0.300831]
try:
    from google.colab import drive; drive.mount("/content/drive", force_remount=False)
except Exception: pass
DEV = torch.device("cuda" if torch.cuda.is_available() else "cpu")
os.makedirs(ARCH, exist_ok=True)

# ------------------------------- context -------------------------------
SP   = json.load(open(f"{SAVE}/plantseg_splits.json"))
I768 = json.load(open(f"{C768}/index.json"))
row  = I768["row_of"]
CLEAN_VAL = sorted(set(SP["clean_val"]) & set(I768["stems"]))
DSDIR = "/content/my_images/plantseg"
meta = {os.path.splitext(r["Name"])[0]: r for r in csv.DictReader(
        open(f"{DSDIR}/Metadata.csv", encoding="utf-8-sig", newline=""))}
frac = {k: float(v["Mask ratio"]) for k, v in meta.items()}
def quart(s):
    f = frac[s]
    return 0 if f <= FROZEN[0] else 1 if f <= FROZEN[1] else 2 if f <= FROZEN[2] else 3
QV = np.array([quart(s) for s in CLEAN_VAL])

MEAN, STD = [0.485,0.456,0.406], [0.229,0.224,0.225]
NORM = transforms.Normalize(MEAN, STD)
_mm = {}
def cache():
    if not _mm:
        _mm["i"] = np.load(I768["img_path"], mmap_mode="r")
        _mm["m"] = np.load(I768["msk_path"], mmap_mode="r")
    return _mm["i"], _mm["m"]
def strip_detail(im):
    return im.resize((TRAIN_RES,)*2, Image.BILINEAR).resize((FULL_RES,)*2, Image.BILINEAR)

class EvalSet(Dataset):
    def __init__(s, st, up): s.st, s.up = list(st), up
    def __len__(s): return len(s.st)
    def __getitem__(s, i):
        I, M = cache(); r = row[s.st[i]]
        img = Image.fromarray(I[r])
        if s.up: img = strip_detail(img)
        y = torch.from_numpy((np.asarray(M[r]) > 0).astype(np.float32)).unsqueeze(0)
        return NORM(TF.to_tensor(img)), y

SPEC = dict(channels=(64,256,512,1024,2048), decoder=(512,256,128,64))
class DoubleConv(nn.Module):
    def __init__(s,i,o):
        super().__init__()
        s.net = nn.Sequential(nn.Conv2d(i,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True),
                              nn.Conv2d(o,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True))
    def forward(s,x): return s.net(x)
class Up(nn.Module):
    def __init__(s,i,k,o):
        super().__init__(); s.reduce=nn.Conv2d(i,i//2,1,bias=False); s.conv=DoubleConv(i//2+k,o)
    def forward(s,x,skip):
        x = F.interpolate(x, size=skip.shape[-2:], mode="bilinear", align_corners=False)
        return s.conv(torch.cat([s.reduce(x), skip], 1))
class ResUNet(nn.Module):
    def __init__(s):
        super().__init__()
        base = models.resnet50(weights=None)
        c0,c1,c2,c3,c4 = SPEC["channels"]; d1,d2,d3,d4 = SPEC["decoder"]
        L = list(base.children())
        s.stem=nn.Sequential(*L[:3]); s.pool=L[3]
        s.l1,s.l2,s.l3,s.l4 = L[4],L[5],L[6],L[7]
        s.u1,s.u2,s.u3,s.u4 = Up(c4,c3,d1),Up(d1,c2,d2),Up(d2,c1,d3),Up(d3,c0,d4)
        s.head = nn.Sequential(nn.Conv2d(d4,32,3,padding=1,bias=False),
                               nn.BatchNorm2d(32), nn.ReLU(True), nn.Conv2d(32,1,1))
    def forward(s,x):
        size=x.shape[-2:]
        x0=s.stem(x); x1=s.l1(s.pool(x0)); x2=s.l2(x1); x3=s.l3(x2); x4=s.l4(x3)
        d=s.u1(x4,x3); d=s.u2(d,x2); d=s.u3(d,x1); d=s.u4(d,x0)
        return s.head(F.interpolate(d,size=size,mode="bilinear",align_corners=False))

@torch.no_grad()
def prob_768(net, x, mode):
    if mode == "full768":
        return torch.sigmoid(net(x).float())
    if mode == "whole384":
        d = F.interpolate(x, size=(TRAIN_RES,)*2, mode="bilinear", align_corners=False)
        p = torch.sigmoid(net(d).float())
        return F.interpolate(p, size=(FULL_RES,)*2, mode="bilinear", align_corners=False)
    acc = torch.zeros(x.shape[0], 1, FULL_RES, FULL_RES, device=x.device)
    cnt = torch.zeros_like(acc)
    ORIG = (0, (FULL_RES-TRAIN_RES)//2, FULL_RES-TRAIN_RES)
    for oy in ORIG:
        for ox in ORIG:
            t = x[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES]
            acc[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += torch.sigmoid(net(t).float())
            cnt[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += 1
    return acc/cnt.clamp(min=1)

THS = [round(t, 2) for t in np.arange(0.10, 0.91, 0.05)]
@torch.no_grad()
def score_per_image(bp, mode, up):
    net = ResUNet().to(DEV); net.load_state_dict(torch.load(bp, map_location=DEV)); net.eval()
    dl = DataLoader(EvalSet(CLEAN_VAL, up), BATCH_EVAL, shuffle=False,
                    num_workers=min(2, os.cpu_count() or 1), pin_memory=DEV.type == "cuda")
    n = len(CLEAN_VAL)
    per = np.zeros((n, len(THS))); Ipx = np.zeros((n, len(THS)))
    P = np.zeros((n, len(THS))); G = np.zeros(n); k = 0
    for x, y in dl:
        x, y = x.to(DEV, non_blocking=True), y.to(DEV, non_blocking=True)
        with torch.amp.autocast("cuda", enabled=DEV.type == "cuda"):
            p = prob_768(net, x, mode)
        b = x.shape[0]; yf = y.view(b, -1); pf = p.view(b, -1)
        G[k:k+b] = yf.sum(1).cpu().numpy()
        for j, t in enumerate(THS):
            pr = (pf > t).float(); it = (pr*yf).sum(1); pp = pr.sum(1)
            Ipx[k:k+b, j] = it.cpu().numpy(); P[k:k+b, j] = pp.cpu().numpy()
            per[k:k+b, j] = (it/(pp+yf.sum(1)-it).clamp(min=1)).cpu().numpy()
        k += b
    pooled = Ipx.sum(0)/np.maximum(P.sum(0)+G.sum()-Ipx.sum(0), 1)
    j = int(np.argmax(pooled))
    del net; torch.cuda.empty_cache()
    return dict(th=THS[j], pooled=float(pooled[j]), per=per[:, j])

def sha(p, chunk=1 << 24):
    h = hashlib.sha256()
    with open(p, "rb") as f:
        while (b := f.read(chunk)): h.update(b)
    return h.hexdigest()[:16]

# -------------------------- which runs to handle -----------------------
# summary.json is written only at the END of cell D, so its presence is
# the completion test -- safer than trusting a step count.
done = sorted(glob.glob(f"{STAGE}/*/summary.json"), key=os.path.getmtime)
if not done:
    raise SystemExit(f"no completed run under {STAGE}")
target = RUN_NAME or os.path.basename(os.path.dirname(done[-1]))
print(f"target run: {target}   ({len(done)} completed run(s) on Drive)\n")

man_p = f"{SAVE}/weights_manifest.json"
man = json.load(open(man_p)) if os.path.exists(man_p) else {}

for sm in done:
    d = os.path.dirname(sm); name = os.path.basename(d)
    J = json.load(open(sm))
    bp, ck = f"{d}/best.pt", f"{d}/ckpt.pt"
    if not os.path.exists(bp):
        print(f"{name}: no best.pt, skipped"); continue
    is_target = (name == target)

    # 1. per-image vectors
    npz = f"{d}/per_image.npz"
    need = True
    if os.path.exists(npz):
        try:
            z = np.load(npz, allow_pickle=True); need = list(z["stems"]) != CLEAN_VAL
        except Exception: need = True
    if need:
        t = time.time()
        r = score_per_image(bp, J["select_on"], J["input"] == "crop_768_up")
        np.savez(npz, per=r["per"], th=r["th"], pooled=r["pooled"],
                 mode=J["select_on"], stems=np.array(CLEAN_VAL))
        print(f"{name}: scored per image in {time.time()-t:.0f}s  "
              f"(mode {J['select_on']}, th {r['th']}, pooled {r['pooled']:.4f})")
    else:
        print(f"{name}: per_image.npz already current")

    # 2. archive
    zp = f"{ARCH}/{name}_results.zip"
    if is_target or not os.path.exists(zp):
        with zipfile.ZipFile(zp, "w", zipfile.ZIP_DEFLATED) as z:
            for f in ("summary.json", "history.json", "curves.png",
                      "clean_val_gain.npy", "per_image.npz"):
                p = f"{d}/{f}"
                if os.path.exists(p): z.write(p, arcname=f"{name}/{f}")
        print(f"{name}: archive {os.path.getsize(zp)/1e6:.2f} MB")

    # 3. manifest
    if is_target or name not in man:
        man[name] = dict(file=f"{name}__best.pt", bytes=os.path.getsize(bp),
                         sha256_16=sha(bp), best_val_pooled=J.get("best_val_pooled"),
                         gpu=J.get("gpu", "not recorded"), batch=J.get("batch"),
                         input=J["input"], loss=J["loss"], jitter=J["jitter"],
                         seed=J["seed"], floor=J.get("floor"))

    # 4. prune -- only a COMPLETED run, which summary.json proves
    if DELETE_CKPT and os.path.exists(ck):
        mb = os.path.getsize(ck)/1e6
        os.remove(ck); print(f"{name}: removed ckpt.pt (freed {mb:.0f} MB)")

json.dump(man, open(man_p, "w"), indent=1)
print(f"\nmanifest: {man_p}  ({len(man)} run(s))")

# ------------------------------ drive usage ----------------------------
def dsz(p):
    return sum(os.path.getsize(os.path.join(r, f))
               for r, _, fs in os.walk(p) for f in fs) if os.path.isdir(p) else 0
tot = 0
print("\nDrive usage")
for lab, p in (("stage2", STAGE), ("stage2_archive", ARCH),
               ("runs (stage 1)", f"{SAVE}/runs"), ("cache768", f"{SAVE}/cache768")):
    s = dsz(p); tot += s
    print(f"   {lab:18}{s/1e9:7.2f} GB")
print(f"   {'TRACKED TOTAL':18}{tot/1e9:7.2f} GB"
      + ("   !! approaching a 15 GB free Drive" if tot/1e9 > 11 else ""))

# ------------------------------- download ------------------------------
if DOWNLOAD:
    d = f"{STAGE}/{target}"
    loc = f"/content/{target}__best.pt"
    if not os.path.exists(loc): shutil.copyfile(f"{d}/best.pt", loc)
    print(f"\n{target}  best.pt {os.path.getsize(loc)/1e6:.1f} MB  "
          f"sha256[:16] {man[target]['sha256_16']}")
    try:
        from google.colab import files
        files.download(f"{ARCH}/{target}_results.zip")
        time.sleep(2)
        files.download(loc)
        print("downloads queued. Verify locally:")
        print(f"   shasum -a 256 {target}__best.pt | cut -c1-16"
              f"   ->  {man[target]['sha256_16']}")
    except Exception as e:
        print(f"(browser download unavailable: {type(e).__name__})")
    print(f"also on Drive:\n   {ARCH}/{target}_results.zip\n   {d}/best.pt")
```

```python
# =====================================================================
#  cell E -- score every finished Stage 2 run per image, then compare.
#
#  Cell D saves per-quartile MEANS, which cannot support a paired test.
#  This reloads each best.pt and writes the per-image IoU vector over
#  clean_val, at that run's own selected mode and threshold. No retraining.
#
#  Then, against the S-384 baseline, for each other run:
#    - paired per-image differences over the same 688 images
#    - an image-level paired bootstrap on Q0  (the PRIMARY inference:
#      688 paired images carry far more power than a 3-seed t-interval,
#      which at 3 seeds needs a seed-sd below 0.008 to clear +0.02)
#    - the same split by realized detail gain, declared in advance
#    - the PILOT rule: kill only if the mean Q0 difference is <= 0.
#      Killing anything that misses the +0.02 gate at one seed discards
#      a real +0.02 effect half the time; the <= 0 rule, 10% of the time.
#
#  A single-seed result here is a FILTER, never a result. Nothing from
#  this cell goes in the paper without seeds 43 and 44.
# =====================================================================
import os, csv, json, glob, time
import numpy as np
from PIL import Image
import torch, torch.nn as nn, torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models
import torchvision.transforms.functional as TF

SAVE, C768 = "/content/drive/MyDrive", "/content/cache768"
STAGE   = f"{SAVE}/stage2"
BASE    = "S-384_seed42"           # the baseline every run is paired against
TRAIN_RES, FULL_RES = 384, 768
BATCH_EVAL = 4
NBOOT   = 10000
FROZEN  = [0.059468, 0.138448, 0.300831]
KILL_AT = 0.0                      # pilot rule, declared before looking
try:
    from google.colab import drive; drive.mount("/content/drive", force_remount=False)
except Exception: pass
DEV = torch.device("cuda" if torch.cuda.is_available() else "cpu")

SP = json.load(open(f"{SAVE}/plantseg_splits.json"))
I768 = json.load(open(f"{C768}/index.json"))
row = I768["row_of"]
CLEAN_VAL = sorted(set(SP["clean_val"]) & set(I768["stems"]))
DSDIR = "/content/my_images/plantseg"
meta = {os.path.splitext(r["Name"])[0]: r for r in csv.DictReader(
        open(f"{DSDIR}/Metadata.csv", encoding="utf-8-sig", newline=""))}
frac = {k: float(v["Mask ratio"]) for k, v in meta.items()}
def quart(s):
    f = frac[s]
    return 0 if f <= FROZEN[0] else 1 if f <= FROZEN[1] else 2 if f <= FROZEN[2] else 3
QV = np.array([quart(s) for s in CLEAN_VAL])
res = {k: tuple(int(v) for v in meta[k]["Resolution"].lower().split("x")[:2])
       for k in CLEAN_VAL}
GAIN = np.array([min(min(res[s]), FULL_RES)/TRAIN_RES for s in CLEAN_VAL])
print(f"clean_val {len(CLEAN_VAL)}  quartiles {[int((QV==q).sum()) for q in range(4)]}")

MEAN, STD = [0.485,0.456,0.406], [0.229,0.224,0.225]
NORM = transforms.Normalize(MEAN, STD)
_mm = {}
def cache():
    if not _mm:
        _mm["i"] = np.load(I768["img_path"], mmap_mode="r")
        _mm["m"] = np.load(I768["msk_path"], mmap_mode="r")
    return _mm["i"], _mm["m"]
def strip_detail(im):
    return im.resize((TRAIN_RES,)*2, Image.BILINEAR).resize((FULL_RES,)*2, Image.BILINEAR)

class EvalSet(Dataset):
    def __init__(s, st, up): s.st, s.up = list(st), up
    def __len__(s): return len(s.st)
    def __getitem__(s, i):
        I, M = cache(); r = row[s.st[i]]
        img = Image.fromarray(I[r])
        if s.up: img = strip_detail(img)
        y = torch.from_numpy((np.asarray(M[r]) > 0).astype(np.float32)).unsqueeze(0)
        return NORM(TF.to_tensor(img)), y

SPEC = dict(channels=(64,256,512,1024,2048), decoder=(512,256,128,64))
class DoubleConv(nn.Module):
    def __init__(s,i,o):
        super().__init__()
        s.net = nn.Sequential(nn.Conv2d(i,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True),
                              nn.Conv2d(o,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True))
    def forward(s,x): return s.net(x)
class Up(nn.Module):
    def __init__(s,i,k,o):
        super().__init__(); s.reduce=nn.Conv2d(i,i//2,1,bias=False); s.conv=DoubleConv(i//2+k,o)
    def forward(s,x,skip):
        x = F.interpolate(x, size=skip.shape[-2:], mode="bilinear", align_corners=False)
        return s.conv(torch.cat([s.reduce(x), skip], 1))
class ResUNet(nn.Module):
    def __init__(s):
        super().__init__()
        base = models.resnet50(weights=None)
        c0,c1,c2,c3,c4 = SPEC["channels"]; d1,d2,d3,d4 = SPEC["decoder"]
        L = list(base.children())
        s.stem=nn.Sequential(*L[:3]); s.pool=L[3]
        s.l1,s.l2,s.l3,s.l4 = L[4],L[5],L[6],L[7]
        s.u1,s.u2,s.u3,s.u4 = Up(c4,c3,d1),Up(d1,c2,d2),Up(d2,c1,d3),Up(d3,c0,d4)
        s.head = nn.Sequential(nn.Conv2d(d4,32,3,padding=1,bias=False),
                               nn.BatchNorm2d(32), nn.ReLU(True), nn.Conv2d(32,1,1))
    def forward(s,x):
        size=x.shape[-2:]
        x0=s.stem(x); x1=s.l1(s.pool(x0)); x2=s.l2(x1); x3=s.l3(x2); x4=s.l4(x3)
        d=s.u1(x4,x3); d=s.u2(d,x2); d=s.u3(d,x1); d=s.u4(d,x0)
        return s.head(F.interpolate(d,size=size,mode="bilinear",align_corners=False))

@torch.no_grad()
def prob_768(net, x, mode):
    if mode == "full768":
        return torch.sigmoid(net(x).float())
    if mode == "whole384":
        d = F.interpolate(x, size=(TRAIN_RES,)*2, mode="bilinear", align_corners=False)
        p = torch.sigmoid(net(d).float())
        return F.interpolate(p, size=(FULL_RES,)*2, mode="bilinear", align_corners=False)
    acc = torch.zeros(x.shape[0], 1, FULL_RES, FULL_RES, device=x.device)
    cnt = torch.zeros_like(acc)
    ORIG = (0, (FULL_RES-TRAIN_RES)//2, FULL_RES-TRAIN_RES)
    for oy in ORIG:
        for ox in ORIG:
            t = x[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES]
            acc[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += torch.sigmoid(net(t).float())
            cnt[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += 1
    return acc/cnt.clamp(min=1)

THS = [round(t, 2) for t in np.arange(0.10, 0.91, 0.05)]
@torch.no_grad()
def score(bp, mode, up):
    net = ResUNet().to(DEV); net.load_state_dict(torch.load(bp, map_location=DEV)); net.eval()
    dl = DataLoader(EvalSet(CLEAN_VAL, up), BATCH_EVAL, shuffle=False,
                    num_workers=min(2, os.cpu_count() or 1), pin_memory=DEV.type == "cuda")
    n = len(CLEAN_VAL)
    per = np.zeros((n, len(THS))); Ipx = np.zeros((n, len(THS)))
    P = np.zeros((n, len(THS))); G = np.zeros(n); k = 0
    for x, y in dl:
        x, y = x.to(DEV, non_blocking=True), y.to(DEV, non_blocking=True)
        with torch.amp.autocast("cuda", enabled=DEV.type == "cuda"):
            p = prob_768(net, x, mode)
        b = x.shape[0]; yf = y.view(b, -1); pf = p.view(b, -1)
        G[k:k+b] = yf.sum(1).cpu().numpy()
        for j, t in enumerate(THS):
            pr = (pf > t).float(); it = (pr*yf).sum(1); pp = pr.sum(1)
            Ipx[k:k+b, j] = it.cpu().numpy(); P[k:k+b, j] = pp.cpu().numpy()
            per[k:k+b, j] = (it/(pp+yf.sum(1)-it).clamp(min=1)).cpu().numpy()
        k += b
    pooled = Ipx.sum(0)/np.maximum(P.sum(0)+G.sum()-Ipx.sum(0), 1)
    j = int(np.argmax(pooled))
    del net; torch.cuda.empty_cache()
    return dict(th=THS[j], pooled=float(pooled[j]), per=per[:, j])

# ------------------------- score what is finished ----------------------
runs = {}
for sm in sorted(glob.glob(f"{STAGE}/*/summary.json")):
    d = os.path.dirname(sm); name = os.path.basename(d)
    if not os.path.exists(f"{d}/best.pt"):
        print(f"   {name}: no best.pt, skipped"); continue
    J = json.load(open(sm))
    npz = f"{d}/per_image.npz"
    if os.path.exists(npz):
        z = np.load(npz, allow_pickle=True)
        if list(z["stems"]) == CLEAN_VAL:
            runs[name] = dict(per=z["per"], th=float(z["th"]), pooled=float(z["pooled"]),
                              mode=str(z["mode"]), cfg=J); print(f"   {name}: cached"); continue
    t = time.time()
    r = score(f"{d}/best.pt", J["select_on"], J["input"] == "crop_768_up")
    np.savez(npz, per=r["per"], th=r["th"], pooled=r["pooled"],
             mode=J["select_on"], stems=np.array(CLEAN_VAL))
    runs[name] = dict(**r, mode=J["select_on"], cfg=J)
    print(f"   {name}: scored in {time.time()-t:.0f}s  "
          f"mode {J['select_on']} th {r['th']} pooled {r['pooled']:.4f}")

if BASE not in runs:
    raise SystemExit(f"baseline {BASE} not scored yet -- run it first")
print(f"\n{len(runs)} run(s) scored; baseline {BASE}")

# --------------------------- paired comparison -------------------------
rng = np.random.default_rng(0)
b_per = runs[BASE]["per"]; q0 = QV == 0
bands = {"Q0 all": q0, "Q0 gain<=1.0": q0 & (GAIN <= 1.0),
         "Q0 gain>=1.5": q0 & (GAIN >= 1.5), "all images": np.ones(len(QV), bool)}

def boot(d, sel):
    v = d[sel]
    if len(v) < 5: return float("nan"), float("nan"), float("nan")
    idx = rng.integers(0, len(v), (NBOOT, len(v)))
    bs = v[idx].mean(1)
    return float(v.mean()), float(np.percentile(bs, 2.5)), float(np.percentile(bs, 97.5))

print("\n" + "=" * 88)
print(f"paired per-image IoU difference vs {BASE}, clean_val, "
      f"{NBOOT} image-level bootstraps")
print("=" * 88)
out = {}
for name, r in runs.items():
    if name == BASE: continue
    d = r["per"] - b_per
    print(f"\n{name}   ({r['cfg']['input']} / {r['cfg']['loss']}"
          f"{' / jitter' if r['cfg']['jitter'] else ''})   "
          f"pooled {r['pooled']:.4f} vs {runs[BASE]['pooled']:.4f}")
    print(f"   {'band':16}{'n':>6}{'mean diff':>12}{'95% CI':>22}")
    row = {}
    for lab, sel in bands.items():
        m, lo, hi = boot(d, sel)
        star = "  *" if (lo > 0 or hi < 0) else ""
        print(f"   {lab:16}{int(sel.sum()):>6}{m:>12.4f}"
              f"   [{lo:+.4f}, {hi:+.4f}]{star}")
        row[lab] = dict(n=int(sel.sum()), mean=m, lo=lo, hi=hi)
    mq0 = row["Q0 all"]["mean"]
    verdict = ("KILL -- Q0 moves the wrong way" if mq0 <= KILL_AT else
               "KEEP -- run seeds 43 and 44")
    print(f"   PILOT ({'<=' if mq0 <= KILL_AT else '>'} {KILL_AT:+.2f}): {verdict}")
    if row["Q0 gain>=1.5"]["n"] >= 20 and row["Q0 gain<=1.0"]["n"] >= 20:
        sep = row["Q0 gain>=1.5"]["mean"] - row["Q0 gain<=1.0"]["mean"]
        print(f"   mechanism check: high-detail minus no-detail = {sep:+.4f}"
              f"   {'(consistent with resolution)' if sep > 0 else '(not resolution)'}")
        row["mechanism_sep"] = sep
    out[name] = row

print("\n" + "=" * 88)
print("* marks a CI excluding zero. That is an image-level interval: it says the")
print("difference is not sampling noise over these 688 images, NOT that it would")
print("survive another training seed. Only seeds 43 and 44 answer that, and the")
print("gate is judged on the three-seed paired mean, not on anything here.")
json.dump(dict(baseline=BASE, kill_at=KILL_AT, nboot=NBOOT, results=out),
          open(f"{SAVE}/stage2_compare.json", "w"), indent=1, default=float)
print(f"\nwrote {SAVE}/stage2_compare.json")
```

```python
# =====================================================================
#  cell G -- what is the detail-gain band actually measuring?
#  CPU only, seconds. Reads the per_image.npz files cell F writes.
#
#  Why this cell exists.  S-768c showed a +0.0505 Q0 separation between
#  high-detail and no-detail images, which looked like resolution doing
#  the work.  S-768up -- identical crop geometry, all detail above 384
#  destroyed -- showed +0.0529.  The control removes MOST of the detail
#  from the high-gain band and almost NONE from the no-gain band (a
#  natively small image has no detail above 384 to lose), so if the
#  separation were detail it had to shrink.  It did not.  The band is a
#  proxy for something else.  This cell tests the three candidates:
#
#    A  difference of differences, with a CI, so the null is quantified
#    B  regression to the mean -- do high-gain Q0 images simply START
#       lower under the baseline, so any noisy change gains more there?
#    C  composition -- are the two bands different images (species,
#       disease, lesion size, absolute lesion pixel count)?
#
#  B is tested two ways: stratified (separation recomputed inside
#  baseline-IoU terciles) and by regression (band coefficient adjusted
#  for baseline IoU).  If the band effect survives both, it is about the
#  images themselves and C says which way.
# =====================================================================
import os, csv, json
from collections import Counter
import numpy as np

SAVE   = "/content/drive/MyDrive"
STAGE  = f"{SAVE}/stage2"
DSDIR  = "/content/my_images/plantseg"
FROZEN = [0.059468, 0.138448, 0.300831]
BASE   = "S-384_seed42"
ARMS   = ["S-768c_seed42", "S-768up_seed42"]
NBOOT  = 10000
HI_CUT, LO_CUT = 1.5, 1.0          # the bands, declared before any run
try:
    from google.colab import drive; drive.mount("/content/drive", force_remount=False)
except Exception: pass

# ------------------------------ metadata ------------------------------
rows = list(csv.DictReader(open(f"{DSDIR}/Metadata.csv", encoding="utf-8-sig", newline="")))
meta = {os.path.splitext(r["Name"])[0]: r for r in rows}
cols = list(rows[0].keys())

def pick(*pats):
    for c in cols:
        lc = c.lower()
        if any(p in lc for p in pats) and "mask" not in lc and "name" not in lc:
            return c
    return None
CPLANT = pick("plant", "species", "crop", "host")
CDIS   = pick("disease")
print("metadata columns:", cols)
print(f"using plant column {CPLANT!r}, disease column {CDIS!r}\n")

def load(name):
    z = np.load(f"{STAGE}/{name}/per_image.npz", allow_pickle=True)
    return [str(s) for s in z["stems"]], np.asarray(z["per"], dtype=np.float64)

stems, base = load(BASE)
d = {}
for a in ARMS:
    s, v = load(a)
    if s != stems:
        raise SystemExit(f"{a} was scored on a different image list -- rerun cell F")
    d[a] = v - base                      # paired per-image difference

frac = {k: float(v["Mask ratio"]) for k, v in meta.items()}
def quart(s):
    f = frac[s]
    return 0 if f <= FROZEN[0] else 1 if f <= FROZEN[1] else 2 if f <= FROZEN[2] else 3

def wh(s):
    a, b_ = meta[s]["Resolution"].lower().split("x")[:2]
    return int(float(a.strip())), int(float(b_.strip()))

QV    = np.array([quart(s) for s in stems])
SHORT = np.array([min(wh(s)) for s in stems], dtype=np.float64)
AREA  = np.array([wh(s)[0] * wh(s)[1] for s in stems], dtype=np.float64)
FR    = np.array([frac[s] for s in stems])
GAIN  = np.minimum(SHORT, 768.0) / 384.0
q0    = QV == 0
HI    = q0 & (GAIN >= HI_CUT)
LO    = q0 & (GAIN <= LO_CUT)
ih, il = np.flatnonzero(HI), np.flatnonzero(LO)
print(f"clean_val {len(stems)}   Q0 {int(q0.sum())}   "
      f"high-gain {len(ih)}   no-gain {len(il)}   middle {int(q0.sum())-len(ih)-len(il)}")

rng = np.random.default_rng(0)
bh = rng.integers(0, len(ih), (NBOOT, len(ih)))
bl = rng.integers(0, len(il), (NBOOT, len(il)))
def ci(v): return float(np.percentile(v, 2.5)), float(np.percentile(v, 97.5))

# ------------- A  difference of differences, with a CI ----------------
print("\n" + "=" * 78)
print("A  band separation per arm, and the difference between them")
print("=" * 78)
sep, sepb = {}, {}
for a in ARMS:
    sep[a]  = float(d[a][ih].mean() - d[a][il].mean())
    sepb[a] = d[a][ih][bh].mean(1) - d[a][il][bl].mean(1)
    lo, hi = ci(sepb[a])
    print(f"   {a:18} sep {sep[a]:+.4f}   95% CI [{lo:+.4f}, {hi:+.4f}]")
dod   = sep[ARMS[0]] - sep[ARMS[1]]
dod_b = sepb[ARMS[0]] - sepb[ARMS[1]]
dod_lo, dod_hi = ci(dod_b)
print(f"\n   detail effect = sep(real detail) - sep(detail stripped)")
print(f"                 = {dod:+.4f}   95% CI [{dod_lo:+.4f}, {dod_hi:+.4f}]"
      f"   {'NOT zero' if (dod_lo > 0 or dod_hi < 0) else 'indistinguishable from zero'}")
print(f"   the control keeps crop geometry and destroys detail above 384, so a")
print(f"   detail mechanism requires this to be clearly positive.  the interval's")
print(f"   upper end is the largest detail effect this study can still rule IN.")

# ------------- B  regression to the mean -------------------------------
print("\n" + "=" * 78)
print("B  does the high-gain band simply start lower under the baseline?")
print("=" * 78)
mh, ml = float(base[HI].mean()), float(base[LO].mean())
print(f"   baseline {BASE} mean per-image IoU:")
print(f"      high-gain {mh:.4f} (n={len(ih)})   no-gain {ml:.4f} (n={len(il)})"
      f"   difference {mh-ml:+.4f}")
print(f"   a LOWER high-gain baseline would make any noisy change look better there.\n")

bq0   = base[q0]
edges = np.quantile(bq0, [0.0, 1/3, 2/3, 1.0]); edges[-1] += 1e-9
TBIN  = np.digitize(base, edges[1:-1])        # 0,1,2 over all images
print(f"   separation recomputed INSIDE baseline-IoU terciles of Q0")
print(f"   (tercile edges {edges[1]:.3f}, {edges[2]:.3f})")
print(f"      {'tercile':10}{'n hi':>6}{'n lo':>6}" + "".join(f"{a.split('_')[0]:>16}" for a in ARMS))
strat = {a: [] for a in ARMS}
for t in range(3):
    sh, sl = HI & (TBIN == t), LO & (TBIN == t)
    cells = []
    for a in ARMS:
        if sh.sum() >= 8 and sl.sum() >= 8:
            v = float(d[a][sh].mean() - d[a][sl].mean()); strat[a].append((v, int(sh.sum()+sl.sum())))
            cells.append(f"{v:>+16.4f}")
        else:
            cells.append(f"{'too few':>16}")
    print(f"      {t:<10}{int(sh.sum()):>6}{int(sl.sum()):>6}" + "".join(cells))
for a in ARMS:
    if strat[a]:
        w  = np.array([n for _, n in strat[a]], dtype=float)
        av = float(np.average([v for v, _ in strat[a]], weights=w))
        print(f"      {a:18} weighted within-tercile sep {av:+.4f}"
              f"   (unstratified {sep[a]:+.4f})")

sel = HI | LO
X   = np.column_stack([np.ones(int(sel.sum())), HI[sel].astype(float), base[sel]])
print(f"\n   least squares  diff ~ 1 + high_gain + baseline_IoU   (n={int(sel.sum())})")
for a in ARMS:
    c, *_ = np.linalg.lstsq(X, d[a][sel], rcond=None)
    print(f"      {a:18} band {c[1]:+.4f} adjusted   (raw {sep[a]:+.4f})"
          f"   baseline slope {c[2]:+.3f}")
print(f"   a band coefficient that collapses toward 0 once baseline IoU is in the")
print(f"   model means the band was standing in for how hard the images are.")

# ------------- C  are they different images at all? -------------------
print("\n" + "=" * 78)
print("C  composition of the two bands")
print("=" * 78)
SA = np.array(stems)
lesion_px = FR * AREA
for lab, s_ in (("high-gain", HI), ("no-gain", LO)):
    print(f"   {lab:10} short side median {np.median(SHORT[s_]):>7.0f}"
          f"   lesion fraction median {np.median(FR[s_]):.5f}"
          f"   lesion pixels in the ORIGINAL median {np.median(lesion_px[s_]):>9.0f}")
for col, lab in ((CPLANT, "plant"), (CDIS, "disease")):
    if not col: continue
    print(f"\n   most common {lab} per band")
    for bl_, s_ in (("high-gain", HI), ("no-gain", LO)):
        c = Counter(meta[s][col] for s in SA[s_])
        tot = sum(c.values())
        print(f"      {bl_:10} " + ",  ".join(f"{k} {v} ({100*v/tot:.0f}%)"
                                              for k, v in c.most_common(4)))
    ch = Counter(meta[s][col] for s in SA[HI]); cl = Counter(meta[s][col] for s in SA[LO])
    keys = set(ch) | set(cl)
    overlap = sum(min(ch.get(k, 0)/max(len(ih), 1), cl.get(k, 0)/max(len(il), 1)) for k in keys)
    print(f"      share of the two bands' {lab} mix that overlaps: {100*overlap:.0f}%"
          f"   (100% = same mix, 0% = disjoint)")

out = dict(sep=sep, detail_effect=dict(value=dod, lo=dod_lo, hi=dod_hi),
           baseline=dict(high=mh, low=ml, diff=mh-ml),
           n=dict(hi=len(ih), lo=len(il), q0=int(q0.sum())))
json.dump(out, open(f"{SAVE}/stage2_band_check.json", "w"), indent=1, default=float)
print(f"\nwrote {SAVE}/stage2_band_check.json")
```

```python
# =====================================================================
#  cell D -- Stage 2: supervised tests of components 1 and 2.
#  One cell, three switches, 25,000 steps, resume-safe. Validation only;
#  the test split is not touched until Stage 5.
#
#      INPUT_MODE in {full_384, crop_768, crop_768_up}
#      FG_LOSS    in {pooled, class_balanced, image_balanced}
#      JITTER     in {False, True}
#
#  (full_384, pooled, False) is the S-384 baseline.
#
#  crop_768_up is the control the crop statistics made necessary: uniform
#  384 crops of a 768 image carry 1.50x the Q0 lesion density of the
#  whole image, because uniform crop ORIGINS over-sample the centre (a
#  centre pixel sits in 384 of 385 origins per axis, a corner pixel in 1).
#  So crop_768 changes resolution AND density at once. crop_768_up takes
#  the 768 image, downsamples it to 384 and upsamples it back, destroying
#  exactly the detail above 384 and nothing else, then crops identically.
#  crop_768 beating crop_768_up isolates resolution; a tie says the gain
#  is crop geometry.
#
#  Everything reads ONE cache: /content/cache768 (537 labelled + 846 val).
#  The 441 cache is not needed, so prep.py need not be run this session.
# =====================================================================
import os, csv, json, math, time, random, shutil, zipfile, collections
import numpy as np
import matplotlib; matplotlib.use("Agg")
import matplotlib.pyplot as plt
from PIL import Image, ImageFilter
import torch, torch.nn as nn, torch.optim as optim, torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models
import torchvision.transforms.functional as TF

# ------------------------- the three switches --------------------------
INPUT_MODE = "full_384"        # full_384 | crop_768 | crop_768_up
FG_LOSS    = "class_balanced"          # pooled | class_balanced | image_balanced
JITTER     = False             # True only for S-768c-jit
TRAIN_SEED = 42
# -----------------------------------------------------------------------
FLOOR_FRAC = 0.01              # declared from check 0f: binds on 8.9% of
                               # non-empty Q0 crops, caps amplification 21.5x
SELECT_ON  = {"full_384": "whole384", "crop_768": "full768",
              "crop_768_up": "full768"}[INPUT_MODE]
MAX_STEPS  = 25000
EVAL_EVERY = 1000
SHARE_EVERY = 500              # per-quartile lesion-loss share
CKPT_EVERY = 2500
LR_PATIENCE = 3                # patience counts EVAL_EVERY blocks, now 1000
TRAIN_RES  = 384
FULL_RES   = 768
BATCH      = 4                 # matches the existing pipeline; batch 8
                               # doubles every run to ~2.2h and Stage 2 to ~42h
BATCH_EVAL = 4                 # 768 input on a T4
LR_ENCODER, LR_DECODER, WEIGHT_DECAY = 1e-5, 1e-4, 1e-4
FROZEN     = [0.059468, 0.138448, 0.300831]
SAVE, C768 = "/content/drive/MyDrive", "/content/cache768"
NUM_WORKERS = min(2, os.cpu_count() or 1)
FRESH      = True              # wipes this run's folder. Set back to False
                               # once it is past step 2500 so a dropped
                               # session resumes instead of restarting.

TAG = {"full_384": "384", "crop_768": "768c", "crop_768_up": "768up"}[INPUT_MODE]
if FG_LOSS == "class_balanced":  TAG += "-cb"
if FG_LOSS == "image_balanced":  TAG += "-bal"
if JITTER:                       TAG += "-jit"
RUN = f"S-{TAG}_seed{TRAIN_SEED}"
OUT = f"{SAVE}/stage2/{RUN}"
if FRESH and os.path.isdir(OUT): shutil.rmtree(OUT, ignore_errors=True)
os.makedirs(OUT, exist_ok=True)

random.seed(TRAIN_SEED); np.random.seed(TRAIN_SEED)
torch.manual_seed(TRAIN_SEED); torch.cuda.manual_seed_all(TRAIN_SEED)
torch.backends.cudnn.deterministic = True; torch.backends.cudnn.benchmark = False
DEV = torch.device("cuda" if torch.cuda.is_available() else "cpu")
try:
    from google.colab import drive; drive.mount("/content/drive", force_remount=False)
except Exception: pass
print(f"{RUN}\ndevice {DEV} {torch.cuda.get_device_name(0) if DEV.type=='cuda' else ''}")
print(f"input {INPUT_MODE} | loss {FG_LOSS} | jitter {JITTER} | "
      f"select on {SELECT_ON} | floor {FLOOR_FRAC} | batch {BATCH}")

# --------------------------- data and asserts --------------------------
SP = json.load(open(f"{SAVE}/plantseg_splits.json"))
if max(abs(a-b) for a, b in zip(SP["quartile_cuts"], FROZEN)) > 1e-6:
    raise SystemExit(f"cuts {SP['quartile_cuts']} != frozen {FROZEN}")
if {k: len(v) for k, v in SP["official_splits"].items()} != \
   {"train": 5367, "val": 846, "test": 1561}:
    raise SystemExit("split sizes are not 5367/846/1561")
I768 = json.load(open(f"{C768}/index.json"))
if I768["load_size"] != FULL_RES:
    raise SystemExit(f"cache is {I768['load_size']}, expected {FULL_RES}")
row = I768["row_of"]
LAB = sorted(I768["labelled"])
CLEAN_VAL = sorted(set(SP["clean_val"]) & set(I768["stems"]))
if len(LAB) != 537: raise SystemExit(f"{len(LAB)} labelled, expected 537")
DSDIR = "/content/my_images/plantseg"
meta = {os.path.splitext(r["Name"])[0]: r for r in csv.DictReader(
        open(f"{DSDIR}/Metadata.csv", encoding="utf-8-sig", newline=""))}
frac = {k: float(v["Mask ratio"]) for k, v in meta.items()}
def quart(s):
    f = frac[s]
    return 0 if f <= FROZEN[0] else 1 if f <= FROZEN[1] else 2 if f <= FROZEN[2] else 3
QV = np.array([quart(s) for s in CLEAN_VAL])
print(f"labelled {len(LAB)}   clean_val {len(CLEAN_VAL)} "
      f"(quartiles {[int((QV==q).sum()) for q in range(4)]})")
print("   gates are judged on clean_val, not the full 846: full val carries")
print("   17.14% near-duplicates of training images.")

MEAN, STD = [0.485,0.456,0.406], [0.229,0.224,0.225]
NORM = transforms.Normalize(MEAN, STD)
JIT  = transforms.ColorJitter(0.4, 0.4, 0.4, 0.1)
_mm = {}
def cache():
    if not _mm:
        _mm["i"] = np.load(I768["img_path"], mmap_mode="r")
        _mm["m"] = np.load(I768["msk_path"], mmap_mode="r")
    return _mm["i"], _mm["m"]

def strip_detail(im):
    """768 -> 384 -> 768. Destroys exactly the detail above 384, keeps
       geometry, size and lesion density identical."""
    return im.resize((TRAIN_RES,)*2, Image.BILINEAR).resize((FULL_RES,)*2, Image.BILINEAR)

def photometric(img):
    if random.random() < 0.8: img = JIT(img)
    if random.random() < 0.2: img = TF.to_grayscale(img, num_output_channels=3)
    if random.random() < 0.3:
        img = img.filter(ImageFilter.GaussianBlur(radius=random.uniform(0.1, 1.5)))
    return img

def geometric(img, msk):
    if random.random() < 0.5: img, msk = TF.hflip(img), TF.hflip(msk)
    if random.random() < 0.5: img, msk = TF.vflip(img), TF.vflip(msk)
    k = random.randint(0, 3)
    if k:
        img = img.rotate(90*k, expand=True, resample=Image.BILINEAR)
        msk = msk.rotate(90*k, expand=True, resample=Image.NEAREST)
    return img, msk

def rrc(img, msk, scale=(0.5, 1.0), ratio=(3/4, 4/3)):
    w, h = img.size; area = w*h
    for _ in range(10):
        t = area*random.uniform(*scale)
        ar = math.exp(random.uniform(math.log(ratio[0]), math.log(ratio[1])))
        cw, ch = int(round(math.sqrt(t*ar))), int(round(math.sqrt(t/ar)))
        if cw <= w and ch <= h:
            x, y = random.randint(0, w-cw), random.randint(0, h-ch)
            return img.crop((x, y, x+cw, y+ch)), msk.crop((x, y, x+cw, y+ch))
    return img, msk

class Labelled(Dataset):
    """full_384   : whole image, rrc + geometric, resized to 384
       crop_768   : uniform 384 window of the 768 image (the crop IS the
                    spatial augmentation, so rrc is not applied on top)
       crop_768_up: the same window of a detail-stripped 768 image"""
    def __init__(s, st): s.st = list(st)
    def __len__(s): return len(s.st)
    def __getitem__(s, i):
        I, M = cache(); r = row[s.st[i]]
        img = Image.fromarray(I[r]); msk = Image.fromarray(M[r])
        if INPUT_MODE == "crop_768_up": img = strip_detail(img)
        if INPUT_MODE == "full_384":
            img, msk = rrc(img, msk)
            img, msk = geometric(img, msk)
            img = img.resize((TRAIN_RES,)*2, Image.BILINEAR)
            msk = msk.resize((TRAIN_RES,)*2, Image.NEAREST)
        else:
            side = (random.randint(256, FULL_RES) if JITTER else TRAIN_RES)
            x = random.randint(0, FULL_RES-side); y = random.randint(0, FULL_RES-side)
            img = img.crop((x, y, x+side, y+side)); msk = msk.crop((x, y, x+side, y+side))
            img, msk = geometric(img, msk)
            if side != TRAIN_RES:
                img = img.resize((TRAIN_RES,)*2, Image.BILINEAR)
                msk = msk.resize((TRAIN_RES,)*2, Image.NEAREST)
        y_ = torch.from_numpy((np.array(msk) > 0).astype(np.float32)).unsqueeze(0)
        return NORM(TF.to_tensor(photometric(img))), y_, quart(s.st[i])

class EvalSet(Dataset):
    """input follows INPUT_MODE; the TARGET is always the 768 mask."""
    def __init__(s, st): s.st = list(st)
    def __len__(s): return len(s.st)
    def __getitem__(s, i):
        I, M = cache(); r = row[s.st[i]]
        img = Image.fromarray(I[r])
        if INPUT_MODE == "crop_768_up": img = strip_detail(img)
        y = torch.from_numpy((np.asarray(M[r]) > 0).astype(np.float32)).unsqueeze(0)
        return NORM(TF.to_tensor(img)), y            # image at 768, mask at 768

def seed_worker(_):
    s = torch.initial_seed() % 2**32; np.random.seed(s); random.seed(s)
LK = dict(num_workers=NUM_WORKERS, pin_memory=DEV.type == "cuda",
          worker_init_fn=seed_worker, persistent_workers=NUM_WORKERS > 0)
g = torch.Generator(); g.manual_seed(TRAIN_SEED)
loader_l = DataLoader(Labelled(LAB), BATCH, shuffle=True, drop_last=True,
                      generator=g, **LK)
val_loader = DataLoader(EvalSet(CLEAN_VAL), BATCH_EVAL, shuffle=False, **LK)
def cycle(dl):
    while True:
        for b in dl: yield b

# ------------------------------- model ---------------------------------
SPEC = dict(channels=(64,256,512,1024,2048), decoder=(512,256,128,64))
class DoubleConv(nn.Module):
    def __init__(s,i,o):
        super().__init__()
        s.net = nn.Sequential(nn.Conv2d(i,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True),
                              nn.Conv2d(o,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True))
    def forward(s,x): return s.net(x)
class Up(nn.Module):
    def __init__(s,i,k,o):
        super().__init__(); s.reduce=nn.Conv2d(i,i//2,1,bias=False); s.conv=DoubleConv(i//2+k,o)
    def forward(s,x,skip):
        x = F.interpolate(x, size=skip.shape[-2:], mode="bilinear", align_corners=False)
        return s.conv(torch.cat([s.reduce(x), skip], 1))
class ResUNet(nn.Module):
    def __init__(s):
        super().__init__()
        base = models.resnet50(weights=models.ResNet50_Weights.DEFAULT)
        c0,c1,c2,c3,c4 = SPEC["channels"]; d1,d2,d3,d4 = SPEC["decoder"]
        L = list(base.children())
        s.stem=nn.Sequential(*L[:3]); s.pool=L[3]
        s.l1,s.l2,s.l3,s.l4 = L[4],L[5],L[6],L[7]
        s.u1,s.u2,s.u3,s.u4 = Up(c4,c3,d1),Up(d1,c2,d2),Up(d2,c1,d3),Up(d3,c0,d4)
        s.head = nn.Sequential(nn.Conv2d(d4,32,3,padding=1,bias=False),
                               nn.BatchNorm2d(32), nn.ReLU(True), nn.Conv2d(32,1,1))
    def enc_params(s):
        for m in (s.stem,s.l1,s.l2,s.l3,s.l4): yield from m.parameters()
    def dec_params(s):
        for m in (s.u1,s.u2,s.u3,s.u4,s.head): yield from m.parameters()
    def forward(s,x):
        size=x.shape[-2:]
        x0=s.stem(x); x1=s.l1(s.pool(x0)); x2=s.l2(x1); x3=s.l3(x2); x4=s.l4(x3)
        d=s.u1(x4,x3); d=s.u2(d,x2); d=s.u3(d,x1); d=s.u4(d,x0)
        return s.head(F.interpolate(d,size=size,mode="bilinear",align_corners=False))

# ------------------------------- losses --------------------------------
bce_none = nn.BCEWithLogitsLoss(reduction="none")
def tversky(logits, tgt, eps=1.0):
    p = torch.sigmoid(logits)
    tp = (p*tgt).sum((1,2,3)); fp = (p*(1-tgt)).sum((1,2,3)); fn = ((1-p)*tgt).sum((1,2,3))
    return 1 - (tp+eps)/(tp+0.5*fp+0.5*fn+eps)

def fg_term(logits, tgt):
    """returns (per-image BCE term, per-image lesion-only contribution).
       All three forms are scaled so a uniform per-pixel loss c gives c,
       so the learning rate and the BCE/Tversky balance do not shift
       between configurations. The plan's formula writes the sum of the
       two class means; using their average is the same comparison at the
       same scale."""
    e = bce_none(logits, tgt)
    sf = tgt.sum((1,2,3)); sb = (1-tgt).sum((1,2,3))
    ef = (e*tgt).sum((1,2,3)); eb = (e*(1-tgt)).sum((1,2,3))
    if FG_LOSS == "pooled":
        return e.mean((1,2,3)), ef/tgt[0].numel()
    if FG_LOSS == "class_balanced":                 # batch-level class means
        n = logits.shape[0]
        mf = ef.sum()/sf.sum().clamp(min=1.0); mb = eb.sum()/sb.sum().clamp(min=1.0)
        return (0.5*(mf+mb)).expand(n), 0.5*ef/sf.sum().clamp(min=1.0)
    floor = FLOOR_FRAC*tgt[0].numel()               # per-image, with the floor
    tf = ef/torch.clamp(sf, min=floor)
    tb = eb/sb.clamp(min=1.0)
    return 0.5*(tf+tb), 0.5*tf

def sup_loss(logits, tgt):
    b, lesion = fg_term(logits, tgt)
    return b + tversky(logits, tgt), lesion

# --------------------------- 768 inference -----------------------------
@torch.no_grad()
def prob_768(net, x, mode):
    """x is the model's input image at 768. Returns a 768 probability map."""
    if mode == "full768":
        return torch.sigmoid(net(x).float())
    if mode == "whole384":
        d = F.interpolate(x, size=(TRAIN_RES,)*2, mode="bilinear", align_corners=False)
        p = torch.sigmoid(net(d).float())
        return F.interpolate(p, size=(FULL_RES,)*2, mode="bilinear", align_corners=False)
    # genuinely overlapping: stride 192 gives 3x3 tiles at 50% overlap.
    # Origins (0, 384) alone would ABUT, not overlap, and blending would
    # be a no-op -- the seam is exactly where a tiled model is weakest.
    acc = torch.zeros(x.shape[0], 1, FULL_RES, FULL_RES, device=x.device)
    cnt = torch.zeros_like(acc)
    ORIG = (0, (FULL_RES-TRAIN_RES)//2, FULL_RES-TRAIN_RES)
    for oy in ORIG:
        for ox in ORIG:
            t = x[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES]
            acc[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += torch.sigmoid(net(t).float())
            cnt[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += 1
    return acc/cnt.clamp(min=1)                      # blend, threshold afterwards

THS = [round(t, 2) for t in np.arange(0.10, 0.91, 0.05)]
@torch.no_grad()
def evaluate(net, mode):
    net.eval()
    nI = len(CLEAN_VAL); per = np.zeros((nI, len(THS))); P = np.zeros((nI, len(THS)))
    Ipx = np.zeros((nI, len(THS))); G = np.zeros(nI); k = 0
    for x, y in val_loader:
        x, y = x.to(DEV, non_blocking=True), y.to(DEV, non_blocking=True)
        with torch.amp.autocast("cuda", enabled=DEV.type == "cuda"):
            p = prob_768(net, x, mode)
        b = x.shape[0]; yf = y.view(b, -1)
        G[k:k+b] = yf.sum(1).cpu().numpy()
        pf = p.view(b, -1)
        for j, t in enumerate(THS):
            pr = (pf > t).float()
            it = (pr*yf).sum(1); pp = pr.sum(1)
            Ipx[k:k+b, j] = it.cpu().numpy(); P[k:k+b, j] = pp.cpu().numpy()
            per[k:k+b, j] = (it/(pp+yf.sum(1)-it).clamp(min=1)).cpu().numpy()
        k += b
    pooled = Ipx.sum(0)/np.maximum(P.sum(0)+G.sum()-Ipx.sum(0), 1)
    j = int(np.argmax(pooled)); t = THS[j]
    byq = {q: float(per[QV == q, j].mean()) for q in range(4)}
    return dict(th=t, pooled=float(pooled[j]), perimg=float(per[:, j].mean()),
                byq=byq, pred_frac=float(P[:, j].sum()/(len(CLEAN_VAL)*FULL_RES*FULL_RES)))

# --------------------------- resume or start ---------------------------
model = ResUNet().to(DEV)
opt = optim.AdamW([{"params": list(model.enc_params()), "lr": LR_ENCODER},
                   {"params": list(model.dec_params()), "lr": LR_DECODER}],
                  weight_decay=WEIGHT_DECAY)
sched = optim.lr_scheduler.ReduceLROnPlateau(opt, mode="max", patience=LR_PATIENCE, factor=0.5)
scaler = torch.amp.GradScaler("cuda")
step, best, history, shares = 0, -1.0, [], []
CK, BEST = f"{OUT}/ckpt.pt", f"{OUT}/best.pt"
if os.path.exists(CK):
    c = torch.load(CK, map_location=DEV, weights_only=False)
    model.load_state_dict(c["model"]); opt.load_state_dict(c["opt"])
    sched.load_state_dict(c["sched"]); scaler.load_state_dict(c["scaler"])
    step, best, history, shares = c["step"], c["best"], c["history"], c.get("shares", [])
    random.setstate(c["py_rng"]); np.random.set_state(c["np_rng"])
    torch.set_rng_state(c["t_rng"].cpu())
    if c.get("cuda_rng") is not None and DEV.type == "cuda":
        torch.cuda.set_rng_state(c["cuda_rng"].cpu())
    print(f"RESUMED from step {step} (best {best:.4f})")
else:
    print("starting fresh")

def save_ckpt():
    torch.save(dict(model=model.state_dict(), opt=opt.state_dict(), sched=sched.state_dict(),
                    scaler=scaler.state_dict(), step=step, best=best, history=history,
                    shares=shares, py_rng=random.getstate(), np_rng=np.random.get_state(),
                    t_rng=torch.get_rng_state(),
                    cuda_rng=torch.cuda.get_rng_state() if DEV.type == "cuda" else None), CK)

# ------------------------------- train ---------------------------------
it = cycle(loader_l)
print("=" * 76); print(f"{RUN}  --  {len(LAB)} labelled, {MAX_STEPS} steps"); print("=" * 76)
t0 = t_last = time.time()
r_loss = r_n = 0.0; q_les = np.zeros(4); q_cnt = np.zeros(4)
while step < MAX_STEPS:
    model.train()
    x, y, q = next(it)
    x, y = x.to(DEV, non_blocking=True), y.to(DEV, non_blocking=True)
    opt.zero_grad(set_to_none=True)
    with torch.amp.autocast("cuda", enabled=DEV.type == "cuda"):
        lg = model(x)
        vec, lesion = sup_loss(lg.float(), y)
        loss = vec.mean()
    scaler.scale(loss).backward(); scaler.step(opt); scaler.update()
    step += 1
    r_loss += loss.item(); r_n += 1
    lz = lesion.detach().float().cpu().numpy(); qz = q.numpy()
    for b in range(len(qz)): q_les[qz[b]] += lz[b]; q_cnt[qz[b]] += 1

    if step % SHARE_EVERY == 0:
        tot = max(q_les.sum(), 1e-12)
        sh = (q_les/tot).round(4).tolist()
        shares.append(dict(step=step, share=sh, seen=q_cnt.astype(int).tolist()))
        print(f"   [step {step}] lesion-loss share by quartile {sh}"
              f"  (images seen {q_cnt.astype(int).tolist()})")
        q_les[:] = 0; q_cnt[:] = 0

    if step % EVAL_EVERY == 0:
        m = evaluate(model, SELECT_ON)
        prev = opt.param_groups[-1]["lr"]; sched.step(m["pooled"])
        cur = opt.param_groups[-1]["lr"]
        saved = ""
        if m["pooled"] > best:
            best = m["pooled"]; torch.save(model.state_dict(), BEST); saved = "  <-- saved"
        history.append(dict(step=step, train_loss=r_loss/max(r_n, 1), lr=cur, **m))
        sps = EVAL_EVERY/(time.time()-t_last); t_last = time.time()
        print(f"step {step:6d} | loss {r_loss/max(r_n,1):.4f} | val pooled {m['pooled']:.4f} "
              f"@{m['th']} | per-img {m['perimg']:.4f} | Q0 {m['byq'][0]:.4f} "
              f"Q3 {m['byq'][3]:.4f} | best {best:.4f} | {sps:.2f} it/s | "
              f"ETA {(MAX_STEPS-step)/sps/3600:.1f}h{saved}"
              + (f"  [lr {prev:.1e}->{cur:.1e}]" if cur != prev else ""))
        r_loss = r_n = 0.0
        json.dump(dict(run=RUN, input=INPUT_MODE, loss=FG_LOSS, jitter=JITTER,
                       seed=TRAIN_SEED, floor=FLOOR_FRAC, select_on=SELECT_ON,
                       batch=BATCH, history=history, shares=shares),
                  open(f"{OUT}/history.json", "w"), indent=1)
    if step % CKPT_EVERY == 0:
        save_ckpt(); print(f"    checkpointed at step {step}")
save_ckpt()
print(f"\ntraining done in {(time.time()-t0)/3600:.2f}h, best val pooled {best:.4f}")

# ------------------- final: all three inference modes ------------------
net = ResUNet().to(DEV); net.load_state_dict(torch.load(BEST, map_location=DEV)); net.eval()
print("\n" + "=" * 76)
print("clean_val under each inference mode (best.pt was selected on "
      f"{SELECT_ON})")
print("=" * 76)
print(f"{'mode':12}{'th':>6}{'pooled':>9}{'per-img':>9}"
      f"{'Q0':>8}{'Q1':>8}{'Q2':>8}{'Q3':>8}{'Q3-Q0':>8}")
modes = ["whole384", "full768", "tiles384"]
final = {}
for md in modes:
    m = evaluate(net, md); final[md] = m
    print(f"{md:12}{m['th']:>6.2f}{m['pooled']:>9.4f}{m['perimg']:>9.4f}"
          + "".join(f"{m['byq'][q]:>8.4f}" for q in range(4))
          + f"{m['byq'][3]-m['byq'][0]:>8.4f}")

# the stratified check: does any Q0 gain track the realized detail gain?
res = {k: tuple(int(v) for v in meta[k]["Resolution"].lower().split("x")[:2])
       for k in CLEAN_VAL}
gain = np.array([min(min(res[s]), FULL_RES)/TRAIN_RES for s in CLEAN_VAL])
q0 = QV == 0
print(f"\nclean_val Q0 split by realized worst-axis detail gain at {FULL_RES}:")
for lab, sel in (("<= 1.0 (no gain)", q0 & (gain <= 1.0)),
                 ("1.0 - 1.5", q0 & (gain > 1.0) & (gain < 1.5)),
                 (">= 1.5", q0 & (gain >= 1.5))):
    print(f"   {lab:18} n={int(sel.sum()):4d}")
print("   per-image Q0 IoU within each band is in the saved json; compare it")
print("   against the matching S-384 run, paired by image.")

np.save(f"{OUT}/clean_val_gain.npy", gain)
json.dump(dict(run=RUN, input=INPUT_MODE, loss=FG_LOSS, jitter=JITTER, seed=TRAIN_SEED,
               floor=FLOOR_FRAC, select_on=SELECT_ON, batch=BATCH, best_val_pooled=best,
               clean_val=CLEAN_VAL, quartiles=QV.tolist(), final=final,
               history=history, shares=shares),
          open(f"{OUT}/summary.json", "w"), indent=1)
xs = [h["step"] for h in history]
plt.figure(figsize=(14, 4))
plt.subplot(1, 3, 1); plt.plot(xs, [h["train_loss"] for h in history]); plt.title("train loss")
plt.subplot(1, 3, 2); plt.plot(xs, [h["pooled"] for h in history]); plt.title("val pooled IoU")
plt.subplot(1, 3, 3)
for q in range(4): plt.plot(xs, [h["byq"][q] for h in history], label=f"Q{q}")
plt.legend(fontsize=7); plt.title("val per-image IoU by quartile")
plt.tight_layout(); plt.savefig(f"{OUT}/curves.png", dpi=110); plt.close()
print(f"\nwrote {OUT}/summary.json, best.pt, curves.png")
print(">>> download best.pt to local disk before the session ends <<<")
```

```python
# =====================================================================
#  cell F -- run this ONCE after every cell D training. Standalone: it
#  reads nothing from cell D's variables, so it also works in a fresh
#  session or after a runtime restart.
#
#  For the target run it:
#    1. scores it per image on clean_val  -> per_image.npz
#       (the vectors every paired test needs; cell D saves only means)
#    2. archives summary / history / curves / gain / per_image to Drive
#    3. records config, GPU and a checksum in weights_manifest.json
#    4. deletes ckpt.pt, but ONLY if the run completed. 490 MB each, and
#       19 of them would overrun a free Drive.
#    5. prints what Drive is holding
#    6. downloads the archive and best.pt
#
#  Any OTHER run still missing per_image.npz or an archive is also
#  brought up to date, without being downloaded again.
# =====================================================================
import os, csv, json, glob, time, zipfile, shutil, hashlib
import numpy as np
from PIL import Image
import torch, torch.nn as nn, torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models
import torchvision.transforms.functional as TF

RUN_NAME    = None       # None = the most recently finished run in stage2/
DELETE_CKPT = True       # False if you intend to extend a run later
DOWNLOAD    = True
SAVE, C768  = "/content/drive/MyDrive", "/content/cache768"
STAGE, ARCH = f"{SAVE}/stage2", f"{SAVE}/stage2_archive"
TRAIN_RES, FULL_RES = 384, 768
BATCH_EVAL  = 12 if torch.cuda.is_available() else 2
FROZEN      = [0.059468, 0.138448, 0.300831]
try:
    from google.colab import drive; drive.mount("/content/drive", force_remount=False)
except Exception: pass
DEV = torch.device("cuda" if torch.cuda.is_available() else "cpu")
os.makedirs(ARCH, exist_ok=True)

# ------------------------------- context -------------------------------
SP   = json.load(open(f"{SAVE}/plantseg_splits.json"))
I768 = json.load(open(f"{C768}/index.json"))
row  = I768["row_of"]
CLEAN_VAL = sorted(set(SP["clean_val"]) & set(I768["stems"]))
DSDIR = "/content/my_images/plantseg"
meta = {os.path.splitext(r["Name"])[0]: r for r in csv.DictReader(
        open(f"{DSDIR}/Metadata.csv", encoding="utf-8-sig", newline=""))}
frac = {k: float(v["Mask ratio"]) for k, v in meta.items()}
def quart(s):
    f = frac[s]
    return 0 if f <= FROZEN[0] else 1 if f <= FROZEN[1] else 2 if f <= FROZEN[2] else 3
QV = np.array([quart(s) for s in CLEAN_VAL])

MEAN, STD = [0.485,0.456,0.406], [0.229,0.224,0.225]
NORM = transforms.Normalize(MEAN, STD)
_mm = {}
def cache():
    if not _mm:
        _mm["i"] = np.load(I768["img_path"], mmap_mode="r")
        _mm["m"] = np.load(I768["msk_path"], mmap_mode="r")
    return _mm["i"], _mm["m"]
def strip_detail(im):
    return im.resize((TRAIN_RES,)*2, Image.BILINEAR).resize((FULL_RES,)*2, Image.BILINEAR)

class EvalSet(Dataset):
    def __init__(s, st, up): s.st, s.up = list(st), up
    def __len__(s): return len(s.st)
    def __getitem__(s, i):
        I, M = cache(); r = row[s.st[i]]
        img = Image.fromarray(I[r])
        if s.up: img = strip_detail(img)
        y = torch.from_numpy((np.asarray(M[r]) > 0).astype(np.float32)).unsqueeze(0)
        return NORM(TF.to_tensor(img)), y

SPEC = dict(channels=(64,256,512,1024,2048), decoder=(512,256,128,64))
class DoubleConv(nn.Module):
    def __init__(s,i,o):
        super().__init__()
        s.net = nn.Sequential(nn.Conv2d(i,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True),
                              nn.Conv2d(o,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True))
    def forward(s,x): return s.net(x)
class Up(nn.Module):
    def __init__(s,i,k,o):
        super().__init__(); s.reduce=nn.Conv2d(i,i//2,1,bias=False); s.conv=DoubleConv(i//2+k,o)
    def forward(s,x,skip):
        x = F.interpolate(x, size=skip.shape[-2:], mode="bilinear", align_corners=False)
        return s.conv(torch.cat([s.reduce(x), skip], 1))
class ResUNet(nn.Module):
    def __init__(s):
        super().__init__()
        base = models.resnet50(weights=None)
        c0,c1,c2,c3,c4 = SPEC["channels"]; d1,d2,d3,d4 = SPEC["decoder"]
        L = list(base.children())
        s.stem=nn.Sequential(*L[:3]); s.pool=L[3]
        s.l1,s.l2,s.l3,s.l4 = L[4],L[5],L[6],L[7]
        s.u1,s.u2,s.u3,s.u4 = Up(c4,c3,d1),Up(d1,c2,d2),Up(d2,c1,d3),Up(d3,c0,d4)
        s.head = nn.Sequential(nn.Conv2d(d4,32,3,padding=1,bias=False),
                               nn.BatchNorm2d(32), nn.ReLU(True), nn.Conv2d(32,1,1))
    def forward(s,x):
        size=x.shape[-2:]
        x0=s.stem(x); x1=s.l1(s.pool(x0)); x2=s.l2(x1); x3=s.l3(x2); x4=s.l4(x3)
        d=s.u1(x4,x3); d=s.u2(d,x2); d=s.u3(d,x1); d=s.u4(d,x0)
        return s.head(F.interpolate(d,size=size,mode="bilinear",align_corners=False))

@torch.no_grad()
def prob_768(net, x, mode):
    if mode == "full768":
        return torch.sigmoid(net(x).float())
    if mode == "whole384":
        d = F.interpolate(x, size=(TRAIN_RES,)*2, mode="bilinear", align_corners=False)
        p = torch.sigmoid(net(d).float())
        return F.interpolate(p, size=(FULL_RES,)*2, mode="bilinear", align_corners=False)
    acc = torch.zeros(x.shape[0], 1, FULL_RES, FULL_RES, device=x.device)
    cnt = torch.zeros_like(acc)
    ORIG = (0, (FULL_RES-TRAIN_RES)//2, FULL_RES-TRAIN_RES)
    for oy in ORIG:
        for ox in ORIG:
            t = x[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES]
            acc[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += torch.sigmoid(net(t).float())
            cnt[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += 1
    return acc/cnt.clamp(min=1)

THS = [round(t, 2) for t in np.arange(0.10, 0.91, 0.05)]
@torch.no_grad()
def score_per_image(bp, mode, up):
    net = ResUNet().to(DEV); net.load_state_dict(torch.load(bp, map_location=DEV)); net.eval()
    dl = DataLoader(EvalSet(CLEAN_VAL, up), BATCH_EVAL, shuffle=False,
                    num_workers=min(2, os.cpu_count() or 1), pin_memory=DEV.type == "cuda")
    n = len(CLEAN_VAL)
    per = np.zeros((n, len(THS))); Ipx = np.zeros((n, len(THS)))
    P = np.zeros((n, len(THS))); G = np.zeros(n); k = 0
    for x, y in dl:
        x, y = x.to(DEV, non_blocking=True), y.to(DEV, non_blocking=True)
        with torch.amp.autocast("cuda", enabled=DEV.type == "cuda"):
            p = prob_768(net, x, mode)
        b = x.shape[0]; yf = y.view(b, -1); pf = p.view(b, -1)
        G[k:k+b] = yf.sum(1).cpu().numpy()
        for j, t in enumerate(THS):
            pr = (pf > t).float(); it = (pr*yf).sum(1); pp = pr.sum(1)
            Ipx[k:k+b, j] = it.cpu().numpy(); P[k:k+b, j] = pp.cpu().numpy()
            per[k:k+b, j] = (it/(pp+yf.sum(1)-it).clamp(min=1)).cpu().numpy()
        k += b
    pooled = Ipx.sum(0)/np.maximum(P.sum(0)+G.sum()-Ipx.sum(0), 1)
    j = int(np.argmax(pooled))
    del net; torch.cuda.empty_cache()
    return dict(th=THS[j], pooled=float(pooled[j]), per=per[:, j])

def sha(p, chunk=1 << 24):
    h = hashlib.sha256()
    with open(p, "rb") as f:
        while (b := f.read(chunk)): h.update(b)
    return h.hexdigest()[:16]

# -------------------------- which runs to handle -----------------------
# summary.json is written only at the END of cell D, so its presence is
# the completion test -- safer than trusting a step count.
done = sorted(glob.glob(f"{STAGE}/*/summary.json"), key=os.path.getmtime)
if not done:
    raise SystemExit(f"no completed run under {STAGE}")
target = RUN_NAME or os.path.basename(os.path.dirname(done[-1]))
print(f"target run: {target}   ({len(done)} completed run(s) on Drive)\n")

man_p = f"{SAVE}/weights_manifest.json"
man = json.load(open(man_p)) if os.path.exists(man_p) else {}

for sm in done:
    d = os.path.dirname(sm); name = os.path.basename(d)
    J = json.load(open(sm))
    bp, ck = f"{d}/best.pt", f"{d}/ckpt.pt"
    if not os.path.exists(bp):
        print(f"{name}: no best.pt, skipped"); continue
    is_target = (name == target)

    # 1. per-image vectors
    npz = f"{d}/per_image.npz"
    need = True
    if os.path.exists(npz):
        try:
            z = np.load(npz, allow_pickle=True); need = list(z["stems"]) != CLEAN_VAL
        except Exception: need = True
    if need:
        t = time.time()
        r = score_per_image(bp, J["select_on"], J["input"] == "crop_768_up")
        np.savez(npz, per=r["per"], th=r["th"], pooled=r["pooled"],
                 mode=J["select_on"], stems=np.array(CLEAN_VAL))
        print(f"{name}: scored per image in {time.time()-t:.0f}s  "
              f"(mode {J['select_on']}, th {r['th']}, pooled {r['pooled']:.4f})")
    else:
        print(f"{name}: per_image.npz already current")

    # 2. archive
    zp = f"{ARCH}/{name}_results.zip"
    if is_target or not os.path.exists(zp):
        with zipfile.ZipFile(zp, "w", zipfile.ZIP_DEFLATED) as z:
            for f in ("summary.json", "history.json", "curves.png",
                      "clean_val_gain.npy", "per_image.npz"):
                p = f"{d}/{f}"
                if os.path.exists(p): z.write(p, arcname=f"{name}/{f}")
        print(f"{name}: archive {os.path.getsize(zp)/1e6:.2f} MB")

    # 3. manifest
    if is_target or name not in man:
        man[name] = dict(file=f"{name}__best.pt", bytes=os.path.getsize(bp),
                         sha256_16=sha(bp), best_val_pooled=J.get("best_val_pooled"),
                         gpu=J.get("gpu", "not recorded"), batch=J.get("batch"),
                         input=J["input"], loss=J["loss"], jitter=J["jitter"],
                         seed=J["seed"], floor=J.get("floor"))

    # 4. prune -- only a COMPLETED run, which summary.json proves
    if DELETE_CKPT and os.path.exists(ck):
        mb = os.path.getsize(ck)/1e6
        os.remove(ck); print(f"{name}: removed ckpt.pt (freed {mb:.0f} MB)")

json.dump(man, open(man_p, "w"), indent=1)
print(f"\nmanifest: {man_p}  ({len(man)} run(s))")

# ------------------------------ drive usage ----------------------------
def dsz(p):
    return sum(os.path.getsize(os.path.join(r, f))
               for r, _, fs in os.walk(p) for f in fs) if os.path.isdir(p) else 0
tot = 0
print("\nDrive usage")
for lab, p in (("stage2", STAGE), ("stage2_archive", ARCH),
               ("runs (stage 1)", f"{SAVE}/runs"), ("cache768", f"{SAVE}/cache768")):
    s = dsz(p); tot += s
    print(f"   {lab:18}{s/1e9:7.2f} GB")
print(f"   {'TRACKED TOTAL':18}{tot/1e9:7.2f} GB"
      + ("   !! approaching a 15 GB free Drive" if tot/1e9 > 11 else ""))

# ------------------------------- download ------------------------------
if DOWNLOAD:
    d = f"{STAGE}/{target}"
    loc = f"/content/{target}__best.pt"
    if not os.path.exists(loc): shutil.copyfile(f"{d}/best.pt", loc)
    print(f"\n{target}  best.pt {os.path.getsize(loc)/1e6:.1f} MB  "
          f"sha256[:16] {man[target]['sha256_16']}")
    try:
        from google.colab import files
        files.download(f"{ARCH}/{target}_results.zip")
        time.sleep(2)
        files.download(loc)
        print("downloads queued. Verify locally:")
        print(f"   shasum -a 256 {target}__best.pt | cut -c1-16"
              f"   ->  {man[target]['sha256_16']}")
    except Exception as e:
        print(f"(browser download unavailable: {type(e).__name__})")
    print(f"also on Drive:\n   {ARCH}/{target}_results.zip\n   {d}/best.pt")
```

```python
# =====================================================================
#  cell E -- score every finished Stage 2 run per image, then compare.
#
#  Cell D saves per-quartile MEANS, which cannot support a paired test.
#  This reloads each best.pt and writes the per-image IoU vector over
#  clean_val, at that run's own selected mode and threshold. No retraining.
#
#  Then, against the S-384 baseline, for each other run:
#    - paired per-image differences over the same 688 images
#    - an image-level paired bootstrap on Q0  (the PRIMARY inference:
#      688 paired images carry far more power than a 3-seed t-interval,
#      which at 3 seeds needs a seed-sd below 0.008 to clear +0.02)
#    - the same split by realized detail gain, declared in advance
#    - the PILOT rule: kill only if the mean Q0 difference is <= 0.
#      Killing anything that misses the +0.02 gate at one seed discards
#      a real +0.02 effect half the time; the <= 0 rule, 10% of the time.
#
#  A single-seed result here is a FILTER, never a result. Nothing from
#  this cell goes in the paper without seeds 43 and 44.
# =====================================================================
import os, csv, json, glob, time
import numpy as np
from PIL import Image
import torch, torch.nn as nn, torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models
import torchvision.transforms.functional as TF

SAVE, C768 = "/content/drive/MyDrive", "/content/cache768"
STAGE   = f"{SAVE}/stage2"
BASE    = "S-384_seed42"           # the baseline every run is paired against
TRAIN_RES, FULL_RES = 384, 768
BATCH_EVAL = 4
NBOOT   = 10000
FROZEN  = [0.059468, 0.138448, 0.300831]
KILL_AT = 0.0                      # pilot rule, declared before looking
try:
    from google.colab import drive; drive.mount("/content/drive", force_remount=False)
except Exception: pass
DEV = torch.device("cuda" if torch.cuda.is_available() else "cpu")

SP = json.load(open(f"{SAVE}/plantseg_splits.json"))
I768 = json.load(open(f"{C768}/index.json"))
row = I768["row_of"]
CLEAN_VAL = sorted(set(SP["clean_val"]) & set(I768["stems"]))
DSDIR = "/content/my_images/plantseg"
meta = {os.path.splitext(r["Name"])[0]: r for r in csv.DictReader(
        open(f"{DSDIR}/Metadata.csv", encoding="utf-8-sig", newline=""))}
frac = {k: float(v["Mask ratio"]) for k, v in meta.items()}
def quart(s):
    f = frac[s]
    return 0 if f <= FROZEN[0] else 1 if f <= FROZEN[1] else 2 if f <= FROZEN[2] else 3
QV = np.array([quart(s) for s in CLEAN_VAL])
res = {k: tuple(int(v) for v in meta[k]["Resolution"].lower().split("x")[:2])
       for k in CLEAN_VAL}
GAIN = np.array([min(min(res[s]), FULL_RES)/TRAIN_RES for s in CLEAN_VAL])
print(f"clean_val {len(CLEAN_VAL)}  quartiles {[int((QV==q).sum()) for q in range(4)]}")

MEAN, STD = [0.485,0.456,0.406], [0.229,0.224,0.225]
NORM = transforms.Normalize(MEAN, STD)
_mm = {}
def cache():
    if not _mm:
        _mm["i"] = np.load(I768["img_path"], mmap_mode="r")
        _mm["m"] = np.load(I768["msk_path"], mmap_mode="r")
    return _mm["i"], _mm["m"]
def strip_detail(im):
    return im.resize((TRAIN_RES,)*2, Image.BILINEAR).resize((FULL_RES,)*2, Image.BILINEAR)

class EvalSet(Dataset):
    def __init__(s, st, up): s.st, s.up = list(st), up
    def __len__(s): return len(s.st)
    def __getitem__(s, i):
        I, M = cache(); r = row[s.st[i]]
        img = Image.fromarray(I[r])
        if s.up: img = strip_detail(img)
        y = torch.from_numpy((np.asarray(M[r]) > 0).astype(np.float32)).unsqueeze(0)
        return NORM(TF.to_tensor(img)), y

SPEC = dict(channels=(64,256,512,1024,2048), decoder=(512,256,128,64))
class DoubleConv(nn.Module):
    def __init__(s,i,o):
        super().__init__()
        s.net = nn.Sequential(nn.Conv2d(i,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True),
                              nn.Conv2d(o,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True))
    def forward(s,x): return s.net(x)
class Up(nn.Module):
    def __init__(s,i,k,o):
        super().__init__(); s.reduce=nn.Conv2d(i,i//2,1,bias=False); s.conv=DoubleConv(i//2+k,o)
    def forward(s,x,skip):
        x = F.interpolate(x, size=skip.shape[-2:], mode="bilinear", align_corners=False)
        return s.conv(torch.cat([s.reduce(x), skip], 1))
class ResUNet(nn.Module):
    def __init__(s):
        super().__init__()
        base = models.resnet50(weights=None)
        c0,c1,c2,c3,c4 = SPEC["channels"]; d1,d2,d3,d4 = SPEC["decoder"]
        L = list(base.children())
        s.stem=nn.Sequential(*L[:3]); s.pool=L[3]
        s.l1,s.l2,s.l3,s.l4 = L[4],L[5],L[6],L[7]
        s.u1,s.u2,s.u3,s.u4 = Up(c4,c3,d1),Up(d1,c2,d2),Up(d2,c1,d3),Up(d3,c0,d4)
        s.head = nn.Sequential(nn.Conv2d(d4,32,3,padding=1,bias=False),
                               nn.BatchNorm2d(32), nn.ReLU(True), nn.Conv2d(32,1,1))
    def forward(s,x):
        size=x.shape[-2:]
        x0=s.stem(x); x1=s.l1(s.pool(x0)); x2=s.l2(x1); x3=s.l3(x2); x4=s.l4(x3)
        d=s.u1(x4,x3); d=s.u2(d,x2); d=s.u3(d,x1); d=s.u4(d,x0)
        return s.head(F.interpolate(d,size=size,mode="bilinear",align_corners=False))

@torch.no_grad()
def prob_768(net, x, mode):
    if mode == "full768":
        return torch.sigmoid(net(x).float())
    if mode == "whole384":
        d = F.interpolate(x, size=(TRAIN_RES,)*2, mode="bilinear", align_corners=False)
        p = torch.sigmoid(net(d).float())
        return F.interpolate(p, size=(FULL_RES,)*2, mode="bilinear", align_corners=False)
    acc = torch.zeros(x.shape[0], 1, FULL_RES, FULL_RES, device=x.device)
    cnt = torch.zeros_like(acc)
    ORIG = (0, (FULL_RES-TRAIN_RES)//2, FULL_RES-TRAIN_RES)
    for oy in ORIG:
        for ox in ORIG:
            t = x[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES]
            acc[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += torch.sigmoid(net(t).float())
            cnt[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += 1
    return acc/cnt.clamp(min=1)

THS = [round(t, 2) for t in np.arange(0.10, 0.91, 0.05)]
@torch.no_grad()
def score(bp, mode, up):
    net = ResUNet().to(DEV); net.load_state_dict(torch.load(bp, map_location=DEV)); net.eval()
    dl = DataLoader(EvalSet(CLEAN_VAL, up), BATCH_EVAL, shuffle=False,
                    num_workers=min(2, os.cpu_count() or 1), pin_memory=DEV.type == "cuda")
    n = len(CLEAN_VAL)
    per = np.zeros((n, len(THS))); Ipx = np.zeros((n, len(THS)))
    P = np.zeros((n, len(THS))); G = np.zeros(n); k = 0
    for x, y in dl:
        x, y = x.to(DEV, non_blocking=True), y.to(DEV, non_blocking=True)
        with torch.amp.autocast("cuda", enabled=DEV.type == "cuda"):
            p = prob_768(net, x, mode)
        b = x.shape[0]; yf = y.view(b, -1); pf = p.view(b, -1)
        G[k:k+b] = yf.sum(1).cpu().numpy()
        for j, t in enumerate(THS):
            pr = (pf > t).float(); it = (pr*yf).sum(1); pp = pr.sum(1)
            Ipx[k:k+b, j] = it.cpu().numpy(); P[k:k+b, j] = pp.cpu().numpy()
            per[k:k+b, j] = (it/(pp+yf.sum(1)-it).clamp(min=1)).cpu().numpy()
        k += b
    pooled = Ipx.sum(0)/np.maximum(P.sum(0)+G.sum()-Ipx.sum(0), 1)
    j = int(np.argmax(pooled))
    del net; torch.cuda.empty_cache()
    return dict(th=THS[j], pooled=float(pooled[j]), per=per[:, j])

# ------------------------- score what is finished ----------------------
runs = {}
for sm in sorted(glob.glob(f"{STAGE}/*/summary.json")):
    d = os.path.dirname(sm); name = os.path.basename(d)
    if not os.path.exists(f"{d}/best.pt"):
        print(f"   {name}: no best.pt, skipped"); continue
    J = json.load(open(sm))
    npz = f"{d}/per_image.npz"
    if os.path.exists(npz):
        z = np.load(npz, allow_pickle=True)
        if list(z["stems"]) == CLEAN_VAL:
            runs[name] = dict(per=z["per"], th=float(z["th"]), pooled=float(z["pooled"]),
                              mode=str(z["mode"]), cfg=J); print(f"   {name}: cached"); continue
    t = time.time()
    r = score(f"{d}/best.pt", J["select_on"], J["input"] == "crop_768_up")
    np.savez(npz, per=r["per"], th=r["th"], pooled=r["pooled"],
             mode=J["select_on"], stems=np.array(CLEAN_VAL))
    runs[name] = dict(**r, mode=J["select_on"], cfg=J)
    print(f"   {name}: scored in {time.time()-t:.0f}s  "
          f"mode {J['select_on']} th {r['th']} pooled {r['pooled']:.4f}")

if BASE not in runs:
    raise SystemExit(f"baseline {BASE} not scored yet -- run it first")
print(f"\n{len(runs)} run(s) scored; baseline {BASE}")

# --------------------------- paired comparison -------------------------
rng = np.random.default_rng(0)
b_per = runs[BASE]["per"]; q0 = QV == 0
bands = {"Q0 all": q0, "Q0 gain<=1.0": q0 & (GAIN <= 1.0),
         "Q0 gain>=1.5": q0 & (GAIN >= 1.5), "all images": np.ones(len(QV), bool)}

def boot(d, sel):
    v = d[sel]
    if len(v) < 5: return float("nan"), float("nan"), float("nan")
    idx = rng.integers(0, len(v), (NBOOT, len(v)))
    bs = v[idx].mean(1)
    return float(v.mean()), float(np.percentile(bs, 2.5)), float(np.percentile(bs, 97.5))

print("\n" + "=" * 88)
print(f"paired per-image IoU difference vs {BASE}, clean_val, "
      f"{NBOOT} image-level bootstraps")
print("=" * 88)
out = {}
for name, r in runs.items():
    if name == BASE: continue
    d = r["per"] - b_per
    print(f"\n{name}   ({r['cfg']['input']} / {r['cfg']['loss']}"
          f"{' / jitter' if r['cfg']['jitter'] else ''})   "
          f"pooled {r['pooled']:.4f} vs {runs[BASE]['pooled']:.4f}")
    print(f"   {'band':16}{'n':>6}{'mean diff':>12}{'95% CI':>22}")
    row = {}
    for lab, sel in bands.items():
        m, lo, hi = boot(d, sel)
        star = "  *" if (lo > 0 or hi < 0) else ""
        print(f"   {lab:16}{int(sel.sum()):>6}{m:>12.4f}"
              f"   [{lo:+.4f}, {hi:+.4f}]{star}")
        row[lab] = dict(n=int(sel.sum()), mean=m, lo=lo, hi=hi)
    mq0 = row["Q0 all"]["mean"]
    verdict = ("KILL -- Q0 moves the wrong way" if mq0 <= KILL_AT else
               "KEEP -- run seeds 43 and 44")
    print(f"   PILOT ({'<=' if mq0 <= KILL_AT else '>'} {KILL_AT:+.2f}): {verdict}")
    if row["Q0 gain>=1.5"]["n"] >= 20 and row["Q0 gain<=1.0"]["n"] >= 20:
        sep = row["Q0 gain>=1.5"]["mean"] - row["Q0 gain<=1.0"]["mean"]
        print(f"   mechanism check: high-detail minus no-detail = {sep:+.4f}"
              f"   {'(consistent with resolution)' if sep > 0 else '(not resolution)'}")
        row["mechanism_sep"] = sep
    out[name] = row

print("\n" + "=" * 88)
print("* marks a CI excluding zero. That is an image-level interval: it says the")
print("difference is not sampling noise over these 688 images, NOT that it would")
print("survive another training seed. Only seeds 43 and 44 answer that, and the")
print("gate is judged on the three-seed paired mean, not on anything here.")
json.dump(dict(baseline=BASE, kill_at=KILL_AT, nboot=NBOOT, results=out),
          open(f"{SAVE}/stage2_compare.json", "w"), indent=1, default=float)
print(f"\nwrote {SAVE}/stage2_compare.json")
```

```python
# =====================================================================
#  cell D -- Stage 2: supervised tests of components 1 and 2.
#  One cell, three switches, 25,000 steps, resume-safe. Validation only;
#  the test split is not touched until Stage 5.
#
#      INPUT_MODE in {full_384, crop_768, crop_768_up}
#      FG_LOSS    in {pooled, class_balanced, image_balanced}
#      JITTER     in {False, True}
#
#  (full_384, pooled, False) is the S-384 baseline.
#
#  crop_768_up is the control the crop statistics made necessary: uniform
#  384 crops of a 768 image carry 1.50x the Q0 lesion density of the
#  whole image, because uniform crop ORIGINS over-sample the centre (a
#  centre pixel sits in 384 of 385 origins per axis, a corner pixel in 1).
#  So crop_768 changes resolution AND density at once. crop_768_up takes
#  the 768 image, downsamples it to 384 and upsamples it back, destroying
#  exactly the detail above 384 and nothing else, then crops identically.
#  crop_768 beating crop_768_up isolates resolution; a tie says the gain
#  is crop geometry.
#
#  Everything reads ONE cache: /content/cache768 (537 labelled + 846 val).
#  The 441 cache is not needed, so prep.py need not be run this session.
# =====================================================================
import os, csv, json, math, time, random, shutil, zipfile, collections
import numpy as np
import matplotlib; matplotlib.use("Agg")
import matplotlib.pyplot as plt
from PIL import Image, ImageFilter
import torch, torch.nn as nn, torch.optim as optim, torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models
import torchvision.transforms.functional as TF

# ------------------------- the three switches --------------------------
INPUT_MODE = "crop_768"        # full_384 | crop_768 | crop_768_up
FG_LOSS    = "class_balanced"          # pooled | class_balanced | image_balanced
JITTER     = False             # True only for S-768c-jit
TRAIN_SEED = 42
# -----------------------------------------------------------------------
FLOOR_FRAC = 0.01              # declared from check 0f: binds on 8.9% of
                               # non-empty Q0 crops, caps amplification 21.5x
SELECT_ON  = {"full_384": "whole384", "crop_768": "full768",
              "crop_768_up": "full768"}[INPUT_MODE]
MAX_STEPS  = 25000
EVAL_EVERY = 1000
SHARE_EVERY = 500              # per-quartile lesion-loss share
CKPT_EVERY = 2500
LR_PATIENCE = 3                # patience counts EVAL_EVERY blocks, now 1000
TRAIN_RES  = 384
FULL_RES   = 768
BATCH      = 4                 # matches the existing pipeline; batch 8
                               # doubles every run to ~2.2h and Stage 2 to ~42h
BATCH_EVAL = 4                 # 768 input on a T4
LR_ENCODER, LR_DECODER, WEIGHT_DECAY = 1e-5, 1e-4, 1e-4
FROZEN     = [0.059468, 0.138448, 0.300831]
SAVE, C768 = "/content/drive/MyDrive", "/content/cache768"
NUM_WORKERS = min(2, os.cpu_count() or 1)
FRESH      = True              # wipes this run's folder. Set back to False
                               # once it is past step 2500 so a dropped
                               # session resumes instead of restarting.

TAG = {"full_384": "384", "crop_768": "768c", "crop_768_up": "768up"}[INPUT_MODE]
if FG_LOSS == "class_balanced":  TAG += "-cb"
if FG_LOSS == "image_balanced":  TAG += "-bal"
if JITTER:                       TAG += "-jit"
RUN = f"S-{TAG}_seed{TRAIN_SEED}"
OUT = f"{SAVE}/stage2/{RUN}"
if FRESH and os.path.isdir(OUT): shutil.rmtree(OUT, ignore_errors=True)
os.makedirs(OUT, exist_ok=True)

random.seed(TRAIN_SEED); np.random.seed(TRAIN_SEED)
torch.manual_seed(TRAIN_SEED); torch.cuda.manual_seed_all(TRAIN_SEED)
torch.backends.cudnn.deterministic = True; torch.backends.cudnn.benchmark = False
DEV = torch.device("cuda" if torch.cuda.is_available() else "cpu")
try:
    from google.colab import drive; drive.mount("/content/drive", force_remount=False)
except Exception: pass
print(f"{RUN}\ndevice {DEV} {torch.cuda.get_device_name(0) if DEV.type=='cuda' else ''}")
print(f"input {INPUT_MODE} | loss {FG_LOSS} | jitter {JITTER} | "
      f"select on {SELECT_ON} | floor {FLOOR_FRAC} | batch {BATCH}")

# --------------------------- data and asserts --------------------------
SP = json.load(open(f"{SAVE}/plantseg_splits.json"))
if max(abs(a-b) for a, b in zip(SP["quartile_cuts"], FROZEN)) > 1e-6:
    raise SystemExit(f"cuts {SP['quartile_cuts']} != frozen {FROZEN}")
if {k: len(v) for k, v in SP["official_splits"].items()} != \
   {"train": 5367, "val": 846, "test": 1561}:
    raise SystemExit("split sizes are not 5367/846/1561")
I768 = json.load(open(f"{C768}/index.json"))
if I768["load_size"] != FULL_RES:
    raise SystemExit(f"cache is {I768['load_size']}, expected {FULL_RES}")
row = I768["row_of"]
LAB = sorted(I768["labelled"])
CLEAN_VAL = sorted(set(SP["clean_val"]) & set(I768["stems"]))
if len(LAB) != 537: raise SystemExit(f"{len(LAB)} labelled, expected 537")
DSDIR = "/content/my_images/plantseg"
meta = {os.path.splitext(r["Name"])[0]: r for r in csv.DictReader(
        open(f"{DSDIR}/Metadata.csv", encoding="utf-8-sig", newline=""))}
frac = {k: float(v["Mask ratio"]) for k, v in meta.items()}
def quart(s):
    f = frac[s]
    return 0 if f <= FROZEN[0] else 1 if f <= FROZEN[1] else 2 if f <= FROZEN[2] else 3
QV = np.array([quart(s) for s in CLEAN_VAL])
print(f"labelled {len(LAB)}   clean_val {len(CLEAN_VAL)} "
      f"(quartiles {[int((QV==q).sum()) for q in range(4)]})")
print("   gates are judged on clean_val, not the full 846: full val carries")
print("   17.14% near-duplicates of training images.")

MEAN, STD = [0.485,0.456,0.406], [0.229,0.224,0.225]
NORM = transforms.Normalize(MEAN, STD)
JIT  = transforms.ColorJitter(0.4, 0.4, 0.4, 0.1)
_mm = {}
def cache():
    if not _mm:
        _mm["i"] = np.load(I768["img_path"], mmap_mode="r")
        _mm["m"] = np.load(I768["msk_path"], mmap_mode="r")
    return _mm["i"], _mm["m"]

def strip_detail(im):
    """768 -> 384 -> 768. Destroys exactly the detail above 384, keeps
       geometry, size and lesion density identical."""
    return im.resize((TRAIN_RES,)*2, Image.BILINEAR).resize((FULL_RES,)*2, Image.BILINEAR)

def photometric(img):
    if random.random() < 0.8: img = JIT(img)
    if random.random() < 0.2: img = TF.to_grayscale(img, num_output_channels=3)
    if random.random() < 0.3:
        img = img.filter(ImageFilter.GaussianBlur(radius=random.uniform(0.1, 1.5)))
    return img

def geometric(img, msk):
    if random.random() < 0.5: img, msk = TF.hflip(img), TF.hflip(msk)
    if random.random() < 0.5: img, msk = TF.vflip(img), TF.vflip(msk)
    k = random.randint(0, 3)
    if k:
        img = img.rotate(90*k, expand=True, resample=Image.BILINEAR)
        msk = msk.rotate(90*k, expand=True, resample=Image.NEAREST)
    return img, msk

def rrc(img, msk, scale=(0.5, 1.0), ratio=(3/4, 4/3)):
    w, h = img.size; area = w*h
    for _ in range(10):
        t = area*random.uniform(*scale)
        ar = math.exp(random.uniform(math.log(ratio[0]), math.log(ratio[1])))
        cw, ch = int(round(math.sqrt(t*ar))), int(round(math.sqrt(t/ar)))
        if cw <= w and ch <= h:
            x, y = random.randint(0, w-cw), random.randint(0, h-ch)
            return img.crop((x, y, x+cw, y+ch)), msk.crop((x, y, x+cw, y+ch))
    return img, msk

class Labelled(Dataset):
    """full_384   : whole image, rrc + geometric, resized to 384
       crop_768   : uniform 384 window of the 768 image (the crop IS the
                    spatial augmentation, so rrc is not applied on top)
       crop_768_up: the same window of a detail-stripped 768 image"""
    def __init__(s, st): s.st = list(st)
    def __len__(s): return len(s.st)
    def __getitem__(s, i):
        I, M = cache(); r = row[s.st[i]]
        img = Image.fromarray(I[r]); msk = Image.fromarray(M[r])
        if INPUT_MODE == "crop_768_up": img = strip_detail(img)
        if INPUT_MODE == "full_384":
            img, msk = rrc(img, msk)
            img, msk = geometric(img, msk)
            img = img.resize((TRAIN_RES,)*2, Image.BILINEAR)
            msk = msk.resize((TRAIN_RES,)*2, Image.NEAREST)
        else:
            side = (random.randint(256, FULL_RES) if JITTER else TRAIN_RES)
            x = random.randint(0, FULL_RES-side); y = random.randint(0, FULL_RES-side)
            img = img.crop((x, y, x+side, y+side)); msk = msk.crop((x, y, x+side, y+side))
            img, msk = geometric(img, msk)
            if side != TRAIN_RES:
                img = img.resize((TRAIN_RES,)*2, Image.BILINEAR)
                msk = msk.resize((TRAIN_RES,)*2, Image.NEAREST)
        y_ = torch.from_numpy((np.array(msk) > 0).astype(np.float32)).unsqueeze(0)
        return NORM(TF.to_tensor(photometric(img))), y_, quart(s.st[i])

class EvalSet(Dataset):
    """input follows INPUT_MODE; the TARGET is always the 768 mask."""
    def __init__(s, st): s.st = list(st)
    def __len__(s): return len(s.st)
    def __getitem__(s, i):
        I, M = cache(); r = row[s.st[i]]
        img = Image.fromarray(I[r])
        if INPUT_MODE == "crop_768_up": img = strip_detail(img)
        y = torch.from_numpy((np.asarray(M[r]) > 0).astype(np.float32)).unsqueeze(0)
        return NORM(TF.to_tensor(img)), y            # image at 768, mask at 768

def seed_worker(_):
    s = torch.initial_seed() % 2**32; np.random.seed(s); random.seed(s)
LK = dict(num_workers=NUM_WORKERS, pin_memory=DEV.type == "cuda",
          worker_init_fn=seed_worker, persistent_workers=NUM_WORKERS > 0)
g = torch.Generator(); g.manual_seed(TRAIN_SEED)
loader_l = DataLoader(Labelled(LAB), BATCH, shuffle=True, drop_last=True,
                      generator=g, **LK)
val_loader = DataLoader(EvalSet(CLEAN_VAL), BATCH_EVAL, shuffle=False, **LK)
def cycle(dl):
    while True:
        for b in dl: yield b

# ------------------------------- model ---------------------------------
SPEC = dict(channels=(64,256,512,1024,2048), decoder=(512,256,128,64))
class DoubleConv(nn.Module):
    def __init__(s,i,o):
        super().__init__()
        s.net = nn.Sequential(nn.Conv2d(i,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True),
                              nn.Conv2d(o,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True))
    def forward(s,x): return s.net(x)
class Up(nn.Module):
    def __init__(s,i,k,o):
        super().__init__(); s.reduce=nn.Conv2d(i,i//2,1,bias=False); s.conv=DoubleConv(i//2+k,o)
    def forward(s,x,skip):
        x = F.interpolate(x, size=skip.shape[-2:], mode="bilinear", align_corners=False)
        return s.conv(torch.cat([s.reduce(x), skip], 1))
class ResUNet(nn.Module):
    def __init__(s):
        super().__init__()
        base = models.resnet50(weights=models.ResNet50_Weights.DEFAULT)
        c0,c1,c2,c3,c4 = SPEC["channels"]; d1,d2,d3,d4 = SPEC["decoder"]
        L = list(base.children())
        s.stem=nn.Sequential(*L[:3]); s.pool=L[3]
        s.l1,s.l2,s.l3,s.l4 = L[4],L[5],L[6],L[7]
        s.u1,s.u2,s.u3,s.u4 = Up(c4,c3,d1),Up(d1,c2,d2),Up(d2,c1,d3),Up(d3,c0,d4)
        s.head = nn.Sequential(nn.Conv2d(d4,32,3,padding=1,bias=False),
                               nn.BatchNorm2d(32), nn.ReLU(True), nn.Conv2d(32,1,1))
    def enc_params(s):
        for m in (s.stem,s.l1,s.l2,s.l3,s.l4): yield from m.parameters()
    def dec_params(s):
        for m in (s.u1,s.u2,s.u3,s.u4,s.head): yield from m.parameters()
    def forward(s,x):
        size=x.shape[-2:]
        x0=s.stem(x); x1=s.l1(s.pool(x0)); x2=s.l2(x1); x3=s.l3(x2); x4=s.l4(x3)
        d=s.u1(x4,x3); d=s.u2(d,x2); d=s.u3(d,x1); d=s.u4(d,x0)
        return s.head(F.interpolate(d,size=size,mode="bilinear",align_corners=False))

# ------------------------------- losses --------------------------------
bce_none = nn.BCEWithLogitsLoss(reduction="none")
def tversky(logits, tgt, eps=1.0):
    p = torch.sigmoid(logits)
    tp = (p*tgt).sum((1,2,3)); fp = (p*(1-tgt)).sum((1,2,3)); fn = ((1-p)*tgt).sum((1,2,3))
    return 1 - (tp+eps)/(tp+0.5*fp+0.5*fn+eps)

def fg_term(logits, tgt):
    """returns (per-image BCE term, per-image lesion-only contribution).
       All three forms are scaled so a uniform per-pixel loss c gives c,
       so the learning rate and the BCE/Tversky balance do not shift
       between configurations. The plan's formula writes the sum of the
       two class means; using their average is the same comparison at the
       same scale."""
    e = bce_none(logits, tgt)
    sf = tgt.sum((1,2,3)); sb = (1-tgt).sum((1,2,3))
    ef = (e*tgt).sum((1,2,3)); eb = (e*(1-tgt)).sum((1,2,3))
    if FG_LOSS == "pooled":
        return e.mean((1,2,3)), ef/tgt[0].numel()
    if FG_LOSS == "class_balanced":                 # batch-level class means
        n = logits.shape[0]
        mf = ef.sum()/sf.sum().clamp(min=1.0); mb = eb.sum()/sb.sum().clamp(min=1.0)
        return (0.5*(mf+mb)).expand(n), 0.5*ef/sf.sum().clamp(min=1.0)
    floor = FLOOR_FRAC*tgt[0].numel()               # per-image, with the floor
    tf = ef/torch.clamp(sf, min=floor)
    tb = eb/sb.clamp(min=1.0)
    return 0.5*(tf+tb), 0.5*tf

def sup_loss(logits, tgt):
    b, lesion = fg_term(logits, tgt)
    return b + tversky(logits, tgt), lesion

# --------------------------- 768 inference -----------------------------
@torch.no_grad()
def prob_768(net, x, mode):
    """x is the model's input image at 768. Returns a 768 probability map."""
    if mode == "full768":
        return torch.sigmoid(net(x).float())
    if mode == "whole384":
        d = F.interpolate(x, size=(TRAIN_RES,)*2, mode="bilinear", align_corners=False)
        p = torch.sigmoid(net(d).float())
        return F.interpolate(p, size=(FULL_RES,)*2, mode="bilinear", align_corners=False)
    # genuinely overlapping: stride 192 gives 3x3 tiles at 50% overlap.
    # Origins (0, 384) alone would ABUT, not overlap, and blending would
    # be a no-op -- the seam is exactly where a tiled model is weakest.
    acc = torch.zeros(x.shape[0], 1, FULL_RES, FULL_RES, device=x.device)
    cnt = torch.zeros_like(acc)
    ORIG = (0, (FULL_RES-TRAIN_RES)//2, FULL_RES-TRAIN_RES)
    for oy in ORIG:
        for ox in ORIG:
            t = x[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES]
            acc[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += torch.sigmoid(net(t).float())
            cnt[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += 1
    return acc/cnt.clamp(min=1)                      # blend, threshold afterwards

THS = [round(t, 2) for t in np.arange(0.10, 0.91, 0.05)]
@torch.no_grad()
def evaluate(net, mode):
    net.eval()
    nI = len(CLEAN_VAL); per = np.zeros((nI, len(THS))); P = np.zeros((nI, len(THS)))
    Ipx = np.zeros((nI, len(THS))); G = np.zeros(nI); k = 0
    for x, y in val_loader:
        x, y = x.to(DEV, non_blocking=True), y.to(DEV, non_blocking=True)
        with torch.amp.autocast("cuda", enabled=DEV.type == "cuda"):
            p = prob_768(net, x, mode)
        b = x.shape[0]; yf = y.view(b, -1)
        G[k:k+b] = yf.sum(1).cpu().numpy()
        pf = p.view(b, -1)
        for j, t in enumerate(THS):
            pr = (pf > t).float()
            it = (pr*yf).sum(1); pp = pr.sum(1)
            Ipx[k:k+b, j] = it.cpu().numpy(); P[k:k+b, j] = pp.cpu().numpy()
            per[k:k+b, j] = (it/(pp+yf.sum(1)-it).clamp(min=1)).cpu().numpy()
        k += b
    pooled = Ipx.sum(0)/np.maximum(P.sum(0)+G.sum()-Ipx.sum(0), 1)
    j = int(np.argmax(pooled)); t = THS[j]
    byq = {q: float(per[QV == q, j].mean()) for q in range(4)}
    return dict(th=t, pooled=float(pooled[j]), perimg=float(per[:, j].mean()),
                byq=byq, pred_frac=float(P[:, j].sum()/(len(CLEAN_VAL)*FULL_RES*FULL_RES)))

# --------------------------- resume or start ---------------------------
model = ResUNet().to(DEV)
opt = optim.AdamW([{"params": list(model.enc_params()), "lr": LR_ENCODER},
                   {"params": list(model.dec_params()), "lr": LR_DECODER}],
                  weight_decay=WEIGHT_DECAY)
sched = optim.lr_scheduler.ReduceLROnPlateau(opt, mode="max", patience=LR_PATIENCE, factor=0.5)
scaler = torch.amp.GradScaler("cuda")
step, best, history, shares = 0, -1.0, [], []
CK, BEST = f"{OUT}/ckpt.pt", f"{OUT}/best.pt"
if os.path.exists(CK):
    c = torch.load(CK, map_location=DEV, weights_only=False)
    model.load_state_dict(c["model"]); opt.load_state_dict(c["opt"])
    sched.load_state_dict(c["sched"]); scaler.load_state_dict(c["scaler"])
    step, best, history, shares = c["step"], c["best"], c["history"], c.get("shares", [])
    random.setstate(c["py_rng"]); np.random.set_state(c["np_rng"])
    torch.set_rng_state(c["t_rng"].cpu())
    if c.get("cuda_rng") is not None and DEV.type == "cuda":
        torch.cuda.set_rng_state(c["cuda_rng"].cpu())
    print(f"RESUMED from step {step} (best {best:.4f})")
else:
    print("starting fresh")

def save_ckpt():
    torch.save(dict(model=model.state_dict(), opt=opt.state_dict(), sched=sched.state_dict(),
                    scaler=scaler.state_dict(), step=step, best=best, history=history,
                    shares=shares, py_rng=random.getstate(), np_rng=np.random.get_state(),
                    t_rng=torch.get_rng_state(),
                    cuda_rng=torch.cuda.get_rng_state() if DEV.type == "cuda" else None), CK)

# ------------------------------- train ---------------------------------
it = cycle(loader_l)
print("=" * 76); print(f"{RUN}  --  {len(LAB)} labelled, {MAX_STEPS} steps"); print("=" * 76)
t0 = t_last = time.time()
r_loss = r_n = 0.0; q_les = np.zeros(4); q_cnt = np.zeros(4)
while step < MAX_STEPS:
    model.train()
    x, y, q = next(it)
    x, y = x.to(DEV, non_blocking=True), y.to(DEV, non_blocking=True)
    opt.zero_grad(set_to_none=True)
    with torch.amp.autocast("cuda", enabled=DEV.type == "cuda"):
        lg = model(x)
        vec, lesion = sup_loss(lg.float(), y)
        loss = vec.mean()
    scaler.scale(loss).backward(); scaler.step(opt); scaler.update()
    step += 1
    r_loss += loss.item(); r_n += 1
    lz = lesion.detach().float().cpu().numpy(); qz = q.numpy()
    for b in range(len(qz)): q_les[qz[b]] += lz[b]; q_cnt[qz[b]] += 1

    if step % SHARE_EVERY == 0:
        tot = max(q_les.sum(), 1e-12)
        sh = (q_les/tot).round(4).tolist()
        shares.append(dict(step=step, share=sh, seen=q_cnt.astype(int).tolist()))
        print(f"   [step {step}] lesion-loss share by quartile {sh}"
              f"  (images seen {q_cnt.astype(int).tolist()})")
        q_les[:] = 0; q_cnt[:] = 0

    if step % EVAL_EVERY == 0:
        m = evaluate(model, SELECT_ON)
        prev = opt.param_groups[-1]["lr"]; sched.step(m["pooled"])
        cur = opt.param_groups[-1]["lr"]
        saved = ""
        if m["pooled"] > best:
            best = m["pooled"]; torch.save(model.state_dict(), BEST); saved = "  <-- saved"
        history.append(dict(step=step, train_loss=r_loss/max(r_n, 1), lr=cur, **m))
        sps = EVAL_EVERY/(time.time()-t_last); t_last = time.time()
        print(f"step {step:6d} | loss {r_loss/max(r_n,1):.4f} | val pooled {m['pooled']:.4f} "
              f"@{m['th']} | per-img {m['perimg']:.4f} | Q0 {m['byq'][0]:.4f} "
              f"Q3 {m['byq'][3]:.4f} | best {best:.4f} | {sps:.2f} it/s | "
              f"ETA {(MAX_STEPS-step)/sps/3600:.1f}h{saved}"
              + (f"  [lr {prev:.1e}->{cur:.1e}]" if cur != prev else ""))
        r_loss = r_n = 0.0
        json.dump(dict(run=RUN, input=INPUT_MODE, loss=FG_LOSS, jitter=JITTER,
                       seed=TRAIN_SEED, floor=FLOOR_FRAC, select_on=SELECT_ON,
                       batch=BATCH, history=history, shares=shares),
                  open(f"{OUT}/history.json", "w"), indent=1)
    if step % CKPT_EVERY == 0:
        save_ckpt(); print(f"    checkpointed at step {step}")
save_ckpt()
print(f"\ntraining done in {(time.time()-t0)/3600:.2f}h, best val pooled {best:.4f}")

# ------------------- final: all three inference modes ------------------
net = ResUNet().to(DEV); net.load_state_dict(torch.load(BEST, map_location=DEV)); net.eval()
print("\n" + "=" * 76)
print("clean_val under each inference mode (best.pt was selected on "
      f"{SELECT_ON})")
print("=" * 76)
print(f"{'mode':12}{'th':>6}{'pooled':>9}{'per-img':>9}"
      f"{'Q0':>8}{'Q1':>8}{'Q2':>8}{'Q3':>8}{'Q3-Q0':>8}")
modes = ["whole384", "full768", "tiles384"]
final = {}
for md in modes:
    m = evaluate(net, md); final[md] = m
    print(f"{md:12}{m['th']:>6.2f}{m['pooled']:>9.4f}{m['perimg']:>9.4f}"
          + "".join(f"{m['byq'][q]:>8.4f}" for q in range(4))
          + f"{m['byq'][3]-m['byq'][0]:>8.4f}")

# the stratified check: does any Q0 gain track the realized detail gain?
res = {k: tuple(int(v) for v in meta[k]["Resolution"].lower().split("x")[:2])
       for k in CLEAN_VAL}
gain = np.array([min(min(res[s]), FULL_RES)/TRAIN_RES for s in CLEAN_VAL])
q0 = QV == 0
print(f"\nclean_val Q0 split by realized worst-axis detail gain at {FULL_RES}:")
for lab, sel in (("<= 1.0 (no gain)", q0 & (gain <= 1.0)),
                 ("1.0 - 1.5", q0 & (gain > 1.0) & (gain < 1.5)),
                 (">= 1.5", q0 & (gain >= 1.5))):
    print(f"   {lab:18} n={int(sel.sum()):4d}")
print("   per-image Q0 IoU within each band is in the saved json; compare it")
print("   against the matching S-384 run, paired by image.")

np.save(f"{OUT}/clean_val_gain.npy", gain)
json.dump(dict(run=RUN, input=INPUT_MODE, loss=FG_LOSS, jitter=JITTER, seed=TRAIN_SEED,
               floor=FLOOR_FRAC, select_on=SELECT_ON, batch=BATCH, best_val_pooled=best,
               clean_val=CLEAN_VAL, quartiles=QV.tolist(), final=final,
               history=history, shares=shares),
          open(f"{OUT}/summary.json", "w"), indent=1)
xs = [h["step"] for h in history]
plt.figure(figsize=(14, 4))
plt.subplot(1, 3, 1); plt.plot(xs, [h["train_loss"] for h in history]); plt.title("train loss")
plt.subplot(1, 3, 2); plt.plot(xs, [h["pooled"] for h in history]); plt.title("val pooled IoU")
plt.subplot(1, 3, 3)
for q in range(4): plt.plot(xs, [h["byq"][q] for h in history], label=f"Q{q}")
plt.legend(fontsize=7); plt.title("val per-image IoU by quartile")
plt.tight_layout(); plt.savefig(f"{OUT}/curves.png", dpi=110); plt.close()
print(f"\nwrote {OUT}/summary.json, best.pt, curves.png")
print(">>> download best.pt to local disk before the session ends <<<")
```

```python
# =====================================================================
#  cell F -- run this ONCE after every cell D training. Standalone: it
#  reads nothing from cell D's variables, so it also works in a fresh
#  session or after a runtime restart.
#
#  For the target run it:
#    1. scores it per image on clean_val  -> per_image.npz
#       (the vectors every paired test needs; cell D saves only means)
#    2. archives summary / history / curves / gain / per_image to Drive
#    3. records config, GPU and a checksum in weights_manifest.json
#    4. deletes ckpt.pt, but ONLY if the run completed. 490 MB each, and
#       19 of them would overrun a free Drive.
#    5. prints what Drive is holding
#    6. downloads the archive and best.pt
#
#  Any OTHER run still missing per_image.npz or an archive is also
#  brought up to date, without being downloaded again.
# =====================================================================
import os, csv, json, glob, time, zipfile, shutil, hashlib
import numpy as np
from PIL import Image
import torch, torch.nn as nn, torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models
import torchvision.transforms.functional as TF

RUN_NAME    = None       # None = the most recently finished run in stage2/
DELETE_CKPT = True       # False if you intend to extend a run later
DOWNLOAD    = True
SAVE, C768  = "/content/drive/MyDrive", "/content/cache768"
STAGE, ARCH = f"{SAVE}/stage2", f"{SAVE}/stage2_archive"
TRAIN_RES, FULL_RES = 384, 768
BATCH_EVAL  = 12 if torch.cuda.is_available() else 2
FROZEN      = [0.059468, 0.138448, 0.300831]
try:
    from google.colab import drive; drive.mount("/content/drive", force_remount=False)
except Exception: pass
DEV = torch.device("cuda" if torch.cuda.is_available() else "cpu")
os.makedirs(ARCH, exist_ok=True)

# ------------------------------- context -------------------------------
SP   = json.load(open(f"{SAVE}/plantseg_splits.json"))
I768 = json.load(open(f"{C768}/index.json"))
row  = I768["row_of"]
CLEAN_VAL = sorted(set(SP["clean_val"]) & set(I768["stems"]))
DSDIR = "/content/my_images/plantseg"
meta = {os.path.splitext(r["Name"])[0]: r for r in csv.DictReader(
        open(f"{DSDIR}/Metadata.csv", encoding="utf-8-sig", newline=""))}
frac = {k: float(v["Mask ratio"]) for k, v in meta.items()}
def quart(s):
    f = frac[s]
    return 0 if f <= FROZEN[0] else 1 if f <= FROZEN[1] else 2 if f <= FROZEN[2] else 3
QV = np.array([quart(s) for s in CLEAN_VAL])

MEAN, STD = [0.485,0.456,0.406], [0.229,0.224,0.225]
NORM = transforms.Normalize(MEAN, STD)
_mm = {}
def cache():
    if not _mm:
        _mm["i"] = np.load(I768["img_path"], mmap_mode="r")
        _mm["m"] = np.load(I768["msk_path"], mmap_mode="r")
    return _mm["i"], _mm["m"]
def strip_detail(im):
    return im.resize((TRAIN_RES,)*2, Image.BILINEAR).resize((FULL_RES,)*2, Image.BILINEAR)

class EvalSet(Dataset):
    def __init__(s, st, up): s.st, s.up = list(st), up
    def __len__(s): return len(s.st)
    def __getitem__(s, i):
        I, M = cache(); r = row[s.st[i]]
        img = Image.fromarray(I[r])
        if s.up: img = strip_detail(img)
        y = torch.from_numpy((np.asarray(M[r]) > 0).astype(np.float32)).unsqueeze(0)
        return NORM(TF.to_tensor(img)), y

SPEC = dict(channels=(64,256,512,1024,2048), decoder=(512,256,128,64))
class DoubleConv(nn.Module):
    def __init__(s,i,o):
        super().__init__()
        s.net = nn.Sequential(nn.Conv2d(i,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True),
                              nn.Conv2d(o,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True))
    def forward(s,x): return s.net(x)
class Up(nn.Module):
    def __init__(s,i,k,o):
        super().__init__(); s.reduce=nn.Conv2d(i,i//2,1,bias=False); s.conv=DoubleConv(i//2+k,o)
    def forward(s,x,skip):
        x = F.interpolate(x, size=skip.shape[-2:], mode="bilinear", align_corners=False)
        return s.conv(torch.cat([s.reduce(x), skip], 1))
class ResUNet(nn.Module):
    def __init__(s):
        super().__init__()
        base = models.resnet50(weights=None)
        c0,c1,c2,c3,c4 = SPEC["channels"]; d1,d2,d3,d4 = SPEC["decoder"]
        L = list(base.children())
        s.stem=nn.Sequential(*L[:3]); s.pool=L[3]
        s.l1,s.l2,s.l3,s.l4 = L[4],L[5],L[6],L[7]
        s.u1,s.u2,s.u3,s.u4 = Up(c4,c3,d1),Up(d1,c2,d2),Up(d2,c1,d3),Up(d3,c0,d4)
        s.head = nn.Sequential(nn.Conv2d(d4,32,3,padding=1,bias=False),
                               nn.BatchNorm2d(32), nn.ReLU(True), nn.Conv2d(32,1,1))
    def forward(s,x):
        size=x.shape[-2:]
        x0=s.stem(x); x1=s.l1(s.pool(x0)); x2=s.l2(x1); x3=s.l3(x2); x4=s.l4(x3)
        d=s.u1(x4,x3); d=s.u2(d,x2); d=s.u3(d,x1); d=s.u4(d,x0)
        return s.head(F.interpolate(d,size=size,mode="bilinear",align_corners=False))

@torch.no_grad()
def prob_768(net, x, mode):
    if mode == "full768":
        return torch.sigmoid(net(x).float())
    if mode == "whole384":
        d = F.interpolate(x, size=(TRAIN_RES,)*2, mode="bilinear", align_corners=False)
        p = torch.sigmoid(net(d).float())
        return F.interpolate(p, size=(FULL_RES,)*2, mode="bilinear", align_corners=False)
    acc = torch.zeros(x.shape[0], 1, FULL_RES, FULL_RES, device=x.device)
    cnt = torch.zeros_like(acc)
    ORIG = (0, (FULL_RES-TRAIN_RES)//2, FULL_RES-TRAIN_RES)
    for oy in ORIG:
        for ox in ORIG:
            t = x[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES]
            acc[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += torch.sigmoid(net(t).float())
            cnt[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += 1
    return acc/cnt.clamp(min=1)

THS = [round(t, 2) for t in np.arange(0.10, 0.91, 0.05)]
@torch.no_grad()
def score_per_image(bp, mode, up):
    net = ResUNet().to(DEV); net.load_state_dict(torch.load(bp, map_location=DEV)); net.eval()
    dl = DataLoader(EvalSet(CLEAN_VAL, up), BATCH_EVAL, shuffle=False,
                    num_workers=min(2, os.cpu_count() or 1), pin_memory=DEV.type == "cuda")
    n = len(CLEAN_VAL)
    per = np.zeros((n, len(THS))); Ipx = np.zeros((n, len(THS)))
    P = np.zeros((n, len(THS))); G = np.zeros(n); k = 0
    for x, y in dl:
        x, y = x.to(DEV, non_blocking=True), y.to(DEV, non_blocking=True)
        with torch.amp.autocast("cuda", enabled=DEV.type == "cuda"):
            p = prob_768(net, x, mode)
        b = x.shape[0]; yf = y.view(b, -1); pf = p.view(b, -1)
        G[k:k+b] = yf.sum(1).cpu().numpy()
        for j, t in enumerate(THS):
            pr = (pf > t).float(); it = (pr*yf).sum(1); pp = pr.sum(1)
            Ipx[k:k+b, j] = it.cpu().numpy(); P[k:k+b, j] = pp.cpu().numpy()
            per[k:k+b, j] = (it/(pp+yf.sum(1)-it).clamp(min=1)).cpu().numpy()
        k += b
    pooled = Ipx.sum(0)/np.maximum(P.sum(0)+G.sum()-Ipx.sum(0), 1)
    j = int(np.argmax(pooled))
    del net; torch.cuda.empty_cache()
    return dict(th=THS[j], pooled=float(pooled[j]), per=per[:, j])

def sha(p, chunk=1 << 24):
    h = hashlib.sha256()
    with open(p, "rb") as f:
        while (b := f.read(chunk)): h.update(b)
    return h.hexdigest()[:16]

# -------------------------- which runs to handle -----------------------
# summary.json is written only at the END of cell D, so its presence is
# the completion test -- safer than trusting a step count.
done = sorted(glob.glob(f"{STAGE}/*/summary.json"), key=os.path.getmtime)
if not done:
    raise SystemExit(f"no completed run under {STAGE}")
target = RUN_NAME or os.path.basename(os.path.dirname(done[-1]))
print(f"target run: {target}   ({len(done)} completed run(s) on Drive)\n")

man_p = f"{SAVE}/weights_manifest.json"
man = json.load(open(man_p)) if os.path.exists(man_p) else {}

for sm in done:
    d = os.path.dirname(sm); name = os.path.basename(d)
    J = json.load(open(sm))
    bp, ck = f"{d}/best.pt", f"{d}/ckpt.pt"
    if not os.path.exists(bp):
        print(f"{name}: no best.pt, skipped"); continue
    is_target = (name == target)

    # 1. per-image vectors
    npz = f"{d}/per_image.npz"
    need = True
    if os.path.exists(npz):
        try:
            z = np.load(npz, allow_pickle=True); need = list(z["stems"]) != CLEAN_VAL
        except Exception: need = True
    if need:
        t = time.time()
        r = score_per_image(bp, J["select_on"], J["input"] == "crop_768_up")
        np.savez(npz, per=r["per"], th=r["th"], pooled=r["pooled"],
                 mode=J["select_on"], stems=np.array(CLEAN_VAL))
        print(f"{name}: scored per image in {time.time()-t:.0f}s  "
              f"(mode {J['select_on']}, th {r['th']}, pooled {r['pooled']:.4f})")
    else:
        print(f"{name}: per_image.npz already current")

    # 2. archive
    zp = f"{ARCH}/{name}_results.zip"
    if is_target or not os.path.exists(zp):
        with zipfile.ZipFile(zp, "w", zipfile.ZIP_DEFLATED) as z:
            for f in ("summary.json", "history.json", "curves.png",
                      "clean_val_gain.npy", "per_image.npz"):
                p = f"{d}/{f}"
                if os.path.exists(p): z.write(p, arcname=f"{name}/{f}")
        print(f"{name}: archive {os.path.getsize(zp)/1e6:.2f} MB")

    # 3. manifest
    if is_target or name not in man:
        man[name] = dict(file=f"{name}__best.pt", bytes=os.path.getsize(bp),
                         sha256_16=sha(bp), best_val_pooled=J.get("best_val_pooled"),
                         gpu=J.get("gpu", "not recorded"), batch=J.get("batch"),
                         input=J["input"], loss=J["loss"], jitter=J["jitter"],
                         seed=J["seed"], floor=J.get("floor"))

    # 4. prune -- only a COMPLETED run, which summary.json proves
    if DELETE_CKPT and os.path.exists(ck):
        mb = os.path.getsize(ck)/1e6
        os.remove(ck); print(f"{name}: removed ckpt.pt (freed {mb:.0f} MB)")

json.dump(man, open(man_p, "w"), indent=1)
print(f"\nmanifest: {man_p}  ({len(man)} run(s))")

# ------------------------------ drive usage ----------------------------
def dsz(p):
    return sum(os.path.getsize(os.path.join(r, f))
               for r, _, fs in os.walk(p) for f in fs) if os.path.isdir(p) else 0
tot = 0
print("\nDrive usage")
for lab, p in (("stage2", STAGE), ("stage2_archive", ARCH),
               ("runs (stage 1)", f"{SAVE}/runs"), ("cache768", f"{SAVE}/cache768")):
    s = dsz(p); tot += s
    print(f"   {lab:18}{s/1e9:7.2f} GB")
print(f"   {'TRACKED TOTAL':18}{tot/1e9:7.2f} GB"
      + ("   !! approaching a 15 GB free Drive" if tot/1e9 > 11 else ""))

# ------------------------------- download ------------------------------
if DOWNLOAD:
    d = f"{STAGE}/{target}"
    loc = f"/content/{target}__best.pt"
    if not os.path.exists(loc): shutil.copyfile(f"{d}/best.pt", loc)
    print(f"\n{target}  best.pt {os.path.getsize(loc)/1e6:.1f} MB  "
          f"sha256[:16] {man[target]['sha256_16']}")
    try:
        from google.colab import files
        files.download(f"{ARCH}/{target}_results.zip")
        time.sleep(2)
        files.download(loc)
        print("downloads queued. Verify locally:")
        print(f"   shasum -a 256 {target}__best.pt | cut -c1-16"
              f"   ->  {man[target]['sha256_16']}")
    except Exception as e:
        print(f"(browser download unavailable: {type(e).__name__})")
    print(f"also on Drive:\n   {ARCH}/{target}_results.zip\n   {d}/best.pt")
```

```python
# =====================================================================
#  cell E -- score every finished Stage 2 run per image, then compare.
#
#  Cell D saves per-quartile MEANS, which cannot support a paired test.
#  This reloads each best.pt and writes the per-image IoU vector over
#  clean_val, at that run's own selected mode and threshold. No retraining.
#
#  Then, against the S-384 baseline, for each other run:
#    - paired per-image differences over the same 688 images
#    - an image-level paired bootstrap on Q0  (the PRIMARY inference:
#      688 paired images carry far more power than a 3-seed t-interval,
#      which at 3 seeds needs a seed-sd below 0.008 to clear +0.02)
#    - the same split by realized detail gain, declared in advance
#    - the PILOT rule: kill only if the mean Q0 difference is <= 0.
#      Killing anything that misses the +0.02 gate at one seed discards
#      a real +0.02 effect half the time; the <= 0 rule, 10% of the time.
#
#  A single-seed result here is a FILTER, never a result. Nothing from
#  this cell goes in the paper without seeds 43 and 44.
# =====================================================================
import os, csv, json, glob, time
import numpy as np
from PIL import Image
import torch, torch.nn as nn, torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models
import torchvision.transforms.functional as TF

SAVE, C768 = "/content/drive/MyDrive", "/content/cache768"
STAGE   = f"{SAVE}/stage2"
BASE    = "S-384_seed42"           # the baseline every run is paired against
TRAIN_RES, FULL_RES = 384, 768
BATCH_EVAL = 4
NBOOT   = 10000
FROZEN  = [0.059468, 0.138448, 0.300831]
KILL_AT = 0.0                      # pilot rule, declared before looking
try:
    from google.colab import drive; drive.mount("/content/drive", force_remount=False)
except Exception: pass
DEV = torch.device("cuda" if torch.cuda.is_available() else "cpu")

SP = json.load(open(f"{SAVE}/plantseg_splits.json"))
I768 = json.load(open(f"{C768}/index.json"))
row = I768["row_of"]
CLEAN_VAL = sorted(set(SP["clean_val"]) & set(I768["stems"]))
DSDIR = "/content/my_images/plantseg"
meta = {os.path.splitext(r["Name"])[0]: r for r in csv.DictReader(
        open(f"{DSDIR}/Metadata.csv", encoding="utf-8-sig", newline=""))}
frac = {k: float(v["Mask ratio"]) for k, v in meta.items()}
def quart(s):
    f = frac[s]
    return 0 if f <= FROZEN[0] else 1 if f <= FROZEN[1] else 2 if f <= FROZEN[2] else 3
QV = np.array([quart(s) for s in CLEAN_VAL])
res = {k: tuple(int(v) for v in meta[k]["Resolution"].lower().split("x")[:2])
       for k in CLEAN_VAL}
GAIN = np.array([min(min(res[s]), FULL_RES)/TRAIN_RES for s in CLEAN_VAL])
print(f"clean_val {len(CLEAN_VAL)}  quartiles {[int((QV==q).sum()) for q in range(4)]}")

MEAN, STD = [0.485,0.456,0.406], [0.229,0.224,0.225]
NORM = transforms.Normalize(MEAN, STD)
_mm = {}
def cache():
    if not _mm:
        _mm["i"] = np.load(I768["img_path"], mmap_mode="r")
        _mm["m"] = np.load(I768["msk_path"], mmap_mode="r")
    return _mm["i"], _mm["m"]
def strip_detail(im):
    return im.resize((TRAIN_RES,)*2, Image.BILINEAR).resize((FULL_RES,)*2, Image.BILINEAR)

class EvalSet(Dataset):
    def __init__(s, st, up): s.st, s.up = list(st), up
    def __len__(s): return len(s.st)
    def __getitem__(s, i):
        I, M = cache(); r = row[s.st[i]]
        img = Image.fromarray(I[r])
        if s.up: img = strip_detail(img)
        y = torch.from_numpy((np.asarray(M[r]) > 0).astype(np.float32)).unsqueeze(0)
        return NORM(TF.to_tensor(img)), y

SPEC = dict(channels=(64,256,512,1024,2048), decoder=(512,256,128,64))
class DoubleConv(nn.Module):
    def __init__(s,i,o):
        super().__init__()
        s.net = nn.Sequential(nn.Conv2d(i,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True),
                              nn.Conv2d(o,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True))
    def forward(s,x): return s.net(x)
class Up(nn.Module):
    def __init__(s,i,k,o):
        super().__init__(); s.reduce=nn.Conv2d(i,i//2,1,bias=False); s.conv=DoubleConv(i//2+k,o)
    def forward(s,x,skip):
        x = F.interpolate(x, size=skip.shape[-2:], mode="bilinear", align_corners=False)
        return s.conv(torch.cat([s.reduce(x), skip], 1))
class ResUNet(nn.Module):
    def __init__(s):
        super().__init__()
        base = models.resnet50(weights=None)
        c0,c1,c2,c3,c4 = SPEC["channels"]; d1,d2,d3,d4 = SPEC["decoder"]
        L = list(base.children())
        s.stem=nn.Sequential(*L[:3]); s.pool=L[3]
        s.l1,s.l2,s.l3,s.l4 = L[4],L[5],L[6],L[7]
        s.u1,s.u2,s.u3,s.u4 = Up(c4,c3,d1),Up(d1,c2,d2),Up(d2,c1,d3),Up(d3,c0,d4)
        s.head = nn.Sequential(nn.Conv2d(d4,32,3,padding=1,bias=False),
                               nn.BatchNorm2d(32), nn.ReLU(True), nn.Conv2d(32,1,1))
    def forward(s,x):
        size=x.shape[-2:]
        x0=s.stem(x); x1=s.l1(s.pool(x0)); x2=s.l2(x1); x3=s.l3(x2); x4=s.l4(x3)
        d=s.u1(x4,x3); d=s.u2(d,x2); d=s.u3(d,x1); d=s.u4(d,x0)
        return s.head(F.interpolate(d,size=size,mode="bilinear",align_corners=False))

@torch.no_grad()
def prob_768(net, x, mode):
    if mode == "full768":
        return torch.sigmoid(net(x).float())
    if mode == "whole384":
        d = F.interpolate(x, size=(TRAIN_RES,)*2, mode="bilinear", align_corners=False)
        p = torch.sigmoid(net(d).float())
        return F.interpolate(p, size=(FULL_RES,)*2, mode="bilinear", align_corners=False)
    acc = torch.zeros(x.shape[0], 1, FULL_RES, FULL_RES, device=x.device)
    cnt = torch.zeros_like(acc)
    ORIG = (0, (FULL_RES-TRAIN_RES)//2, FULL_RES-TRAIN_RES)
    for oy in ORIG:
        for ox in ORIG:
            t = x[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES]
            acc[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += torch.sigmoid(net(t).float())
            cnt[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += 1
    return acc/cnt.clamp(min=1)

THS = [round(t, 2) for t in np.arange(0.10, 0.91, 0.05)]
@torch.no_grad()
def score(bp, mode, up):
    net = ResUNet().to(DEV); net.load_state_dict(torch.load(bp, map_location=DEV)); net.eval()
    dl = DataLoader(EvalSet(CLEAN_VAL, up), BATCH_EVAL, shuffle=False,
                    num_workers=min(2, os.cpu_count() or 1), pin_memory=DEV.type == "cuda")
    n = len(CLEAN_VAL)
    per = np.zeros((n, len(THS))); Ipx = np.zeros((n, len(THS)))
    P = np.zeros((n, len(THS))); G = np.zeros(n); k = 0
    for x, y in dl:
        x, y = x.to(DEV, non_blocking=True), y.to(DEV, non_blocking=True)
        with torch.amp.autocast("cuda", enabled=DEV.type == "cuda"):
            p = prob_768(net, x, mode)
        b = x.shape[0]; yf = y.view(b, -1); pf = p.view(b, -1)
        G[k:k+b] = yf.sum(1).cpu().numpy()
        for j, t in enumerate(THS):
            pr = (pf > t).float(); it = (pr*yf).sum(1); pp = pr.sum(1)
            Ipx[k:k+b, j] = it.cpu().numpy(); P[k:k+b, j] = pp.cpu().numpy()
            per[k:k+b, j] = (it/(pp+yf.sum(1)-it).clamp(min=1)).cpu().numpy()
        k += b
    pooled = Ipx.sum(0)/np.maximum(P.sum(0)+G.sum()-Ipx.sum(0), 1)
    j = int(np.argmax(pooled))
    del net; torch.cuda.empty_cache()
    return dict(th=THS[j], pooled=float(pooled[j]), per=per[:, j])

# ------------------------- score what is finished ----------------------
runs = {}
for sm in sorted(glob.glob(f"{STAGE}/*/summary.json")):
    d = os.path.dirname(sm); name = os.path.basename(d)
    if not os.path.exists(f"{d}/best.pt"):
        print(f"   {name}: no best.pt, skipped"); continue
    J = json.load(open(sm))
    npz = f"{d}/per_image.npz"
    if os.path.exists(npz):
        z = np.load(npz, allow_pickle=True)
        if list(z["stems"]) == CLEAN_VAL:
            runs[name] = dict(per=z["per"], th=float(z["th"]), pooled=float(z["pooled"]),
                              mode=str(z["mode"]), cfg=J); print(f"   {name}: cached"); continue
    t = time.time()
    r = score(f"{d}/best.pt", J["select_on"], J["input"] == "crop_768_up")
    np.savez(npz, per=r["per"], th=r["th"], pooled=r["pooled"],
             mode=J["select_on"], stems=np.array(CLEAN_VAL))
    runs[name] = dict(**r, mode=J["select_on"], cfg=J)
    print(f"   {name}: scored in {time.time()-t:.0f}s  "
          f"mode {J['select_on']} th {r['th']} pooled {r['pooled']:.4f}")

if BASE not in runs:
    raise SystemExit(f"baseline {BASE} not scored yet -- run it first")
print(f"\n{len(runs)} run(s) scored; baseline {BASE}")

# --------------------------- paired comparison -------------------------
rng = np.random.default_rng(0)
b_per = runs[BASE]["per"]; q0 = QV == 0
bands = {"Q0 all": q0, "Q0 gain<=1.0": q0 & (GAIN <= 1.0),
         "Q0 gain>=1.5": q0 & (GAIN >= 1.5), "all images": np.ones(len(QV), bool)}

def boot(d, sel):
    v = d[sel]
    if len(v) < 5: return float("nan"), float("nan"), float("nan")
    idx = rng.integers(0, len(v), (NBOOT, len(v)))
    bs = v[idx].mean(1)
    return float(v.mean()), float(np.percentile(bs, 2.5)), float(np.percentile(bs, 97.5))

print("\n" + "=" * 88)
print(f"paired per-image IoU difference vs {BASE}, clean_val, "
      f"{NBOOT} image-level bootstraps")
print("=" * 88)
out = {}
for name, r in runs.items():
    if name == BASE: continue
    d = r["per"] - b_per
    print(f"\n{name}   ({r['cfg']['input']} / {r['cfg']['loss']}"
          f"{' / jitter' if r['cfg']['jitter'] else ''})   "
          f"pooled {r['pooled']:.4f} vs {runs[BASE]['pooled']:.4f}")
    print(f"   {'band':16}{'n':>6}{'mean diff':>12}{'95% CI':>22}")
    row = {}
    for lab, sel in bands.items():
        m, lo, hi = boot(d, sel)
        star = "  *" if (lo > 0 or hi < 0) else ""
        print(f"   {lab:16}{int(sel.sum()):>6}{m:>12.4f}"
              f"   [{lo:+.4f}, {hi:+.4f}]{star}")
        row[lab] = dict(n=int(sel.sum()), mean=m, lo=lo, hi=hi)
    mq0 = row["Q0 all"]["mean"]
    verdict = ("KILL -- Q0 moves the wrong way" if mq0 <= KILL_AT else
               "KEEP -- run seeds 43 and 44")
    print(f"   PILOT ({'<=' if mq0 <= KILL_AT else '>'} {KILL_AT:+.2f}): {verdict}")
    if row["Q0 gain>=1.5"]["n"] >= 20 and row["Q0 gain<=1.0"]["n"] >= 20:
        sep = row["Q0 gain>=1.5"]["mean"] - row["Q0 gain<=1.0"]["mean"]
        print(f"   mechanism check: high-detail minus no-detail = {sep:+.4f}"
              f"   {'(consistent with resolution)' if sep > 0 else '(not resolution)'}")
        row["mechanism_sep"] = sep
    out[name] = row

print("\n" + "=" * 88)
print("* marks a CI excluding zero. That is an image-level interval: it says the")
print("difference is not sampling noise over these 688 images, NOT that it would")
print("survive another training seed. Only seeds 43 and 44 answer that, and the")
print("gate is judged on the three-seed paired mean, not on anything here.")
json.dump(dict(baseline=BASE, kill_at=KILL_AT, nboot=NBOOT, results=out),
          open(f"{SAVE}/stage2_compare.json", "w"), indent=1, default=float)
print(f"\nwrote {SAVE}/stage2_compare.json")
```

```python
# =====================================================================
#  cell D -- Stage 2: supervised tests of components 1 and 2.
#  One cell, three switches, 25,000 steps, resume-safe. Validation only;
#  the test split is not touched until Stage 5.
#
#      INPUT_MODE in {full_384, crop_768, crop_768_up}
#      FG_LOSS    in {pooled, class_balanced, image_balanced}
#      JITTER     in {False, True}
#
#  (full_384, pooled, False) is the S-384 baseline.
#
#  crop_768_up is the control the crop statistics made necessary: uniform
#  384 crops of a 768 image carry 1.50x the Q0 lesion density of the
#  whole image, because uniform crop ORIGINS over-sample the centre (a
#  centre pixel sits in 384 of 385 origins per axis, a corner pixel in 1).
#  So crop_768 changes resolution AND density at once. crop_768_up takes
#  the 768 image, downsamples it to 384 and upsamples it back, destroying
#  exactly the detail above 384 and nothing else, then crops identically.
#  crop_768 beating crop_768_up isolates resolution; a tie says the gain
#  is crop geometry.
#
#  Everything reads ONE cache: /content/cache768 (537 labelled + 846 val).
#  The 441 cache is not needed, so prep.py need not be run this session.
# =====================================================================
import os, csv, json, math, time, random, shutil, zipfile, collections
import numpy as np
import matplotlib; matplotlib.use("Agg")
import matplotlib.pyplot as plt
from PIL import Image, ImageFilter
import torch, torch.nn as nn, torch.optim as optim, torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models
import torchvision.transforms.functional as TF

# ------------------------- the three switches --------------------------
INPUT_MODE = "full_384"        # full_384 | crop_768 | crop_768_up
FG_LOSS    = "class_balanced"          # pooled | class_balanced | image_balanced
JITTER     = False             # True only for S-768c-jit
TRAIN_SEED = 43
# -----------------------------------------------------------------------
FLOOR_FRAC = 0.01              # declared from check 0f: binds on 8.9% of
                               # non-empty Q0 crops, caps amplification 21.5x
SELECT_ON  = {"full_384": "whole384", "crop_768": "full768",
              "crop_768_up": "full768"}[INPUT_MODE]
MAX_STEPS  = 25000
EVAL_EVERY = 1000
SHARE_EVERY = 500              # per-quartile lesion-loss share
CKPT_EVERY = 2500
LR_PATIENCE = 3                # patience counts EVAL_EVERY blocks, now 1000
TRAIN_RES  = 384
FULL_RES   = 768
BATCH      = 4                 # matches the existing pipeline; batch 8
                               # doubles every run to ~2.2h and Stage 2 to ~42h
BATCH_EVAL = 4                 # 768 input on a T4
LR_ENCODER, LR_DECODER, WEIGHT_DECAY = 1e-5, 1e-4, 1e-4
FROZEN     = [0.059468, 0.138448, 0.300831]
SAVE, C768 = "/content/drive/MyDrive", "/content/cache768"
NUM_WORKERS = min(2, os.cpu_count() or 1)
FRESH      = True              # wipes this run's folder. Set back to False
                               # once it is past step 2500 so a dropped
                               # session resumes instead of restarting.

TAG = {"full_384": "384", "crop_768": "768c", "crop_768_up": "768up"}[INPUT_MODE]
if FG_LOSS == "class_balanced":  TAG += "-cb"
if FG_LOSS == "image_balanced":  TAG += "-bal"
if JITTER:                       TAG += "-jit"
RUN = f"S-{TAG}_seed{TRAIN_SEED}"
OUT = f"{SAVE}/stage2/{RUN}"
if FRESH and os.path.isdir(OUT): shutil.rmtree(OUT, ignore_errors=True)
os.makedirs(OUT, exist_ok=True)

random.seed(TRAIN_SEED); np.random.seed(TRAIN_SEED)
torch.manual_seed(TRAIN_SEED); torch.cuda.manual_seed_all(TRAIN_SEED)
torch.backends.cudnn.deterministic = True; torch.backends.cudnn.benchmark = False
DEV = torch.device("cuda" if torch.cuda.is_available() else "cpu")
try:
    from google.colab import drive; drive.mount("/content/drive", force_remount=False)
except Exception: pass
print(f"{RUN}\ndevice {DEV} {torch.cuda.get_device_name(0) if DEV.type=='cuda' else ''}")
print(f"input {INPUT_MODE} | loss {FG_LOSS} | jitter {JITTER} | "
      f"select on {SELECT_ON} | floor {FLOOR_FRAC} | batch {BATCH}")

# --------------------------- data and asserts --------------------------
SP = json.load(open(f"{SAVE}/plantseg_splits.json"))
if max(abs(a-b) for a, b in zip(SP["quartile_cuts"], FROZEN)) > 1e-6:
    raise SystemExit(f"cuts {SP['quartile_cuts']} != frozen {FROZEN}")
if {k: len(v) for k, v in SP["official_splits"].items()} != \
   {"train": 5367, "val": 846, "test": 1561}:
    raise SystemExit("split sizes are not 5367/846/1561")
I768 = json.load(open(f"{C768}/index.json"))
if I768["load_size"] != FULL_RES:
    raise SystemExit(f"cache is {I768['load_size']}, expected {FULL_RES}")
row = I768["row_of"]
LAB = sorted(I768["labelled"])
CLEAN_VAL = sorted(set(SP["clean_val"]) & set(I768["stems"]))
if len(LAB) != 537: raise SystemExit(f"{len(LAB)} labelled, expected 537")
DSDIR = "/content/my_images/plantseg"
meta = {os.path.splitext(r["Name"])[0]: r for r in csv.DictReader(
        open(f"{DSDIR}/Metadata.csv", encoding="utf-8-sig", newline=""))}
frac = {k: float(v["Mask ratio"]) for k, v in meta.items()}
def quart(s):
    f = frac[s]
    return 0 if f <= FROZEN[0] else 1 if f <= FROZEN[1] else 2 if f <= FROZEN[2] else 3
QV = np.array([quart(s) for s in CLEAN_VAL])
print(f"labelled {len(LAB)}   clean_val {len(CLEAN_VAL)} "
      f"(quartiles {[int((QV==q).sum()) for q in range(4)]})")
print("   gates are judged on clean_val, not the full 846: full val carries")
print("   17.14% near-duplicates of training images.")

MEAN, STD = [0.485,0.456,0.406], [0.229,0.224,0.225]
NORM = transforms.Normalize(MEAN, STD)
JIT  = transforms.ColorJitter(0.4, 0.4, 0.4, 0.1)
_mm = {}
def cache():
    if not _mm:
        _mm["i"] = np.load(I768["img_path"], mmap_mode="r")
        _mm["m"] = np.load(I768["msk_path"], mmap_mode="r")
    return _mm["i"], _mm["m"]

def strip_detail(im):
    """768 -> 384 -> 768. Destroys exactly the detail above 384, keeps
       geometry, size and lesion density identical."""
    return im.resize((TRAIN_RES,)*2, Image.BILINEAR).resize((FULL_RES,)*2, Image.BILINEAR)

def photometric(img):
    if random.random() < 0.8: img = JIT(img)
    if random.random() < 0.2: img = TF.to_grayscale(img, num_output_channels=3)
    if random.random() < 0.3:
        img = img.filter(ImageFilter.GaussianBlur(radius=random.uniform(0.1, 1.5)))
    return img

def geometric(img, msk):
    if random.random() < 0.5: img, msk = TF.hflip(img), TF.hflip(msk)
    if random.random() < 0.5: img, msk = TF.vflip(img), TF.vflip(msk)
    k = random.randint(0, 3)
    if k:
        img = img.rotate(90*k, expand=True, resample=Image.BILINEAR)
        msk = msk.rotate(90*k, expand=True, resample=Image.NEAREST)
    return img, msk

def rrc(img, msk, scale=(0.5, 1.0), ratio=(3/4, 4/3)):
    w, h = img.size; area = w*h
    for _ in range(10):
        t = area*random.uniform(*scale)
        ar = math.exp(random.uniform(math.log(ratio[0]), math.log(ratio[1])))
        cw, ch = int(round(math.sqrt(t*ar))), int(round(math.sqrt(t/ar)))
        if cw <= w and ch <= h:
            x, y = random.randint(0, w-cw), random.randint(0, h-ch)
            return img.crop((x, y, x+cw, y+ch)), msk.crop((x, y, x+cw, y+ch))
    return img, msk

class Labelled(Dataset):
    """full_384   : whole image, rrc + geometric, resized to 384
       crop_768   : uniform 384 window of the 768 image (the crop IS the
                    spatial augmentation, so rrc is not applied on top)
       crop_768_up: the same window of a detail-stripped 768 image"""
    def __init__(s, st): s.st = list(st)
    def __len__(s): return len(s.st)
    def __getitem__(s, i):
        I, M = cache(); r = row[s.st[i]]
        img = Image.fromarray(I[r]); msk = Image.fromarray(M[r])
        if INPUT_MODE == "crop_768_up": img = strip_detail(img)
        if INPUT_MODE == "full_384":
            img, msk = rrc(img, msk)
            img, msk = geometric(img, msk)
            img = img.resize((TRAIN_RES,)*2, Image.BILINEAR)
            msk = msk.resize((TRAIN_RES,)*2, Image.NEAREST)
        else:
            side = (random.randint(256, FULL_RES) if JITTER else TRAIN_RES)
            x = random.randint(0, FULL_RES-side); y = random.randint(0, FULL_RES-side)
            img = img.crop((x, y, x+side, y+side)); msk = msk.crop((x, y, x+side, y+side))
            img, msk = geometric(img, msk)
            if side != TRAIN_RES:
                img = img.resize((TRAIN_RES,)*2, Image.BILINEAR)
                msk = msk.resize((TRAIN_RES,)*2, Image.NEAREST)
        y_ = torch.from_numpy((np.array(msk) > 0).astype(np.float32)).unsqueeze(0)
        return NORM(TF.to_tensor(photometric(img))), y_, quart(s.st[i])

class EvalSet(Dataset):
    """input follows INPUT_MODE; the TARGET is always the 768 mask."""
    def __init__(s, st): s.st = list(st)
    def __len__(s): return len(s.st)
    def __getitem__(s, i):
        I, M = cache(); r = row[s.st[i]]
        img = Image.fromarray(I[r])
        if INPUT_MODE == "crop_768_up": img = strip_detail(img)
        y = torch.from_numpy((np.asarray(M[r]) > 0).astype(np.float32)).unsqueeze(0)
        return NORM(TF.to_tensor(img)), y            # image at 768, mask at 768

def seed_worker(_):
    s = torch.initial_seed() % 2**32; np.random.seed(s); random.seed(s)
LK = dict(num_workers=NUM_WORKERS, pin_memory=DEV.type == "cuda",
          worker_init_fn=seed_worker, persistent_workers=NUM_WORKERS > 0)
g = torch.Generator(); g.manual_seed(TRAIN_SEED)
loader_l = DataLoader(Labelled(LAB), BATCH, shuffle=True, drop_last=True,
                      generator=g, **LK)
val_loader = DataLoader(EvalSet(CLEAN_VAL), BATCH_EVAL, shuffle=False, **LK)
def cycle(dl):
    while True:
        for b in dl: yield b

# ------------------------------- model ---------------------------------
SPEC = dict(channels=(64,256,512,1024,2048), decoder=(512,256,128,64))
class DoubleConv(nn.Module):
    def __init__(s,i,o):
        super().__init__()
        s.net = nn.Sequential(nn.Conv2d(i,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True),
                              nn.Conv2d(o,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True))
    def forward(s,x): return s.net(x)
class Up(nn.Module):
    def __init__(s,i,k,o):
        super().__init__(); s.reduce=nn.Conv2d(i,i//2,1,bias=False); s.conv=DoubleConv(i//2+k,o)
    def forward(s,x,skip):
        x = F.interpolate(x, size=skip.shape[-2:], mode="bilinear", align_corners=False)
        return s.conv(torch.cat([s.reduce(x), skip], 1))
class ResUNet(nn.Module):
    def __init__(s):
        super().__init__()
        base = models.resnet50(weights=models.ResNet50_Weights.DEFAULT)
        c0,c1,c2,c3,c4 = SPEC["channels"]; d1,d2,d3,d4 = SPEC["decoder"]
        L = list(base.children())
        s.stem=nn.Sequential(*L[:3]); s.pool=L[3]
        s.l1,s.l2,s.l3,s.l4 = L[4],L[5],L[6],L[7]
        s.u1,s.u2,s.u3,s.u4 = Up(c4,c3,d1),Up(d1,c2,d2),Up(d2,c1,d3),Up(d3,c0,d4)
        s.head = nn.Sequential(nn.Conv2d(d4,32,3,padding=1,bias=False),
                               nn.BatchNorm2d(32), nn.ReLU(True), nn.Conv2d(32,1,1))
    def enc_params(s):
        for m in (s.stem,s.l1,s.l2,s.l3,s.l4): yield from m.parameters()
    def dec_params(s):
        for m in (s.u1,s.u2,s.u3,s.u4,s.head): yield from m.parameters()
    def forward(s,x):
        size=x.shape[-2:]
        x0=s.stem(x); x1=s.l1(s.pool(x0)); x2=s.l2(x1); x3=s.l3(x2); x4=s.l4(x3)
        d=s.u1(x4,x3); d=s.u2(d,x2); d=s.u3(d,x1); d=s.u4(d,x0)
        return s.head(F.interpolate(d,size=size,mode="bilinear",align_corners=False))

# ------------------------------- losses --------------------------------
bce_none = nn.BCEWithLogitsLoss(reduction="none")
def tversky(logits, tgt, eps=1.0):
    p = torch.sigmoid(logits)
    tp = (p*tgt).sum((1,2,3)); fp = (p*(1-tgt)).sum((1,2,3)); fn = ((1-p)*tgt).sum((1,2,3))
    return 1 - (tp+eps)/(tp+0.5*fp+0.5*fn+eps)

def fg_term(logits, tgt):
    """returns (per-image BCE term, per-image lesion-only contribution).
       All three forms are scaled so a uniform per-pixel loss c gives c,
       so the learning rate and the BCE/Tversky balance do not shift
       between configurations. The plan's formula writes the sum of the
       two class means; using their average is the same comparison at the
       same scale."""
    e = bce_none(logits, tgt)
    sf = tgt.sum((1,2,3)); sb = (1-tgt).sum((1,2,3))
    ef = (e*tgt).sum((1,2,3)); eb = (e*(1-tgt)).sum((1,2,3))
    if FG_LOSS == "pooled":
        return e.mean((1,2,3)), ef/tgt[0].numel()
    if FG_LOSS == "class_balanced":                 # batch-level class means
        n = logits.shape[0]
        mf = ef.sum()/sf.sum().clamp(min=1.0); mb = eb.sum()/sb.sum().clamp(min=1.0)
        return (0.5*(mf+mb)).expand(n), 0.5*ef/sf.sum().clamp(min=1.0)
    floor = FLOOR_FRAC*tgt[0].numel()               # per-image, with the floor
    tf = ef/torch.clamp(sf, min=floor)
    tb = eb/sb.clamp(min=1.0)
    return 0.5*(tf+tb), 0.5*tf

def sup_loss(logits, tgt):
    b, lesion = fg_term(logits, tgt)
    return b + tversky(logits, tgt), lesion

# --------------------------- 768 inference -----------------------------
@torch.no_grad()
def prob_768(net, x, mode):
    """x is the model's input image at 768. Returns a 768 probability map."""
    if mode == "full768":
        return torch.sigmoid(net(x).float())
    if mode == "whole384":
        d = F.interpolate(x, size=(TRAIN_RES,)*2, mode="bilinear", align_corners=False)
        p = torch.sigmoid(net(d).float())
        return F.interpolate(p, size=(FULL_RES,)*2, mode="bilinear", align_corners=False)
    # genuinely overlapping: stride 192 gives 3x3 tiles at 50% overlap.
    # Origins (0, 384) alone would ABUT, not overlap, and blending would
    # be a no-op -- the seam is exactly where a tiled model is weakest.
    acc = torch.zeros(x.shape[0], 1, FULL_RES, FULL_RES, device=x.device)
    cnt = torch.zeros_like(acc)
    ORIG = (0, (FULL_RES-TRAIN_RES)//2, FULL_RES-TRAIN_RES)
    for oy in ORIG:
        for ox in ORIG:
            t = x[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES]
            acc[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += torch.sigmoid(net(t).float())
            cnt[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += 1
    return acc/cnt.clamp(min=1)                      # blend, threshold afterwards

THS = [round(t, 2) for t in np.arange(0.10, 0.91, 0.05)]
@torch.no_grad()
def evaluate(net, mode):
    net.eval()
    nI = len(CLEAN_VAL); per = np.zeros((nI, len(THS))); P = np.zeros((nI, len(THS)))
    Ipx = np.zeros((nI, len(THS))); G = np.zeros(nI); k = 0
    for x, y in val_loader:
        x, y = x.to(DEV, non_blocking=True), y.to(DEV, non_blocking=True)
        with torch.amp.autocast("cuda", enabled=DEV.type == "cuda"):
            p = prob_768(net, x, mode)
        b = x.shape[0]; yf = y.view(b, -1)
        G[k:k+b] = yf.sum(1).cpu().numpy()
        pf = p.view(b, -1)
        for j, t in enumerate(THS):
            pr = (pf > t).float()
            it = (pr*yf).sum(1); pp = pr.sum(1)
            Ipx[k:k+b, j] = it.cpu().numpy(); P[k:k+b, j] = pp.cpu().numpy()
            per[k:k+b, j] = (it/(pp+yf.sum(1)-it).clamp(min=1)).cpu().numpy()
        k += b
    pooled = Ipx.sum(0)/np.maximum(P.sum(0)+G.sum()-Ipx.sum(0), 1)
    j = int(np.argmax(pooled)); t = THS[j]
    byq = {q: float(per[QV == q, j].mean()) for q in range(4)}
    return dict(th=t, pooled=float(pooled[j]), perimg=float(per[:, j].mean()),
                byq=byq, pred_frac=float(P[:, j].sum()/(len(CLEAN_VAL)*FULL_RES*FULL_RES)))

# --------------------------- resume or start ---------------------------
model = ResUNet().to(DEV)
opt = optim.AdamW([{"params": list(model.enc_params()), "lr": LR_ENCODER},
                   {"params": list(model.dec_params()), "lr": LR_DECODER}],
                  weight_decay=WEIGHT_DECAY)
sched = optim.lr_scheduler.ReduceLROnPlateau(opt, mode="max", patience=LR_PATIENCE, factor=0.5)
scaler = torch.amp.GradScaler("cuda")
step, best, history, shares = 0, -1.0, [], []
CK, BEST = f"{OUT}/ckpt.pt", f"{OUT}/best.pt"
if os.path.exists(CK):
    c = torch.load(CK, map_location=DEV, weights_only=False)
    model.load_state_dict(c["model"]); opt.load_state_dict(c["opt"])
    sched.load_state_dict(c["sched"]); scaler.load_state_dict(c["scaler"])
    step, best, history, shares = c["step"], c["best"], c["history"], c.get("shares", [])
    random.setstate(c["py_rng"]); np.random.set_state(c["np_rng"])
    torch.set_rng_state(c["t_rng"].cpu())
    if c.get("cuda_rng") is not None and DEV.type == "cuda":
        torch.cuda.set_rng_state(c["cuda_rng"].cpu())
    print(f"RESUMED from step {step} (best {best:.4f})")
else:
    print("starting fresh")

def save_ckpt():
    torch.save(dict(model=model.state_dict(), opt=opt.state_dict(), sched=sched.state_dict(),
                    scaler=scaler.state_dict(), step=step, best=best, history=history,
                    shares=shares, py_rng=random.getstate(), np_rng=np.random.get_state(),
                    t_rng=torch.get_rng_state(),
                    cuda_rng=torch.cuda.get_rng_state() if DEV.type == "cuda" else None), CK)

# ------------------------------- train ---------------------------------
it = cycle(loader_l)
print("=" * 76); print(f"{RUN}  --  {len(LAB)} labelled, {MAX_STEPS} steps"); print("=" * 76)
t0 = t_last = time.time()
r_loss = r_n = 0.0; q_les = np.zeros(4); q_cnt = np.zeros(4)
while step < MAX_STEPS:
    model.train()
    x, y, q = next(it)
    x, y = x.to(DEV, non_blocking=True), y.to(DEV, non_blocking=True)
    opt.zero_grad(set_to_none=True)
    with torch.amp.autocast("cuda", enabled=DEV.type == "cuda"):
        lg = model(x)
        vec, lesion = sup_loss(lg.float(), y)
        loss = vec.mean()
    scaler.scale(loss).backward(); scaler.step(opt); scaler.update()
    step += 1
    r_loss += loss.item(); r_n += 1
    lz = lesion.detach().float().cpu().numpy(); qz = q.numpy()
    for b in range(len(qz)): q_les[qz[b]] += lz[b]; q_cnt[qz[b]] += 1

    if step % SHARE_EVERY == 0:
        tot = max(q_les.sum(), 1e-12)
        sh = (q_les/tot).round(4).tolist()
        shares.append(dict(step=step, share=sh, seen=q_cnt.astype(int).tolist()))
        print(f"   [step {step}] lesion-loss share by quartile {sh}"
              f"  (images seen {q_cnt.astype(int).tolist()})")
        q_les[:] = 0; q_cnt[:] = 0

    if step % EVAL_EVERY == 0:
        m = evaluate(model, SELECT_ON)
        prev = opt.param_groups[-1]["lr"]; sched.step(m["pooled"])
        cur = opt.param_groups[-1]["lr"]
        saved = ""
        if m["pooled"] > best:
            best = m["pooled"]; torch.save(model.state_dict(), BEST); saved = "  <-- saved"
        history.append(dict(step=step, train_loss=r_loss/max(r_n, 1), lr=cur, **m))
        sps = EVAL_EVERY/(time.time()-t_last); t_last = time.time()
        print(f"step {step:6d} | loss {r_loss/max(r_n,1):.4f} | val pooled {m['pooled']:.4f} "
              f"@{m['th']} | per-img {m['perimg']:.4f} | Q0 {m['byq'][0]:.4f} "
              f"Q3 {m['byq'][3]:.4f} | best {best:.4f} | {sps:.2f} it/s | "
              f"ETA {(MAX_STEPS-step)/sps/3600:.1f}h{saved}"
              + (f"  [lr {prev:.1e}->{cur:.1e}]" if cur != prev else ""))
        r_loss = r_n = 0.0
        json.dump(dict(run=RUN, input=INPUT_MODE, loss=FG_LOSS, jitter=JITTER,
                       seed=TRAIN_SEED, floor=FLOOR_FRAC, select_on=SELECT_ON,
                       batch=BATCH, history=history, shares=shares),
                  open(f"{OUT}/history.json", "w"), indent=1)
    if step % CKPT_EVERY == 0:
        save_ckpt(); print(f"    checkpointed at step {step}")
save_ckpt()
print(f"\ntraining done in {(time.time()-t0)/3600:.2f}h, best val pooled {best:.4f}")

# ------------------- final: all three inference modes ------------------
net = ResUNet().to(DEV); net.load_state_dict(torch.load(BEST, map_location=DEV)); net.eval()
print("\n" + "=" * 76)
print("clean_val under each inference mode (best.pt was selected on "
      f"{SELECT_ON})")
print("=" * 76)
print(f"{'mode':12}{'th':>6}{'pooled':>9}{'per-img':>9}"
      f"{'Q0':>8}{'Q1':>8}{'Q2':>8}{'Q3':>8}{'Q3-Q0':>8}")
modes = ["whole384", "full768", "tiles384"]
final = {}
for md in modes:
    m = evaluate(net, md); final[md] = m
    print(f"{md:12}{m['th']:>6.2f}{m['pooled']:>9.4f}{m['perimg']:>9.4f}"
          + "".join(f"{m['byq'][q]:>8.4f}" for q in range(4))
          + f"{m['byq'][3]-m['byq'][0]:>8.4f}")

# the stratified check: does any Q0 gain track the realized detail gain?
res = {k: tuple(int(v) for v in meta[k]["Resolution"].lower().split("x")[:2])
       for k in CLEAN_VAL}
gain = np.array([min(min(res[s]), FULL_RES)/TRAIN_RES for s in CLEAN_VAL])
q0 = QV == 0
print(f"\nclean_val Q0 split by realized worst-axis detail gain at {FULL_RES}:")
for lab, sel in (("<= 1.0 (no gain)", q0 & (gain <= 1.0)),
                 ("1.0 - 1.5", q0 & (gain > 1.0) & (gain < 1.5)),
                 (">= 1.5", q0 & (gain >= 1.5))):
    print(f"   {lab:18} n={int(sel.sum()):4d}")
print("   per-image Q0 IoU within each band is in the saved json; compare it")
print("   against the matching S-384 run, paired by image.")

np.save(f"{OUT}/clean_val_gain.npy", gain)
json.dump(dict(run=RUN, input=INPUT_MODE, loss=FG_LOSS, jitter=JITTER, seed=TRAIN_SEED,
               floor=FLOOR_FRAC, select_on=SELECT_ON, batch=BATCH, best_val_pooled=best,
               clean_val=CLEAN_VAL, quartiles=QV.tolist(), final=final,
               history=history, shares=shares),
          open(f"{OUT}/summary.json", "w"), indent=1)
xs = [h["step"] for h in history]
plt.figure(figsize=(14, 4))
plt.subplot(1, 3, 1); plt.plot(xs, [h["train_loss"] for h in history]); plt.title("train loss")
plt.subplot(1, 3, 2); plt.plot(xs, [h["pooled"] for h in history]); plt.title("val pooled IoU")
plt.subplot(1, 3, 3)
for q in range(4): plt.plot(xs, [h["byq"][q] for h in history], label=f"Q{q}")
plt.legend(fontsize=7); plt.title("val per-image IoU by quartile")
plt.tight_layout(); plt.savefig(f"{OUT}/curves.png", dpi=110); plt.close()
print(f"\nwrote {OUT}/summary.json, best.pt, curves.png")
print(">>> download best.pt to local disk before the session ends <<<")
```

```python
# =====================================================================
#  cell F -- run this ONCE after every cell D training. Standalone: it
#  reads nothing from cell D's variables, so it also works in a fresh
#  session or after a runtime restart.
#
#  For the target run it:
#    1. scores it per image on clean_val  -> per_image.npz
#       (the vectors every paired test needs; cell D saves only means)
#    2. archives summary / history / curves / gain / per_image to Drive
#    3. records config, GPU and a checksum in weights_manifest.json
#    4. deletes ckpt.pt, but ONLY if the run completed. 490 MB each, and
#       19 of them would overrun a free Drive.
#    5. prints what Drive is holding
#    6. downloads the archive and best.pt
#
#  Any OTHER run still missing per_image.npz or an archive is also
#  brought up to date, without being downloaded again.
# =====================================================================
import os, csv, json, glob, time, zipfile, shutil, hashlib
import numpy as np
from PIL import Image
import torch, torch.nn as nn, torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models
import torchvision.transforms.functional as TF

RUN_NAME    = None       # None = the most recently finished run in stage2/
DELETE_CKPT = True       # False if you intend to extend a run later
DOWNLOAD    = True
SAVE, C768  = "/content/drive/MyDrive", "/content/cache768"
STAGE, ARCH = f"{SAVE}/stage2", f"{SAVE}/stage2_archive"
TRAIN_RES, FULL_RES = 384, 768
BATCH_EVAL  = 12 if torch.cuda.is_available() else 2
FROZEN      = [0.059468, 0.138448, 0.300831]
try:
    from google.colab import drive; drive.mount("/content/drive", force_remount=False)
except Exception: pass
DEV = torch.device("cuda" if torch.cuda.is_available() else "cpu")
os.makedirs(ARCH, exist_ok=True)

# ------------------------------- context -------------------------------
SP   = json.load(open(f"{SAVE}/plantseg_splits.json"))
I768 = json.load(open(f"{C768}/index.json"))
row  = I768["row_of"]
CLEAN_VAL = sorted(set(SP["clean_val"]) & set(I768["stems"]))
DSDIR = "/content/my_images/plantseg"
meta = {os.path.splitext(r["Name"])[0]: r for r in csv.DictReader(
        open(f"{DSDIR}/Metadata.csv", encoding="utf-8-sig", newline=""))}
frac = {k: float(v["Mask ratio"]) for k, v in meta.items()}
def quart(s):
    f = frac[s]
    return 0 if f <= FROZEN[0] else 1 if f <= FROZEN[1] else 2 if f <= FROZEN[2] else 3
QV = np.array([quart(s) for s in CLEAN_VAL])

MEAN, STD = [0.485,0.456,0.406], [0.229,0.224,0.225]
NORM = transforms.Normalize(MEAN, STD)
_mm = {}
def cache():
    if not _mm:
        _mm["i"] = np.load(I768["img_path"], mmap_mode="r")
        _mm["m"] = np.load(I768["msk_path"], mmap_mode="r")
    return _mm["i"], _mm["m"]
def strip_detail(im):
    return im.resize((TRAIN_RES,)*2, Image.BILINEAR).resize((FULL_RES,)*2, Image.BILINEAR)

class EvalSet(Dataset):
    def __init__(s, st, up): s.st, s.up = list(st), up
    def __len__(s): return len(s.st)
    def __getitem__(s, i):
        I, M = cache(); r = row[s.st[i]]
        img = Image.fromarray(I[r])
        if s.up: img = strip_detail(img)
        y = torch.from_numpy((np.asarray(M[r]) > 0).astype(np.float32)).unsqueeze(0)
        return NORM(TF.to_tensor(img)), y

SPEC = dict(channels=(64,256,512,1024,2048), decoder=(512,256,128,64))
class DoubleConv(nn.Module):
    def __init__(s,i,o):
        super().__init__()
        s.net = nn.Sequential(nn.Conv2d(i,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True),
                              nn.Conv2d(o,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True))
    def forward(s,x): return s.net(x)
class Up(nn.Module):
    def __init__(s,i,k,o):
        super().__init__(); s.reduce=nn.Conv2d(i,i//2,1,bias=False); s.conv=DoubleConv(i//2+k,o)
    def forward(s,x,skip):
        x = F.interpolate(x, size=skip.shape[-2:], mode="bilinear", align_corners=False)
        return s.conv(torch.cat([s.reduce(x), skip], 1))
class ResUNet(nn.Module):
    def __init__(s):
        super().__init__()
        base = models.resnet50(weights=None)
        c0,c1,c2,c3,c4 = SPEC["channels"]; d1,d2,d3,d4 = SPEC["decoder"]
        L = list(base.children())
        s.stem=nn.Sequential(*L[:3]); s.pool=L[3]
        s.l1,s.l2,s.l3,s.l4 = L[4],L[5],L[6],L[7]
        s.u1,s.u2,s.u3,s.u4 = Up(c4,c3,d1),Up(d1,c2,d2),Up(d2,c1,d3),Up(d3,c0,d4)
        s.head = nn.Sequential(nn.Conv2d(d4,32,3,padding=1,bias=False),
                               nn.BatchNorm2d(32), nn.ReLU(True), nn.Conv2d(32,1,1))
    def forward(s,x):
        size=x.shape[-2:]
        x0=s.stem(x); x1=s.l1(s.pool(x0)); x2=s.l2(x1); x3=s.l3(x2); x4=s.l4(x3)
        d=s.u1(x4,x3); d=s.u2(d,x2); d=s.u3(d,x1); d=s.u4(d,x0)
        return s.head(F.interpolate(d,size=size,mode="bilinear",align_corners=False))

@torch.no_grad()
def prob_768(net, x, mode):
    if mode == "full768":
        return torch.sigmoid(net(x).float())
    if mode == "whole384":
        d = F.interpolate(x, size=(TRAIN_RES,)*2, mode="bilinear", align_corners=False)
        p = torch.sigmoid(net(d).float())
        return F.interpolate(p, size=(FULL_RES,)*2, mode="bilinear", align_corners=False)
    acc = torch.zeros(x.shape[0], 1, FULL_RES, FULL_RES, device=x.device)
    cnt = torch.zeros_like(acc)
    ORIG = (0, (FULL_RES-TRAIN_RES)//2, FULL_RES-TRAIN_RES)
    for oy in ORIG:
        for ox in ORIG:
            t = x[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES]
            acc[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += torch.sigmoid(net(t).float())
            cnt[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += 1
    return acc/cnt.clamp(min=1)

THS = [round(t, 2) for t in np.arange(0.10, 0.91, 0.05)]
@torch.no_grad()
def score_per_image(bp, mode, up):
    net = ResUNet().to(DEV); net.load_state_dict(torch.load(bp, map_location=DEV)); net.eval()
    dl = DataLoader(EvalSet(CLEAN_VAL, up), BATCH_EVAL, shuffle=False,
                    num_workers=min(2, os.cpu_count() or 1), pin_memory=DEV.type == "cuda")
    n = len(CLEAN_VAL)
    per = np.zeros((n, len(THS))); Ipx = np.zeros((n, len(THS)))
    P = np.zeros((n, len(THS))); G = np.zeros(n); k = 0
    for x, y in dl:
        x, y = x.to(DEV, non_blocking=True), y.to(DEV, non_blocking=True)
        with torch.amp.autocast("cuda", enabled=DEV.type == "cuda"):
            p = prob_768(net, x, mode)
        b = x.shape[0]; yf = y.view(b, -1); pf = p.view(b, -1)
        G[k:k+b] = yf.sum(1).cpu().numpy()
        for j, t in enumerate(THS):
            pr = (pf > t).float(); it = (pr*yf).sum(1); pp = pr.sum(1)
            Ipx[k:k+b, j] = it.cpu().numpy(); P[k:k+b, j] = pp.cpu().numpy()
            per[k:k+b, j] = (it/(pp+yf.sum(1)-it).clamp(min=1)).cpu().numpy()
        k += b
    pooled = Ipx.sum(0)/np.maximum(P.sum(0)+G.sum()-Ipx.sum(0), 1)
    j = int(np.argmax(pooled))
    del net; torch.cuda.empty_cache()
    return dict(th=THS[j], pooled=float(pooled[j]), per=per[:, j])

def sha(p, chunk=1 << 24):
    h = hashlib.sha256()
    with open(p, "rb") as f:
        while (b := f.read(chunk)): h.update(b)
    return h.hexdigest()[:16]

# -------------------------- which runs to handle -----------------------
# summary.json is written only at the END of cell D, so its presence is
# the completion test -- safer than trusting a step count.
done = sorted(glob.glob(f"{STAGE}/*/summary.json"), key=os.path.getmtime)
if not done:
    raise SystemExit(f"no completed run under {STAGE}")
target = RUN_NAME or os.path.basename(os.path.dirname(done[-1]))
print(f"target run: {target}   ({len(done)} completed run(s) on Drive)\n")

man_p = f"{SAVE}/weights_manifest.json"
man = json.load(open(man_p)) if os.path.exists(man_p) else {}

for sm in done:
    d = os.path.dirname(sm); name = os.path.basename(d)
    J = json.load(open(sm))
    bp, ck = f"{d}/best.pt", f"{d}/ckpt.pt"
    if not os.path.exists(bp):
        print(f"{name}: no best.pt, skipped"); continue
    is_target = (name == target)

    # 1. per-image vectors
    npz = f"{d}/per_image.npz"
    need = True
    if os.path.exists(npz):
        try:
            z = np.load(npz, allow_pickle=True); need = list(z["stems"]) != CLEAN_VAL
        except Exception: need = True
    if need:
        t = time.time()
        r = score_per_image(bp, J["select_on"], J["input"] == "crop_768_up")
        np.savez(npz, per=r["per"], th=r["th"], pooled=r["pooled"],
                 mode=J["select_on"], stems=np.array(CLEAN_VAL))
        print(f"{name}: scored per image in {time.time()-t:.0f}s  "
              f"(mode {J['select_on']}, th {r['th']}, pooled {r['pooled']:.4f})")
    else:
        print(f"{name}: per_image.npz already current")

    # 2. archive
    zp = f"{ARCH}/{name}_results.zip"
    if is_target or not os.path.exists(zp):
        with zipfile.ZipFile(zp, "w", zipfile.ZIP_DEFLATED) as z:
            for f in ("summary.json", "history.json", "curves.png",
                      "clean_val_gain.npy", "per_image.npz"):
                p = f"{d}/{f}"
                if os.path.exists(p): z.write(p, arcname=f"{name}/{f}")
        print(f"{name}: archive {os.path.getsize(zp)/1e6:.2f} MB")

    # 3. manifest
    if is_target or name not in man:
        man[name] = dict(file=f"{name}__best.pt", bytes=os.path.getsize(bp),
                         sha256_16=sha(bp), best_val_pooled=J.get("best_val_pooled"),
                         gpu=J.get("gpu", "not recorded"), batch=J.get("batch"),
                         input=J["input"], loss=J["loss"], jitter=J["jitter"],
                         seed=J["seed"], floor=J.get("floor"))

    # 4. prune -- only a COMPLETED run, which summary.json proves
    if DELETE_CKPT and os.path.exists(ck):
        mb = os.path.getsize(ck)/1e6
        os.remove(ck); print(f"{name}: removed ckpt.pt (freed {mb:.0f} MB)")

json.dump(man, open(man_p, "w"), indent=1)
print(f"\nmanifest: {man_p}  ({len(man)} run(s))")

# ------------------------------ drive usage ----------------------------
def dsz(p):
    return sum(os.path.getsize(os.path.join(r, f))
               for r, _, fs in os.walk(p) for f in fs) if os.path.isdir(p) else 0
tot = 0
print("\nDrive usage")
for lab, p in (("stage2", STAGE), ("stage2_archive", ARCH),
               ("runs (stage 1)", f"{SAVE}/runs"), ("cache768", f"{SAVE}/cache768")):
    s = dsz(p); tot += s
    print(f"   {lab:18}{s/1e9:7.2f} GB")
print(f"   {'TRACKED TOTAL':18}{tot/1e9:7.2f} GB"
      + ("   !! approaching a 15 GB free Drive" if tot/1e9 > 11 else ""))

# ------------------------------- download ------------------------------
if DOWNLOAD:
    d = f"{STAGE}/{target}"
    loc = f"/content/{target}__best.pt"
    if not os.path.exists(loc): shutil.copyfile(f"{d}/best.pt", loc)
    print(f"\n{target}  best.pt {os.path.getsize(loc)/1e6:.1f} MB  "
          f"sha256[:16] {man[target]['sha256_16']}")
    try:
        from google.colab import files
        files.download(f"{ARCH}/{target}_results.zip")
        time.sleep(2)
        files.download(loc)
        print("downloads queued. Verify locally:")
        print(f"   shasum -a 256 {target}__best.pt | cut -c1-16"
              f"   ->  {man[target]['sha256_16']}")
    except Exception as e:
        print(f"(browser download unavailable: {type(e).__name__})")
    print(f"also on Drive:\n   {ARCH}/{target}_results.zip\n   {d}/best.pt")
```

```python
# =====================================================================
#  cell E -- score every finished Stage 2 run per image, then compare.
#
#  Cell D saves per-quartile MEANS, which cannot support a paired test.
#  This reloads each best.pt and writes the per-image IoU vector over
#  clean_val, at that run's own selected mode and threshold. No retraining.
#
#  Then, against the S-384 baseline, for each other run:
#    - paired per-image differences over the same 688 images
#    - an image-level paired bootstrap on Q0  (the PRIMARY inference:
#      688 paired images carry far more power than a 3-seed t-interval,
#      which at 3 seeds needs a seed-sd below 0.008 to clear +0.02)
#    - the same split by realized detail gain, declared in advance
#    - the PILOT rule: kill only if the mean Q0 difference is <= 0.
#      Killing anything that misses the +0.02 gate at one seed discards
#      a real +0.02 effect half the time; the <= 0 rule, 10% of the time.
#
#  A single-seed result here is a FILTER, never a result. Nothing from
#  this cell goes in the paper without seeds 43 and 44.
# =====================================================================
import os, csv, json, glob, time
import numpy as np
from PIL import Image
import torch, torch.nn as nn, torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models
import torchvision.transforms.functional as TF

SAVE, C768 = "/content/drive/MyDrive", "/content/cache768"
STAGE   = f"{SAVE}/stage2"
BASE    = "S-384_seed42"           # the baseline every run is paired against
TRAIN_RES, FULL_RES = 384, 768
BATCH_EVAL = 4
NBOOT   = 10000
FROZEN  = [0.059468, 0.138448, 0.300831]
KILL_AT = 0.0                      # pilot rule, declared before looking
try:
    from google.colab import drive; drive.mount("/content/drive", force_remount=False)
except Exception: pass
DEV = torch.device("cuda" if torch.cuda.is_available() else "cpu")

SP = json.load(open(f"{SAVE}/plantseg_splits.json"))
I768 = json.load(open(f"{C768}/index.json"))
row = I768["row_of"]
CLEAN_VAL = sorted(set(SP["clean_val"]) & set(I768["stems"]))
DSDIR = "/content/my_images/plantseg"
meta = {os.path.splitext(r["Name"])[0]: r for r in csv.DictReader(
        open(f"{DSDIR}/Metadata.csv", encoding="utf-8-sig", newline=""))}
frac = {k: float(v["Mask ratio"]) for k, v in meta.items()}
def quart(s):
    f = frac[s]
    return 0 if f <= FROZEN[0] else 1 if f <= FROZEN[1] else 2 if f <= FROZEN[2] else 3
QV = np.array([quart(s) for s in CLEAN_VAL])
res = {k: tuple(int(v) for v in meta[k]["Resolution"].lower().split("x")[:2])
       for k in CLEAN_VAL}
GAIN = np.array([min(min(res[s]), FULL_RES)/TRAIN_RES for s in CLEAN_VAL])
print(f"clean_val {len(CLEAN_VAL)}  quartiles {[int((QV==q).sum()) for q in range(4)]}")

MEAN, STD = [0.485,0.456,0.406], [0.229,0.224,0.225]
NORM = transforms.Normalize(MEAN, STD)
_mm = {}
def cache():
    if not _mm:
        _mm["i"] = np.load(I768["img_path"], mmap_mode="r")
        _mm["m"] = np.load(I768["msk_path"], mmap_mode="r")
    return _mm["i"], _mm["m"]
def strip_detail(im):
    return im.resize((TRAIN_RES,)*2, Image.BILINEAR).resize((FULL_RES,)*2, Image.BILINEAR)

class EvalSet(Dataset):
    def __init__(s, st, up): s.st, s.up = list(st), up
    def __len__(s): return len(s.st)
    def __getitem__(s, i):
        I, M = cache(); r = row[s.st[i]]
        img = Image.fromarray(I[r])
        if s.up: img = strip_detail(img)
        y = torch.from_numpy((np.asarray(M[r]) > 0).astype(np.float32)).unsqueeze(0)
        return NORM(TF.to_tensor(img)), y

SPEC = dict(channels=(64,256,512,1024,2048), decoder=(512,256,128,64))
class DoubleConv(nn.Module):
    def __init__(s,i,o):
        super().__init__()
        s.net = nn.Sequential(nn.Conv2d(i,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True),
                              nn.Conv2d(o,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True))
    def forward(s,x): return s.net(x)
class Up(nn.Module):
    def __init__(s,i,k,o):
        super().__init__(); s.reduce=nn.Conv2d(i,i//2,1,bias=False); s.conv=DoubleConv(i//2+k,o)
    def forward(s,x,skip):
        x = F.interpolate(x, size=skip.shape[-2:], mode="bilinear", align_corners=False)
        return s.conv(torch.cat([s.reduce(x), skip], 1))
class ResUNet(nn.Module):
    def __init__(s):
        super().__init__()
        base = models.resnet50(weights=None)
        c0,c1,c2,c3,c4 = SPEC["channels"]; d1,d2,d3,d4 = SPEC["decoder"]
        L = list(base.children())
        s.stem=nn.Sequential(*L[:3]); s.pool=L[3]
        s.l1,s.l2,s.l3,s.l4 = L[4],L[5],L[6],L[7]
        s.u1,s.u2,s.u3,s.u4 = Up(c4,c3,d1),Up(d1,c2,d2),Up(d2,c1,d3),Up(d3,c0,d4)
        s.head = nn.Sequential(nn.Conv2d(d4,32,3,padding=1,bias=False),
                               nn.BatchNorm2d(32), nn.ReLU(True), nn.Conv2d(32,1,1))
    def forward(s,x):
        size=x.shape[-2:]
        x0=s.stem(x); x1=s.l1(s.pool(x0)); x2=s.l2(x1); x3=s.l3(x2); x4=s.l4(x3)
        d=s.u1(x4,x3); d=s.u2(d,x2); d=s.u3(d,x1); d=s.u4(d,x0)
        return s.head(F.interpolate(d,size=size,mode="bilinear",align_corners=False))

@torch.no_grad()
def prob_768(net, x, mode):
    if mode == "full768":
        return torch.sigmoid(net(x).float())
    if mode == "whole384":
        d = F.interpolate(x, size=(TRAIN_RES,)*2, mode="bilinear", align_corners=False)
        p = torch.sigmoid(net(d).float())
        return F.interpolate(p, size=(FULL_RES,)*2, mode="bilinear", align_corners=False)
    acc = torch.zeros(x.shape[0], 1, FULL_RES, FULL_RES, device=x.device)
    cnt = torch.zeros_like(acc)
    ORIG = (0, (FULL_RES-TRAIN_RES)//2, FULL_RES-TRAIN_RES)
    for oy in ORIG:
        for ox in ORIG:
            t = x[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES]
            acc[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += torch.sigmoid(net(t).float())
            cnt[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += 1
    return acc/cnt.clamp(min=1)

THS = [round(t, 2) for t in np.arange(0.10, 0.91, 0.05)]
@torch.no_grad()
def score(bp, mode, up):
    net = ResUNet().to(DEV); net.load_state_dict(torch.load(bp, map_location=DEV)); net.eval()
    dl = DataLoader(EvalSet(CLEAN_VAL, up), BATCH_EVAL, shuffle=False,
                    num_workers=min(2, os.cpu_count() or 1), pin_memory=DEV.type == "cuda")
    n = len(CLEAN_VAL)
    per = np.zeros((n, len(THS))); Ipx = np.zeros((n, len(THS)))
    P = np.zeros((n, len(THS))); G = np.zeros(n); k = 0
    for x, y in dl:
        x, y = x.to(DEV, non_blocking=True), y.to(DEV, non_blocking=True)
        with torch.amp.autocast("cuda", enabled=DEV.type == "cuda"):
            p = prob_768(net, x, mode)
        b = x.shape[0]; yf = y.view(b, -1); pf = p.view(b, -1)
        G[k:k+b] = yf.sum(1).cpu().numpy()
        for j, t in enumerate(THS):
            pr = (pf > t).float(); it = (pr*yf).sum(1); pp = pr.sum(1)
            Ipx[k:k+b, j] = it.cpu().numpy(); P[k:k+b, j] = pp.cpu().numpy()
            per[k:k+b, j] = (it/(pp+yf.sum(1)-it).clamp(min=1)).cpu().numpy()
        k += b
    pooled = Ipx.sum(0)/np.maximum(P.sum(0)+G.sum()-Ipx.sum(0), 1)
    j = int(np.argmax(pooled))
    del net; torch.cuda.empty_cache()
    return dict(th=THS[j], pooled=float(pooled[j]), per=per[:, j])

# ------------------------- score what is finished ----------------------
runs = {}
for sm in sorted(glob.glob(f"{STAGE}/*/summary.json")):
    d = os.path.dirname(sm); name = os.path.basename(d)
    if not os.path.exists(f"{d}/best.pt"):
        print(f"   {name}: no best.pt, skipped"); continue
    J = json.load(open(sm))
    npz = f"{d}/per_image.npz"
    if os.path.exists(npz):
        z = np.load(npz, allow_pickle=True)
        if list(z["stems"]) == CLEAN_VAL:
            runs[name] = dict(per=z["per"], th=float(z["th"]), pooled=float(z["pooled"]),
                              mode=str(z["mode"]), cfg=J); print(f"   {name}: cached"); continue
    t = time.time()
    r = score(f"{d}/best.pt", J["select_on"], J["input"] == "crop_768_up")
    np.savez(npz, per=r["per"], th=r["th"], pooled=r["pooled"],
             mode=J["select_on"], stems=np.array(CLEAN_VAL))
    runs[name] = dict(**r, mode=J["select_on"], cfg=J)
    print(f"   {name}: scored in {time.time()-t:.0f}s  "
          f"mode {J['select_on']} th {r['th']} pooled {r['pooled']:.4f}")

if BASE not in runs:
    raise SystemExit(f"baseline {BASE} not scored yet -- run it first")
print(f"\n{len(runs)} run(s) scored; baseline {BASE}")

# --------------------------- paired comparison -------------------------
rng = np.random.default_rng(0)
b_per = runs[BASE]["per"]; q0 = QV == 0
bands = {"Q0 all": q0, "Q0 gain<=1.0": q0 & (GAIN <= 1.0),
         "Q0 gain>=1.5": q0 & (GAIN >= 1.5), "all images": np.ones(len(QV), bool)}

def boot(d, sel):
    v = d[sel]
    if len(v) < 5: return float("nan"), float("nan"), float("nan")
    idx = rng.integers(0, len(v), (NBOOT, len(v)))
    bs = v[idx].mean(1)
    return float(v.mean()), float(np.percentile(bs, 2.5)), float(np.percentile(bs, 97.5))

print("\n" + "=" * 88)
print(f"paired per-image IoU difference vs {BASE}, clean_val, "
      f"{NBOOT} image-level bootstraps")
print("=" * 88)
out = {}
for name, r in runs.items():
    if name == BASE: continue
    d = r["per"] - b_per
    print(f"\n{name}   ({r['cfg']['input']} / {r['cfg']['loss']}"
          f"{' / jitter' if r['cfg']['jitter'] else ''})   "
          f"pooled {r['pooled']:.4f} vs {runs[BASE]['pooled']:.4f}")
    print(f"   {'band':16}{'n':>6}{'mean diff':>12}{'95% CI':>22}")
    row = {}
    for lab, sel in bands.items():
        m, lo, hi = boot(d, sel)
        star = "  *" if (lo > 0 or hi < 0) else ""
        print(f"   {lab:16}{int(sel.sum()):>6}{m:>12.4f}"
              f"   [{lo:+.4f}, {hi:+.4f}]{star}")
        row[lab] = dict(n=int(sel.sum()), mean=m, lo=lo, hi=hi)
    mq0 = row["Q0 all"]["mean"]
    verdict = ("KILL -- Q0 moves the wrong way" if mq0 <= KILL_AT else
               "KEEP -- run seeds 43 and 44")
    print(f"   PILOT ({'<=' if mq0 <= KILL_AT else '>'} {KILL_AT:+.2f}): {verdict}")
    if row["Q0 gain>=1.5"]["n"] >= 20 and row["Q0 gain<=1.0"]["n"] >= 20:
        sep = row["Q0 gain>=1.5"]["mean"] - row["Q0 gain<=1.0"]["mean"]
        print(f"   mechanism check: high-detail minus no-detail = {sep:+.4f}"
              f"   {'(consistent with resolution)' if sep > 0 else '(not resolution)'}")
        row["mechanism_sep"] = sep
    out[name] = row

print("\n" + "=" * 88)
print("* marks a CI excluding zero. That is an image-level interval: it says the")
print("difference is not sampling noise over these 688 images, NOT that it would")
print("survive another training seed. Only seeds 43 and 44 answer that, and the")
print("gate is judged on the three-seed paired mean, not on anything here.")
json.dump(dict(baseline=BASE, kill_at=KILL_AT, nboot=NBOOT, results=out),
          open(f"{SAVE}/stage2_compare.json", "w"), indent=1, default=float)
print(f"\nwrote {SAVE}/stage2_compare.json")
```

```python
# =====================================================================
#  cell D -- Stage 2: supervised tests of components 1 and 2.
#  One cell, three switches, 25,000 steps, resume-safe. Validation only;
#  the test split is not touched until Stage 5.
#
#      INPUT_MODE in {full_384, crop_768, crop_768_up}
#      FG_LOSS    in {pooled, class_balanced, image_balanced}
#      JITTER     in {False, True}
#
#  (full_384, pooled, False) is the S-384 baseline.
#
#  crop_768_up is the control the crop statistics made necessary: uniform
#  384 crops of a 768 image carry 1.50x the Q0 lesion density of the
#  whole image, because uniform crop ORIGINS over-sample the centre (a
#  centre pixel sits in 384 of 385 origins per axis, a corner pixel in 1).
#  So crop_768 changes resolution AND density at once. crop_768_up takes
#  the 768 image, downsamples it to 384 and upsamples it back, destroying
#  exactly the detail above 384 and nothing else, then crops identically.
#  crop_768 beating crop_768_up isolates resolution; a tie says the gain
#  is crop geometry.
#
#  Everything reads ONE cache: /content/cache768 (537 labelled + 846 val).
#  The 441 cache is not needed, so prep.py need not be run this session.
# =====================================================================
import os, csv, json, math, time, random, shutil, zipfile, collections
import numpy as np
import matplotlib; matplotlib.use("Agg")
import matplotlib.pyplot as plt
from PIL import Image, ImageFilter
import torch, torch.nn as nn, torch.optim as optim, torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models
import torchvision.transforms.functional as TF

# ------------------------- the three switches --------------------------
INPUT_MODE = "full_384"        # full_384 | crop_768 | crop_768_up
FG_LOSS    = "class_balanced"          # pooled | class_balanced | image_balanced
JITTER     = False             # True only for S-768c-jit
TRAIN_SEED = 44
# -----------------------------------------------------------------------
FLOOR_FRAC = 0.01              # declared from check 0f: binds on 8.9% of
                               # non-empty Q0 crops, caps amplification 21.5x
SELECT_ON  = {"full_384": "whole384", "crop_768": "full768",
              "crop_768_up": "full768"}[INPUT_MODE]
MAX_STEPS  = 25000
EVAL_EVERY = 1000
SHARE_EVERY = 500              # per-quartile lesion-loss share
CKPT_EVERY = 2500
LR_PATIENCE = 3                # patience counts EVAL_EVERY blocks, now 1000
TRAIN_RES  = 384
FULL_RES   = 768
BATCH      = 4                 # matches the existing pipeline; batch 8
                               # doubles every run to ~2.2h and Stage 2 to ~42h
BATCH_EVAL = 4                 # 768 input on a T4
LR_ENCODER, LR_DECODER, WEIGHT_DECAY = 1e-5, 1e-4, 1e-4
FROZEN     = [0.059468, 0.138448, 0.300831]
SAVE, C768 = "/content/drive/MyDrive", "/content/cache768"
NUM_WORKERS = min(2, os.cpu_count() or 1)
FRESH      = True              # wipes this run's folder. Set back to False
                               # once it is past step 2500 so a dropped
                               # session resumes instead of restarting.

TAG = {"full_384": "384", "crop_768": "768c", "crop_768_up": "768up"}[INPUT_MODE]
if FG_LOSS == "class_balanced":  TAG += "-cb"
if FG_LOSS == "image_balanced":  TAG += "-bal"
if JITTER:                       TAG += "-jit"
RUN = f"S-{TAG}_seed{TRAIN_SEED}"
OUT = f"{SAVE}/stage2/{RUN}"
if FRESH and os.path.isdir(OUT): shutil.rmtree(OUT, ignore_errors=True)
os.makedirs(OUT, exist_ok=True)

random.seed(TRAIN_SEED); np.random.seed(TRAIN_SEED)
torch.manual_seed(TRAIN_SEED); torch.cuda.manual_seed_all(TRAIN_SEED)
torch.backends.cudnn.deterministic = True; torch.backends.cudnn.benchmark = False
DEV = torch.device("cuda" if torch.cuda.is_available() else "cpu")
try:
    from google.colab import drive; drive.mount("/content/drive", force_remount=False)
except Exception: pass
print(f"{RUN}\ndevice {DEV} {torch.cuda.get_device_name(0) if DEV.type=='cuda' else ''}")
print(f"input {INPUT_MODE} | loss {FG_LOSS} | jitter {JITTER} | "
      f"select on {SELECT_ON} | floor {FLOOR_FRAC} | batch {BATCH}")

# --------------------------- data and asserts --------------------------
SP = json.load(open(f"{SAVE}/plantseg_splits.json"))
if max(abs(a-b) for a, b in zip(SP["quartile_cuts"], FROZEN)) > 1e-6:
    raise SystemExit(f"cuts {SP['quartile_cuts']} != frozen {FROZEN}")
if {k: len(v) for k, v in SP["official_splits"].items()} != \
   {"train": 5367, "val": 846, "test": 1561}:
    raise SystemExit("split sizes are not 5367/846/1561")
I768 = json.load(open(f"{C768}/index.json"))
if I768["load_size"] != FULL_RES:
    raise SystemExit(f"cache is {I768['load_size']}, expected {FULL_RES}")
row = I768["row_of"]
LAB = sorted(I768["labelled"])
CLEAN_VAL = sorted(set(SP["clean_val"]) & set(I768["stems"]))
if len(LAB) != 537: raise SystemExit(f"{len(LAB)} labelled, expected 537")
DSDIR = "/content/my_images/plantseg"
meta = {os.path.splitext(r["Name"])[0]: r for r in csv.DictReader(
        open(f"{DSDIR}/Metadata.csv", encoding="utf-8-sig", newline=""))}
frac = {k: float(v["Mask ratio"]) for k, v in meta.items()}
def quart(s):
    f = frac[s]
    return 0 if f <= FROZEN[0] else 1 if f <= FROZEN[1] else 2 if f <= FROZEN[2] else 3
QV = np.array([quart(s) for s in CLEAN_VAL])
print(f"labelled {len(LAB)}   clean_val {len(CLEAN_VAL)} "
      f"(quartiles {[int((QV==q).sum()) for q in range(4)]})")
print("   gates are judged on clean_val, not the full 846: full val carries")
print("   17.14% near-duplicates of training images.")

MEAN, STD = [0.485,0.456,0.406], [0.229,0.224,0.225]
NORM = transforms.Normalize(MEAN, STD)
JIT  = transforms.ColorJitter(0.4, 0.4, 0.4, 0.1)
_mm = {}
def cache():
    if not _mm:
        _mm["i"] = np.load(I768["img_path"], mmap_mode="r")
        _mm["m"] = np.load(I768["msk_path"], mmap_mode="r")
    return _mm["i"], _mm["m"]

def strip_detail(im):
    """768 -> 384 -> 768. Destroys exactly the detail above 384, keeps
       geometry, size and lesion density identical."""
    return im.resize((TRAIN_RES,)*2, Image.BILINEAR).resize((FULL_RES,)*2, Image.BILINEAR)

def photometric(img):
    if random.random() < 0.8: img = JIT(img)
    if random.random() < 0.2: img = TF.to_grayscale(img, num_output_channels=3)
    if random.random() < 0.3:
        img = img.filter(ImageFilter.GaussianBlur(radius=random.uniform(0.1, 1.5)))
    return img

def geometric(img, msk):
    if random.random() < 0.5: img, msk = TF.hflip(img), TF.hflip(msk)
    if random.random() < 0.5: img, msk = TF.vflip(img), TF.vflip(msk)
    k = random.randint(0, 3)
    if k:
        img = img.rotate(90*k, expand=True, resample=Image.BILINEAR)
        msk = msk.rotate(90*k, expand=True, resample=Image.NEAREST)
    return img, msk

def rrc(img, msk, scale=(0.5, 1.0), ratio=(3/4, 4/3)):
    w, h = img.size; area = w*h
    for _ in range(10):
        t = area*random.uniform(*scale)
        ar = math.exp(random.uniform(math.log(ratio[0]), math.log(ratio[1])))
        cw, ch = int(round(math.sqrt(t*ar))), int(round(math.sqrt(t/ar)))
        if cw <= w and ch <= h:
            x, y = random.randint(0, w-cw), random.randint(0, h-ch)
            return img.crop((x, y, x+cw, y+ch)), msk.crop((x, y, x+cw, y+ch))
    return img, msk

class Labelled(Dataset):
    """full_384   : whole image, rrc + geometric, resized to 384
       crop_768   : uniform 384 window of the 768 image (the crop IS the
                    spatial augmentation, so rrc is not applied on top)
       crop_768_up: the same window of a detail-stripped 768 image"""
    def __init__(s, st): s.st = list(st)
    def __len__(s): return len(s.st)
    def __getitem__(s, i):
        I, M = cache(); r = row[s.st[i]]
        img = Image.fromarray(I[r]); msk = Image.fromarray(M[r])
        if INPUT_MODE == "crop_768_up": img = strip_detail(img)
        if INPUT_MODE == "full_384":
            img, msk = rrc(img, msk)
            img, msk = geometric(img, msk)
            img = img.resize((TRAIN_RES,)*2, Image.BILINEAR)
            msk = msk.resize((TRAIN_RES,)*2, Image.NEAREST)
        else:
            side = (random.randint(256, FULL_RES) if JITTER else TRAIN_RES)
            x = random.randint(0, FULL_RES-side); y = random.randint(0, FULL_RES-side)
            img = img.crop((x, y, x+side, y+side)); msk = msk.crop((x, y, x+side, y+side))
            img, msk = geometric(img, msk)
            if side != TRAIN_RES:
                img = img.resize((TRAIN_RES,)*2, Image.BILINEAR)
                msk = msk.resize((TRAIN_RES,)*2, Image.NEAREST)
        y_ = torch.from_numpy((np.array(msk) > 0).astype(np.float32)).unsqueeze(0)
        return NORM(TF.to_tensor(photometric(img))), y_, quart(s.st[i])

class EvalSet(Dataset):
    """input follows INPUT_MODE; the TARGET is always the 768 mask."""
    def __init__(s, st): s.st = list(st)
    def __len__(s): return len(s.st)
    def __getitem__(s, i):
        I, M = cache(); r = row[s.st[i]]
        img = Image.fromarray(I[r])
        if INPUT_MODE == "crop_768_up": img = strip_detail(img)
        y = torch.from_numpy((np.asarray(M[r]) > 0).astype(np.float32)).unsqueeze(0)
        return NORM(TF.to_tensor(img)), y            # image at 768, mask at 768

def seed_worker(_):
    s = torch.initial_seed() % 2**32; np.random.seed(s); random.seed(s)
LK = dict(num_workers=NUM_WORKERS, pin_memory=DEV.type == "cuda",
          worker_init_fn=seed_worker, persistent_workers=NUM_WORKERS > 0)
g = torch.Generator(); g.manual_seed(TRAIN_SEED)
loader_l = DataLoader(Labelled(LAB), BATCH, shuffle=True, drop_last=True,
                      generator=g, **LK)
val_loader = DataLoader(EvalSet(CLEAN_VAL), BATCH_EVAL, shuffle=False, **LK)
def cycle(dl):
    while True:
        for b in dl: yield b

# ------------------------------- model ---------------------------------
SPEC = dict(channels=(64,256,512,1024,2048), decoder=(512,256,128,64))
class DoubleConv(nn.Module):
    def __init__(s,i,o):
        super().__init__()
        s.net = nn.Sequential(nn.Conv2d(i,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True),
                              nn.Conv2d(o,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True))
    def forward(s,x): return s.net(x)
class Up(nn.Module):
    def __init__(s,i,k,o):
        super().__init__(); s.reduce=nn.Conv2d(i,i//2,1,bias=False); s.conv=DoubleConv(i//2+k,o)
    def forward(s,x,skip):
        x = F.interpolate(x, size=skip.shape[-2:], mode="bilinear", align_corners=False)
        return s.conv(torch.cat([s.reduce(x), skip], 1))
class ResUNet(nn.Module):
    def __init__(s):
        super().__init__()
        base = models.resnet50(weights=models.ResNet50_Weights.DEFAULT)
        c0,c1,c2,c3,c4 = SPEC["channels"]; d1,d2,d3,d4 = SPEC["decoder"]
        L = list(base.children())
        s.stem=nn.Sequential(*L[:3]); s.pool=L[3]
        s.l1,s.l2,s.l3,s.l4 = L[4],L[5],L[6],L[7]
        s.u1,s.u2,s.u3,s.u4 = Up(c4,c3,d1),Up(d1,c2,d2),Up(d2,c1,d3),Up(d3,c0,d4)
        s.head = nn.Sequential(nn.Conv2d(d4,32,3,padding=1,bias=False),
                               nn.BatchNorm2d(32), nn.ReLU(True), nn.Conv2d(32,1,1))
    def enc_params(s):
        for m in (s.stem,s.l1,s.l2,s.l3,s.l4): yield from m.parameters()
    def dec_params(s):
        for m in (s.u1,s.u2,s.u3,s.u4,s.head): yield from m.parameters()
    def forward(s,x):
        size=x.shape[-2:]
        x0=s.stem(x); x1=s.l1(s.pool(x0)); x2=s.l2(x1); x3=s.l3(x2); x4=s.l4(x3)
        d=s.u1(x4,x3); d=s.u2(d,x2); d=s.u3(d,x1); d=s.u4(d,x0)
        return s.head(F.interpolate(d,size=size,mode="bilinear",align_corners=False))

# ------------------------------- losses --------------------------------
bce_none = nn.BCEWithLogitsLoss(reduction="none")
def tversky(logits, tgt, eps=1.0):
    p = torch.sigmoid(logits)
    tp = (p*tgt).sum((1,2,3)); fp = (p*(1-tgt)).sum((1,2,3)); fn = ((1-p)*tgt).sum((1,2,3))
    return 1 - (tp+eps)/(tp+0.5*fp+0.5*fn+eps)

def fg_term(logits, tgt):
    """returns (per-image BCE term, per-image lesion-only contribution).
       All three forms are scaled so a uniform per-pixel loss c gives c,
       so the learning rate and the BCE/Tversky balance do not shift
       between configurations. The plan's formula writes the sum of the
       two class means; using their average is the same comparison at the
       same scale."""
    e = bce_none(logits, tgt)
    sf = tgt.sum((1,2,3)); sb = (1-tgt).sum((1,2,3))
    ef = (e*tgt).sum((1,2,3)); eb = (e*(1-tgt)).sum((1,2,3))
    if FG_LOSS == "pooled":
        return e.mean((1,2,3)), ef/tgt[0].numel()
    if FG_LOSS == "class_balanced":                 # batch-level class means
        n = logits.shape[0]
        mf = ef.sum()/sf.sum().clamp(min=1.0); mb = eb.sum()/sb.sum().clamp(min=1.0)
        return (0.5*(mf+mb)).expand(n), 0.5*ef/sf.sum().clamp(min=1.0)
    floor = FLOOR_FRAC*tgt[0].numel()               # per-image, with the floor
    tf = ef/torch.clamp(sf, min=floor)
    tb = eb/sb.clamp(min=1.0)
    return 0.5*(tf+tb), 0.5*tf

def sup_loss(logits, tgt):
    b, lesion = fg_term(logits, tgt)
    return b + tversky(logits, tgt), lesion

# --------------------------- 768 inference -----------------------------
@torch.no_grad()
def prob_768(net, x, mode):
    """x is the model's input image at 768. Returns a 768 probability map."""
    if mode == "full768":
        return torch.sigmoid(net(x).float())
    if mode == "whole384":
        d = F.interpolate(x, size=(TRAIN_RES,)*2, mode="bilinear", align_corners=False)
        p = torch.sigmoid(net(d).float())
        return F.interpolate(p, size=(FULL_RES,)*2, mode="bilinear", align_corners=False)
    # genuinely overlapping: stride 192 gives 3x3 tiles at 50% overlap.
    # Origins (0, 384) alone would ABUT, not overlap, and blending would
    # be a no-op -- the seam is exactly where a tiled model is weakest.
    acc = torch.zeros(x.shape[0], 1, FULL_RES, FULL_RES, device=x.device)
    cnt = torch.zeros_like(acc)
    ORIG = (0, (FULL_RES-TRAIN_RES)//2, FULL_RES-TRAIN_RES)
    for oy in ORIG:
        for ox in ORIG:
            t = x[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES]
            acc[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += torch.sigmoid(net(t).float())
            cnt[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += 1
    return acc/cnt.clamp(min=1)                      # blend, threshold afterwards

THS = [round(t, 2) for t in np.arange(0.10, 0.91, 0.05)]
@torch.no_grad()
def evaluate(net, mode):
    net.eval()
    nI = len(CLEAN_VAL); per = np.zeros((nI, len(THS))); P = np.zeros((nI, len(THS)))
    Ipx = np.zeros((nI, len(THS))); G = np.zeros(nI); k = 0
    for x, y in val_loader:
        x, y = x.to(DEV, non_blocking=True), y.to(DEV, non_blocking=True)
        with torch.amp.autocast("cuda", enabled=DEV.type == "cuda"):
            p = prob_768(net, x, mode)
        b = x.shape[0]; yf = y.view(b, -1)
        G[k:k+b] = yf.sum(1).cpu().numpy()
        pf = p.view(b, -1)
        for j, t in enumerate(THS):
            pr = (pf > t).float()
            it = (pr*yf).sum(1); pp = pr.sum(1)
            Ipx[k:k+b, j] = it.cpu().numpy(); P[k:k+b, j] = pp.cpu().numpy()
            per[k:k+b, j] = (it/(pp+yf.sum(1)-it).clamp(min=1)).cpu().numpy()
        k += b
    pooled = Ipx.sum(0)/np.maximum(P.sum(0)+G.sum()-Ipx.sum(0), 1)
    j = int(np.argmax(pooled)); t = THS[j]
    byq = {q: float(per[QV == q, j].mean()) for q in range(4)}
    return dict(th=t, pooled=float(pooled[j]), perimg=float(per[:, j].mean()),
                byq=byq, pred_frac=float(P[:, j].sum()/(len(CLEAN_VAL)*FULL_RES*FULL_RES)))

# --------------------------- resume or start ---------------------------
model = ResUNet().to(DEV)
opt = optim.AdamW([{"params": list(model.enc_params()), "lr": LR_ENCODER},
                   {"params": list(model.dec_params()), "lr": LR_DECODER}],
                  weight_decay=WEIGHT_DECAY)
sched = optim.lr_scheduler.ReduceLROnPlateau(opt, mode="max", patience=LR_PATIENCE, factor=0.5)
scaler = torch.amp.GradScaler("cuda")
step, best, history, shares = 0, -1.0, [], []
CK, BEST = f"{OUT}/ckpt.pt", f"{OUT}/best.pt"
if os.path.exists(CK):
    c = torch.load(CK, map_location=DEV, weights_only=False)
    model.load_state_dict(c["model"]); opt.load_state_dict(c["opt"])
    sched.load_state_dict(c["sched"]); scaler.load_state_dict(c["scaler"])
    step, best, history, shares = c["step"], c["best"], c["history"], c.get("shares", [])
    random.setstate(c["py_rng"]); np.random.set_state(c["np_rng"])
    torch.set_rng_state(c["t_rng"].cpu())
    if c.get("cuda_rng") is not None and DEV.type == "cuda":
        torch.cuda.set_rng_state(c["cuda_rng"].cpu())
    print(f"RESUMED from step {step} (best {best:.4f})")
else:
    print("starting fresh")

def save_ckpt():
    torch.save(dict(model=model.state_dict(), opt=opt.state_dict(), sched=sched.state_dict(),
                    scaler=scaler.state_dict(), step=step, best=best, history=history,
                    shares=shares, py_rng=random.getstate(), np_rng=np.random.get_state(),
                    t_rng=torch.get_rng_state(),
                    cuda_rng=torch.cuda.get_rng_state() if DEV.type == "cuda" else None), CK)

# ------------------------------- train ---------------------------------
it = cycle(loader_l)
print("=" * 76); print(f"{RUN}  --  {len(LAB)} labelled, {MAX_STEPS} steps"); print("=" * 76)
t0 = t_last = time.time()
r_loss = r_n = 0.0; q_les = np.zeros(4); q_cnt = np.zeros(4)
while step < MAX_STEPS:
    model.train()
    x, y, q = next(it)
    x, y = x.to(DEV, non_blocking=True), y.to(DEV, non_blocking=True)
    opt.zero_grad(set_to_none=True)
    with torch.amp.autocast("cuda", enabled=DEV.type == "cuda"):
        lg = model(x)
        vec, lesion = sup_loss(lg.float(), y)
        loss = vec.mean()
    scaler.scale(loss).backward(); scaler.step(opt); scaler.update()
    step += 1
    r_loss += loss.item(); r_n += 1
    lz = lesion.detach().float().cpu().numpy(); qz = q.numpy()
    for b in range(len(qz)): q_les[qz[b]] += lz[b]; q_cnt[qz[b]] += 1

    if step % SHARE_EVERY == 0:
        tot = max(q_les.sum(), 1e-12)
        sh = (q_les/tot).round(4).tolist()
        shares.append(dict(step=step, share=sh, seen=q_cnt.astype(int).tolist()))
        print(f"   [step {step}] lesion-loss share by quartile {sh}"
              f"  (images seen {q_cnt.astype(int).tolist()})")
        q_les[:] = 0; q_cnt[:] = 0

    if step % EVAL_EVERY == 0:
        m = evaluate(model, SELECT_ON)
        prev = opt.param_groups[-1]["lr"]; sched.step(m["pooled"])
        cur = opt.param_groups[-1]["lr"]
        saved = ""
        if m["pooled"] > best:
            best = m["pooled"]; torch.save(model.state_dict(), BEST); saved = "  <-- saved"
        history.append(dict(step=step, train_loss=r_loss/max(r_n, 1), lr=cur, **m))
        sps = EVAL_EVERY/(time.time()-t_last); t_last = time.time()
        print(f"step {step:6d} | loss {r_loss/max(r_n,1):.4f} | val pooled {m['pooled']:.4f} "
              f"@{m['th']} | per-img {m['perimg']:.4f} | Q0 {m['byq'][0]:.4f} "
              f"Q3 {m['byq'][3]:.4f} | best {best:.4f} | {sps:.2f} it/s | "
              f"ETA {(MAX_STEPS-step)/sps/3600:.1f}h{saved}"
              + (f"  [lr {prev:.1e}->{cur:.1e}]" if cur != prev else ""))
        r_loss = r_n = 0.0
        json.dump(dict(run=RUN, input=INPUT_MODE, loss=FG_LOSS, jitter=JITTER,
                       seed=TRAIN_SEED, floor=FLOOR_FRAC, select_on=SELECT_ON,
                       batch=BATCH, history=history, shares=shares),
                  open(f"{OUT}/history.json", "w"), indent=1)
    if step % CKPT_EVERY == 0:
        save_ckpt(); print(f"    checkpointed at step {step}")
save_ckpt()
print(f"\ntraining done in {(time.time()-t0)/3600:.2f}h, best val pooled {best:.4f}")

# ------------------- final: all three inference modes ------------------
net = ResUNet().to(DEV); net.load_state_dict(torch.load(BEST, map_location=DEV)); net.eval()
print("\n" + "=" * 76)
print("clean_val under each inference mode (best.pt was selected on "
      f"{SELECT_ON})")
print("=" * 76)
print(f"{'mode':12}{'th':>6}{'pooled':>9}{'per-img':>9}"
      f"{'Q0':>8}{'Q1':>8}{'Q2':>8}{'Q3':>8}{'Q3-Q0':>8}")
modes = ["whole384", "full768", "tiles384"]
final = {}
for md in modes:
    m = evaluate(net, md); final[md] = m
    print(f"{md:12}{m['th']:>6.2f}{m['pooled']:>9.4f}{m['perimg']:>9.4f}"
          + "".join(f"{m['byq'][q]:>8.4f}" for q in range(4))
          + f"{m['byq'][3]-m['byq'][0]:>8.4f}")

# the stratified check: does any Q0 gain track the realized detail gain?
res = {k: tuple(int(v) for v in meta[k]["Resolution"].lower().split("x")[:2])
       for k in CLEAN_VAL}
gain = np.array([min(min(res[s]), FULL_RES)/TRAIN_RES for s in CLEAN_VAL])
q0 = QV == 0
print(f"\nclean_val Q0 split by realized worst-axis detail gain at {FULL_RES}:")
for lab, sel in (("<= 1.0 (no gain)", q0 & (gain <= 1.0)),
                 ("1.0 - 1.5", q0 & (gain > 1.0) & (gain < 1.5)),
                 (">= 1.5", q0 & (gain >= 1.5))):
    print(f"   {lab:18} n={int(sel.sum()):4d}")
print("   per-image Q0 IoU within each band is in the saved json; compare it")
print("   against the matching S-384 run, paired by image.")

np.save(f"{OUT}/clean_val_gain.npy", gain)
json.dump(dict(run=RUN, input=INPUT_MODE, loss=FG_LOSS, jitter=JITTER, seed=TRAIN_SEED,
               floor=FLOOR_FRAC, select_on=SELECT_ON, batch=BATCH, best_val_pooled=best,
               clean_val=CLEAN_VAL, quartiles=QV.tolist(), final=final,
               history=history, shares=shares),
          open(f"{OUT}/summary.json", "w"), indent=1)
xs = [h["step"] for h in history]
plt.figure(figsize=(14, 4))

plt.subplot(1, 3, 1); plt.plot(xs, [h["train_loss"] for h in history]); plt.title("train loss")
plt.subplot(1, 3, 2); plt.plot(xs, [h["pooled"] for h in history]); plt.title("val pooled IoU")
plt.subplot(1, 3, 3)
for q in range(4): plt.plot(xs, [h["byq"][q] for h in history], label=f"Q{q}")
plt.legend(fontsize=7); plt.title("val per-image IoU by quartile")
plt.tight_layout(); plt.savefig(f"{OUT}/curves.png", dpi=110); plt.close()
print(f"\nwrote {OUT}/summary.json, best.pt, curves.png")
print(">>> download best.pt to local disk before the session ends <<<")
```

```python
# =====================================================================
#  cell F -- run this ONCE after every cell D training. Standalone: it
#  reads nothing from cell D's variables, so it also works in a fresh
#  session or after a runtime restart.
#
#  For the target run it:
#    1. scores it per image on clean_val  -> per_image.npz
#       (the vectors every paired test needs; cell D saves only means)
#    2. archives summary / history / curves / gain / per_image to Drive
#    3. records config, GPU and a checksum in weights_manifest.json
#    4. deletes ckpt.pt, but ONLY if the run completed. 490 MB each, and
#       19 of them would overrun a free Drive.
#    5. prints what Drive is holding
#    6. downloads the archive and best.pt
#
#  Any OTHER run still missing per_image.npz or an archive is also
#  brought up to date, without being downloaded again.
# =====================================================================
import os, csv, json, glob, time, zipfile, shutil, hashlib
import numpy as np
from PIL import Image
import torch, torch.nn as nn, torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models
import torchvision.transforms.functional as TF

RUN_NAME    = None       # None = the most recently finished run in stage2/
DELETE_CKPT = True       # False if you intend to extend a run later
DOWNLOAD    = True
SAVE, C768  = "/content/drive/MyDrive", "/content/cache768"
STAGE, ARCH = f"{SAVE}/stage2", f"{SAVE}/stage2_archive"
TRAIN_RES, FULL_RES = 384, 768
BATCH_EVAL  = 12 if torch.cuda.is_available() else 2
FROZEN      = [0.059468, 0.138448, 0.300831]
try:
    from google.colab import drive; drive.mount("/content/drive", force_remount=False)
except Exception: pass
DEV = torch.device("cuda" if torch.cuda.is_available() else "cpu")
os.makedirs(ARCH, exist_ok=True)

# ------------------------------- context -------------------------------
SP   = json.load(open(f"{SAVE}/plantseg_splits.json"))
I768 = json.load(open(f"{C768}/index.json"))
row  = I768["row_of"]
CLEAN_VAL = sorted(set(SP["clean_val"]) & set(I768["stems"]))
DSDIR = "/content/my_images/plantseg"
meta = {os.path.splitext(r["Name"])[0]: r for r in csv.DictReader(
        open(f"{DSDIR}/Metadata.csv", encoding="utf-8-sig", newline=""))}
frac = {k: float(v["Mask ratio"]) for k, v in meta.items()}
def quart(s):
    f = frac[s]
    return 0 if f <= FROZEN[0] else 1 if f <= FROZEN[1] else 2 if f <= FROZEN[2] else 3
QV = np.array([quart(s) for s in CLEAN_VAL])

MEAN, STD = [0.485,0.456,0.406], [0.229,0.224,0.225]
NORM = transforms.Normalize(MEAN, STD)
_mm = {}
def cache():
    if not _mm:
        _mm["i"] = np.load(I768["img_path"], mmap_mode="r")
        _mm["m"] = np.load(I768["msk_path"], mmap_mode="r")
    return _mm["i"], _mm["m"]
def strip_detail(im):
    return im.resize((TRAIN_RES,)*2, Image.BILINEAR).resize((FULL_RES,)*2, Image.BILINEAR)

class EvalSet(Dataset):
    def __init__(s, st, up): s.st, s.up = list(st), up
    def __len__(s): return len(s.st)
    def __getitem__(s, i):
        I, M = cache(); r = row[s.st[i]]
        img = Image.fromarray(I[r])
        if s.up: img = strip_detail(img)
        y = torch.from_numpy((np.asarray(M[r]) > 0).astype(np.float32)).unsqueeze(0)
        return NORM(TF.to_tensor(img)), y

SPEC = dict(channels=(64,256,512,1024,2048), decoder=(512,256,128,64))
class DoubleConv(nn.Module):
    def __init__(s,i,o):
        super().__init__()
        s.net = nn.Sequential(nn.Conv2d(i,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True),
                              nn.Conv2d(o,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True))
    def forward(s,x): return s.net(x)
class Up(nn.Module):
    def __init__(s,i,k,o):
        super().__init__(); s.reduce=nn.Conv2d(i,i//2,1,bias=False); s.conv=DoubleConv(i//2+k,o)
    def forward(s,x,skip):
        x = F.interpolate(x, size=skip.shape[-2:], mode="bilinear", align_corners=False)
        return s.conv(torch.cat([s.reduce(x), skip], 1))
class ResUNet(nn.Module):
    def __init__(s):
        super().__init__()
        base = models.resnet50(weights=None)
        c0,c1,c2,c3,c4 = SPEC["channels"]; d1,d2,d3,d4 = SPEC["decoder"]
        L = list(base.children())
        s.stem=nn.Sequential(*L[:3]); s.pool=L[3]
        s.l1,s.l2,s.l3,s.l4 = L[4],L[5],L[6],L[7]
        s.u1,s.u2,s.u3,s.u4 = Up(c4,c3,d1),Up(d1,c2,d2),Up(d2,c1,d3),Up(d3,c0,d4)
        s.head = nn.Sequential(nn.Conv2d(d4,32,3,padding=1,bias=False),
                               nn.BatchNorm2d(32), nn.ReLU(True), nn.Conv2d(32,1,1))
    def forward(s,x):
        size=x.shape[-2:]
        x0=s.stem(x); x1=s.l1(s.pool(x0)); x2=s.l2(x1); x3=s.l3(x2); x4=s.l4(x3)
        d=s.u1(x4,x3); d=s.u2(d,x2); d=s.u3(d,x1); d=s.u4(d,x0)
        return s.head(F.interpolate(d,size=size,mode="bilinear",align_corners=False))

@torch.no_grad()
def prob_768(net, x, mode):
    if mode == "full768":
        return torch.sigmoid(net(x).float())
    if mode == "whole384":
        d = F.interpolate(x, size=(TRAIN_RES,)*2, mode="bilinear", align_corners=False)
        p = torch.sigmoid(net(d).float())
        return F.interpolate(p, size=(FULL_RES,)*2, mode="bilinear", align_corners=False)
    acc = torch.zeros(x.shape[0], 1, FULL_RES, FULL_RES, device=x.device)
    cnt = torch.zeros_like(acc)
    ORIG = (0, (FULL_RES-TRAIN_RES)//2, FULL_RES-TRAIN_RES)
    for oy in ORIG:
        for ox in ORIG:
            t = x[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES]
            acc[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += torch.sigmoid(net(t).float())
            cnt[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += 1
    return acc/cnt.clamp(min=1)

THS = [round(t, 2) for t in np.arange(0.10, 0.91, 0.05)]
@torch.no_grad()
def score_per_image(bp, mode, up):
    net = ResUNet().to(DEV); net.load_state_dict(torch.load(bp, map_location=DEV)); net.eval()
    dl = DataLoader(EvalSet(CLEAN_VAL, up), BATCH_EVAL, shuffle=False,
                    num_workers=min(2, os.cpu_count() or 1), pin_memory=DEV.type == "cuda")
    n = len(CLEAN_VAL)
    per = np.zeros((n, len(THS))); Ipx = np.zeros((n, len(THS)))
    P = np.zeros((n, len(THS))); G = np.zeros(n); k = 0
    for x, y in dl:
        x, y = x.to(DEV, non_blocking=True), y.to(DEV, non_blocking=True)
        with torch.amp.autocast("cuda", enabled=DEV.type == "cuda"):
            p = prob_768(net, x, mode)
        b = x.shape[0]; yf = y.view(b, -1); pf = p.view(b, -1)
        G[k:k+b] = yf.sum(1).cpu().numpy()
        for j, t in enumerate(THS):
            pr = (pf > t).float(); it = (pr*yf).sum(1); pp = pr.sum(1)
            Ipx[k:k+b, j] = it.cpu().numpy(); P[k:k+b, j] = pp.cpu().numpy()
            per[k:k+b, j] = (it/(pp+yf.sum(1)-it).clamp(min=1)).cpu().numpy()
        k += b
    pooled = Ipx.sum(0)/np.maximum(P.sum(0)+G.sum()-Ipx.sum(0), 1)
    j = int(np.argmax(pooled))
    del net; torch.cuda.empty_cache()
    return dict(th=THS[j], pooled=float(pooled[j]), per=per[:, j])

def sha(p, chunk=1 << 24):
    h = hashlib.sha256()
    with open(p, "rb") as f:
        while (b := f.read(chunk)): h.update(b)
    return h.hexdigest()[:16]

# -------------------------- which runs to handle -----------------------
# summary.json is written only at the END of cell D, so its presence is
# the completion test -- safer than trusting a step count.
done = sorted(glob.glob(f"{STAGE}/*/summary.json"), key=os.path.getmtime)
if not done:
    raise SystemExit(f"no completed run under {STAGE}")
target = RUN_NAME or os.path.basename(os.path.dirname(done[-1]))
print(f"target run: {target}   ({len(done)} completed run(s) on Drive)\n")

man_p = f"{SAVE}/weights_manifest.json"
man = json.load(open(man_p)) if os.path.exists(man_p) else {}

for sm in done:
    d = os.path.dirname(sm); name = os.path.basename(d)
    J = json.load(open(sm))
    bp, ck = f"{d}/best.pt", f"{d}/ckpt.pt"
    if not os.path.exists(bp):
        print(f"{name}: no best.pt, skipped"); continue
    is_target = (name == target)

    # 1. per-image vectors
    npz = f"{d}/per_image.npz"
    need = True
    if os.path.exists(npz):
        try:
            z = np.load(npz, allow_pickle=True); need = list(z["stems"]) != CLEAN_VAL
        except Exception: need = True
    if need:
        t = time.time()
        r = score_per_image(bp, J["select_on"], J["input"] == "crop_768_up")
        np.savez(npz, per=r["per"], th=r["th"], pooled=r["pooled"],
                 mode=J["select_on"], stems=np.array(CLEAN_VAL))
        print(f"{name}: scored per image in {time.time()-t:.0f}s  "
              f"(mode {J['select_on']}, th {r['th']}, pooled {r['pooled']:.4f})")
    else:
        print(f"{name}: per_image.npz already current")

    # 2. archive
    zp = f"{ARCH}/{name}_results.zip"
    if is_target or not os.path.exists(zp):
        with zipfile.ZipFile(zp, "w", zipfile.ZIP_DEFLATED) as z:
            for f in ("summary.json", "history.json", "curves.png",
                      "clean_val_gain.npy", "per_image.npz"):
                p = f"{d}/{f}"
                if os.path.exists(p): z.write(p, arcname=f"{name}/{f}")
        print(f"{name}: archive {os.path.getsize(zp)/1e6:.2f} MB")

    # 3. manifest
    if is_target or name not in man:
        man[name] = dict(file=f"{name}__best.pt", bytes=os.path.getsize(bp),
                         sha256_16=sha(bp), best_val_pooled=J.get("best_val_pooled"),
                         gpu=J.get("gpu", "not recorded"), batch=J.get("batch"),
                         input=J["input"], loss=J["loss"], jitter=J["jitter"],
                         seed=J["seed"], floor=J.get("floor"))

    # 4. prune -- only a COMPLETED run, which summary.json proves
    if DELETE_CKPT and os.path.exists(ck):
        mb = os.path.getsize(ck)/1e6
        os.remove(ck); print(f"{name}: removed ckpt.pt (freed {mb:.0f} MB)")

json.dump(man, open(man_p, "w"), indent=1)
print(f"\nmanifest: {man_p}  ({len(man)} run(s))")

# ------------------------------ drive usage ----------------------------
def dsz(p):
    return sum(os.path.getsize(os.path.join(r, f))
               for r, _, fs in os.walk(p) for f in fs) if os.path.isdir(p) else 0
tot = 0
print("\nDrive usage")
for lab, p in (("stage2", STAGE), ("stage2_archive", ARCH),
               ("runs (stage 1)", f"{SAVE}/runs"), ("cache768", f"{SAVE}/cache768")):
    s = dsz(p); tot += s
    print(f"   {lab:18}{s/1e9:7.2f} GB")
print(f"   {'TRACKED TOTAL':18}{tot/1e9:7.2f} GB"
      + ("   !! approaching a 15 GB free Drive" if tot/1e9 > 11 else ""))

# ------------------------------- download ------------------------------
if DOWNLOAD:
    d = f"{STAGE}/{target}"
    loc = f"/content/{target}__best.pt"
    if not os.path.exists(loc): shutil.copyfile(f"{d}/best.pt", loc)
    print(f"\n{target}  best.pt {os.path.getsize(loc)/1e6:.1f} MB  "
          f"sha256[:16] {man[target]['sha256_16']}")
    try:
        from google.colab import files
        files.download(f"{ARCH}/{target}_results.zip")
        time.sleep(2)
        files.download(loc)
        print("downloads queued. Verify locally:")
        print(f"   shasum -a 256 {target}__best.pt | cut -c1-16"
              f"   ->  {man[target]['sha256_16']}")
    except Exception as e:
        print(f"(browser download unavailable: {type(e).__name__})")
    print(f"also on Drive:\n   {ARCH}/{target}_results.zip\n   {d}/best.pt")
```

```python
# =====================================================================
#  cell E -- score every finished Stage 2 run per image, then compare.
#
#  Cell D saves per-quartile MEANS, which cannot support a paired test.
#  This reloads each best.pt and writes the per-image IoU vector over
#  clean_val, at that run's own selected mode and threshold. No retraining.
#
#  Then, against the S-384 baseline, for each other run:
#    - paired per-image differences over the same 688 images
#    - an image-level paired bootstrap on Q0  (the PRIMARY inference:
#      688 paired images carry far more power than a 3-seed t-interval,
#      which at 3 seeds needs a seed-sd below 0.008 to clear +0.02)
#    - the same split by realized detail gain, declared in advance
#    - the PILOT rule: kill only if the mean Q0 difference is <= 0.
#      Killing anything that misses the +0.02 gate at one seed discards
#      a real +0.02 effect half the time; the <= 0 rule, 10% of the time.
#
#  A single-seed result here is a FILTER, never a result. Nothing from
#  this cell goes in the paper without seeds 43 and 44.
# =====================================================================
import os, csv, json, glob, time
import numpy as np
from PIL import Image
import torch, torch.nn as nn, torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models
import torchvision.transforms.functional as TF

SAVE, C768 = "/content/drive/MyDrive", "/content/cache768"
STAGE   = f"{SAVE}/stage2"
BASE    = "S-384_seed42"           # the baseline every run is paired against
TRAIN_RES, FULL_RES = 384, 768
BATCH_EVAL = 4
NBOOT   = 10000
FROZEN  = [0.059468, 0.138448, 0.300831]
KILL_AT = 0.0                      # pilot rule, declared before looking
try:
    from google.colab import drive; drive.mount("/content/drive", force_remount=False)
except Exception: pass
DEV = torch.device("cuda" if torch.cuda.is_available() else "cpu")

SP = json.load(open(f"{SAVE}/plantseg_splits.json"))
I768 = json.load(open(f"{C768}/index.json"))
row = I768["row_of"]
CLEAN_VAL = sorted(set(SP["clean_val"]) & set(I768["stems"]))
DSDIR = "/content/my_images/plantseg"
meta = {os.path.splitext(r["Name"])[0]: r for r in csv.DictReader(
        open(f"{DSDIR}/Metadata.csv", encoding="utf-8-sig", newline=""))}
frac = {k: float(v["Mask ratio"]) for k, v in meta.items()}
def quart(s):
    f = frac[s]
    return 0 if f <= FROZEN[0] else 1 if f <= FROZEN[1] else 2 if f <= FROZEN[2] else 3
QV = np.array([quart(s) for s in CLEAN_VAL])
res = {k: tuple(int(v) for v in meta[k]["Resolution"].lower().split("x")[:2])
       for k in CLEAN_VAL}
GAIN = np.array([min(min(res[s]), FULL_RES)/TRAIN_RES for s in CLEAN_VAL])
print(f"clean_val {len(CLEAN_VAL)}  quartiles {[int((QV==q).sum()) for q in range(4)]}")

MEAN, STD = [0.485,0.456,0.406], [0.229,0.224,0.225]
NORM = transforms.Normalize(MEAN, STD)
_mm = {}
def cache():
    if not _mm:
        _mm["i"] = np.load(I768["img_path"], mmap_mode="r")
        _mm["m"] = np.load(I768["msk_path"], mmap_mode="r")
    return _mm["i"], _mm["m"]
def strip_detail(im):
    return im.resize((TRAIN_RES,)*2, Image.BILINEAR).resize((FULL_RES,)*2, Image.BILINEAR)

class EvalSet(Dataset):
    def __init__(s, st, up): s.st, s.up = list(st), up
    def __len__(s): return len(s.st)
    def __getitem__(s, i):
        I, M = cache(); r = row[s.st[i]]
        img = Image.fromarray(I[r])
        if s.up: img = strip_detail(img)
        y = torch.from_numpy((np.asarray(M[r]) > 0).astype(np.float32)).unsqueeze(0)
        return NORM(TF.to_tensor(img)), y

SPEC = dict(channels=(64,256,512,1024,2048), decoder=(512,256,128,64))
class DoubleConv(nn.Module):
    def __init__(s,i,o):
        super().__init__()
        s.net = nn.Sequential(nn.Conv2d(i,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True),
                              nn.Conv2d(o,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True))
    def forward(s,x): return s.net(x)
class Up(nn.Module):
    def __init__(s,i,k,o):
        super().__init__(); s.reduce=nn.Conv2d(i,i//2,1,bias=False); s.conv=DoubleConv(i//2+k,o)
    def forward(s,x,skip):
        x = F.interpolate(x, size=skip.shape[-2:], mode="bilinear", align_corners=False)
        return s.conv(torch.cat([s.reduce(x), skip], 1))
class ResUNet(nn.Module):
    def __init__(s):
        super().__init__()
        base = models.resnet50(weights=None)
        c0,c1,c2,c3,c4 = SPEC["channels"]; d1,d2,d3,d4 = SPEC["decoder"]
        L = list(base.children())
        s.stem=nn.Sequential(*L[:3]); s.pool=L[3]
        s.l1,s.l2,s.l3,s.l4 = L[4],L[5],L[6],L[7]
        s.u1,s.u2,s.u3,s.u4 = Up(c4,c3,d1),Up(d1,c2,d2),Up(d2,c1,d3),Up(d3,c0,d4)
        s.head = nn.Sequential(nn.Conv2d(d4,32,3,padding=1,bias=False),
                               nn.BatchNorm2d(32), nn.ReLU(True), nn.Conv2d(32,1,1))
    def forward(s,x):
        size=x.shape[-2:]
        x0=s.stem(x); x1=s.l1(s.pool(x0)); x2=s.l2(x1); x3=s.l3(x2); x4=s.l4(x3)
        d=s.u1(x4,x3); d=s.u2(d,x2); d=s.u3(d,x1); d=s.u4(d,x0)
        return s.head(F.interpolate(d,size=size,mode="bilinear",align_corners=False))

@torch.no_grad()
def prob_768(net, x, mode):
    if mode == "full768":
        return torch.sigmoid(net(x).float())
    if mode == "whole384":
        d = F.interpolate(x, size=(TRAIN_RES,)*2, mode="bilinear", align_corners=False)
        p = torch.sigmoid(net(d).float())
        return F.interpolate(p, size=(FULL_RES,)*2, mode="bilinear", align_corners=False)
    acc = torch.zeros(x.shape[0], 1, FULL_RES, FULL_RES, device=x.device)
    cnt = torch.zeros_like(acc)
    ORIG = (0, (FULL_RES-TRAIN_RES)//2, FULL_RES-TRAIN_RES)
    for oy in ORIG:
        for ox in ORIG:
            t = x[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES]
            acc[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += torch.sigmoid(net(t).float())
            cnt[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += 1
    return acc/cnt.clamp(min=1)

THS = [round(t, 2) for t in np.arange(0.10, 0.91, 0.05)]
@torch.no_grad()
def score(bp, mode, up):
    net = ResUNet().to(DEV); net.load_state_dict(torch.load(bp, map_location=DEV)); net.eval()
    dl = DataLoader(EvalSet(CLEAN_VAL, up), BATCH_EVAL, shuffle=False,
                    num_workers=min(2, os.cpu_count() or 1), pin_memory=DEV.type == "cuda")
    n = len(CLEAN_VAL)
    per = np.zeros((n, len(THS))); Ipx = np.zeros((n, len(THS)))
    P = np.zeros((n, len(THS))); G = np.zeros(n); k = 0
    for x, y in dl:
        x, y = x.to(DEV, non_blocking=True), y.to(DEV, non_blocking=True)
        with torch.amp.autocast("cuda", enabled=DEV.type == "cuda"):
            p = prob_768(net, x, mode)
        b = x.shape[0]; yf = y.view(b, -1); pf = p.view(b, -1)
        G[k:k+b] = yf.sum(1).cpu().numpy()
        for j, t in enumerate(THS):
            pr = (pf > t).float(); it = (pr*yf).sum(1); pp = pr.sum(1)
            Ipx[k:k+b, j] = it.cpu().numpy(); P[k:k+b, j] = pp.cpu().numpy()
            per[k:k+b, j] = (it/(pp+yf.sum(1)-it).clamp(min=1)).cpu().numpy()
        k += b
    pooled = Ipx.sum(0)/np.maximum(P.sum(0)+G.sum()-Ipx.sum(0), 1)
    j = int(np.argmax(pooled))
    del net; torch.cuda.empty_cache()
    return dict(th=THS[j], pooled=float(pooled[j]), per=per[:, j])

# ------------------------- score what is finished ----------------------
runs = {}
for sm in sorted(glob.glob(f"{STAGE}/*/summary.json")):
    d = os.path.dirname(sm); name = os.path.basename(d)
    if not os.path.exists(f"{d}/best.pt"):
        print(f"   {name}: no best.pt, skipped"); continue
    J = json.load(open(sm))
    npz = f"{d}/per_image.npz"
    if os.path.exists(npz):
        z = np.load(npz, allow_pickle=True)
        if list(z["stems"]) == CLEAN_VAL:
            runs[name] = dict(per=z["per"], th=float(z["th"]), pooled=float(z["pooled"]),
                              mode=str(z["mode"]), cfg=J); print(f"   {name}: cached"); continue
    t = time.time()
    r = score(f"{d}/best.pt", J["select_on"], J["input"] == "crop_768_up")
    np.savez(npz, per=r["per"], th=r["th"], pooled=r["pooled"],
             mode=J["select_on"], stems=np.array(CLEAN_VAL))
    runs[name] = dict(**r, mode=J["select_on"], cfg=J)
    print(f"   {name}: scored in {time.time()-t:.0f}s  "
          f"mode {J['select_on']} th {r['th']} pooled {r['pooled']:.4f}")

if BASE not in runs:
    raise SystemExit(f"baseline {BASE} not scored yet -- run it first")
print(f"\n{len(runs)} run(s) scored; baseline {BASE}")

# --------------------------- paired comparison -------------------------
rng = np.random.default_rng(0)
b_per = runs[BASE]["per"]; q0 = QV == 0
bands = {"Q0 all": q0, "Q0 gain<=1.0": q0 & (GAIN <= 1.0),
         "Q0 gain>=1.5": q0 & (GAIN >= 1.5), "all images": np.ones(len(QV), bool)}

def boot(d, sel):
    v = d[sel]
    if len(v) < 5: return float("nan"), float("nan"), float("nan")
    idx = rng.integers(0, len(v), (NBOOT, len(v)))
    bs = v[idx].mean(1)
    return float(v.mean()), float(np.percentile(bs, 2.5)), float(np.percentile(bs, 97.5))

print("\n" + "=" * 88)
print(f"paired per-image IoU difference vs {BASE}, clean_val, "
      f"{NBOOT} image-level bootstraps")
print("=" * 88)
out = {}
for name, r in runs.items():
    if name == BASE: continue
    d = r["per"] - b_per
    print(f"\n{name}   ({r['cfg']['input']} / {r['cfg']['loss']}"
          f"{' / jitter' if r['cfg']['jitter'] else ''})   "
          f"pooled {r['pooled']:.4f} vs {runs[BASE]['pooled']:.4f}")
    print(f"   {'band':16}{'n':>6}{'mean diff':>12}{'95% CI':>22}")
    row = {}
    for lab, sel in bands.items():
        m, lo, hi = boot(d, sel)
        star = "  *" if (lo > 0 or hi < 0) else ""
        print(f"   {lab:16}{int(sel.sum()):>6}{m:>12.4f}"
              f"   [{lo:+.4f}, {hi:+.4f}]{star}")
        row[lab] = dict(n=int(sel.sum()), mean=m, lo=lo, hi=hi)
    mq0 = row["Q0 all"]["mean"]
    verdict = ("KILL -- Q0 moves the wrong way" if mq0 <= KILL_AT else
               "KEEP -- run seeds 43 and 44")
    print(f"   PILOT ({'<=' if mq0 <= KILL_AT else '>'} {KILL_AT:+.2f}): {verdict}")
    if row["Q0 gain>=1.5"]["n"] >= 20 and row["Q0 gain<=1.0"]["n"] >= 20:
        sep = row["Q0 gain>=1.5"]["mean"] - row["Q0 gain<=1.0"]["mean"]
        print(f"   mechanism check: high-detail minus no-detail = {sep:+.4f}"
              f"   {'(consistent with resolution)' if sep > 0 else '(not resolution)'}")
        row["mechanism_sep"] = sep
    out[name] = row

print("\n" + "=" * 88)
print("* marks a CI excluding zero. That is an image-level interval: it says the")
print("difference is not sampling noise over these 688 images, NOT that it would")
print("survive another training seed. Only seeds 43 and 44 answer that, and the")
print("gate is judged on the three-seed paired mean, not on anything here.")
json.dump(dict(baseline=BASE, kill_at=KILL_AT, nboot=NBOOT, results=out),
          open(f"{SAVE}/stage2_compare.json", "w"), indent=1, default=float)
print(f"\nwrote {SAVE}/stage2_compare.json")
```

```python
# =====================================================================
#  cell D -- Stage 2: supervised tests of components 1 and 2.
#  One cell, three switches, 25,000 steps, resume-safe. Validation only;
#  the test split is not touched until Stage 5.
#
#      INPUT_MODE in {full_384, crop_768, crop_768_up}
#      FG_LOSS    in {pooled, class_balanced, image_balanced}
#      JITTER     in {False, True}
#
#  (full_384, pooled, False) is the S-384 baseline.
#
#  crop_768_up is the control the crop statistics made necessary: uniform
#  384 crops of a 768 image carry 1.50x the Q0 lesion density of the
#  whole image, because uniform crop ORIGINS over-sample the centre (a
#  centre pixel sits in 384 of 385 origins per axis, a corner pixel in 1).
#  So crop_768 changes resolution AND density at once. crop_768_up takes
#  the 768 image, downsamples it to 384 and upsamples it back, destroying
#  exactly the detail above 384 and nothing else, then crops identically.
#  crop_768 beating crop_768_up isolates resolution; a tie says the gain
#  is crop geometry.
#
#  Everything reads ONE cache: /content/cache768 (537 labelled + 846 val).
#  The 441 cache is not needed, so prep.py need not be run this session.
# =====================================================================
import os, csv, json, math, time, random, shutil, zipfile, collections
import numpy as np
import matplotlib; matplotlib.use("Agg")
import matplotlib.pyplot as plt
from PIL import Image, ImageFilter
import torch, torch.nn as nn, torch.optim as optim, torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models
import torchvision.transforms.functional as TF

# ------------------------- the three switches --------------------------
INPUT_MODE = "full_384"        # full_384 | crop_768 | crop_768_up
FG_LOSS    = "pooled"          # pooled | class_balanced | image_balanced
JITTER     = False             # True only for S-768c-jit
TRAIN_SEED = 43
# -----------------------------------------------------------------------
FLOOR_FRAC = 0.01              # declared from check 0f: binds on 8.9% of
                               # non-empty Q0 crops, caps amplification 21.5x
SELECT_ON  = {"full_384": "whole384", "crop_768": "full768",
              "crop_768_up": "full768"}[INPUT_MODE]
MAX_STEPS  = 25000
EVAL_EVERY = 1000
SHARE_EVERY = 500              # per-quartile lesion-loss share
CKPT_EVERY = 2500
LR_PATIENCE = 3                # patience counts EVAL_EVERY blocks, now 1000
TRAIN_RES  = 384
FULL_RES   = 768
BATCH      = 4                 # matches the existing pipeline; batch 8
                               # doubles every run to ~2.2h and Stage 2 to ~42h
BATCH_EVAL = 4                 # 768 input on a T4
LR_ENCODER, LR_DECODER, WEIGHT_DECAY = 1e-5, 1e-4, 1e-4
FROZEN     = [0.059468, 0.138448, 0.300831]
SAVE, C768 = "/content/drive/MyDrive", "/content/cache768"
NUM_WORKERS = min(2, os.cpu_count() or 1)
FRESH      = True              # wipes this run's folder. Set back to False
                               # once it is past step 2500 so a dropped
                               # session resumes instead of restarting.

TAG = {"full_384": "384", "crop_768": "768c", "crop_768_up": "768up"}[INPUT_MODE]
if FG_LOSS == "class_balanced":  TAG += "-cb"
if FG_LOSS == "image_balanced":  TAG += "-bal"
if JITTER:                       TAG += "-jit"
RUN = f"S-{TAG}_seed{TRAIN_SEED}"
OUT = f"{SAVE}/stage2/{RUN}"
if FRESH and os.path.isdir(OUT): shutil.rmtree(OUT, ignore_errors=True)
os.makedirs(OUT, exist_ok=True)

random.seed(TRAIN_SEED); np.random.seed(TRAIN_SEED)
torch.manual_seed(TRAIN_SEED); torch.cuda.manual_seed_all(TRAIN_SEED)
torch.backends.cudnn.deterministic = True; torch.backends.cudnn.benchmark = False
DEV = torch.device("cuda" if torch.cuda.is_available() else "cpu")
try:
    from google.colab import drive; drive.mount("/content/drive", force_remount=False)
except Exception: pass
print(f"{RUN}\ndevice {DEV} {torch.cuda.get_device_name(0) if DEV.type=='cuda' else ''}")
print(f"input {INPUT_MODE} | loss {FG_LOSS} | jitter {JITTER} | "
      f"select on {SELECT_ON} | floor {FLOOR_FRAC} | batch {BATCH}")

# --------------------------- data and asserts --------------------------
SP = json.load(open(f"{SAVE}/plantseg_splits.json"))
if max(abs(a-b) for a, b in zip(SP["quartile_cuts"], FROZEN)) > 1e-6:
    raise SystemExit(f"cuts {SP['quartile_cuts']} != frozen {FROZEN}")
if {k: len(v) for k, v in SP["official_splits"].items()} != \
   {"train": 5367, "val": 846, "test": 1561}:
    raise SystemExit("split sizes are not 5367/846/1561")
I768 = json.load(open(f"{C768}/index.json"))
if I768["load_size"] != FULL_RES:
    raise SystemExit(f"cache is {I768['load_size']}, expected {FULL_RES}")
row = I768["row_of"]
LAB = sorted(I768["labelled"])
CLEAN_VAL = sorted(set(SP["clean_val"]) & set(I768["stems"]))
if len(LAB) != 537: raise SystemExit(f"{len(LAB)} labelled, expected 537")
DSDIR = "/content/my_images/plantseg"
meta = {os.path.splitext(r["Name"])[0]: r for r in csv.DictReader(
        open(f"{DSDIR}/Metadata.csv", encoding="utf-8-sig", newline=""))}
frac = {k: float(v["Mask ratio"]) for k, v in meta.items()}
def quart(s):
    f = frac[s]
    return 0 if f <= FROZEN[0] else 1 if f <= FROZEN[1] else 2 if f <= FROZEN[2] else 3
QV = np.array([quart(s) for s in CLEAN_VAL])
print(f"labelled {len(LAB)}   clean_val {len(CLEAN_VAL)} "
      f"(quartiles {[int((QV==q).sum()) for q in range(4)]})")
print("   gates are judged on clean_val, not the full 846: full val carries")
print("   17.14% near-duplicates of training images.")

MEAN, STD = [0.485,0.456,0.406], [0.229,0.224,0.225]
NORM = transforms.Normalize(MEAN, STD)
JIT  = transforms.ColorJitter(0.4, 0.4, 0.4, 0.1)
_mm = {}
def cache():
    if not _mm:
        _mm["i"] = np.load(I768["img_path"], mmap_mode="r")
        _mm["m"] = np.load(I768["msk_path"], mmap_mode="r")
    return _mm["i"], _mm["m"]

def strip_detail(im):
    """768 -> 384 -> 768. Destroys exactly the detail above 384, keeps
       geometry, size and lesion density identical."""
    return im.resize((TRAIN_RES,)*2, Image.BILINEAR).resize((FULL_RES,)*2, Image.BILINEAR)

def photometric(img):
    if random.random() < 0.8: img = JIT(img)
    if random.random() < 0.2: img = TF.to_grayscale(img, num_output_channels=3)
    if random.random() < 0.3:
        img = img.filter(ImageFilter.GaussianBlur(radius=random.uniform(0.1, 1.5)))
    return img

def geometric(img, msk):
    if random.random() < 0.5: img, msk = TF.hflip(img), TF.hflip(msk)
    if random.random() < 0.5: img, msk = TF.vflip(img), TF.vflip(msk)
    k = random.randint(0, 3)
    if k:
        img = img.rotate(90*k, expand=True, resample=Image.BILINEAR)
        msk = msk.rotate(90*k, expand=True, resample=Image.NEAREST)
    return img, msk

def rrc(img, msk, scale=(0.5, 1.0), ratio=(3/4, 4/3)):
    w, h = img.size; area = w*h
    for _ in range(10):
        t = area*random.uniform(*scale)
        ar = math.exp(random.uniform(math.log(ratio[0]), math.log(ratio[1])))
        cw, ch = int(round(math.sqrt(t*ar))), int(round(math.sqrt(t/ar)))
        if cw <= w and ch <= h:
            x, y = random.randint(0, w-cw), random.randint(0, h-ch)
            return img.crop((x, y, x+cw, y+ch)), msk.crop((x, y, x+cw, y+ch))
    return img, msk

class Labelled(Dataset):
    """full_384   : whole image, rrc + geometric, resized to 384
       crop_768   : uniform 384 window of the 768 image (the crop IS the
                    spatial augmentation, so rrc is not applied on top)
       crop_768_up: the same window of a detail-stripped 768 image"""
    def __init__(s, st): s.st = list(st)
    def __len__(s): return len(s.st)
    def __getitem__(s, i):
        I, M = cache(); r = row[s.st[i]]
        img = Image.fromarray(I[r]); msk = Image.fromarray(M[r])
        if INPUT_MODE == "crop_768_up": img = strip_detail(img)
        if INPUT_MODE == "full_384":
            img, msk = rrc(img, msk)
            img, msk = geometric(img, msk)
            img = img.resize((TRAIN_RES,)*2, Image.BILINEAR)
            msk = msk.resize((TRAIN_RES,)*2, Image.NEAREST)
        else:
            side = (random.randint(256, FULL_RES) if JITTER else TRAIN_RES)
            x = random.randint(0, FULL_RES-side); y = random.randint(0, FULL_RES-side)
            img = img.crop((x, y, x+side, y+side)); msk = msk.crop((x, y, x+side, y+side))
            img, msk = geometric(img, msk)
            if side != TRAIN_RES:
                img = img.resize((TRAIN_RES,)*2, Image.BILINEAR)
                msk = msk.resize((TRAIN_RES,)*2, Image.NEAREST)
        y_ = torch.from_numpy((np.array(msk) > 0).astype(np.float32)).unsqueeze(0)
        return NORM(TF.to_tensor(photometric(img))), y_, quart(s.st[i])

class EvalSet(Dataset):
    """input follows INPUT_MODE; the TARGET is always the 768 mask."""
    def __init__(s, st): s.st = list(st)
    def __len__(s): return len(s.st)
    def __getitem__(s, i):
        I, M = cache(); r = row[s.st[i]]
        img = Image.fromarray(I[r])
        if INPUT_MODE == "crop_768_up": img = strip_detail(img)
        y = torch.from_numpy((np.asarray(M[r]) > 0).astype(np.float32)).unsqueeze(0)
        return NORM(TF.to_tensor(img)), y            # image at 768, mask at 768

def seed_worker(_):
    s = torch.initial_seed() % 2**32; np.random.seed(s); random.seed(s)
LK = dict(num_workers=NUM_WORKERS, pin_memory=DEV.type == "cuda",
          worker_init_fn=seed_worker, persistent_workers=NUM_WORKERS > 0)
g = torch.Generator(); g.manual_seed(TRAIN_SEED)
loader_l = DataLoader(Labelled(LAB), BATCH, shuffle=True, drop_last=True,
                      generator=g, **LK)
val_loader = DataLoader(EvalSet(CLEAN_VAL), BATCH_EVAL, shuffle=False, **LK)
def cycle(dl):
    while True:
        for b in dl: yield b

# ------------------------------- model ---------------------------------
SPEC = dict(channels=(64,256,512,1024,2048), decoder=(512,256,128,64))
class DoubleConv(nn.Module):
    def __init__(s,i,o):
        super().__init__()
        s.net = nn.Sequential(nn.Conv2d(i,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True),
                              nn.Conv2d(o,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True))
    def forward(s,x): return s.net(x)
class Up(nn.Module):
    def __init__(s,i,k,o):
        super().__init__(); s.reduce=nn.Conv2d(i,i//2,1,bias=False); s.conv=DoubleConv(i//2+k,o)
    def forward(s,x,skip):
        x = F.interpolate(x, size=skip.shape[-2:], mode="bilinear", align_corners=False)
        return s.conv(torch.cat([s.reduce(x), skip], 1))
class ResUNet(nn.Module):
    def __init__(s):
        super().__init__()
        base = models.resnet50(weights=models.ResNet50_Weights.DEFAULT)
        c0,c1,c2,c3,c4 = SPEC["channels"]; d1,d2,d3,d4 = SPEC["decoder"]
        L = list(base.children())
        s.stem=nn.Sequential(*L[:3]); s.pool=L[3]
        s.l1,s.l2,s.l3,s.l4 = L[4],L[5],L[6],L[7]
        s.u1,s.u2,s.u3,s.u4 = Up(c4,c3,d1),Up(d1,c2,d2),Up(d2,c1,d3),Up(d3,c0,d4)
        s.head = nn.Sequential(nn.Conv2d(d4,32,3,padding=1,bias=False),
                               nn.BatchNorm2d(32), nn.ReLU(True), nn.Conv2d(32,1,1))
    def enc_params(s):
        for m in (s.stem,s.l1,s.l2,s.l3,s.l4): yield from m.parameters()
    def dec_params(s):
        for m in (s.u1,s.u2,s.u3,s.u4,s.head): yield from m.parameters()
    def forward(s,x):
        size=x.shape[-2:]
        x0=s.stem(x); x1=s.l1(s.pool(x0)); x2=s.l2(x1); x3=s.l3(x2); x4=s.l4(x3)
        d=s.u1(x4,x3); d=s.u2(d,x2); d=s.u3(d,x1); d=s.u4(d,x0)
        return s.head(F.interpolate(d,size=size,mode="bilinear",align_corners=False))

# ------------------------------- losses --------------------------------
bce_none = nn.BCEWithLogitsLoss(reduction="none")
def tversky(logits, tgt, eps=1.0):
    p = torch.sigmoid(logits)
    tp = (p*tgt).sum((1,2,3)); fp = (p*(1-tgt)).sum((1,2,3)); fn = ((1-p)*tgt).sum((1,2,3))
    return 1 - (tp+eps)/(tp+0.5*fp+0.5*fn+eps)

def fg_term(logits, tgt):
    """returns (per-image BCE term, per-image lesion-only contribution).
       All three forms are scaled so a uniform per-pixel loss c gives c,
       so the learning rate and the BCE/Tversky balance do not shift
       between configurations. The plan's formula writes the sum of the
       two class means; using their average is the same comparison at the
       same scale."""
    e = bce_none(logits, tgt)
    sf = tgt.sum((1,2,3)); sb = (1-tgt).sum((1,2,3))
    ef = (e*tgt).sum((1,2,3)); eb = (e*(1-tgt)).sum((1,2,3))
    if FG_LOSS == "pooled":
        return e.mean((1,2,3)), ef/tgt[0].numel()
    if FG_LOSS == "class_balanced":                 # batch-level class means
        n = logits.shape[0]
        mf = ef.sum()/sf.sum().clamp(min=1.0); mb = eb.sum()/sb.sum().clamp(min=1.0)
        return (0.5*(mf+mb)).expand(n), 0.5*ef/sf.sum().clamp(min=1.0)
    floor = FLOOR_FRAC*tgt[0].numel()               # per-image, with the floor
    tf = ef/torch.clamp(sf, min=floor)
    tb = eb/sb.clamp(min=1.0)
    return 0.5*(tf+tb), 0.5*tf

def sup_loss(logits, tgt):
    b, lesion = fg_term(logits, tgt)
    return b + tversky(logits, tgt), lesion

# --------------------------- 768 inference -----------------------------
@torch.no_grad()
def prob_768(net, x, mode):
    """x is the model's input image at 768. Returns a 768 probability map."""
    if mode == "full768":
        return torch.sigmoid(net(x).float())
    if mode == "whole384":
        d = F.interpolate(x, size=(TRAIN_RES,)*2, mode="bilinear", align_corners=False)
        p = torch.sigmoid(net(d).float())
        return F.interpolate(p, size=(FULL_RES,)*2, mode="bilinear", align_corners=False)
    # genuinely overlapping: stride 192 gives 3x3 tiles at 50% overlap.
    # Origins (0, 384) alone would ABUT, not overlap, and blending would
    # be a no-op -- the seam is exactly where a tiled model is weakest.
    acc = torch.zeros(x.shape[0], 1, FULL_RES, FULL_RES, device=x.device)
    cnt = torch.zeros_like(acc)
    ORIG = (0, (FULL_RES-TRAIN_RES)//2, FULL_RES-TRAIN_RES)
    for oy in ORIG:
        for ox in ORIG:
            t = x[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES]
            acc[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += torch.sigmoid(net(t).float())
            cnt[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += 1
    return acc/cnt.clamp(min=1)                      # blend, threshold afterwards

THS = [round(t, 2) for t in np.arange(0.10, 0.91, 0.05)]
@torch.no_grad()
def evaluate(net, mode):
    net.eval()
    nI = len(CLEAN_VAL); per = np.zeros((nI, len(THS))); P = np.zeros((nI, len(THS)))
    Ipx = np.zeros((nI, len(THS))); G = np.zeros(nI); k = 0
    for x, y in val_loader:
        x, y = x.to(DEV, non_blocking=True), y.to(DEV, non_blocking=True)
        with torch.amp.autocast("cuda", enabled=DEV.type == "cuda"):
            p = prob_768(net, x, mode)
        b = x.shape[0]; yf = y.view(b, -1)
        G[k:k+b] = yf.sum(1).cpu().numpy()
        pf = p.view(b, -1)
        for j, t in enumerate(THS):
            pr = (pf > t).float()
            it = (pr*yf).sum(1); pp = pr.sum(1)
            Ipx[k:k+b, j] = it.cpu().numpy(); P[k:k+b, j] = pp.cpu().numpy()
            per[k:k+b, j] = (it/(pp+yf.sum(1)-it).clamp(min=1)).cpu().numpy()
        k += b
    pooled = Ipx.sum(0)/np.maximum(P.sum(0)+G.sum()-Ipx.sum(0), 1)
    j = int(np.argmax(pooled)); t = THS[j]
    byq = {q: float(per[QV == q, j].mean()) for q in range(4)}
    return dict(th=t, pooled=float(pooled[j]), perimg=float(per[:, j].mean()),
                byq=byq, pred_frac=float(P[:, j].sum()/(len(CLEAN_VAL)*FULL_RES*FULL_RES)))

# --------------------------- resume or start ---------------------------
model = ResUNet().to(DEV)
opt = optim.AdamW([{"params": list(model.enc_params()), "lr": LR_ENCODER},
                   {"params": list(model.dec_params()), "lr": LR_DECODER}],
                  weight_decay=WEIGHT_DECAY)
sched = optim.lr_scheduler.ReduceLROnPlateau(opt, mode="max", patience=LR_PATIENCE, factor=0.5)
scaler = torch.amp.GradScaler("cuda")
step, best, history, shares = 0, -1.0, [], []
CK, BEST = f"{OUT}/ckpt.pt", f"{OUT}/best.pt"
if os.path.exists(CK):
    c = torch.load(CK, map_location=DEV, weights_only=False)
    model.load_state_dict(c["model"]); opt.load_state_dict(c["opt"])
    sched.load_state_dict(c["sched"]); scaler.load_state_dict(c["scaler"])
    step, best, history, shares = c["step"], c["best"], c["history"], c.get("shares", [])
    random.setstate(c["py_rng"]); np.random.set_state(c["np_rng"])
    torch.set_rng_state(c["t_rng"].cpu())
    if c.get("cuda_rng") is not None and DEV.type == "cuda":
        torch.cuda.set_rng_state(c["cuda_rng"].cpu())
    print(f"RESUMED from step {step} (best {best:.4f})")
else:
    print("starting fresh")

def save_ckpt():
    torch.save(dict(model=model.state_dict(), opt=opt.state_dict(), sched=sched.state_dict(),
                    scaler=scaler.state_dict(), step=step, best=best, history=history,
                    shares=shares, py_rng=random.getstate(), np_rng=np.random.get_state(),
                    t_rng=torch.get_rng_state(),
                    cuda_rng=torch.cuda.get_rng_state() if DEV.type == "cuda" else None), CK)

# ------------------------------- train ---------------------------------
it = cycle(loader_l)
print("=" * 76); print(f"{RUN}  --  {len(LAB)} labelled, {MAX_STEPS} steps"); print("=" * 76)
t0 = t_last = time.time()
r_loss = r_n = 0.0; q_les = np.zeros(4); q_cnt = np.zeros(4)
while step < MAX_STEPS:
    model.train()
    x, y, q = next(it)
    x, y = x.to(DEV, non_blocking=True), y.to(DEV, non_blocking=True)
    opt.zero_grad(set_to_none=True)
    with torch.amp.autocast("cuda", enabled=DEV.type == "cuda"):
        lg = model(x)
        vec, lesion = sup_loss(lg.float(), y)
        loss = vec.mean()
    scaler.scale(loss).backward(); scaler.step(opt); scaler.update()
    step += 1
    r_loss += loss.item(); r_n += 1
    lz = lesion.detach().float().cpu().numpy(); qz = q.numpy()
    for b in range(len(qz)): q_les[qz[b]] += lz[b]; q_cnt[qz[b]] += 1

    if step % SHARE_EVERY == 0:
        tot = max(q_les.sum(), 1e-12)
        sh = (q_les/tot).round(4).tolist()
        shares.append(dict(step=step, share=sh, seen=q_cnt.astype(int).tolist()))
        print(f"   [step {step}] lesion-loss share by quartile {sh}"
              f"  (images seen {q_cnt.astype(int).tolist()})")
        q_les[:] = 0; q_cnt[:] = 0

    if step % EVAL_EVERY == 0:
        m = evaluate(model, SELECT_ON)
        prev = opt.param_groups[-1]["lr"]; sched.step(m["pooled"])
        cur = opt.param_groups[-1]["lr"]
        saved = ""
        if m["pooled"] > best:
            best = m["pooled"]; torch.save(model.state_dict(), BEST); saved = "  <-- saved"
        history.append(dict(step=step, train_loss=r_loss/max(r_n, 1), lr=cur, **m))
        sps = EVAL_EVERY/(time.time()-t_last); t_last = time.time()
        print(f"step {step:6d} | loss {r_loss/max(r_n,1):.4f} | val pooled {m['pooled']:.4f} "
              f"@{m['th']} | per-img {m['perimg']:.4f} | Q0 {m['byq'][0]:.4f} "
              f"Q3 {m['byq'][3]:.4f} | best {best:.4f} | {sps:.2f} it/s | "
              f"ETA {(MAX_STEPS-step)/sps/3600:.1f}h{saved}"
              + (f"  [lr {prev:.1e}->{cur:.1e}]" if cur != prev else ""))
        r_loss = r_n = 0.0
        json.dump(dict(run=RUN, input=INPUT_MODE, loss=FG_LOSS, jitter=JITTER,
                       seed=TRAIN_SEED, floor=FLOOR_FRAC, select_on=SELECT_ON,
                       batch=BATCH, history=history, shares=shares),
                  open(f"{OUT}/history.json", "w"), indent=1)
    if step % CKPT_EVERY == 0:
        save_ckpt(); print(f"    checkpointed at step {step}")
save_ckpt()
print(f"\ntraining done in {(time.time()-t0)/3600:.2f}h, best val pooled {best:.4f}")

# ------------------- final: all three inference modes ------------------
net = ResUNet().to(DEV); net.load_state_dict(torch.load(BEST, map_location=DEV)); net.eval()
print("\n" + "=" * 76)
print("clean_val under each inference mode (best.pt was selected on "
      f"{SELECT_ON})")
print("=" * 76)
print(f"{'mode':12}{'th':>6}{'pooled':>9}{'per-img':>9}"
      f"{'Q0':>8}{'Q1':>8}{'Q2':>8}{'Q3':>8}{'Q3-Q0':>8}")
modes = ["whole384", "full768", "tiles384"]
final = {}
for md in modes:
    m = evaluate(net, md); final[md] = m
    print(f"{md:12}{m['th']:>6.2f}{m['pooled']:>9.4f}{m['perimg']:>9.4f}"
          + "".join(f"{m['byq'][q]:>8.4f}" for q in range(4))
          + f"{m['byq'][3]-m['byq'][0]:>8.4f}")

# the stratified check: does any Q0 gain track the realized detail gain?
res = {k: tuple(int(v) for v in meta[k]["Resolution"].lower().split("x")[:2])
       for k in CLEAN_VAL}
gain = np.array([min(min(res[s]), FULL_RES)/TRAIN_RES for s in CLEAN_VAL])
q0 = QV == 0
print(f"\nclean_val Q0 split by realized worst-axis detail gain at {FULL_RES}:")
for lab, sel in (("<= 1.0 (no gain)", q0 & (gain <= 1.0)),
                 ("1.0 - 1.5", q0 & (gain > 1.0) & (gain < 1.5)),
                 (">= 1.5", q0 & (gain >= 1.5))):
    print(f"   {lab:18} n={int(sel.sum()):4d}")
print("   per-image Q0 IoU within each band is in the saved json; compare it")
print("   against the matching S-384 run, paired by image.")

np.save(f"{OUT}/clean_val_gain.npy", gain)
json.dump(dict(run=RUN, input=INPUT_MODE, loss=FG_LOSS, jitter=JITTER, seed=TRAIN_SEED,
               floor=FLOOR_FRAC, select_on=SELECT_ON, batch=BATCH, best_val_pooled=best,
               clean_val=CLEAN_VAL, quartiles=QV.tolist(), final=final,
               history=history, shares=shares),
          open(f"{OUT}/summary.json", "w"), indent=1)
xs = [h["step"] for h in history]
plt.figure(figsize=(14, 4))
plt.subplot(1, 3, 1); plt.plot(xs, [h["train_loss"] for h in history]); plt.title("train loss")
plt.subplot(1, 3, 2); plt.plot(xs, [h["pooled"] for h in history]); plt.title("val pooled IoU")
plt.subplot(1, 3, 3)
for q in range(4): plt.plot(xs, [h["byq"][q] for h in history], label=f"Q{q}")
plt.legend(fontsize=7); plt.title("val per-image IoU by quartile")
plt.tight_layout(); plt.savefig(f"{OUT}/curves.png", dpi=110); plt.close()
print(f"\nwrote {OUT}/summary.json, best.pt, curves.png")
print(">>> download best.pt to local disk before the session ends <<<")
```

```python
# =====================================================================
#  cell F -- run this ONCE after every cell D training. Standalone: it
#  reads nothing from cell D's variables, so it also works in a fresh
#  session or after a runtime restart.
#
#  For the target run it:
#    1. scores it per image on clean_val  -> per_image.npz
#       (the vectors every paired test needs; cell D saves only means)
#    2. archives summary / history / curves / gain / per_image to Drive
#    3. records config, GPU and a checksum in weights_manifest.json
#    4. deletes ckpt.pt, but ONLY if the run completed. 490 MB each, and
#       19 of them would overrun a free Drive.
#    5. prints what Drive is holding
#    6. downloads the archive and best.pt
#
#  Any OTHER run still missing per_image.npz or an archive is also
#  brought up to date, without being downloaded again.
# =====================================================================
import os, csv, json, glob, time, zipfile, shutil, hashlib
import numpy as np
from PIL import Image
import torch, torch.nn as nn, torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models
import torchvision.transforms.functional as TF

RUN_NAME    = None       # None = the most recently finished run in stage2/
DELETE_CKPT = True       # False if you intend to extend a run later
DOWNLOAD    = True
SAVE, C768  = "/content/drive/MyDrive", "/content/cache768"
STAGE, ARCH = f"{SAVE}/stage2", f"{SAVE}/stage2_archive"
TRAIN_RES, FULL_RES = 384, 768
BATCH_EVAL  = 12 if torch.cuda.is_available() else 2
FROZEN      = [0.059468, 0.138448, 0.300831]
try:
    from google.colab import drive; drive.mount("/content/drive", force_remount=False)
except Exception: pass
DEV = torch.device("cuda" if torch.cuda.is_available() else "cpu")
os.makedirs(ARCH, exist_ok=True)

# ------------------------------- context -------------------------------
SP   = json.load(open(f"{SAVE}/plantseg_splits.json"))
I768 = json.load(open(f"{C768}/index.json"))
row  = I768["row_of"]
CLEAN_VAL = sorted(set(SP["clean_val"]) & set(I768["stems"]))
DSDIR = "/content/my_images/plantseg"
meta = {os.path.splitext(r["Name"])[0]: r for r in csv.DictReader(
        open(f"{DSDIR}/Metadata.csv", encoding="utf-8-sig", newline=""))}
frac = {k: float(v["Mask ratio"]) for k, v in meta.items()}
def quart(s):
    f = frac[s]
    return 0 if f <= FROZEN[0] else 1 if f <= FROZEN[1] else 2 if f <= FROZEN[2] else 3
QV = np.array([quart(s) for s in CLEAN_VAL])

MEAN, STD = [0.485,0.456,0.406], [0.229,0.224,0.225]
NORM = transforms.Normalize(MEAN, STD)
_mm = {}
def cache():
    if not _mm:
        _mm["i"] = np.load(I768["img_path"], mmap_mode="r")
        _mm["m"] = np.load(I768["msk_path"], mmap_mode="r")
    return _mm["i"], _mm["m"]
def strip_detail(im):
    return im.resize((TRAIN_RES,)*2, Image.BILINEAR).resize((FULL_RES,)*2, Image.BILINEAR)

class EvalSet(Dataset):
    def __init__(s, st, up): s.st, s.up = list(st), up
    def __len__(s): return len(s.st)
    def __getitem__(s, i):
        I, M = cache(); r = row[s.st[i]]
        img = Image.fromarray(I[r])
        if s.up: img = strip_detail(img)
        y = torch.from_numpy((np.asarray(M[r]) > 0).astype(np.float32)).unsqueeze(0)
        return NORM(TF.to_tensor(img)), y

SPEC = dict(channels=(64,256,512,1024,2048), decoder=(512,256,128,64))
class DoubleConv(nn.Module):
    def __init__(s,i,o):
        super().__init__()
        s.net = nn.Sequential(nn.Conv2d(i,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True),
                              nn.Conv2d(o,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True))
    def forward(s,x): return s.net(x)
class Up(nn.Module):
    def __init__(s,i,k,o):
        super().__init__(); s.reduce=nn.Conv2d(i,i//2,1,bias=False); s.conv=DoubleConv(i//2+k,o)
    def forward(s,x,skip):
        x = F.interpolate(x, size=skip.shape[-2:], mode="bilinear", align_corners=False)
        return s.conv(torch.cat([s.reduce(x), skip], 1))
class ResUNet(nn.Module):
    def __init__(s):
        super().__init__()
        base = models.resnet50(weights=None)
        c0,c1,c2,c3,c4 = SPEC["channels"]; d1,d2,d3,d4 = SPEC["decoder"]
        L = list(base.children())
        s.stem=nn.Sequential(*L[:3]); s.pool=L[3]
        s.l1,s.l2,s.l3,s.l4 = L[4],L[5],L[6],L[7]
        s.u1,s.u2,s.u3,s.u4 = Up(c4,c3,d1),Up(d1,c2,d2),Up(d2,c1,d3),Up(d3,c0,d4)
        s.head = nn.Sequential(nn.Conv2d(d4,32,3,padding=1,bias=False),
                               nn.BatchNorm2d(32), nn.ReLU(True), nn.Conv2d(32,1,1))
    def forward(s,x):
        size=x.shape[-2:]
        x0=s.stem(x); x1=s.l1(s.pool(x0)); x2=s.l2(x1); x3=s.l3(x2); x4=s.l4(x3)
        d=s.u1(x4,x3); d=s.u2(d,x2); d=s.u3(d,x1); d=s.u4(d,x0)
        return s.head(F.interpolate(d,size=size,mode="bilinear",align_corners=False))

@torch.no_grad()
def prob_768(net, x, mode):
    if mode == "full768":
        return torch.sigmoid(net(x).float())
    if mode == "whole384":
        d = F.interpolate(x, size=(TRAIN_RES,)*2, mode="bilinear", align_corners=False)
        p = torch.sigmoid(net(d).float())
        return F.interpolate(p, size=(FULL_RES,)*2, mode="bilinear", align_corners=False)
    acc = torch.zeros(x.shape[0], 1, FULL_RES, FULL_RES, device=x.device)
    cnt = torch.zeros_like(acc)
    ORIG = (0, (FULL_RES-TRAIN_RES)//2, FULL_RES-TRAIN_RES)
    for oy in ORIG:
        for ox in ORIG:
            t = x[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES]
            acc[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += torch.sigmoid(net(t).float())
            cnt[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += 1
    return acc/cnt.clamp(min=1)

THS = [round(t, 2) for t in np.arange(0.10, 0.91, 0.05)]
@torch.no_grad()
def score_per_image(bp, mode, up):
    net = ResUNet().to(DEV); net.load_state_dict(torch.load(bp, map_location=DEV)); net.eval()
    dl = DataLoader(EvalSet(CLEAN_VAL, up), BATCH_EVAL, shuffle=False,
                    num_workers=min(2, os.cpu_count() or 1), pin_memory=DEV.type == "cuda")
    n = len(CLEAN_VAL)
    per = np.zeros((n, len(THS))); Ipx = np.zeros((n, len(THS)))
    P = np.zeros((n, len(THS))); G = np.zeros(n); k = 0
    for x, y in dl:
        x, y = x.to(DEV, non_blocking=True), y.to(DEV, non_blocking=True)
        with torch.amp.autocast("cuda", enabled=DEV.type == "cuda"):
            p = prob_768(net, x, mode)
        b = x.shape[0]; yf = y.view(b, -1); pf = p.view(b, -1)
        G[k:k+b] = yf.sum(1).cpu().numpy()
        for j, t in enumerate(THS):
            pr = (pf > t).float(); it = (pr*yf).sum(1); pp = pr.sum(1)
            Ipx[k:k+b, j] = it.cpu().numpy(); P[k:k+b, j] = pp.cpu().numpy()
            per[k:k+b, j] = (it/(pp+yf.sum(1)-it).clamp(min=1)).cpu().numpy()
        k += b
    pooled = Ipx.sum(0)/np.maximum(P.sum(0)+G.sum()-Ipx.sum(0), 1)
    j = int(np.argmax(pooled))
    del net; torch.cuda.empty_cache()
    return dict(th=THS[j], pooled=float(pooled[j]), per=per[:, j])

def sha(p, chunk=1 << 24):
    h = hashlib.sha256()
    with open(p, "rb") as f:
        while (b := f.read(chunk)): h.update(b)
    return h.hexdigest()[:16]

# -------------------------- which runs to handle -----------------------
# summary.json is written only at the END of cell D, so its presence is
# the completion test -- safer than trusting a step count.
done = sorted(glob.glob(f"{STAGE}/*/summary.json"), key=os.path.getmtime)
if not done:
    raise SystemExit(f"no completed run under {STAGE}")
target = RUN_NAME or os.path.basename(os.path.dirname(done[-1]))
print(f"target run: {target}   ({len(done)} completed run(s) on Drive)\n")

man_p = f"{SAVE}/weights_manifest.json"
man = json.load(open(man_p)) if os.path.exists(man_p) else {}

for sm in done:
    d = os.path.dirname(sm); name = os.path.basename(d)
    J = json.load(open(sm))
    bp, ck = f"{d}/best.pt", f"{d}/ckpt.pt"
    if not os.path.exists(bp):
        print(f"{name}: no best.pt, skipped"); continue
    is_target = (name == target)

    # 1. per-image vectors
    npz = f"{d}/per_image.npz"
    need = True
    if os.path.exists(npz):
        try:
            z = np.load(npz, allow_pickle=True); need = list(z["stems"]) != CLEAN_VAL
        except Exception: need = True
    if need:
        t = time.time()
        r = score_per_image(bp, J["select_on"], J["input"] == "crop_768_up")
        np.savez(npz, per=r["per"], th=r["th"], pooled=r["pooled"],
                 mode=J["select_on"], stems=np.array(CLEAN_VAL))
        print(f"{name}: scored per image in {time.time()-t:.0f}s  "
              f"(mode {J['select_on']}, th {r['th']}, pooled {r['pooled']:.4f})")
    else:
        print(f"{name}: per_image.npz already current")

    # 2. archive
    zp = f"{ARCH}/{name}_results.zip"
    if is_target or not os.path.exists(zp):
        with zipfile.ZipFile(zp, "w", zipfile.ZIP_DEFLATED) as z:
            for f in ("summary.json", "history.json", "curves.png",
                      "clean_val_gain.npy", "per_image.npz"):
                p = f"{d}/{f}"
                if os.path.exists(p): z.write(p, arcname=f"{name}/{f}")
        print(f"{name}: archive {os.path.getsize(zp)/1e6:.2f} MB")

    # 3. manifest
    if is_target or name not in man:
        man[name] = dict(file=f"{name}__best.pt", bytes=os.path.getsize(bp),
                         sha256_16=sha(bp), best_val_pooled=J.get("best_val_pooled"),
                         gpu=J.get("gpu", "not recorded"), batch=J.get("batch"),
                         input=J["input"], loss=J["loss"], jitter=J["jitter"],
                         seed=J["seed"], floor=J.get("floor"))

    # 4. prune -- only a COMPLETED run, which summary.json proves
    if DELETE_CKPT and os.path.exists(ck):
        mb = os.path.getsize(ck)/1e6
        os.remove(ck); print(f"{name}: removed ckpt.pt (freed {mb:.0f} MB)")

json.dump(man, open(man_p, "w"), indent=1)
print(f"\nmanifest: {man_p}  ({len(man)} run(s))")

# ------------------------------ drive usage ----------------------------
def dsz(p):
    return sum(os.path.getsize(os.path.join(r, f))
               for r, _, fs in os.walk(p) for f in fs) if os.path.isdir(p) else 0
tot = 0
print("\nDrive usage")
for lab, p in (("stage2", STAGE), ("stage2_archive", ARCH),
               ("runs (stage 1)", f"{SAVE}/runs"), ("cache768", f"{SAVE}/cache768")):
    s = dsz(p); tot += s
    print(f"   {lab:18}{s/1e9:7.2f} GB")
print(f"   {'TRACKED TOTAL':18}{tot/1e9:7.2f} GB"
      + ("   !! approaching a 15 GB free Drive" if tot/1e9 > 11 else ""))

# ------------------------------- download ------------------------------
if DOWNLOAD:
    d = f"{STAGE}/{target}"
    loc = f"/content/{target}__best.pt"
    if not os.path.exists(loc): shutil.copyfile(f"{d}/best.pt", loc)
    print(f"\n{target}  best.pt {os.path.getsize(loc)/1e6:.1f} MB  "
          f"sha256[:16] {man[target]['sha256_16']}")
    try:
        from google.colab import files
        files.download(f"{ARCH}/{target}_results.zip")
        time.sleep(2)
        files.download(loc)
        print("downloads queued. Verify locally:")
        print(f"   shasum -a 256 {target}__best.pt | cut -c1-16"
              f"   ->  {man[target]['sha256_16']}")
    except Exception as e:
        print(f"(browser download unavailable: {type(e).__name__})")
    print(f"also on Drive:\n   {ARCH}/{target}_results.zip\n   {d}/best.pt")
```

```python
# =====================================================================
#  cell E -- score every finished Stage 2 run per image, then compare.
#
#  Cell D saves per-quartile MEANS, which cannot support a paired test.
#  This reloads each best.pt and writes the per-image IoU vector over
#  clean_val, at that run's own selected mode and threshold. No retraining.
#
#  Then, against the S-384 baseline, for each other run:
#    - paired per-image differences over the same 688 images
#    - an image-level paired bootstrap on Q0  (the PRIMARY inference:
#      688 paired images carry far more power than a 3-seed t-interval,
#      which at 3 seeds needs a seed-sd below 0.008 to clear +0.02)
#    - the same split by realized detail gain, declared in advance
#    - the PILOT rule: kill only if the mean Q0 difference is <= 0.
#      Killing anything that misses the +0.02 gate at one seed discards
#      a real +0.02 effect half the time; the <= 0 rule, 10% of the time.
#
#  A single-seed result here is a FILTER, never a result. Nothing from
#  this cell goes in the paper without seeds 43 and 44.
# =====================================================================
import os, csv, json, glob, time
import numpy as np
from PIL import Image
import torch, torch.nn as nn, torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models
import torchvision.transforms.functional as TF

SAVE, C768 = "/content/drive/MyDrive", "/content/cache768"
STAGE   = f"{SAVE}/stage2"
BASE    = "S-384_seed42"           # the baseline every run is paired against
TRAIN_RES, FULL_RES = 384, 768
BATCH_EVAL = 4
NBOOT   = 10000
FROZEN  = [0.059468, 0.138448, 0.300831]
KILL_AT = 0.0                      # pilot rule, declared before looking
try:
    from google.colab import drive; drive.mount("/content/drive", force_remount=False)
except Exception: pass
DEV = torch.device("cuda" if torch.cuda.is_available() else "cpu")

SP = json.load(open(f"{SAVE}/plantseg_splits.json"))
I768 = json.load(open(f"{C768}/index.json"))
row = I768["row_of"]
CLEAN_VAL = sorted(set(SP["clean_val"]) & set(I768["stems"]))
DSDIR = "/content/my_images/plantseg"
meta = {os.path.splitext(r["Name"])[0]: r for r in csv.DictReader(
        open(f"{DSDIR}/Metadata.csv", encoding="utf-8-sig", newline=""))}
frac = {k: float(v["Mask ratio"]) for k, v in meta.items()}
def quart(s):
    f = frac[s]
    return 0 if f <= FROZEN[0] else 1 if f <= FROZEN[1] else 2 if f <= FROZEN[2] else 3
QV = np.array([quart(s) for s in CLEAN_VAL])
res = {k: tuple(int(v) for v in meta[k]["Resolution"].lower().split("x")[:2])
       for k in CLEAN_VAL}
GAIN = np.array([min(min(res[s]), FULL_RES)/TRAIN_RES for s in CLEAN_VAL])
print(f"clean_val {len(CLEAN_VAL)}  quartiles {[int((QV==q).sum()) for q in range(4)]}")

MEAN, STD = [0.485,0.456,0.406], [0.229,0.224,0.225]
NORM = transforms.Normalize(MEAN, STD)
_mm = {}
def cache():
    if not _mm:
        _mm["i"] = np.load(I768["img_path"], mmap_mode="r")
        _mm["m"] = np.load(I768["msk_path"], mmap_mode="r")
    return _mm["i"], _mm["m"]
def strip_detail(im):
    return im.resize((TRAIN_RES,)*2, Image.BILINEAR).resize((FULL_RES,)*2, Image.BILINEAR)

class EvalSet(Dataset):
    def __init__(s, st, up): s.st, s.up = list(st), up
    def __len__(s): return len(s.st)
    def __getitem__(s, i):
        I, M = cache(); r = row[s.st[i]]
        img = Image.fromarray(I[r])
        if s.up: img = strip_detail(img)
        y = torch.from_numpy((np.asarray(M[r]) > 0).astype(np.float32)).unsqueeze(0)
        return NORM(TF.to_tensor(img)), y

SPEC = dict(channels=(64,256,512,1024,2048), decoder=(512,256,128,64))
class DoubleConv(nn.Module):
    def __init__(s,i,o):
        super().__init__()
        s.net = nn.Sequential(nn.Conv2d(i,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True),
                              nn.Conv2d(o,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True))
    def forward(s,x): return s.net(x)
class Up(nn.Module):
    def __init__(s,i,k,o):
        super().__init__(); s.reduce=nn.Conv2d(i,i//2,1,bias=False); s.conv=DoubleConv(i//2+k,o)
    def forward(s,x,skip):
        x = F.interpolate(x, size=skip.shape[-2:], mode="bilinear", align_corners=False)
        return s.conv(torch.cat([s.reduce(x), skip], 1))
class ResUNet(nn.Module):
    def __init__(s):
        super().__init__()
        base = models.resnet50(weights=None)
        c0,c1,c2,c3,c4 = SPEC["channels"]; d1,d2,d3,d4 = SPEC["decoder"]
        L = list(base.children())
        s.stem=nn.Sequential(*L[:3]); s.pool=L[3]
        s.l1,s.l2,s.l3,s.l4 = L[4],L[5],L[6],L[7]
        s.u1,s.u2,s.u3,s.u4 = Up(c4,c3,d1),Up(d1,c2,d2),Up(d2,c1,d3),Up(d3,c0,d4)
        s.head = nn.Sequential(nn.Conv2d(d4,32,3,padding=1,bias=False),
                               nn.BatchNorm2d(32), nn.ReLU(True), nn.Conv2d(32,1,1))
    def forward(s,x):
        size=x.shape[-2:]
        x0=s.stem(x); x1=s.l1(s.pool(x0)); x2=s.l2(x1); x3=s.l3(x2); x4=s.l4(x3)
        d=s.u1(x4,x3); d=s.u2(d,x2); d=s.u3(d,x1); d=s.u4(d,x0)
        return s.head(F.interpolate(d,size=size,mode="bilinear",align_corners=False))

@torch.no_grad()
def prob_768(net, x, mode):
    if mode == "full768":
        return torch.sigmoid(net(x).float())
    if mode == "whole384":
        d = F.interpolate(x, size=(TRAIN_RES,)*2, mode="bilinear", align_corners=False)
        p = torch.sigmoid(net(d).float())
        return F.interpolate(p, size=(FULL_RES,)*2, mode="bilinear", align_corners=False)
    acc = torch.zeros(x.shape[0], 1, FULL_RES, FULL_RES, device=x.device)
    cnt = torch.zeros_like(acc)
    ORIG = (0, (FULL_RES-TRAIN_RES)//2, FULL_RES-TRAIN_RES)
    for oy in ORIG:
        for ox in ORIG:
            t = x[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES]
            acc[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += torch.sigmoid(net(t).float())
            cnt[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += 1
    return acc/cnt.clamp(min=1)

THS = [round(t, 2) for t in np.arange(0.10, 0.91, 0.05)]
@torch.no_grad()
def score(bp, mode, up):
    net = ResUNet().to(DEV); net.load_state_dict(torch.load(bp, map_location=DEV)); net.eval()
    dl = DataLoader(EvalSet(CLEAN_VAL, up), BATCH_EVAL, shuffle=False,
                    num_workers=min(2, os.cpu_count() or 1), pin_memory=DEV.type == "cuda")
    n = len(CLEAN_VAL)
    per = np.zeros((n, len(THS))); Ipx = np.zeros((n, len(THS)))
    P = np.zeros((n, len(THS))); G = np.zeros(n); k = 0
    for x, y in dl:
        x, y = x.to(DEV, non_blocking=True), y.to(DEV, non_blocking=True)
        with torch.amp.autocast("cuda", enabled=DEV.type == "cuda"):
            p = prob_768(net, x, mode)
        b = x.shape[0]; yf = y.view(b, -1); pf = p.view(b, -1)
        G[k:k+b] = yf.sum(1).cpu().numpy()
        for j, t in enumerate(THS):
            pr = (pf > t).float(); it = (pr*yf).sum(1); pp = pr.sum(1)
            Ipx[k:k+b, j] = it.cpu().numpy(); P[k:k+b, j] = pp.cpu().numpy()
            per[k:k+b, j] = (it/(pp+yf.sum(1)-it).clamp(min=1)).cpu().numpy()
        k += b
    pooled = Ipx.sum(0)/np.maximum(P.sum(0)+G.sum()-Ipx.sum(0), 1)
    j = int(np.argmax(pooled))
    del net; torch.cuda.empty_cache()
    return dict(th=THS[j], pooled=float(pooled[j]), per=per[:, j])

# ------------------------- score what is finished ----------------------
runs = {}
for sm in sorted(glob.glob(f"{STAGE}/*/summary.json")):
    d = os.path.dirname(sm); name = os.path.basename(d)
    if not os.path.exists(f"{d}/best.pt"):
        print(f"   {name}: no best.pt, skipped"); continue
    J = json.load(open(sm))
    npz = f"{d}/per_image.npz"
    if os.path.exists(npz):
        z = np.load(npz, allow_pickle=True)
        if list(z["stems"]) == CLEAN_VAL:
            runs[name] = dict(per=z["per"], th=float(z["th"]), pooled=float(z["pooled"]),
                              mode=str(z["mode"]), cfg=J); print(f"   {name}: cached"); continue
    t = time.time()
    r = score(f"{d}/best.pt", J["select_on"], J["input"] == "crop_768_up")
    np.savez(npz, per=r["per"], th=r["th"], pooled=r["pooled"],
             mode=J["select_on"], stems=np.array(CLEAN_VAL))
    runs[name] = dict(**r, mode=J["select_on"], cfg=J)
    print(f"   {name}: scored in {time.time()-t:.0f}s  "
          f"mode {J['select_on']} th {r['th']} pooled {r['pooled']:.4f}")

if BASE not in runs:
    raise SystemExit(f"baseline {BASE} not scored yet -- run it first")
print(f"\n{len(runs)} run(s) scored; baseline {BASE}")

# --------------------------- paired comparison -------------------------
rng = np.random.default_rng(0)
b_per = runs[BASE]["per"]; q0 = QV == 0
bands = {"Q0 all": q0, "Q0 gain<=1.0": q0 & (GAIN <= 1.0),
         "Q0 gain>=1.5": q0 & (GAIN >= 1.5), "all images": np.ones(len(QV), bool)}

def boot(d, sel):
    v = d[sel]
    if len(v) < 5: return float("nan"), float("nan"), float("nan")
    idx = rng.integers(0, len(v), (NBOOT, len(v)))
    bs = v[idx].mean(1)
    return float(v.mean()), float(np.percentile(bs, 2.5)), float(np.percentile(bs, 97.5))

print("\n" + "=" * 88)
print(f"paired per-image IoU difference vs {BASE}, clean_val, "
      f"{NBOOT} image-level bootstraps")
print("=" * 88)
out = {}
for name, r in runs.items():
    if name == BASE: continue
    d = r["per"] - b_per
    print(f"\n{name}   ({r['cfg']['input']} / {r['cfg']['loss']}"
          f"{' / jitter' if r['cfg']['jitter'] else ''})   "
          f"pooled {r['pooled']:.4f} vs {runs[BASE]['pooled']:.4f}")
    print(f"   {'band':16}{'n':>6}{'mean diff':>12}{'95% CI':>22}")
    row = {}
    for lab, sel in bands.items():
        m, lo, hi = boot(d, sel)
        star = "  *" if (lo > 0 or hi < 0) else ""
        print(f"   {lab:16}{int(sel.sum()):>6}{m:>12.4f}"
              f"   [{lo:+.4f}, {hi:+.4f}]{star}")
        row[lab] = dict(n=int(sel.sum()), mean=m, lo=lo, hi=hi)
    mq0 = row["Q0 all"]["mean"]
    verdict = ("KILL -- Q0 moves the wrong way" if mq0 <= KILL_AT else
               "KEEP -- run seeds 43 and 44")
    print(f"   PILOT ({'<=' if mq0 <= KILL_AT else '>'} {KILL_AT:+.2f}): {verdict}")
    if row["Q0 gain>=1.5"]["n"] >= 20 and row["Q0 gain<=1.0"]["n"] >= 20:
        sep = row["Q0 gain>=1.5"]["mean"] - row["Q0 gain<=1.0"]["mean"]
        print(f"   mechanism check: high-detail minus no-detail = {sep:+.4f}"
              f"   {'(consistent with resolution)' if sep > 0 else '(not resolution)'}")
        row["mechanism_sep"] = sep
    out[name] = row

print("\n" + "=" * 88)
print("* marks a CI excluding zero. That is an image-level interval: it says the")
print("difference is not sampling noise over these 688 images, NOT that it would")
print("survive another training seed. Only seeds 43 and 44 answer that, and the")
print("gate is judged on the three-seed paired mean, not on anything here.")
json.dump(dict(baseline=BASE, kill_at=KILL_AT, nboot=NBOOT, results=out),
          open(f"{SAVE}/stage2_compare.json", "w"), indent=1, default=float)
print(f"\nwrote {SAVE}/stage2_compare.json")
```

```python
# =====================================================================
#  cell D -- Stage 2: supervised tests of components 1 and 2.
#  One cell, three switches, 25,000 steps, resume-safe. Validation only;
#  the test split is not touched until Stage 5.
#
#      INPUT_MODE in {full_384, crop_768, crop_768_up}
#      FG_LOSS    in {pooled, class_balanced, image_balanced}
#      JITTER     in {False, True}
#
#  (full_384, pooled, False) is the S-384 baseline.
#
#  crop_768_up is the control the crop statistics made necessary: uniform
#  384 crops of a 768 image carry 1.50x the Q0 lesion density of the
#  whole image, because uniform crop ORIGINS over-sample the centre (a
#  centre pixel sits in 384 of 385 origins per axis, a corner pixel in 1).
#  So crop_768 changes resolution AND density at once. crop_768_up takes
#  the 768 image, downsamples it to 384 and upsamples it back, destroying
#  exactly the detail above 384 and nothing else, then crops identically.
#  crop_768 beating crop_768_up isolates resolution; a tie says the gain
#  is crop geometry.
#
#  Everything reads ONE cache: /content/cache768 (537 labelled + 846 val).
#  The 441 cache is not needed, so prep.py need not be run this session.
# =====================================================================
import os, csv, json, math, time, random, shutil, zipfile, collections
import numpy as np
import matplotlib; matplotlib.use("Agg")
import matplotlib.pyplot as plt
from PIL import Image, ImageFilter
import torch, torch.nn as nn, torch.optim as optim, torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models
import torchvision.transforms.functional as TF

# ------------------------- the three switches --------------------------
INPUT_MODE = "full_384"        # full_384 | crop_768 | crop_768_up
FG_LOSS    = "pooled"          # pooled | class_balanced | image_balanced
JITTER     = False             # True only for S-768c-jit
TRAIN_SEED = 44
# -----------------------------------------------------------------------
FLOOR_FRAC = 0.01              # declared from check 0f: binds on 8.9% of
                               # non-empty Q0 crops, caps amplification 21.5x
SELECT_ON  = {"full_384": "whole384", "crop_768": "full768",
              "crop_768_up": "full768"}[INPUT_MODE]
MAX_STEPS  = 25000
EVAL_EVERY = 1000
SHARE_EVERY = 500              # per-quartile lesion-loss share
CKPT_EVERY = 2500
LR_PATIENCE = 3                # patience counts EVAL_EVERY blocks, now 1000
TRAIN_RES  = 384
FULL_RES   = 768
BATCH      = 4                 # matches the existing pipeline; batch 8
                               # doubles every run to ~2.2h and Stage 2 to ~42h
BATCH_EVAL = 4                 # 768 input on a T4
LR_ENCODER, LR_DECODER, WEIGHT_DECAY = 1e-5, 1e-4, 1e-4
FROZEN     = [0.059468, 0.138448, 0.300831]
SAVE, C768 = "/content/drive/MyDrive", "/content/cache768"
NUM_WORKERS = min(2, os.cpu_count() or 1)
FRESH      = True              # wipes this run's folder. Set back to False
                               # once it is past step 2500 so a dropped
                               # session resumes instead of restarting.

TAG = {"full_384": "384", "crop_768": "768c", "crop_768_up": "768up"}[INPUT_MODE]
if FG_LOSS == "class_balanced":  TAG += "-cb"
if FG_LOSS == "image_balanced":  TAG += "-bal"
if JITTER:                       TAG += "-jit"
RUN = f"S-{TAG}_seed{TRAIN_SEED}"
OUT = f"{SAVE}/stage2/{RUN}"
if FRESH and os.path.isdir(OUT): shutil.rmtree(OUT, ignore_errors=True)
os.makedirs(OUT, exist_ok=True)

random.seed(TRAIN_SEED); np.random.seed(TRAIN_SEED)
torch.manual_seed(TRAIN_SEED); torch.cuda.manual_seed_all(TRAIN_SEED)
torch.backends.cudnn.deterministic = True; torch.backends.cudnn.benchmark = False
DEV = torch.device("cuda" if torch.cuda.is_available() else "cpu")
try:
    from google.colab import drive; drive.mount("/content/drive", force_remount=False)
except Exception: pass
print(f"{RUN}\ndevice {DEV} {torch.cuda.get_device_name(0) if DEV.type=='cuda' else ''}")
print(f"input {INPUT_MODE} | loss {FG_LOSS} | jitter {JITTER} | "
      f"select on {SELECT_ON} | floor {FLOOR_FRAC} | batch {BATCH}")

# --------------------------- data and asserts --------------------------
SP = json.load(open(f"{SAVE}/plantseg_splits.json"))
if max(abs(a-b) for a, b in zip(SP["quartile_cuts"], FROZEN)) > 1e-6:
    raise SystemExit(f"cuts {SP['quartile_cuts']} != frozen {FROZEN}")
if {k: len(v) for k, v in SP["official_splits"].items()} != \
   {"train": 5367, "val": 846, "test": 1561}:
    raise SystemExit("split sizes are not 5367/846/1561")
I768 = json.load(open(f"{C768}/index.json"))
if I768["load_size"] != FULL_RES:
    raise SystemExit(f"cache is {I768['load_size']}, expected {FULL_RES}")
row = I768["row_of"]
LAB = sorted(I768["labelled"])
CLEAN_VAL = sorted(set(SP["clean_val"]) & set(I768["stems"]))
if len(LAB) != 537: raise SystemExit(f"{len(LAB)} labelled, expected 537")
DSDIR = "/content/my_images/plantseg"
meta = {os.path.splitext(r["Name"])[0]: r for r in csv.DictReader(
        open(f"{DSDIR}/Metadata.csv", encoding="utf-8-sig", newline=""))}
frac = {k: float(v["Mask ratio"]) for k, v in meta.items()}
def quart(s):
    f = frac[s]
    return 0 if f <= FROZEN[0] else 1 if f <= FROZEN[1] else 2 if f <= FROZEN[2] else 3
QV = np.array([quart(s) for s in CLEAN_VAL])
print(f"labelled {len(LAB)}   clean_val {len(CLEAN_VAL)} "
      f"(quartiles {[int((QV==q).sum()) for q in range(4)]})")
print("   gates are judged on clean_val, not the full 846: full val carries")
print("   17.14% near-duplicates of training images.")

MEAN, STD = [0.485,0.456,0.406], [0.229,0.224,0.225]
NORM = transforms.Normalize(MEAN, STD)
JIT  = transforms.ColorJitter(0.4, 0.4, 0.4, 0.1)
_mm = {}
def cache():
    if not _mm:
        _mm["i"] = np.load(I768["img_path"], mmap_mode="r")
        _mm["m"] = np.load(I768["msk_path"], mmap_mode="r")
    return _mm["i"], _mm["m"]

def strip_detail(im):
    """768 -> 384 -> 768. Destroys exactly the detail above 384, keeps
       geometry, size and lesion density identical."""
    return im.resize((TRAIN_RES,)*2, Image.BILINEAR).resize((FULL_RES,)*2, Image.BILINEAR)

def photometric(img):
    if random.random() < 0.8: img = JIT(img)
    if random.random() < 0.2: img = TF.to_grayscale(img, num_output_channels=3)
    if random.random() < 0.3:
        img = img.filter(ImageFilter.GaussianBlur(radius=random.uniform(0.1, 1.5)))
    return img

def geometric(img, msk):
    if random.random() < 0.5: img, msk = TF.hflip(img), TF.hflip(msk)
    if random.random() < 0.5: img, msk = TF.vflip(img), TF.vflip(msk)
    k = random.randint(0, 3)
    if k:
        img = img.rotate(90*k, expand=True, resample=Image.BILINEAR)
        msk = msk.rotate(90*k, expand=True, resample=Image.NEAREST)
    return img, msk

def rrc(img, msk, scale=(0.5, 1.0), ratio=(3/4, 4/3)):
    w, h = img.size; area = w*h
    for _ in range(10):
        t = area*random.uniform(*scale)
        ar = math.exp(random.uniform(math.log(ratio[0]), math.log(ratio[1])))
        cw, ch = int(round(math.sqrt(t*ar))), int(round(math.sqrt(t/ar)))
        if cw <= w and ch <= h:
            x, y = random.randint(0, w-cw), random.randint(0, h-ch)
            return img.crop((x, y, x+cw, y+ch)), msk.crop((x, y, x+cw, y+ch))
    return img, msk

class Labelled(Dataset):
    """full_384   : whole image, rrc + geometric, resized to 384
       crop_768   : uniform 384 window of the 768 image (the crop IS the
                    spatial augmentation, so rrc is not applied on top)
       crop_768_up: the same window of a detail-stripped 768 image"""
    def __init__(s, st): s.st = list(st)
    def __len__(s): return len(s.st)
    def __getitem__(s, i):
        I, M = cache(); r = row[s.st[i]]
        img = Image.fromarray(I[r]); msk = Image.fromarray(M[r])
        if INPUT_MODE == "crop_768_up": img = strip_detail(img)
        if INPUT_MODE == "full_384":
            img, msk = rrc(img, msk)
            img, msk = geometric(img, msk)
            img = img.resize((TRAIN_RES,)*2, Image.BILINEAR)
            msk = msk.resize((TRAIN_RES,)*2, Image.NEAREST)
        else:
            side = (random.randint(256, FULL_RES) if JITTER else TRAIN_RES)
            x = random.randint(0, FULL_RES-side); y = random.randint(0, FULL_RES-side)
            img = img.crop((x, y, x+side, y+side)); msk = msk.crop((x, y, x+side, y+side))
            img, msk = geometric(img, msk)
            if side != TRAIN_RES:
                img = img.resize((TRAIN_RES,)*2, Image.BILINEAR)
                msk = msk.resize((TRAIN_RES,)*2, Image.NEAREST)
        y_ = torch.from_numpy((np.array(msk) > 0).astype(np.float32)).unsqueeze(0)
        return NORM(TF.to_tensor(photometric(img))), y_, quart(s.st[i])

class EvalSet(Dataset):
    """input follows INPUT_MODE; the TARGET is always the 768 mask."""
    def __init__(s, st): s.st = list(st)
    def __len__(s): return len(s.st)
    def __getitem__(s, i):
        I, M = cache(); r = row[s.st[i]]
        img = Image.fromarray(I[r])
        if INPUT_MODE == "crop_768_up": img = strip_detail(img)
        y = torch.from_numpy((np.asarray(M[r]) > 0).astype(np.float32)).unsqueeze(0)
        return NORM(TF.to_tensor(img)), y            # image at 768, mask at 768

def seed_worker(_):
    s = torch.initial_seed() % 2**32; np.random.seed(s); random.seed(s)
LK = dict(num_workers=NUM_WORKERS, pin_memory=DEV.type == "cuda",
          worker_init_fn=seed_worker, persistent_workers=NUM_WORKERS > 0)
g = torch.Generator(); g.manual_seed(TRAIN_SEED)
loader_l = DataLoader(Labelled(LAB), BATCH, shuffle=True, drop_last=True,
                      generator=g, **LK)
val_loader = DataLoader(EvalSet(CLEAN_VAL), BATCH_EVAL, shuffle=False, **LK)
def cycle(dl):
    while True:
        for b in dl: yield b

# ------------------------------- model ---------------------------------
SPEC = dict(channels=(64,256,512,1024,2048), decoder=(512,256,128,64))
class DoubleConv(nn.Module):
    def __init__(s,i,o):
        super().__init__()
        s.net = nn.Sequential(nn.Conv2d(i,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True),
                              nn.Conv2d(o,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True))
    def forward(s,x): return s.net(x)
class Up(nn.Module):
    def __init__(s,i,k,o):
        super().__init__(); s.reduce=nn.Conv2d(i,i//2,1,bias=False); s.conv=DoubleConv(i//2+k,o)
    def forward(s,x,skip):
        x = F.interpolate(x, size=skip.shape[-2:], mode="bilinear", align_corners=False)
        return s.conv(torch.cat([s.reduce(x), skip], 1))
class ResUNet(nn.Module):
    def __init__(s):
        super().__init__()
        base = models.resnet50(weights=models.ResNet50_Weights.DEFAULT)
        c0,c1,c2,c3,c4 = SPEC["channels"]; d1,d2,d3,d4 = SPEC["decoder"]
        L = list(base.children())
        s.stem=nn.Sequential(*L[:3]); s.pool=L[3]
        s.l1,s.l2,s.l3,s.l4 = L[4],L[5],L[6],L[7]
        s.u1,s.u2,s.u3,s.u4 = Up(c4,c3,d1),Up(d1,c2,d2),Up(d2,c1,d3),Up(d3,c0,d4)
        s.head = nn.Sequential(nn.Conv2d(d4,32,3,padding=1,bias=False),
                               nn.BatchNorm2d(32), nn.ReLU(True), nn.Conv2d(32,1,1))
    def enc_params(s):
        for m in (s.stem,s.l1,s.l2,s.l3,s.l4): yield from m.parameters()
    def dec_params(s):
        for m in (s.u1,s.u2,s.u3,s.u4,s.head): yield from m.parameters()
    def forward(s,x):
        size=x.shape[-2:]
        x0=s.stem(x); x1=s.l1(s.pool(x0)); x2=s.l2(x1); x3=s.l3(x2); x4=s.l4(x3)
        d=s.u1(x4,x3); d=s.u2(d,x2); d=s.u3(d,x1); d=s.u4(d,x0)
        return s.head(F.interpolate(d,size=size,mode="bilinear",align_corners=False))

# ------------------------------- losses --------------------------------
bce_none = nn.BCEWithLogitsLoss(reduction="none")
def tversky(logits, tgt, eps=1.0):
    p = torch.sigmoid(logits)
    tp = (p*tgt).sum((1,2,3)); fp = (p*(1-tgt)).sum((1,2,3)); fn = ((1-p)*tgt).sum((1,2,3))
    return 1 - (tp+eps)/(tp+0.5*fp+0.5*fn+eps)

def fg_term(logits, tgt):
    """returns (per-image BCE term, per-image lesion-only contribution).
       All three forms are scaled so a uniform per-pixel loss c gives c,
       so the learning rate and the BCE/Tversky balance do not shift
       between configurations. The plan's formula writes the sum of the
       two class means; using their average is the same comparison at the
       same scale."""
    e = bce_none(logits, tgt)
    sf = tgt.sum((1,2,3)); sb = (1-tgt).sum((1,2,3))
    ef = (e*tgt).sum((1,2,3)); eb = (e*(1-tgt)).sum((1,2,3))
    if FG_LOSS == "pooled":
        return e.mean((1,2,3)), ef/tgt[0].numel()
    if FG_LOSS == "class_balanced":                 # batch-level class means
        n = logits.shape[0]
        mf = ef.sum()/sf.sum().clamp(min=1.0); mb = eb.sum()/sb.sum().clamp(min=1.0)
        return (0.5*(mf+mb)).expand(n), 0.5*ef/sf.sum().clamp(min=1.0)
    floor = FLOOR_FRAC*tgt[0].numel()               # per-image, with the floor
    tf = ef/torch.clamp(sf, min=floor)
    tb = eb/sb.clamp(min=1.0)
    return 0.5*(tf+tb), 0.5*tf

def sup_loss(logits, tgt):
    b, lesion = fg_term(logits, tgt)
    return b + tversky(logits, tgt), lesion

# --------------------------- 768 inference -----------------------------
@torch.no_grad()
def prob_768(net, x, mode):
    """x is the model's input image at 768. Returns a 768 probability map."""
    if mode == "full768":
        return torch.sigmoid(net(x).float())
    if mode == "whole384":
        d = F.interpolate(x, size=(TRAIN_RES,)*2, mode="bilinear", align_corners=False)
        p = torch.sigmoid(net(d).float())
        return F.interpolate(p, size=(FULL_RES,)*2, mode="bilinear", align_corners=False)
    # genuinely overlapping: stride 192 gives 3x3 tiles at 50% overlap.
    # Origins (0, 384) alone would ABUT, not overlap, and blending would
    # be a no-op -- the seam is exactly where a tiled model is weakest.
    acc = torch.zeros(x.shape[0], 1, FULL_RES, FULL_RES, device=x.device)
    cnt = torch.zeros_like(acc)
    ORIG = (0, (FULL_RES-TRAIN_RES)//2, FULL_RES-TRAIN_RES)
    for oy in ORIG:
        for ox in ORIG:
            t = x[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES]
            acc[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += torch.sigmoid(net(t).float())
            cnt[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += 1
    return acc/cnt.clamp(min=1)                      # blend, threshold afterwards

THS = [round(t, 2) for t in np.arange(0.10, 0.91, 0.05)]
@torch.no_grad()
def evaluate(net, mode):
    net.eval()
    nI = len(CLEAN_VAL); per = np.zeros((nI, len(THS))); P = np.zeros((nI, len(THS)))
    Ipx = np.zeros((nI, len(THS))); G = np.zeros(nI); k = 0
    for x, y in val_loader:
        x, y = x.to(DEV, non_blocking=True), y.to(DEV, non_blocking=True)
        with torch.amp.autocast("cuda", enabled=DEV.type == "cuda"):
            p = prob_768(net, x, mode)
        b = x.shape[0]; yf = y.view(b, -1)
        G[k:k+b] = yf.sum(1).cpu().numpy()
        pf = p.view(b, -1)
        for j, t in enumerate(THS):
            pr = (pf > t).float()
            it = (pr*yf).sum(1); pp = pr.sum(1)
            Ipx[k:k+b, j] = it.cpu().numpy(); P[k:k+b, j] = pp.cpu().numpy()
            per[k:k+b, j] = (it/(pp+yf.sum(1)-it).clamp(min=1)).cpu().numpy()
        k += b
    pooled = Ipx.sum(0)/np.maximum(P.sum(0)+G.sum()-Ipx.sum(0), 1)
    j = int(np.argmax(pooled)); t = THS[j]
    byq = {q: float(per[QV == q, j].mean()) for q in range(4)}
    return dict(th=t, pooled=float(pooled[j]), perimg=float(per[:, j].mean()),
                byq=byq, pred_frac=float(P[:, j].sum()/(len(CLEAN_VAL)*FULL_RES*FULL_RES)))

# --------------------------- resume or start ---------------------------
model = ResUNet().to(DEV)
opt = optim.AdamW([{"params": list(model.enc_params()), "lr": LR_ENCODER},
                   {"params": list(model.dec_params()), "lr": LR_DECODER}],
                  weight_decay=WEIGHT_DECAY)
sched = optim.lr_scheduler.ReduceLROnPlateau(opt, mode="max", patience=LR_PATIENCE, factor=0.5)
scaler = torch.amp.GradScaler("cuda")
step, best, history, shares = 0, -1.0, [], []
CK, BEST = f"{OUT}/ckpt.pt", f"{OUT}/best.pt"
if os.path.exists(CK):
    c = torch.load(CK, map_location=DEV, weights_only=False)
    model.load_state_dict(c["model"]); opt.load_state_dict(c["opt"])
    sched.load_state_dict(c["sched"]); scaler.load_state_dict(c["scaler"])
    step, best, history, shares = c["step"], c["best"], c["history"], c.get("shares", [])
    random.setstate(c["py_rng"]); np.random.set_state(c["np_rng"])
    torch.set_rng_state(c["t_rng"].cpu())
    if c.get("cuda_rng") is not None and DEV.type == "cuda":
        torch.cuda.set_rng_state(c["cuda_rng"].cpu())
    print(f"RESUMED from step {step} (best {best:.4f})")
else:
    print("starting fresh")

def save_ckpt():
    torch.save(dict(model=model.state_dict(), opt=opt.state_dict(), sched=sched.state_dict(),
                    scaler=scaler.state_dict(), step=step, best=best, history=history,
                    shares=shares, py_rng=random.getstate(), np_rng=np.random.get_state(),
                    t_rng=torch.get_rng_state(),
                    cuda_rng=torch.cuda.get_rng_state() if DEV.type == "cuda" else None), CK)

# ------------------------------- train ---------------------------------
it = cycle(loader_l)
print("=" * 76); print(f"{RUN}  --  {len(LAB)} labelled, {MAX_STEPS} steps"); print("=" * 76)
t0 = t_last = time.time()
r_loss = r_n = 0.0; q_les = np.zeros(4); q_cnt = np.zeros(4)
while step < MAX_STEPS:
    model.train()
    x, y, q = next(it)
    x, y = x.to(DEV, non_blocking=True), y.to(DEV, non_blocking=True)
    opt.zero_grad(set_to_none=True)
    with torch.amp.autocast("cuda", enabled=DEV.type == "cuda"):
        lg = model(x)
        vec, lesion = sup_loss(lg.float(), y)
        loss = vec.mean()
    scaler.scale(loss).backward(); scaler.step(opt); scaler.update()
    step += 1
    r_loss += loss.item(); r_n += 1
    lz = lesion.detach().float().cpu().numpy(); qz = q.numpy()
    for b in range(len(qz)): q_les[qz[b]] += lz[b]; q_cnt[qz[b]] += 1

    if step % SHARE_EVERY == 0:
        tot = max(q_les.sum(), 1e-12)
        sh = (q_les/tot).round(4).tolist()
        shares.append(dict(step=step, share=sh, seen=q_cnt.astype(int).tolist()))
        print(f"   [step {step}] lesion-loss share by quartile {sh}"
              f"  (images seen {q_cnt.astype(int).tolist()})")
        q_les[:] = 0; q_cnt[:] = 0

    if step % EVAL_EVERY == 0:
        m = evaluate(model, SELECT_ON)
        prev = opt.param_groups[-1]["lr"]; sched.step(m["pooled"])
        cur = opt.param_groups[-1]["lr"]
        saved = ""
        if m["pooled"] > best:
            best = m["pooled"]; torch.save(model.state_dict(), BEST); saved = "  <-- saved"
        history.append(dict(step=step, train_loss=r_loss/max(r_n, 1), lr=cur, **m))
        sps = EVAL_EVERY/(time.time()-t_last); t_last = time.time()
        print(f"step {step:6d} | loss {r_loss/max(r_n,1):.4f} | val pooled {m['pooled']:.4f} "
              f"@{m['th']} | per-img {m['perimg']:.4f} | Q0 {m['byq'][0]:.4f} "
              f"Q3 {m['byq'][3]:.4f} | best {best:.4f} | {sps:.2f} it/s | "
              f"ETA {(MAX_STEPS-step)/sps/3600:.1f}h{saved}"
              + (f"  [lr {prev:.1e}->{cur:.1e}]" if cur != prev else ""))
        r_loss = r_n = 0.0
        json.dump(dict(run=RUN, input=INPUT_MODE, loss=FG_LOSS, jitter=JITTER,
                       seed=TRAIN_SEED, floor=FLOOR_FRAC, select_on=SELECT_ON,
                       batch=BATCH, history=history, shares=shares),
                  open(f"{OUT}/history.json", "w"), indent=1)
    if step % CKPT_EVERY == 0:
        save_ckpt(); print(f"    checkpointed at step {step}")
save_ckpt()
print(f"\ntraining done in {(time.time()-t0)/3600:.2f}h, best val pooled {best:.4f}")

# ------------------- final: all three inference modes ------------------
net = ResUNet().to(DEV); net.load_state_dict(torch.load(BEST, map_location=DEV)); net.eval()
print("\n" + "=" * 76)
print("clean_val under each inference mode (best.pt was selected on "
      f"{SELECT_ON})")
print("=" * 76)
print(f"{'mode':12}{'th':>6}{'pooled':>9}{'per-img':>9}"
      f"{'Q0':>8}{'Q1':>8}{'Q2':>8}{'Q3':>8}{'Q3-Q0':>8}")
modes = ["whole384", "full768", "tiles384"]
final = {}
for md in modes:
    m = evaluate(net, md); final[md] = m
    print(f"{md:12}{m['th']:>6.2f}{m['pooled']:>9.4f}{m['perimg']:>9.4f}"
          + "".join(f"{m['byq'][q]:>8.4f}" for q in range(4))
          + f"{m['byq'][3]-m['byq'][0]:>8.4f}")

# the stratified check: does any Q0 gain track the realized detail gain?
res = {k: tuple(int(v) for v in meta[k]["Resolution"].lower().split("x")[:2])
       for k in CLEAN_VAL}
gain = np.array([min(min(res[s]), FULL_RES)/TRAIN_RES for s in CLEAN_VAL])
q0 = QV == 0
print(f"\nclean_val Q0 split by realized worst-axis detail gain at {FULL_RES}:")
for lab, sel in (("<= 1.0 (no gain)", q0 & (gain <= 1.0)),
                 ("1.0 - 1.5", q0 & (gain > 1.0) & (gain < 1.5)),
                 (">= 1.5", q0 & (gain >= 1.5))):
    print(f"   {lab:18} n={int(sel.sum()):4d}")
print("   per-image Q0 IoU within each band is in the saved json; compare it")
print("   against the matching S-384 run, paired by image.")

np.save(f"{OUT}/clean_val_gain.npy", gain)
json.dump(dict(run=RUN, input=INPUT_MODE, loss=FG_LOSS, jitter=JITTER, seed=TRAIN_SEED,
               floor=FLOOR_FRAC, select_on=SELECT_ON, batch=BATCH, best_val_pooled=best,
               clean_val=CLEAN_VAL, quartiles=QV.tolist(), final=final,
               history=history, shares=shares),
          open(f"{OUT}/summary.json", "w"), indent=1)
xs = [h["step"] for h in history]
plt.figure(figsize=(14, 4))
plt.subplot(1, 3, 1); plt.plot(xs, [h["train_loss"] for h in history]); plt.title("train loss")
plt.subplot(1, 3, 2); plt.plot(xs, [h["pooled"] for h in history]); plt.title("val pooled IoU")
plt.subplot(1, 3, 3)
for q in range(4): plt.plot(xs, [h["byq"][q] for h in history], label=f"Q{q}")
plt.legend(fontsize=7); plt.title("val per-image IoU by quartile")
plt.tight_layout(); plt.savefig(f"{OUT}/curves.png", dpi=110); plt.close()
print(f"\nwrote {OUT}/summary.json, best.pt, curves.png")
print(">>> download best.pt to local disk before the session ends <<<")
```

```python
# =====================================================================
#  cell F -- run this ONCE after every cell D training. Standalone: it
#  reads nothing from cell D's variables, so it also works in a fresh
#  session or after a runtime restart.
#
#  For the target run it:
#    1. scores it per image on clean_val  -> per_image.npz
#       (the vectors every paired test needs; cell D saves only means)
#    2. archives summary / history / curves / gain / per_image to Drive
#    3. records config, GPU and a checksum in weights_manifest.json
#    4. deletes ckpt.pt, but ONLY if the run completed. 490 MB each, and
#       19 of them would overrun a free Drive.
#    5. prints what Drive is holding
#    6. downloads the archive and best.pt
#
#  Any OTHER run still missing per_image.npz or an archive is also
#  brought up to date, without being downloaded again.
# =====================================================================
import os, csv, json, glob, time, zipfile, shutil, hashlib
import numpy as np
from PIL import Image
import torch, torch.nn as nn, torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models
import torchvision.transforms.functional as TF

RUN_NAME    = None       # None = the most recently finished run in stage2/
DELETE_CKPT = True       # False if you intend to extend a run later
DOWNLOAD    = True
SAVE, C768  = "/content/drive/MyDrive", "/content/cache768"
STAGE, ARCH = f"{SAVE}/stage2", f"{SAVE}/stage2_archive"
TRAIN_RES, FULL_RES = 384, 768
BATCH_EVAL  = 12 if torch.cuda.is_available() else 2
FROZEN      = [0.059468, 0.138448, 0.300831]
try:
    from google.colab import drive; drive.mount("/content/drive", force_remount=False)
except Exception: pass
DEV = torch.device("cuda" if torch.cuda.is_available() else "cpu")
os.makedirs(ARCH, exist_ok=True)

# ------------------------------- context -------------------------------
SP   = json.load(open(f"{SAVE}/plantseg_splits.json"))
I768 = json.load(open(f"{C768}/index.json"))
row  = I768["row_of"]
CLEAN_VAL = sorted(set(SP["clean_val"]) & set(I768["stems"]))
DSDIR = "/content/my_images/plantseg"
meta = {os.path.splitext(r["Name"])[0]: r for r in csv.DictReader(
        open(f"{DSDIR}/Metadata.csv", encoding="utf-8-sig", newline=""))}
frac = {k: float(v["Mask ratio"]) for k, v in meta.items()}
def quart(s):
    f = frac[s]
    return 0 if f <= FROZEN[0] else 1 if f <= FROZEN[1] else 2 if f <= FROZEN[2] else 3
QV = np.array([quart(s) for s in CLEAN_VAL])

MEAN, STD = [0.485,0.456,0.406], [0.229,0.224,0.225]
NORM = transforms.Normalize(MEAN, STD)
_mm = {}
def cache():
    if not _mm:
        _mm["i"] = np.load(I768["img_path"], mmap_mode="r")
        _mm["m"] = np.load(I768["msk_path"], mmap_mode="r")
    return _mm["i"], _mm["m"]
def strip_detail(im):
    return im.resize((TRAIN_RES,)*2, Image.BILINEAR).resize((FULL_RES,)*2, Image.BILINEAR)

class EvalSet(Dataset):
    def __init__(s, st, up): s.st, s.up = list(st), up
    def __len__(s): return len(s.st)
    def __getitem__(s, i):
        I, M = cache(); r = row[s.st[i]]
        img = Image.fromarray(I[r])
        if s.up: img = strip_detail(img)
        y = torch.from_numpy((np.asarray(M[r]) > 0).astype(np.float32)).unsqueeze(0)
        return NORM(TF.to_tensor(img)), y

SPEC = dict(channels=(64,256,512,1024,2048), decoder=(512,256,128,64))
class DoubleConv(nn.Module):
    def __init__(s,i,o):
        super().__init__()
        s.net = nn.Sequential(nn.Conv2d(i,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True),
                              nn.Conv2d(o,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True))
    def forward(s,x): return s.net(x)
class Up(nn.Module):
    def __init__(s,i,k,o):
        super().__init__(); s.reduce=nn.Conv2d(i,i//2,1,bias=False); s.conv=DoubleConv(i//2+k,o)
    def forward(s,x,skip):
        x = F.interpolate(x, size=skip.shape[-2:], mode="bilinear", align_corners=False)
        return s.conv(torch.cat([s.reduce(x), skip], 1))
class ResUNet(nn.Module):
    def __init__(s):
        super().__init__()
        base = models.resnet50(weights=None)
        c0,c1,c2,c3,c4 = SPEC["channels"]; d1,d2,d3,d4 = SPEC["decoder"]
        L = list(base.children())
        s.stem=nn.Sequential(*L[:3]); s.pool=L[3]
        s.l1,s.l2,s.l3,s.l4 = L[4],L[5],L[6],L[7]
        s.u1,s.u2,s.u3,s.u4 = Up(c4,c3,d1),Up(d1,c2,d2),Up(d2,c1,d3),Up(d3,c0,d4)
        s.head = nn.Sequential(nn.Conv2d(d4,32,3,padding=1,bias=False),
                               nn.BatchNorm2d(32), nn.ReLU(True), nn.Conv2d(32,1,1))
    def forward(s,x):
        size=x.shape[-2:]
        x0=s.stem(x); x1=s.l1(s.pool(x0)); x2=s.l2(x1); x3=s.l3(x2); x4=s.l4(x3)
        d=s.u1(x4,x3); d=s.u2(d,x2); d=s.u3(d,x1); d=s.u4(d,x0)
        return s.head(F.interpolate(d,size=size,mode="bilinear",align_corners=False))

@torch.no_grad()
def prob_768(net, x, mode):
    if mode == "full768":
        return torch.sigmoid(net(x).float())
    if mode == "whole384":
        d = F.interpolate(x, size=(TRAIN_RES,)*2, mode="bilinear", align_corners=False)
        p = torch.sigmoid(net(d).float())
        return F.interpolate(p, size=(FULL_RES,)*2, mode="bilinear", align_corners=False)
    acc = torch.zeros(x.shape[0], 1, FULL_RES, FULL_RES, device=x.device)
    cnt = torch.zeros_like(acc)
    ORIG = (0, (FULL_RES-TRAIN_RES)//2, FULL_RES-TRAIN_RES)
    for oy in ORIG:
        for ox in ORIG:
            t = x[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES]
            acc[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += torch.sigmoid(net(t).float())
            cnt[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += 1
    return acc/cnt.clamp(min=1)

THS = [round(t, 2) for t in np.arange(0.10, 0.91, 0.05)]
@torch.no_grad()
def score_per_image(bp, mode, up):
    net = ResUNet().to(DEV); net.load_state_dict(torch.load(bp, map_location=DEV)); net.eval()
    dl = DataLoader(EvalSet(CLEAN_VAL, up), BATCH_EVAL, shuffle=False,
                    num_workers=min(2, os.cpu_count() or 1), pin_memory=DEV.type == "cuda")
    n = len(CLEAN_VAL)
    per = np.zeros((n, len(THS))); Ipx = np.zeros((n, len(THS)))
    P = np.zeros((n, len(THS))); G = np.zeros(n); k = 0
    for x, y in dl:
        x, y = x.to(DEV, non_blocking=True), y.to(DEV, non_blocking=True)
        with torch.amp.autocast("cuda", enabled=DEV.type == "cuda"):
            p = prob_768(net, x, mode)
        b = x.shape[0]; yf = y.view(b, -1); pf = p.view(b, -1)
        G[k:k+b] = yf.sum(1).cpu().numpy()
        for j, t in enumerate(THS):
            pr = (pf > t).float(); it = (pr*yf).sum(1); pp = pr.sum(1)
            Ipx[k:k+b, j] = it.cpu().numpy(); P[k:k+b, j] = pp.cpu().numpy()
            per[k:k+b, j] = (it/(pp+yf.sum(1)-it).clamp(min=1)).cpu().numpy()
        k += b
    pooled = Ipx.sum(0)/np.maximum(P.sum(0)+G.sum()-Ipx.sum(0), 1)
    j = int(np.argmax(pooled))
    del net; torch.cuda.empty_cache()
    return dict(th=THS[j], pooled=float(pooled[j]), per=per[:, j])

def sha(p, chunk=1 << 24):
    h = hashlib.sha256()
    with open(p, "rb") as f:
        while (b := f.read(chunk)): h.update(b)
    return h.hexdigest()[:16]

# -------------------------- which runs to handle -----------------------
# summary.json is written only at the END of cell D, so its presence is
# the completion test -- safer than trusting a step count.
done = sorted(glob.glob(f"{STAGE}/*/summary.json"), key=os.path.getmtime)
if not done:
    raise SystemExit(f"no completed run under {STAGE}")
target = RUN_NAME or os.path.basename(os.path.dirname(done[-1]))
print(f"target run: {target}   ({len(done)} completed run(s) on Drive)\n")

man_p = f"{SAVE}/weights_manifest.json"
man = json.load(open(man_p)) if os.path.exists(man_p) else {}

for sm in done:
    d = os.path.dirname(sm); name = os.path.basename(d)
    J = json.load(open(sm))
    bp, ck = f"{d}/best.pt", f"{d}/ckpt.pt"
    if not os.path.exists(bp):
        print(f"{name}: no best.pt, skipped"); continue
    is_target = (name == target)

    # 1. per-image vectors
    npz = f"{d}/per_image.npz"
    need = True
    if os.path.exists(npz):
        try:
            z = np.load(npz, allow_pickle=True); need = list(z["stems"]) != CLEAN_VAL
        except Exception: need = True
    if need:
        t = time.time()
        r = score_per_image(bp, J["select_on"], J["input"] == "crop_768_up")
        np.savez(npz, per=r["per"], th=r["th"], pooled=r["pooled"],
                 mode=J["select_on"], stems=np.array(CLEAN_VAL))
        print(f"{name}: scored per image in {time.time()-t:.0f}s  "
              f"(mode {J['select_on']}, th {r['th']}, pooled {r['pooled']:.4f})")
    else:
        print(f"{name}: per_image.npz already current")

    # 2. archive
    zp = f"{ARCH}/{name}_results.zip"
    if is_target or not os.path.exists(zp):
        with zipfile.ZipFile(zp, "w", zipfile.ZIP_DEFLATED) as z:
            for f in ("summary.json", "history.json", "curves.png",
                      "clean_val_gain.npy", "per_image.npz"):
                p = f"{d}/{f}"
                if os.path.exists(p): z.write(p, arcname=f"{name}/{f}")
        print(f"{name}: archive {os.path.getsize(zp)/1e6:.2f} MB")

    # 3. manifest
    if is_target or name not in man:
        man[name] = dict(file=f"{name}__best.pt", bytes=os.path.getsize(bp),
                         sha256_16=sha(bp), best_val_pooled=J.get("best_val_pooled"),
                         gpu=J.get("gpu", "not recorded"), batch=J.get("batch"),
                         input=J["input"], loss=J["loss"], jitter=J["jitter"],
                         seed=J["seed"], floor=J.get("floor"))

    # 4. prune -- only a COMPLETED run, which summary.json proves
    if DELETE_CKPT and os.path.exists(ck):
        mb = os.path.getsize(ck)/1e6
        os.remove(ck); print(f"{name}: removed ckpt.pt (freed {mb:.0f} MB)")

json.dump(man, open(man_p, "w"), indent=1)
print(f"\nmanifest: {man_p}  ({len(man)} run(s))")

# ------------------------------ drive usage ----------------------------
def dsz(p):
    return sum(os.path.getsize(os.path.join(r, f))
               for r, _, fs in os.walk(p) for f in fs) if os.path.isdir(p) else 0
tot = 0
print("\nDrive usage")
for lab, p in (("stage2", STAGE), ("stage2_archive", ARCH),
               ("runs (stage 1)", f"{SAVE}/runs"), ("cache768", f"{SAVE}/cache768")):
    s = dsz(p); tot += s
    print(f"   {lab:18}{s/1e9:7.2f} GB")
print(f"   {'TRACKED TOTAL':18}{tot/1e9:7.2f} GB"
      + ("   !! approaching a 15 GB free Drive" if tot/1e9 > 11 else ""))

# ------------------------------- download ------------------------------
if DOWNLOAD:
    d = f"{STAGE}/{target}"
    loc = f"/content/{target}__best.pt"
    if not os.path.exists(loc): shutil.copyfile(f"{d}/best.pt", loc)
    print(f"\n{target}  best.pt {os.path.getsize(loc)/1e6:.1f} MB  "
          f"sha256[:16] {man[target]['sha256_16']}")
    try:
        from google.colab import files
        files.download(f"{ARCH}/{target}_results.zip")
        time.sleep(2)
        files.download(loc)
        print("downloads queued. Verify locally:")
        print(f"   shasum -a 256 {target}__best.pt | cut -c1-16"
              f"   ->  {man[target]['sha256_16']}")
    except Exception as e:
        print(f"(browser download unavailable: {type(e).__name__})")
    print(f"also on Drive:\n   {ARCH}/{target}_results.zip\n   {d}/best.pt")
```

```python
# =====================================================================
#  cell E -- score every finished Stage 2 run per image, then compare.
#
#  Cell D saves per-quartile MEANS, which cannot support a paired test.
#  This reloads each best.pt and writes the per-image IoU vector over
#  clean_val, at that run's own selected mode and threshold. No retraining.
#
#  Then, against the S-384 baseline, for each other run:
#    - paired per-image differences over the same 688 images
#    - an image-level paired bootstrap on Q0  (the PRIMARY inference:
#      688 paired images carry far more power than a 3-seed t-interval,
#      which at 3 seeds needs a seed-sd below 0.008 to clear +0.02)
#    - the same split by realized detail gain, declared in advance
#    - the PILOT rule: kill only if the mean Q0 difference is <= 0.
#      Killing anything that misses the +0.02 gate at one seed discards
#      a real +0.02 effect half the time; the <= 0 rule, 10% of the time.
#
#  A single-seed result here is a FILTER, never a result. Nothing from
#  this cell goes in the paper without seeds 43 and 44.
# =====================================================================
import os, csv, json, glob, time
import numpy as np
from PIL import Image
import torch, torch.nn as nn, torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models
import torchvision.transforms.functional as TF

SAVE, C768 = "/content/drive/MyDrive", "/content/cache768"
STAGE   = f"{SAVE}/stage2"
BASE    = "S-384_seed42"           # the baseline every run is paired against
TRAIN_RES, FULL_RES = 384, 768
BATCH_EVAL = 4
NBOOT   = 10000
FROZEN  = [0.059468, 0.138448, 0.300831]
KILL_AT = 0.0                      # pilot rule, declared before looking
try:
    from google.colab import drive; drive.mount("/content/drive", force_remount=False)
except Exception: pass
DEV = torch.device("cuda" if torch.cuda.is_available() else "cpu")

SP = json.load(open(f"{SAVE}/plantseg_splits.json"))
I768 = json.load(open(f"{C768}/index.json"))
row = I768["row_of"]
CLEAN_VAL = sorted(set(SP["clean_val"]) & set(I768["stems"]))
DSDIR = "/content/my_images/plantseg"
meta = {os.path.splitext(r["Name"])[0]: r for r in csv.DictReader(
        open(f"{DSDIR}/Metadata.csv", encoding="utf-8-sig", newline=""))}
frac = {k: float(v["Mask ratio"]) for k, v in meta.items()}
def quart(s):
    f = frac[s]
    return 0 if f <= FROZEN[0] else 1 if f <= FROZEN[1] else 2 if f <= FROZEN[2] else 3
QV = np.array([quart(s) for s in CLEAN_VAL])
res = {k: tuple(int(v) for v in meta[k]["Resolution"].lower().split("x")[:2])
       for k in CLEAN_VAL}
GAIN = np.array([min(min(res[s]), FULL_RES)/TRAIN_RES for s in CLEAN_VAL])
print(f"clean_val {len(CLEAN_VAL)}  quartiles {[int((QV==q).sum()) for q in range(4)]}")

MEAN, STD = [0.485,0.456,0.406], [0.229,0.224,0.225]
NORM = transforms.Normalize(MEAN, STD)
_mm = {}
def cache():
    if not _mm:
        _mm["i"] = np.load(I768["img_path"], mmap_mode="r")
        _mm["m"] = np.load(I768["msk_path"], mmap_mode="r")
    return _mm["i"], _mm["m"]
def strip_detail(im):
    return im.resize((TRAIN_RES,)*2, Image.BILINEAR).resize((FULL_RES,)*2, Image.BILINEAR)

class EvalSet(Dataset):
    def __init__(s, st, up): s.st, s.up = list(st), up
    def __len__(s): return len(s.st)
    def __getitem__(s, i):
        I, M = cache(); r = row[s.st[i]]
        img = Image.fromarray(I[r])
        if s.up: img = strip_detail(img)
        y = torch.from_numpy((np.asarray(M[r]) > 0).astype(np.float32)).unsqueeze(0)
        return NORM(TF.to_tensor(img)), y

SPEC = dict(channels=(64,256,512,1024,2048), decoder=(512,256,128,64))
class DoubleConv(nn.Module):
    def __init__(s,i,o):
        super().__init__()
        s.net = nn.Sequential(nn.Conv2d(i,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True),
                              nn.Conv2d(o,o,3,1,1,bias=False), nn.BatchNorm2d(o), nn.ReLU(True))
    def forward(s,x): return s.net(x)
class Up(nn.Module):
    def __init__(s,i,k,o):
        super().__init__(); s.reduce=nn.Conv2d(i,i//2,1,bias=False); s.conv=DoubleConv(i//2+k,o)
    def forward(s,x,skip):
        x = F.interpolate(x, size=skip.shape[-2:], mode="bilinear", align_corners=False)
        return s.conv(torch.cat([s.reduce(x), skip], 1))
class ResUNet(nn.Module):
    def __init__(s):
        super().__init__()
        base = models.resnet50(weights=None)
        c0,c1,c2,c3,c4 = SPEC["channels"]; d1,d2,d3,d4 = SPEC["decoder"]
        L = list(base.children())
        s.stem=nn.Sequential(*L[:3]); s.pool=L[3]
        s.l1,s.l2,s.l3,s.l4 = L[4],L[5],L[6],L[7]
        s.u1,s.u2,s.u3,s.u4 = Up(c4,c3,d1),Up(d1,c2,d2),Up(d2,c1,d3),Up(d3,c0,d4)
        s.head = nn.Sequential(nn.Conv2d(d4,32,3,padding=1,bias=False),
                               nn.BatchNorm2d(32), nn.ReLU(True), nn.Conv2d(32,1,1))
    def forward(s,x):
        size=x.shape[-2:]
        x0=s.stem(x); x1=s.l1(s.pool(x0)); x2=s.l2(x1); x3=s.l3(x2); x4=s.l4(x3)
        d=s.u1(x4,x3); d=s.u2(d,x2); d=s.u3(d,x1); d=s.u4(d,x0)
        return s.head(F.interpolate(d,size=size,mode="bilinear",align_corners=False))

@torch.no_grad()
def prob_768(net, x, mode):
    if mode == "full768":
        return torch.sigmoid(net(x).float())
    if mode == "whole384":
        d = F.interpolate(x, size=(TRAIN_RES,)*2, mode="bilinear", align_corners=False)
        p = torch.sigmoid(net(d).float())
        return F.interpolate(p, size=(FULL_RES,)*2, mode="bilinear", align_corners=False)
    acc = torch.zeros(x.shape[0], 1, FULL_RES, FULL_RES, device=x.device)
    cnt = torch.zeros_like(acc)
    ORIG = (0, (FULL_RES-TRAIN_RES)//2, FULL_RES-TRAIN_RES)
    for oy in ORIG:
        for ox in ORIG:
            t = x[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES]
            acc[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += torch.sigmoid(net(t).float())
            cnt[:, :, oy:oy+TRAIN_RES, ox:ox+TRAIN_RES] += 1
    return acc/cnt.clamp(min=1)

THS = [round(t, 2) for t in np.arange(0.10, 0.91, 0.05)]
@torch.no_grad()
def score(bp, mode, up):
    net = ResUNet().to(DEV); net.load_state_dict(torch.load(bp, map_location=DEV)); net.eval()
    dl = DataLoader(EvalSet(CLEAN_VAL, up), BATCH_EVAL, shuffle=False,
                    num_workers=min(2, os.cpu_count() or 1), pin_memory=DEV.type == "cuda")
    n = len(CLEAN_VAL)
    per = np.zeros((n, len(THS))); Ipx = np.zeros((n, len(THS)))
    P = np.zeros((n, len(THS))); G = np.zeros(n); k = 0
    for x, y in dl:
        x, y = x.to(DEV, non_blocking=True), y.to(DEV, non_blocking=True)
        with torch.amp.autocast("cuda", enabled=DEV.type == "cuda"):
            p = prob_768(net, x, mode)
        b = x.shape[0]; yf = y.view(b, -1); pf = p.view(b, -1)
        G[k:k+b] = yf.sum(1).cpu().numpy()
        for j, t in enumerate(THS):
            pr = (pf > t).float(); it = (pr*yf).sum(1); pp = pr.sum(1)
            Ipx[k:k+b, j] = it.cpu().numpy(); P[k:k+b, j] = pp.cpu().numpy()
            per[k:k+b, j] = (it/(pp+yf.sum(1)-it).clamp(min=1)).cpu().numpy()
        k += b
    pooled = Ipx.sum(0)/np.maximum(P.sum(0)+G.sum()-Ipx.sum(0), 1)
    j = int(np.argmax(pooled))
    del net; torch.cuda.empty_cache()
    return dict(th=THS[j], pooled=float(pooled[j]), per=per[:, j])

# ------------------------- score what is finished ----------------------
runs = {}
for sm in sorted(glob.glob(f"{STAGE}/*/summary.json")):
    d = os.path.dirname(sm); name = os.path.basename(d)
    if not os.path.exists(f"{d}/best.pt"):
        print(f"   {name}: no best.pt, skipped"); continue
    J = json.load(open(sm))
    npz = f"{d}/per_image.npz"
    if os.path.exists(npz):
        z = np.load(npz, allow_pickle=True)
        if list(z["stems"]) == CLEAN_VAL:
            runs[name] = dict(per=z["per"], th=float(z["th"]), pooled=float(z["pooled"]),
                              mode=str(z["mode"]), cfg=J); print(f"   {name}: cached"); continue
    t = time.time()
    r = score(f"{d}/best.pt", J["select_on"], J["input"] == "crop_768_up")
    np.savez(npz, per=r["per"], th=r["th"], pooled=r["pooled"],
             mode=J["select_on"], stems=np.array(CLEAN_VAL))
    runs[name] = dict(**r, mode=J["select_on"], cfg=J)
    print(f"   {name}: scored in {time.time()-t:.0f}s  "
          f"mode {J['select_on']} th {r['th']} pooled {r['pooled']:.4f}")

if BASE not in runs:
    raise SystemExit(f"baseline {BASE} not scored yet -- run it first")
print(f"\n{len(runs)} run(s) scored; baseline {BASE}")

# --------------------------- paired comparison -------------------------
rng = np.random.default_rng(0)
b_per = runs[BASE]["per"]; q0 = QV == 0
bands = {"Q0 all": q0, "Q0 gain<=1.0": q0 & (GAIN <= 1.0),
         "Q0 gain>=1.5": q0 & (GAIN >= 1.5), "all images": np.ones(len(QV), bool)}

def boot(d, sel):
    v = d[sel]
    if len(v) < 5: return float("nan"), float("nan"), float("nan")
    idx = rng.integers(0, len(v), (NBOOT, len(v)))
    bs = v[idx].mean(1)
    return float(v.mean()), float(np.percentile(bs, 2.5)), float(np.percentile(bs, 97.5))

print("\n" + "=" * 88)
print(f"paired per-image IoU difference vs {BASE}, clean_val, "
      f"{NBOOT} image-level bootstraps")
print("=" * 88)
out = {}
for name, r in runs.items():
    if name == BASE: continue
    d = r["per"] - b_per
    print(f"\n{name}   ({r['cfg']['input']} / {r['cfg']['loss']}"
          f"{' / jitter' if r['cfg']['jitter'] else ''})   "
          f"pooled {r['pooled']:.4f} vs {runs[BASE]['pooled']:.4f}")
    print(f"   {'band':16}{'n':>6}{'mean diff':>12}{'95% CI':>22}")
    row = {}
    for lab, sel in bands.items():
        m, lo, hi = boot(d, sel)
        star = "  *" if (lo > 0 or hi < 0) else ""
        print(f"   {lab:16}{int(sel.sum()):>6}{m:>12.4f}"
              f"   [{lo:+.4f}, {hi:+.4f}]{star}")
        row[lab] = dict(n=int(sel.sum()), mean=m, lo=lo, hi=hi)
    mq0 = row["Q0 all"]["mean"]
    verdict = ("KILL -- Q0 moves the wrong way" if mq0 <= KILL_AT else
               "KEEP -- run seeds 43 and 44")
    print(f"   PILOT ({'<=' if mq0 <= KILL_AT else '>'} {KILL_AT:+.2f}): {verdict}")
    if row["Q0 gain>=1.5"]["n"] >= 20 and row["Q0 gain<=1.0"]["n"] >= 20:
        sep = row["Q0 gain>=1.5"]["mean"] - row["Q0 gain<=1.0"]["mean"]
        print(f"   mechanism check: high-detail minus no-detail = {sep:+.4f}"
              f"   {'(consistent with resolution)' if sep > 0 else '(not resolution)'}")
        row["mechanism_sep"] = sep
    out[name] = row

print("\n" + "=" * 88)
print("* marks a CI excluding zero. That is an image-level interval: it says the")
print("difference is not sampling noise over these 688 images, NOT that it would")
print("survive another training seed. Only seeds 43 and 44 answer that, and the")
print("gate is judged on the three-seed paired mean, not on anything here.")
json.dump(dict(baseline=BASE, kill_at=KILL_AT, nboot=NBOOT, results=out),
          open(f"{SAVE}/stage2_compare.json", "w"), indent=1, default=float)
print(f"\nwrote {SAVE}/stage2_compare.json")
```

```python
# =====================================================================
#  cell H -- re-analyse Stage 2 on PLATEAU MEANS, not the selected
#  checkpoint. CPU only, seconds, no retraining. Reads the history
#  already inside each stage2/<RUN>/summary.json.
#
#  Why.  best.pt is selected by maximising POOLED val IoU, which is
#  dominated by dense images and is nearly flat across steps 6k-25k.
#  Q0 meanwhile climbs through training. So the Q0 you read off the
#  selected checkpoint depends on which arbitrary point of a flat
#  pooled curve happened to win -- S-384_seed42 selected at step 3000
#  and S-384_seed43 at step 15000, and their Q0 differs by +0.0279
#  with nothing but the seed changed.
#
#  The plateau mean averages ~20 evaluations per run, so that
#  selection noise mostly cancels. It is a DIFFERENT estimator of the
#  same quantity, declared here, reported alongside the primary
#  selected-checkpoint numbers, never instead of them.
#
#  Everything is paired WITHIN SEED. A config seed with no matching
#  baseline seed is listed as unpaired and excluded from the gate.
# =====================================================================
import os, json, glob
import numpy as np
from math import sqrt

SAVE     = "/content/drive/MyDrive"
STAGE    = f"{SAVE}/stage2"
BASE_CFG = "S-384"
PLATEAU  = 6000            # average evals at steps >= this
GATE_Q0  = 0.02
TCRIT    = {2: 4.303, 3: 3.182, 4: 2.776, 5: 2.571}
try:
    from google.colab import drive; drive.mount("/content/drive", force_remount=False)
except Exception: pass

def qkey(byq, q):
    """history/final store byq with int keys in python but str keys in json."""
    return float(byq[str(q)] if str(q) in byq else byq[q])

runs = {}
for sm in sorted(glob.glob(f"{STAGE}/*/summary.json")):
    d = json.load(open(sm)); name = d["run"]
    hist = [h for h in d["history"] if h["step"] >= PLATEAU]
    if not hist:
        print(f"!! {name}: no evals at step >= {PLATEAU}, skipped"); continue
    sel  = d["final"][d["select_on"]]
    best = max(d["history"], key=lambda h: h["pooled"])
    runs[name] = dict(
        cfg=d["input"], loss=d["loss"], jitter=d["jitter"], seed=d["seed"],
        sel_step=best["step"], n_plateau=len(hist),
        sel_q0=qkey(sel["byq"], 0), sel_q3=qkey(sel["byq"], 3),
        sel_pooled=float(sel["pooled"]),
        pl_q0=float(np.mean([qkey(h["byq"], 0) for h in hist])),
        pl_q0_sd=float(np.std([qkey(h["byq"], 0) for h in hist], ddof=1)),
        pl_q3=float(np.mean([qkey(h["byq"], 3) for h in hist])),
        pl_pooled=float(np.mean([h["pooled"] for h in hist])))
if not runs:
    raise SystemExit("no summary.json found")

def tag(r):
    t = {"full_384": "S-384", "crop_768": "S-768c", "crop_768_up": "S-768up"}[r["cfg"]]
    if r["loss"] == "class_balanced": t += "-cb"
    if r["loss"] == "image_balanced": t += "-bal"
    if r["jitter"]: t += "-jit"
    return t
by_cfg = {}
for n, r in runs.items(): by_cfg.setdefault(tag(r), {})[r["seed"]] = n

# ------------------------------ per run -------------------------------
print("=" * 104)
print(f"per run: Q0 at the selected checkpoint vs averaged over steps >= {PLATEAU}")
print("=" * 104)
print(f"{'run':22}{'sel step':>9}{'n':>4}{'sel Q0':>9}{'plateau Q0':>12}"
      f"{'its sd':>9}{'sel pooled':>12}{'pl pooled':>11}{'sel Q3':>9}{'pl Q3':>9}")
for cfg in sorted(by_cfg):
    for sd_ in sorted(by_cfg[cfg]):
        r = runs[by_cfg[cfg][sd_]]
        print(f"{by_cfg[cfg][sd_]:22}{r['sel_step']:>9}{r['n_plateau']:>4}"
              f"{r['sel_q0']:>9.4f}{r['pl_q0']:>12.4f}{r['pl_q0_sd']:>9.4f}"
              f"{r['sel_pooled']:>12.4f}{r['pl_pooled']:>11.4f}"
              f"{r['sel_q3']:>9.4f}{r['pl_q3']:>9.4f}")

# ---------------- how much noise does selection add? ------------------
print("\n" + "=" * 104)
print("seed-to-seed spread WITHIN a config (this is the noise any effect must beat)")
print("=" * 104)
print(f"{'config':14}{'seeds':>8}" + "".join(f"{h:>16}" for h in
      ("sel Q0 sd", "plateau Q0 sd", "sel Q0 range", "plateau range")))
for cfg in sorted(by_cfg):
    sds = sorted(by_cfg[cfg])
    if len(sds) < 2:
        print(f"{cfg:14}{len(sds):>8}{'(need 2+)':>16}"); continue
    s = np.array([runs[by_cfg[cfg][k]]["sel_q0"] for k in sds])
    p = np.array([runs[by_cfg[cfg][k]]["pl_q0"] for k in sds])
    print(f"{cfg:14}{len(sds):>8}{s.std(ddof=1):>16.4f}{p.std(ddof=1):>16.4f}"
          f"{s.max()-s.min():>16.4f}{p.max()-p.min():>16.4f}")
print("   if plateau sd is much smaller than selected sd, the selected-checkpoint")
print("   comparison was dominated by WHICH STEP won, not by the seed or the method.")

# -------------- the gate, paired within seed, both ways ---------------
print("\n" + "=" * 104)
print(f"paired WITHIN SEED against {BASE_CFG}   (gate: mean >= {GATE_Q0:+.2f},"
      f" all seeds positive, t-interval excludes 0)")
print("=" * 104)
if BASE_CFG not in by_cfg:
    raise SystemExit(f"baseline {BASE_CFG} not found among {sorted(by_cfg)}")
base = by_cfg[BASE_CFG]
out = {}
for cfg in sorted(by_cfg):
    if cfg == BASE_CFG: continue
    shared  = sorted(set(by_cfg[cfg]) & set(base))
    missing = sorted(set(by_cfg[cfg]) - set(base))
    print(f"\n{cfg}   paired seeds {shared}"
          + (f"   UNPAIRED (no baseline seed): {missing}" if missing else ""))
    if not shared:
        print("   nothing to pair"); continue
    row = {}
    for est, kq in (("selected checkpoint", "sel_q0"), (f"plateau mean", "pl_q0")):
        d = [runs[by_cfg[cfg][k]][kq] - runs[base[k]][kq] for k in shared]
        m = float(np.mean(d))
        line = f"   {est:20} per seed " + " ".join(f"{v:+.4f}" for v in d) + f"   mean {m:+.4f}"
        if len(d) >= 3:
            s = float(np.std(d, ddof=1)); n = len(d)
            half = TCRIT.get(n-1, 2.5)*s/sqrt(n)
            ok = (m >= GATE_Q0) and all(v > 0 for v in d) and (m-half > 0)
            line += (f"   sd {s:.4f}   95% CI [{m-half:+.4f}, {m+half:+.4f}]"
                     f"   ->  {'PASS' if ok else 'FAIL'}")
            row[est] = dict(per_seed=d, mean=m, sd=s, lo=m-half, hi=m+half, pass_=ok)
        else:
            line += f"   ({len(d)} seed(s): no gate, need 3 paired)"
            row[est] = dict(per_seed=d, mean=m)
        print(line)
    out[cfg] = dict(paired_seeds=shared, unpaired=missing, **{k.replace(" ", "_"): v
                                                              for k, v in row.items()})

json.dump(dict(plateau_from=PLATEAU, gate_q0=GATE_Q0, results=out,
               runs={k: v for k, v in runs.items()}),
          open(f"{SAVE}/stage2_plateau.json", "w"), indent=1, default=float)
print(f"\nwrote {SAVE}/stage2_plateau.json")
print("\nthe selected-checkpoint row is the PRIMARY, predeclared estimator.")
print("the plateau row is a secondary estimator declared after seeing that")
print("selection step varies from 3000 to 15000 across runs. report both.")
```
